## 第 1 章 仮定したことと、示したこと ── 見分け方

シミュレーションの結論には二種類ある。模型の部品を組み合わせた結果として**出てきた**もの(示したこと)と、結論の向きに働く部品を、初めから式に**書いておいた**もの(仮定したこと)である。後者が悪いわけではない。経済の模型は、どこかで仮定を置かなければ動かない。問題は、両者を区別しないまま「シミュレーションで示した」と言うことである。作者自身、gj_6 の中で「高分配にすれば、そうなるように予めプログラムしているのであり、こうなるのはあたり前である」と書いて、この区別に自分で気づいている。

このノートでは、区別のために三つの手を使う。

### 1.1 三つの手

**(1) 帳簿を閉じる**。すべての支払いに、払う側と受け取る側を書く。支出の合計と所得の合計が一致しない模型では、「需要が増えた」と言ったとき、そのお金がどこから来たかが言えない。帳簿を閉じると、仮定を一つも足さずに出てくる関係がある(第 2 章の「資本家の所得 = 赤字 ÷ 漏れの率」)。

**(2) 結論を出している式を、中立なものに替える**(切り離しの実験)。結論の向きに働く項を外し、それでも結論が残るかを見る。外すと消えるなら、その結論は、その項の仮定である。ただし、項を外すだけだと模型が壊れることがある。そのときは、項の役割を、もっと素直な部品(情報、外部性、帳簿)で置き換えた模型を作り直す(第 2 章、第 3 章)。

**(3) 結論が臨界の近くにあるかを見る**(感度の地図)。結論がパラメータの値に強く依存する(少し動かすと逆になる)なら、その結論は「示した」とは言えず、「そのパラメータのとき」の話である。逆に、臨界の点(相転移、二つの釣り合いの境目)そのものが、結論になることもある(第 3 章の崖、第 4 章の ρ ≈ 1/N)。

この三つに加えて、**期待値の計算**も使う。乱数を使うシミュレーションの結果が、乱数を期待値に置き換えた紙の上の計算で再現できるなら、その結論は、乱数の相互作用から出たのではなく、期待値の式から出ている(第 3 章)。

### 1.2 これまでのノートで、この手がどう使われたか

- **ノート 1**(micro_economy の解剖)の第 7 章は、帳簿を確かめ(7.1 検算)、三つの誤りを見つけた。同じノートの第 8 章では、最適化が何を満たすべきだったか(労働の需給と利潤率の均等化)を数えた。
- **ノート O**(外作用の模型)の第 4.3 節は、2011 年の模型に帳簿を付け、同じ帳簿でも物差しが違えば評価が違うことを見た。
- **ノート 3**(simbd の経済): ノート 3 の第 1.6 節は、simbd の帳簿が閉じていないこと(支払いが誰の所得にもならない)を見つけ、同じノートの第 8 章でそれを閉じた(地主の払う賃金を雇い人の所得にするだけで、Gini が 0.62 から 0.45 になった)。ノート 3 の第 4 章は、一家離散の条件を帳簿の式から導き、作者のロスカットの値 27.5 が臨界の値 22 のすぐ上にあることを見た(感度の地図)。

### 2.1 元のノートの主張

必需品と贅沢品の二財、富裕層(株主 10 人)と大衆(労働者 90 人)の二階級からなるマクロ経済で、技術と文化への投資、金利(政策金利 + 銀行プレミアム)、公共事業、再分配を動かして、四つのアイデアを確かめている。

1. **分配の負の効果**: 分配すると必需品の需要が増え、供給が限られたところで労働が逼迫して、物価が上がる。まるで金持ちが物価を操作したように見える。技術への投資で抑えようとしても、抑えられなかった。
2. **文化と技術の対立**: 文化が高いと需要が多く、技術だけでは需要は生まれない。選択の余地(文化)があるところに株式の富が生じる。贅沢品の割合は、与えたパラメータでほぼ決まる。
3. **金利と物価**: 文化(選択の余地)があれば銀行プレミアムが抑えられる。金利を上げると、プレミアムが費用として物価を押し上げる(利上げが物価高を招く)。文化は在庫・遊びとして物価の緩衝になる。
4. **債券と公共事業**: 債券は「弱い株式」で、公共事業は強制的な必需品の支出として、金利のショックをやわらげる。

作者は、模型の性格を正直に書いている(最初のセル)。

そして、1 番目の結果について(10 番目のセル)「高分配による必需品需要増は、consume メソッドの中でほぼハードコーディングされている。高分配にすれば、そうなるように予めプログラムしているのであり、こうなるのはあたり前である」と書き、最後のセルで「私自身はあまり成功したシミュレーションでなかったな…と述懐する」と結んでいる。この章の仕事は、(1) どの結論がどの式に書き込まれていたかを一覧にし、(2) 帳簿を閉じた小さな模型で、書き込まずに出てくるものと出てこないものを分けることである。

### 2.2 結論は、どの式から来るか

模型の本体は 5 番目のセル(`#esgj06`)にある。主張ごとに、それを出している行を並べる。

| 主張 | 行(5 番目のセル) | 何が書き込まれているか |
|---|---|---|
| 分配で必需品の需要が増える | `dynamic_req_unit = base_req_unit * (1.0 + 0.8 * np.tanh(income_ratio * 2.0))`、`income_ratio = self.redistribution_rate` | 大衆の必需品の**必要量**そのものが、再分配の率の関数になっている |
| 文化でプレミアムが下がり、技術で上がる | `self.bank_premium = (bank_premium_base * (innovation_friction / (friction_reduction + 0.1))) / public_stability`、`friction_reduction = np.log1p(self.culture_level)`、`innovation_friction = np.log1p(self.tech_level)` | プレミアムが文化の水準と技術の水準の比で決まる。コメントは「技術革新のスピード(tech_level の増加分)」だが、式は増加分ではなく**水準**を使っている |
| 公共事業でプレミアムが下がる | `public_stability = 1.0 + self.public_work_rate * 2.0` | 公共事業の率で、プレミアムを直接割っている |
| 金利が物価に転嫁される | `effective_price_basic = self.price_basic * (1 + effective_price_magnifier * effective_rate)` | 消費者の払う価格に、金利を掛けている |
| 文化が物価の緩衝になる | `adj` の中の `sens = 0.25 / (np.log1p(buf) + 0.1)`(`buf` は文化の水準) | 価格の動きやすさを、文化の水準で割っている |
| 文化が株式の価値を生む | `target_stock = (self.tech_level * 50 + self.culture_level * 50) * (1 + np.log1p(self.culture_level) * 1.0) * ...` | 株式の目標の価値に、文化の水準を掛けている |
| 技術偏重は脅威 | `tech_threat = max(0, ((self.tech_bias + self.public_work_rate) - 0.5) * 7.0)` | 脅威は、技術の水準ではなく、投資の**配分のパラメータ**で決まる |
| 幸福度が労働に効く | `labor_efficiency = 1.0 + 0.3 * np.tanh(self.mass.happiness / 50.0)` | 幸福度で労働力が最大 3 割増える |

つまり、四つの主張は、どれも、その主張の向きに働く項として式に入っている。シミュレーションは、書き込まれた向きを、相互作用で少しぼかして見せている。

**帳簿が閉じていないところ**。主張とは別に、お金の出入りが追えない箇所がある。

- 所得は、産出の価値の 55% が大衆に、45% が富裕層に、固定で配られる(`self.mass.income = total_output_val * 0.55`)。支出がいくらであっても、それが企業の売上として戻ることはない。支出と産出の差のお金は、どこからも来ず、どこへも行かない。
- 必需品の生産は、需要ではなく、固定の目標 1000(と公共事業)に合わせる(`target_basic_adj = (1000.0 * (1.0 - ...)) + public_demand`)。需要の変化は、価格の調整(`adj`)を通してしか生産に届かない。
- 公共事業は国債で賄うことになっているが、記録されるのは支出の 5%(`self.bond_market_val += public_demand * 0.05`)で、国債の利子は誰にも払われない。株式と債券の価値は、別の式(`target_stock`、`target_bond`)で決まり、誰かの貯蓄の積み重ねではない。
- 富裕層は、所得に加えて資産の 1% を使う(`self.rich.income + (self.rich.asset_stock + self.rich.asset_bond)*0.01`)が、資産は減らない。

ノート 1 の第 7 章で micro_economy の帳簿を確かめたときと同じ問題である。帳簿が閉じていないと、「分配すると需要が増える」の**需要のお金**がどこから来たのかが言えない。

### 2.3 批判のまとめ

- 主な結論は、どれも、結論の向きに働く項として書き込まれている(作者自身の診断どおり)。
- 帳簿が閉じていないので、書き込まれた項を外したときに何が残るかを、この模型の中では確かめられない。
- それでも、作者のアイデアの多くは、経済学の筋の通った考えに対応している。分配と需要(カレツキ、ケインズ)、労働が必需品から贅沢品へ移る構造の変化(パシネッティ)、選択肢の数を好む「多様性への愛好」(ディキシットとスティグリッツ)、金利が費用として価格に乗る「費用の経路」(Barth and Ramey 2001)、などである。だから、正しいやり方は、**帳簿を閉じた小さな模型を作り、作者の四つの主張のうち、どれが書き込まずに出てくるかを見る**ことである。

### 2.4 こうすれば良かった ── 帳簿を閉じた二財の模型

部品は少なくする。

- **賃金を 1(物差し)にする**。働くのは労働者で、労働力は 1。
- **必需品**は競争的に作られ、価格は費用 (1 + i_c)/a(a は技術、i_c は運転資金の金利。賃金を先に払うために借り、利子は資本家に入る)。短期には、必需品の部門で働ける人に上限があり(生産能力)、需要がそれを超えると価格が上がって、差額は持ち主(資本家)の**レント**になる。
- **贅沢品**は種類ごとに独占的に競争し、価格は費用の 1/ρ 倍(値上げ幅は種類の間の代替のしやすさで決まる。ディキシット–スティグリッツ)。差額が**利潤**で、株式の価値の源になる。種類の数 n が**文化**で、同じ量の贅沢品でも、種類が多いほど満足が大きい(多様性への愛好)。
- **家計の需要**は、まず必需品の必要量を買い、残り(余裕)の一定の割合を必需品、残りを贅沢品に使う(ストーン–ギアリーの型)。労働者は所得をすべて使う。資本家は所得の割合 s を貯め、そのうち割合 ι で人を雇って研究と文化に投資し、残りを国債や現金で持つ(**漏れ**)。
- **政府**は、資本家の所得に税率 τ をかけて労働者に配り(再分配)、公共事業 G で人を雇い、国債で賄う。国債の利子は資本家に入る。
- **雇用は需要で決まる**(人手が余れば失業、足りなければ物価が上がる)。

すべての支払いが誰かの所得になるので、帳簿が閉じる。Lean の `gj_books` は、この定義から、「資本家の貯蓄から投資を引いた残り = 政府の赤字」(部門の収支の和が 0)を導く。さらに `gj_capitalist_income` は、資本家の税引き後の所得が **赤字 ÷ 漏れの率** に決まることを示す(カレツキの利潤の式の形。「資本家は自分が使った分を稼ぐ」)。

````python
%%writefile -a misc_models.lean
/-! ## 第 2 章 必需品と贅沢品: 帳簿を閉じる

賃金を 1 とする。必需品の売上 EN は、賃金 LN と運転資金の利子 ic·LN とレント rent に分かれる。
贅沢品の売上 EL も同じく LL、ic·LL、利潤 profit に分かれる。働く人の数 L は LN + LL + I + G。
労働者の所得は L + T(T は再分配)、資本家の所得は rent + profit + 利子 − T。
労働者は所得をすべて使い、消費の合計は EN + EL = 労働者の所得 + 資本家の消費 Ck。
このとき、資本家の貯蓄から投資を引いた残りは、ちょうど政府の赤字 G + iB·B になる(部門の収支の和は 0)。 -/

theorem gj_books {EN EL LN LL I G ic iB B rent profit T L Mw Mk Ck : ℝ}
    (hEN : EN = (1 + ic) * LN + rent) (hEL : EL = (1 + ic) * LL + profit)
    (hL : L = LN + LL + I + G) (hMw : Mw = L + T)
    (hMk : Mk = rent + profit + ic * (LN + LL) + iB * B - T)
    (hC : EN + EL = Mw + Ck) :
    Mk - Ck - I = G + iB * B := by
  subst hEN hEL hMw hMk hL
  linarith

/-- 資本家が所得の割合 s を貯め、その割合 ι を投資に回すなら、資本家の(税引き後の)所得は
赤字 ÷ 漏れの率 (1 − ι) s に決まる(カレツキの利潤の式の形)。再分配の率 τ はこの式に出てこない。 -/
theorem gj_capitalist_income {Mk s ι D : ℝ} (h : 0 < (1 - ι) * s)
    (hbal : Mk - (1 - s) * Mk - ι * s * Mk = D) :
    Mk = D / ((1 - ι) * s) := by
  rw [eq_div_iff h.ne']
  linarith

/-- ゴーマンの集計: 二つの家計の、必要量を超えた所得のうち必需品に回す割合 b が同じなら、
二人の間で t だけ所得を移しても、必需品への支出の合計は変わらない。 -/
theorem gj_gorman (p g1 g2 b M1 M2 t : ℝ) :
    (p * g1 + b * (M1 - t - p * g1)) + (p * g2 + b * (M2 + t - p * g2))
      = (p * g1 + b * (M1 - p * g1)) + (p * g2 + b * (M2 - p * g2)) := by
  ring

/-- 割合が違えば(払う側 b1、受け取る側 b2)、必需品への支出の合計は (b2 − b1) t だけ変わる。
再分配で必需品の需要が増えるのは、受け取る側の割合が大きいときだけである。 -/
theorem gj_engel_shift (p g1 g2 b1 b2 M1 M2 t : ℝ) :
    ((p * g1 + b1 * (M1 - t - p * g1)) + (p * g2 + b2 * (M2 + t - p * g2)))
      - ((p * g1 + b1 * (M1 - p * g1)) + (p * g2 + b2 * (M2 - p * g2))) = (b2 - b1) * t := by
  ring
````

````python
%%writefile gj.py
# gj.py: 第 2 章 必需品と贅沢品のマクロ経済を、帳簿を閉じて書き直した小さな模型
# 元のノート(gj_6)のプログラムは使わない。ここで新しく書く。
# 賃金を 1(物差し)とし、一期ごとに、支出 = 所得 となる釣り合いを求める。
from dataclasses import dataclass, replace, asdict
import numpy as np


@dataclass
class GJ:
    """模型のパラメータ。労働者(大衆)は人口の Hw、資本家(富裕層)は Hk。働くのは労働者だけで、労働力は Lmax。"""
    Hw: float = 0.9          # 労働者の人口
    Hk: float = 0.1          # 資本家の人口
    gw: float = 0.3          # 労働者一人あたりの必需品の必要量
    gk: float = 0.5          # 資本家一人あたりの必需品の必要量
    bw: float = 0.3          # 労働者が、必要量を超えた所得のうち必需品に回す割合(限界の割合)
    bk: float = 0.1          # 資本家の同じ割合
    s: float = 0.5           # 資本家の貯蓄の割合
    iota: float = 0.5        # 貯蓄のうち、研究と文化への投資(人を雇う)に回す割合。残りは国債や現金で持つ(漏れ)
    psi: float = 0.0         # 文化(贅沢品の種類の数 n)が貯蓄の割合を下げる強さ: s / (1 + ψ n)
    rho: float = 0.75        # 贅沢品の種類の間の代替のしやすさ。値上げ幅(マークアップ)は 1/ρ
    aL: float = 1.0          # 贅沢品の生産性
    tau: float = 0.2         # 再分配: 資本家の所得への税率。全額を労働者に配る
    G: float = 0.02          # 公共事業(人を雇う)。国債で賄う
    ic: float = 0.03         # 企業の運転資金の金利(賃金を先に払うために借りる。利子は資本家の所得)
    iB: float = 0.02         # 国債の金利(利子は資本家の所得)
    Lmax: float = 1.0        # 労働力


def stone_geary(M, need, b, pN):
    """所得 M の家計の支出。まず必需品の必要量 need を買い、残り(余裕)の割合 b を必需品、1 − b を贅沢品に使う。
    所得が必要量に届かなければ、すべてを必需品に使う。戻り値は (必需品への支出, 贅沢品への支出)。"""
    if M <= pN * need:
        return M, 0.0
    extra = M - pN * need
    return pN * need + b * extra, (1 - b) * extra
````

````python
%%writefile -a gj.py
def period(p, a, n=1.0, Lcap=np.inf, B=1.0, tol=1e-12):
    """一期の釣り合い。a は必需品の生産性(技術)、n は贅沢品の種類の数(文化)、Lcap は必需品の部門で
    すぐに働ける人の上限(短期の生産能力)、B は国債の残高。
    必需品は競争的で、価格は費用 (1 + ic)/a。需要が生産能力を超えると価格が上がり、差額は持ち主(資本家)のレントになる。
    贅沢品は種類ごとに独占的に競争し、価格は費用の 1/ρ 倍。働く人が足りなければ、価格がさらに上がる。
    すべての支払いは誰かの所得になる: 賃金は労働者に、レント・利潤・利子は資本家に。"""
    cN = (1 + p.ic) / a
    cL = (1 + p.ic) / (p.rho * p.aL)
    s = p.s / (1 + p.psi * n)
    pN, pL, Mw, Mk = cN, cL, 0.5, 0.2
    for it in range(20000):
        ENw, ELw = stone_geary(Mw, p.Hw * p.gw, p.bw, pN)
        Ck = (1 - s) * Mk
        ENk, ELk = stone_geary(Ck, p.Hk * p.gk, p.bk, pN)
        I = p.iota * s * Mk
        EN, EL = ENw + ENk, ELw + ELk
        LN = min(EN / pN / a, Lcap)
        pN_new = max(cN, EN / (a * Lcap)) if np.isfinite(Lcap) else cN
        free = max(p.Lmax - LN - I - p.G, 1e-9)
        LL = min(EL / pL / p.aL, free)
        pL_new = max(cL, EL / (p.aL * free))
        L = LN + LL + I + p.G
        rent = EN - (1 + p.ic) * LN          # 必需品の売上から、賃金と運転資金の利子を引いた残り
        profit = EL - (1 + p.ic) * LL        # 贅沢品の売上から、賃金と利子を引いた残り
        interest = p.ic * (LN + LL) + p.iB * B
        Mk_pre = rent + profit + interest
        T = p.tau * Mk_pre
        Mw_new, Mk_new = L + T, Mk_pre - T
        err = abs(Mw_new - Mw) + abs(Mk_new - Mk) + abs(pN_new - pN) + abs(pL_new - pL)
        d = 0.3
        Mw += d * (Mw_new - Mw); Mk += d * (Mk_new - Mk)
        pN += d * (pN_new - pN); pL += d * (pL_new - pL)
        if err < tol:
            break
    # 部門の収支: 労働者は貯めない。資本家の貯蓄の残り(投資に回さない分)と、政府の赤字は一致するはず
    Sk = Mk - Ck - I
    deficit = p.G + p.iB * B
    welfare_lux = (EL / pL) * n ** ((1 - p.rho) / p.rho)   # 種類の数が多いほど、同じ量の贅沢品でも満足が大きい(多様性への愛好)
    return dict(pN=pN, cN=cN, pL=pL, cL=cL, EN=EN, EL=EL, LN=LN, LL=LL, I=I, L=L, unemp=p.Lmax - L,
                rent=rent, profit=profit, interest=interest, Mw=Mw, Mk=Mk, T=T, Ck=Ck, Sk=Sk, deficit=deficit,
                books=Sk - deficit, QN=EN / pN, lux_index=welfare_lux, it=it)


def table(rows, keys, fmt='{:8.3f}'):
    print(' ' * 9 + ''.join(f'{k:>10}' for k in keys))
    for label, r in rows:
        print(f'{label:>9}' + ''.join(fmt.format(r[k]).rjust(10) for k in keys))
````

````python
import gj
from dataclasses import replace
p0 = gj.GJ()
base = gj.period(p0, a=1.5)
print('基準の釣り合い(a = 1.5、n = 1)')
for k in ['pN', 'pL', 'LN', 'LL', 'I', 'L', 'unemp', 'profit', 'interest', 'Mw', 'Mk']:
    print(f'  {k:>8} = {base[k]:.4f}')
print(f'  帳簿: 資本家の貯蓄の残り {base["Sk"]:.6f} − 政府の赤字 {base["deficit"]:.6f} = {base["books"]:.1e}')
print(f'  資本家の所得 = 赤字 / ((1 − ι) s) = {base["deficit"] / ((1 - p0.iota) * p0.s):.4f}')
````

#### 主張 1: 分配の負の効果

再分配の率 τ を 0 から 0.6 まで上げる。長期(必需品の生産能力が需要に合わせて動く)と、短期(生産能力を基準の値で止める)を比べる。さらに、資本家と労働者の「余裕のうち必需品に回す割合」を同じにした場合(0.3 と 0.3)も見る。

````python
K = ['L', 'unemp', 'EN', 'QN', 'pN', 'pL', 'rent', 'Mw', 'Mk']
print('長期(生産能力は需要に合わせる)')
gj.table([(f'τ={t}', gj.period(replace(p0, tau=t), 1.5)) for t in [0.0, 0.2, 0.4, 0.6]], K)
print('短期(必需品の生産能力を基準の値で止める)')
gj.table([(f'τ={t}', gj.period(replace(p0, tau=t), 1.5, Lcap=base['LN'])) for t in [0.0, 0.2, 0.4, 0.6]], K)
print('割合が同じ(資本家も 0.3)、長期')
gj.table([(f'τ={t}', gj.period(replace(p0, tau=t, bk=0.3), 1.5)) for t in [0.0, 0.2, 0.4]], K)
````

**読み方**

- **再分配しても、資本家の所得 Mk は 0.16 のまま変わらない**。資本家の所得は、赤字 ÷ 漏れの率で決まるからである(Lean の `gj_capitalist_income`)。再分配で労働者の所得が増えると、それはすべて使われて誰かの売上になり、最後は資本家の利潤として戻ってくる。再分配で減るのは**失業**である(τ = 0 で 2 割、0.2 で 4%、0.4 で 0)。
- 完全雇用に届いたあと(τ ≥ 0.4)、さらに分配すると、人手が足りず、**贅沢品の価格が上がる**(1.37 → 1.88)。
- 必需品の生産能力を止めた短期では、**必需品の価格が上がり**(0.69 → 0.85)、差額は持ち主のレントとして資本家に入る(0.11)。作者の「分配すると、まるで金持ちが物価を操作しているように見える」は、帳簿を閉じた模型でも**出てくる**。ただし、操作ではなく、短期の生産能力の不足が作るレントである。生産能力が追いつけば(長期)、必需品の価格は戻る。
- 必需品の需要が増える理由は、二つに分けられる。一つは**全体の所得が増えること**(失業が減る。上の表)。もう一つは、**受け取る側(労働者)が、余裕のうち必需品に回す割合が大きいこと**で、割合が同じなら、この効果は消える(Lean の `gj_gorman`。違えば、変化は割合の差 × 移した額 `gj_engel_shift`)。元の模型は、この二つのどちらでもなく、必要量そのものを再分配の率で上げていた。

#### 主張 2: 文化と技術

技術 a を上げる(必需品が少ない人手で作れる)。文化 n(贅沢品の種類の数)も変える。文化が資本家の貯蓄の割合を下げる強さ ψ を、0(文化は需要に効かない)と 0.2 で比べる。

````python
print('失業率(行: 技術 a、列: 文化 n)')
for psi in [0.0, 0.2]:
    print(f'ψ = {psi}')
    print('   a \\ n ' + ''.join(f'{n:7.1f}' for n in [1, 2, 4, 8]))
    for a in [1.5, 3.0, 5.0]:
        print(f'   {a:5.1f}  ' + ''.join(f'{gj.period(replace(p0, psi=psi), a, n=n)["unemp"]:7.3f}' for n in [1, 2, 4, 8]))
print()
K2 = ['unemp', 'LN', 'LL', 'pN', 'pL', 'profit', 'Mk', 'lux_index']
print('ψ = 0.2 のとき')
gj.table([(f'a{a},n{n}', gj.period(replace(p0, psi=0.2), a, n=n)) for a, n in [(1.5, 1), (5, 1), (5, 4), (5, 8)]], K2)
````

**読み方**

- **技術を上げると、失業が増える**(ψ = 0 で、a = 1.5 → 5 で 4% → 18%)。必需品の価格は下がり、必需品に要る人手は減るが、浮いた人手を吸う新しい需要が自動的には生まれない。これはパシネッティの「構造の変化」の条件(生産性が上がったら、新しい需要が浮いた労働を吸わなければ失業になる)そのものである。作者の「技術はそのままでは需要を生まない」は、帳簿を閉じた模型でも**出てくる**。
- **文化(種類の数)は、ψ = 0 なら、失業にも物価にもまったく効かない**(表の行が n によらない)。効くのは満足(`lux_index`、多様性への愛好)だけである。種類が増えても、使うお金の総額は増えないからである。
- 文化が需要を生むのは、**種類が多いと、資本家が貯め込むより使うようになる**(ψ > 0)と仮定したときだけである。ψ = 0.2 なら、文化が増えるほど失業が減り、利潤(株式の価値の源)と資本家の所得が増える。作者の「選択の余地のあるところに株式の富が生じる」は、この仮定(買いたいものがあれば、貯め込まない)を一つ置けば出てくる。逆に、この仮定は、どこかに書いておかなければならない。元の模型では、株式の価値の式に文化の水準を掛けることで、この仮定を見えない形で置いていた。
- 利潤は、贅沢品(独占的な競争)からしか生まれない。必需品は競争的で、短期のレントを除けば利潤がない。作者の 2009 年の見方「債権では社会の富は増えない。株式のみ社会の富を増やせる」は、この模型では「**値上げ幅のある差別化された財だけが、持続する利潤を生む**」と言い換えられる。

#### 主張 3: 金利と物価

二つの金利を別々に上げる。運転資金の金利 i_c(企業の費用になる)と、国債の金利 i_B(国債の持ち主の所得になる)である。国債の残高は 3(所得の 3 年分ほど)にする。

````python
K3 = ['unemp', 'pN', 'pL', 'Mw', 'Mk']
print('運転資金の金利 i_c(費用の経路)')
gj.table([(f'ic={i}', gj.period(replace(p0, ic=i), 1.5)) for i in [0.0, 0.03, 0.06, 0.10]], K3)
print('国債の金利 i_B(利子の所得の経路、国債の残高 B = 3)')
gj.table([(f'iB={i}', gj.period(replace(p0, iB=i), 1.5, B=3.0)) for i in [0.0, 0.01, 0.02, 0.04]], K3)
````

**読み方**

- **費用の経路**: 運転資金の金利を上げると、必需品の価格は費用の分だけ上がり(0.67 → 0.73)、同時に**失業が増える**(0 → 26%)。資本家の所得は赤字で決まっているので、売上のうち利子に取られる分が増えると、同じ資本家の所得を出すのに要る生産が減るからである。物価が上がって景気が悪くなる(スタグフレーションの型)。これは、実証で「利上げの直後に物価がかえって上がる」(価格の謎、price puzzle)の説明の一つとして知られる費用の経路(Barth and Ramey 2001)である。ただし、上がるのは物価の**水準**で、上がり続ける(インフレ)わけではない。
- **利子の所得の経路**: 国債の残高が大きいと、国債の金利を上げることは、政府が国債の持ち主にお金を配ることになる。赤字が増え、資本家の所得が増え、需要が増える。完全雇用に届くと、物価が上がる(i_B = 0.02 → 0.04 で、贅沢品の価格が 1.79 → 2.74)。**国債の多い経済では、利上げが需要を増やし、物価を押し上げることがある**。作者と Gemini さんの「物価抑制のための利上げが、さらなる物価高を招く」は、銀行プレミアムの経路ではなく、この経路で、帳簿を閉じた模型からも出てくる(どれだけ強いかは、国債の持ち主がその利子をどれだけ使うかで決まる)。

#### 主張 3 の続き: 文化は物価の緩衝になるか

必需品の生産性が一期だけ 3 割落ちる(凶作)。生産能力(働ける人)はすぐには増やせない。このときの必需品の価格の跳ね方を、必需品の必要量(余裕の大きさ)と、文化(種類の数)を変えて比べる。

````python
print(' 必要量 gw | 文化 n | 平時の価格 | 凶作の価格 | 跳ね方 | 平時に贅沢品に回る人手の割合')
for gw in [0.2, 0.3, 0.45]:
    for n in [1, 4]:
        q = replace(p0, gw=gw)
        r0 = gj.period(q, 1.5, n=n)
        r1 = gj.period(q, 1.05, n=n, Lcap=r0['LN'])
        print(f'   {gw:4.2f}    |  {n}  |   {r0["pN"]:.3f}   |   {r1["pN"]:.3f}   | ×{r1["pN"] / r0["pN"]:.3f} |   {r0["LL"] / r0["L"]:.3f}')
````

**読み方**

- 凶作のときの価格の跳ね方は、**文化(種類の数)によらない**(n = 1 と 4 で同じ)。
- 跳ね方を小さくするのは、**余裕の大きさ**である。必需品の必要量が小さく、所得の多くが贅沢品に回っている社会(gw = 0.2)では、凶作のとき、贅沢品を買うのをやめて必需品に回す余地があり、価格の跳ね方は小さい(×1.55)。必要量が大きい社会(gw = 0.45)では、×1.66 になる。
- だから、作者の「文化 = バッファ」は、**種類の数**ではなく、**必需品以外に回っている所得と人手の厚み**として読むのが正しい。これはノート I(必需品と贅沢品の宇宙的独立関係)の「贅沢品の厚みが、危機のときの必需品を支える」と同じ形である。種類の数が効くとすれば、危機のときに、贅沢品の部門の人手や設備を必需品に転用できる速さを通してで、それはこの模型にはない別の部品(転用の速さ)である。

#### 主張 4: 債券と公共事業

````python
gj.table([(f'G={g}', gj.period(replace(p0, G=g), 1.5)) for g in [0.0, 0.02, 0.04, 0.08]], ['unemp', 'pL', 'Mw', 'Mk', 'deficit'])
````

- **公共事業は、雇用を増やし(乗数)、資本家の所得も増やす**(G = 0 で失業 47%、0.02 で 4%。資本家の所得は 0.08 → 0.16)。赤字は、最後は資本家の貯蓄として国債の形で戻る(帳簿)。完全雇用を超えると、物価が上がる。
- 作者の「国債を根拠に公共事業した場合は公共事業の方での効果がある」「債券は弱い株式」は、この模型では「国債は、資本家の貯蓄の受け皿(漏れの行き先)で、公共事業は、その漏れを支出に戻す道」と言える。公共事業が「強制的な必需品支出」かどうか(何を作るか)は、雇用と所得には効かない。効くのは、支出の大きさと、誰の所得になるかである。

#### まとめ: 四つの主張のうち、何が残るか

| 主張 | 元の模型 | 帳簿を閉じた模型 |
|---|---|---|
| 分配で必需品の需要が増え、物価が上がる | 必要量を再分配の率で上げる(書き込み) | **残る**。ただし理由は、失業が減ることと、受け取る側の必需品の割合が大きいこと。物価が上がるのは短期の生産能力の不足(レント)か、完全雇用を超えたとき。資本家の所得は減らない |
| 技術だけでは需要は生まれない | 脅威の項(書き込み) | **残る**。浮いた人手を吸う需要がなければ失業になる(パシネッティ) |
| 文化が需要と株式の富を生む | 株式の価値に文化を掛ける(書き込み) | **仮定を一つ置けば残る**(種類が多いと貯め込まない: ψ > 0)。利潤は差別化された財からしか生まれない |
| 利上げが物価高を招く | 価格に金利を掛ける(書き込み) | **残る**。費用の経路(水準が上がり、景気は悪くなる)と、国債の利子の経路(国債が多いと需要が増える) |
| 文化が物価の緩衝になる | 価格の動きやすさを文化で割る(書き込み) | **形を変えて残る**。緩衝は種類の数ではなく、必需品以外に回る余裕の厚み |
| 公共事業がショックをやわらげる | プレミアムを公共事業で割る(書き込み) | **残る**。支出の乗数として。何を作るかは雇用には効かない |

作者のアイデアの多くは、正しい方向を向いていた。足りなかったのは、帳簿を閉じることと、それぞれの効果の**理由**を一つの部品に対応させることだった。

**注: この模型は、資本家に有利な模型に見えるかもしれない**。上の結果では、再分配しても資本家の所得は変わらず、公共事業はむしろ資本家の所得を増やし、政府の赤字は最後は資本家の貯蓄(国債)として戻る。これは、資本家の味方をする主張ではなく、模型の置いた二つの仮定から出る帳簿の帰結である。

1. **労働者は貯めない**(受け取った所得を全部使う)。カレツキの「労働者は得たものを使い、資本家は使ったものを得る」の形で、貯める部門が資本家しかないので、政府の赤字(と投資)の行き先は、資本家の貯蓄しかない。
2. **資本家の所得は、利潤(と国債の利子)だけ**で、利潤の取り分そのものは、需要の構成で決まる。

労働者も貯める(年金基金などを通じて国債を持つ。Pasinetti 1962)ようにすると、資本家の所得はもう赤字だけでは決まらず、再分配で減る(ノート 4 の第 2 章、一般のカレツキの式)。また、この模型で再分配が減らすのは失業で、その得は労働者に入る。どちらの階級に有利かは、模型の仮定の置き方の問題であり、gj の結論は「誰が貯めるか」という仮定とひとまとまりで読む必要がある。

### 2.5 ノート 4 にどう活かすか

ノート 4では、me_c(ノート 1 で作った、最適化で価格を決める経済の小さな版)と、v6(信用循環モデル)をつなぐ。この章の模型からは、次の三つを引き継ぐ。

### 3.1 元のノートの問いと結果

作者の問いは、最初のセルにこう書かれている。

- m = 0.5(良い環境)では非縁故主義が勝つ(30 期後の資金が 778 万 対 340 万。初めは 10 万)、
- m = 1.0 と 2.0 では縁故主義が勝つ(m = 1.0 で 8.1 万 対 2.3 万)、

で、作者は「投資環境がよほど良くなければ、縁故主義が有利であることが示されてしまった」と結んだ。その 2 では、閾値 q と環境 m の平面に 30 期後の成長の等高線を描き、同じ成長率の等高線に、山の左(q が小さい)と右(q が大きい)の二つの枝があることを見て、左を「安定志向(失われた数十年の日本)」、右を「途上国型」と読んだ。

### 3.2 元のノートの模型では、何が結果を決めていたか

作者自身が書くとおり(実験 1 の前の文)、「縁故主義だと完全に情報がわかり非縁故主義だと不完全にしかわからない…というのは私にはモデル化できないので、単純に縁故主義だとリスク許容度が小さく、非縁故主義だとリスク許容度が大きいというプログラムにして」いる。両方とも、事業のリスクを**正しく知っている**。違うのは閾値だけである。このとき、何が勝ち負けを決めるかは、紙の上で計算できる。

**(a) 投資の量は、どちらでも同じ**。投資家は事業をランダムな順に見て、手元に残った資金の (1 − r) の割合(閾値を超えるなら、その 2 割)を投じ、次の事業へ移る。200 の事業を見終わるころには、手元の資金はほぼ残らない。下の計算では、投資されずに残る資金は、どちらの規則でも 0 である(数値の誤差の範囲)。作者が見たかった「投資ボリューム」の違いは、この模型には**ない**。違うのは、どのリスクの事業にお金が回るか(構成)だけである。

**(b) 事業の数も戻りも、経済の大きさによらない**。作者の「経済も大きくなって機会も増え」にあたるしくみ(投資が誰かの売上になり、ほかの事業を成功しやすくする)はない。

**(c) だから、結果は、事業の期待の戻りの符号だけで決まる**。リスク r の事業に 1 を投じたときの期待の戻りは e(r) = (1 − r)^m (1 + 1.5 r) である。これが 1 を超える事業は得、下回る事業は損である。e(r) = 1 になるリスク r*(m)(損益の分かれ目)を計算すると、下のようになる。

````python
%%writefile nepo.py
# nepo.py: 第 3 章 縁故主義と非縁故主義の計算
# 元のノートのプログラムは使わない。元の規則の期待値(3.2)と、情報を戻した新しい模型(3.3〜)を、ここで書く。
import numpy as np
from scipy.optimize import brentq
from scipy.stats import norm


def expected_gross(r, m):
    """元のノートの事業: リスク r の事業は、確率 (1 − r)^m で成功し、成功すると投資額の 1 + 1.5 r 倍が戻る。
    期待の戻り(投資 1 あたり)は (1 − r)^m (1 + 1.5 r)。m は事業の環境(元の PROFIT_RISK_MAG)。"""
    return (1 - r) ** m * (1 + 1.5 * r)


def breakeven_risk(m):
    """期待の戻りがちょうど 1 になるリスク r*(m)。これより安全な事業は得、危ない事業は損。
    どの r でも 1 を下回るなら None。"""
    f = lambda r: expected_gross(r, m) - 1
    if f(1e-9) <= 0 or f(0.999999) >= 0:
        return None if f(1e-6) <= 0 else 1.0
    return brentq(f, 1e-9, 0.999999)
````

````python
%%writefile -a nepo.py
def rule_growth(q, m, n_business=200, n_investor=100, aversion=0.8, seed=0, reps=20):
    """元のノートの実験 1 の投資の規則の、一期あたりの資金の増え方の期待値(成功・失敗の乱数を期待値で置き換える)。
    投資家は事業をランダムな順に見て、手元の資金の (1 − r)(閾値 q を超えるリスクなら、さらに 1 − aversion を掛ける)
    の割合を投じる。事業のリスクの引き直しを reps 回して平均する。
    戻り値は (増え方の期待値, 投資されずに残る資金の割合)。"""
    rng = np.random.default_rng(seed)
    grow = left = 0.0
    for _ in range(reps):
        r = rng.random(n_business)
        f = np.where(r <= q, 1 - r, (1 - r) * (1 - aversion))
        e = expected_gross(r, m)
        for _ in range(n_investor):
            o = rng.permutation(n_business)
            rem = np.concatenate([[1.0], np.cumprod(1 - f[o])[:-1]])   # その事業を見る前に残っている資金
            w = f[o] * rem
            grow += np.sum(w * e[o])
            left += rem[-1] * (1 - f[o][-1])
    n = reps * n_investor
    return grow / n + left / n, left / n
````

````python
import nepo
print('損益の分かれ目のリスク r*(m)(これより安全な事業は得):')
for m in [0.5, 0.75, 1.0, 1.25, 1.5, 2.0]:
    rs = nepo.breakeven_risk(m)
    print(f'  m = {m:4.2f}: r* = ' + (f'{rs:.3f}' if rs else 'なし(どの事業も損)'))
print()
print('元の実験 1 の規則の、一期の資金の増え方の期待値(成功・失敗の乱数を期待値に置き換えたもの)')
print('  m    | 縁故 q=0.3 | 非縁故 q=0.7 | 投資されずに残る資金 | 30 期後(期待値の 30 乗、初め 10 万)')
for m in [0.5, 1.0, 2.0]:
    gn, ln_ = nepo.rule_growth(0.3, m)
    gm, lm = nepo.rule_growth(0.7, m)
    print(f'  {m:3.1f}  |  {gn:.4f}   |   {gm:.4f}    |   {ln_:.1e} / {lm:.1e}   |  {1e5 * gn ** 30:12,.0f} 対 {1e5 * gm ** 30:12,.0f}')

qs = np.linspace(0.05, 0.95, 19)
fig, ax = plt.subplots(figsize=(6.5, 3.8))
for i, m in enumerate([0.5, 0.75, 1.0]):
    g = [nepo.rule_growth(q, m, reps=8)[0] for q in qs]
    ax.plot(qs, g, color=PAL[i], label=f'm = {m}')
    rs = nepo.breakeven_risk(m)
    ax.axvline(rs, color=PAL[i], ls=':', lw=1)
ax.axhline(1, color=INK2, lw=0.8)
ax.set_xlabel('threshold q (SAFE_THRESHOLD)'); ax.set_ylabel('expected growth per period')
ax.set_title('the original rule: growth peaks near the break-even risk r*(m) (dotted)')
ax.legend(fontsize=8); plt.tight_layout(); plt.show()
````

**読み方**

- 期待値の計算(乱数なし)で、元のノートの結果がそのまま出る。m = 0.5 で一期 1.12 対 1.16、m = 1.0 で 0.99 対 0.95、m = 2.0 で 0.80 対 0.69 である(元のノートの 30 期後の平均の資金を一期あたりに直すと、1.125 対 1.156、0.993 対 0.952、0.805 対 0.66)。元の結果は、事業の期待の戻りの計算で説明しつくせる。
- **m = 1.0 で「縁故主義が勝った」とき、どちらの経済も縮んでいる**(初め 10 万が 8 万と 2 万)。m ≥ 1.5 では、どのリスクの事業も損で、投資しないのが一番である。投資をやめられない規則なので、損な事業に投じる量が少ない(閾値の低い)ほうが「勝つ」。
- m = 0.5 では損益の分かれ目が r* = 0.79 にあり、q = 0.7 の非縁故主義のほうが分かれ目に近い。m = 1.0 では r* = 1/3 で、q = 0.3 の縁故主義のほうが近い。**勝つのは、閾値が損益の分かれ目に近いほう**である。
- 図の山の頂(成長の最大)は、どの m でも、損益の分かれ目 r*(m)(点線)の近くにある。閾値をそこまで上げると、得な事業をすべて拾い、損な事業を避けられるからである(m = 0.5 で頂が少し左にあるのは、危ない事業ほど投じる割合 1 − r が小さく、閾値を上げた効果が薄れるため)。

**その 2 の「二つの枝」**。成長は、閾値 q について、損益の分かれ目のあたりに頂のある**単峰の山**になる。単峰の山を水平に切れば、切り口は必ず左右に二つある。m = 1 の場合を式で書くと、「リスク q 以下の事業に等しく投じる」ときの得は q²/4 − q³/2 で、q = 1/3 で最大値 1/108 をとる(Lean の `nepo_gain_pos_iff` と `nepo_ridge`)。だから、二つの枝は、二つの経済の型ではなく、**同じ山の、慎重すぎる側と大胆すぎる側**である。どちらも、頂(損益の分かれ目で線を引く)より悪い。左の枝を「日本」、右を「途上国」と読むには、山の外の理由が要る。

作者の直感の中身(「投資の量が増える」「経済が大きくなって機会が増える」)は、この模型には入っていなかった。それを入れるには、二つのものが要る。**情報**(縁故主義は何を知っているのか)と、**外部性**(投資が他人の機会を作る)である。

Lean で、m = 1 の損益の分かれ目と山の頂、そして「正しく知っている貸し手は損をしない」(情報の値打ち)を確かめておく。

````python
%%writefile -a misc_models.lean
/-! ## 第 3 章 縁故主義: 期待の戻り、尾根、情報の値打ち

元のノートの事業は、リスク r のとき確率 (1 − r)^m で成功し、成功すると 1 + 1.5 r 倍が戻る。
m = 1 のとき、期待の戻りから 1 を引くと r (1/2 − 3r/2) になり、r < 1/3 で正、r > 1/3 で負。 -/

theorem nepo_gain_m1 (r : ℝ) : (1 - r) * (1 + 3 / 2 * r) - 1 = r * (1 / 2 - 3 / 2 * r) := by
  ring

theorem nepo_gain_pos_iff {r : ℝ} (hr : 0 < r) :
    0 < (1 - r) * (1 + 3 / 2 * r) - 1 ↔ r < 1 / 3 := by
  rw [nepo_gain_m1]
  constructor
  · intro h
    by_contra hc
    rw [not_lt] at hc
    nlinarith
  · intro h
    apply mul_pos hr
    linarith

/-- 「リスク q 以下の事業に等しく投じる」ときの、一期の得(m = 1)は
∫₀^q r (1/2 − 3r/2) dr = q²/4 − q³/2。これは q = 1/3 で最大値 1/108 をとる単峰の山で、
同じ得になる q が、山の左右に二つずつある(元のノートその 2 の「二つの枝」)。ここでは山の頂の不等式を確かめる。 -/
theorem nepo_ridge {q : ℝ} (hq : 0 ≤ q) : q ^ 2 / 4 - q ^ 3 / 2 ≤ 1 / 108 := by
  nlinarith [sq_nonneg (q - 1 / 3), mul_nonneg hq (sq_nonneg (q - 1 / 3))]

/-- 情報の値打ち: 事業 i の得 e i − 1 を知っている貸し手は、得の正の事業だけを選べる。
どんな選び方 S の得も、それを超えない。だから、外部性がなければ、正しく知る貸し手(縁故)が、
網の目の中で損をすることはない。 -/
theorem info_value {ι : Type*} [DecidableEq ι] (U S : Finset ι) (hS : S ⊆ U) (e : ι → ℝ) :
    ∑ i ∈ S, (e i - 1) ≤ ∑ i ∈ U.filter (fun i => 1 < e i), (e i - 1) := by
  calc ∑ i ∈ S, (e i - 1) ≤ ∑ i ∈ S, max (e i - 1) 0 :=
        Finset.sum_le_sum fun i _ => le_max_left _ _
    _ ≤ ∑ i ∈ U, max (e i - 1) 0 :=
        Finset.sum_le_sum_of_subset_of_nonneg hS fun i _ _ => le_max_right _ _
    _ = ∑ i ∈ U.filter (fun i => 1 < e i), (e i - 1) := by
        rw [Finset.sum_filter]
        apply Finset.sum_congr rfl
        intro i _
        split_ifs with h
        · exact max_eq_left (by linarith)
        · exact max_eq_right (by linarith)
````

### 3.3 情報を戻す ── 関係融資と市場型融資

金融論には、作者の問いにほぼそのまま当たる比較がある。**関係融資**(relationship lending: 長い付き合いのある借り手に、帳簿に出ない情報を使って貸す。Rajan 1992、Petersen and Rajan 1994)と、**市場型融資**(arm's-length / transaction lending: 誰にでも、帳簿や格付けなどの公の情報だけで貸す)である。縁故主義は、極端な関係融資と読める。

模型を次のように作る。事業のリスク r と戻りは元のノートと同じにする。違いは、貸し手が何を知っているかである。

- **関係融資(縁故)**: 自分の網の目(親族・知り合い)の中の事業について、リスク r を**正しく知る**。網の目は事業全体の割合 ν だけで、その外には貸さない。
- **市場型融資(非縁故)**: すべての事業について、雑音のある信号 s = r + ε(ε は標準偏差 σ の正規分布)だけを見る。信号から期待の戻りを推し量る(ベイズの更新)。

どちらも、見込んだ期待の戻りが 1 + h 以上なら貸し、貸さないお金は現金で持つ(戻り 1)。h は**慎重さ**で、元のノートの「リスク許容度」はこれにあたる。こうすると、元のノートで一つにまとめられていた三つの効果が分かれる。

1. **見分ける効果**(情報): 関係融資は損な事業を避けられる。市場型融資は、雑音のせいで損な事業にも貸し、得な事業を逃す。
2. **広さの効果**(網の目): 関係融資は、網の目の外の得な事業を逃す。
3. **慎重さの効果**: h > 0 なら、得な事業の一部も逃す。

投資の量 x(貸したお金の割合)は、規則から決まる。元のノートと違い、投資の量が結果として変わる。

````python
%%writefile -a nepo.py
R = np.linspace(0.0005, 0.9995, 1000)     # 事業のリスクの格子(リスクは 0〜1 に一様に分布する)
DR = R[1] - R[0]


def demand_boost(x, kappa, x_ref=0.5):
    """投資の外部性: 経済全体の投資の割合 x が高いほど、どの事業も成功しやすい(投資は誰かの売上になる)。
    成功の確率に 1 + κ (x − x_ref) を掛ける。κ = 0 なら外部性なし。"""
    return 1 + kappa * (x - x_ref)


def _market_tables(m, sigma):
    """市場型の貸し手: 事業のリスク r の信号 s = r + ε(ε は標準偏差 σ の正規分布)を見る。
    信号ごとの期待の戻りの推定 E[e | s] と、リスクごとの信号の分布 P(s | r) を返す。"""
    S = np.linspace(-1, 2, 601)
    L = norm.pdf((S[:, None] - R[None, :]) / sigma)
    e = expected_gross(R, m)
    post = (L * e[None, :]).sum(1) / L.sum(1)
    ps = L / L.sum(0)[None, :]
    return post, ps


def lending(kind, m, h=0.0, kappa=0.0, nu=1.0, sigma=0.3, x0=1.0, tol=1e-11):
    """貸し手の規則と、投資の割合 x の釣り合い(皆が x を正しく見込む)を求める。
    kind = 'rel'(関係融資、縁故): 網の目の中の割合 ν の事業について、リスクを正しく知る。網の外には貸さない。
    kind = 'mkt'(市場型融資、非縁故): すべての事業について、雑音のある信号だけを見る。
    どちらも、見込んだ期待の戻りが 1 + h 以上なら貸す(h は慎重さ。h > 0 は元の「リスク許容度が小さい」にあたる)。
    x0 は釣り合いを探す出発点(1 なら楽観、0 なら悲観から)。戻り値は dict(x, G, ...)。G は一期の資金の増え方の期待値。"""
    e = expected_gross(R, m)
    if kind == 'mkt':
        post, ps = _market_tables(m, sigma)
    x = x0
    for _ in range(2000):
        g = demand_boost(x, kappa)
        if kind == 'rel':
            fund = nu * (e * g >= 1 + h)
        else:
            fund = (ps * (post * g >= 1 + h)[:, None]).sum(0)
        x_new = float(np.sum(fund) * DR)
        if abs(x_new - x) < tol:
            break
        x = 0.5 * x + 0.5 * x_new
    g = demand_boost(x, kappa)
    G = (1 - x) + float(np.sum(fund * e * g) * DR)
    bad = float(np.sum(fund * (e * g < 1)) * DR)          # 損になる事業に回った割合
    good_missed = float(np.sum((1 - fund) * (e * g >= 1)) * DR)   # 得になるのに貸されなかった割合
    return dict(x=x, G=G, bad=bad, missed=good_missed, g=g)


def both_equilibria(kind, m, **kw):
    """楽観と悲観の二つの出発点から釣り合いを求める。二つが違えば、釣り合いが二つある。"""
    hi = lending(kind, m, x0=1.0, **kw)
    lo = lending(kind, m, x0=0.0, **kw)
    return hi, lo
````

- 慎重さ(h = 0.1)は、どの環境でも関係融資を弱める。

### 3.4 投資は誰かの売上になる ── 外部性と、二つの釣り合い

作者の直感の残りの半分は、「縁故主義でないほうが経済も大きくなって機会も増え」である。投資されたお金は、賃金や仕入れとして誰かの売上になり、ほかの事業を成功しやすくする。これを、経済全体の投資の割合 x が高いほど、どの事業の成功の確率も 1 + κ(x − 0.5) 倍になる、と入れる。κ が外部性の強さである。貸し手は皆、x を正しく見込むとする(釣り合い)。

外部性があると、**釣り合いが二つになることがある**。皆が貸すと見込めば、事業は成功しやすく、実際に皆が貸す。皆が貸さないと見込めば、事業は成功しにくく、誰も貸さない(全員が現金を持つ)。楽観(x の見込み 1)と悲観(0)の二つの出発点から釣り合いを探して、二つが違えば、二つの釣り合いがある。

````python
ms = np.linspace(0.4, 1.0, 13)
ks = np.linspace(0, 0.5, 11)
Gm = np.zeros((len(ks), len(ms), 2)); Gr = np.zeros_like(Gm)
for i, k in enumerate(ks):
    for j, m in enumerate(ms):
        hi, lo = nepo.both_equilibria('mkt', m, kappa=k)
        Gm[i, j] = hi['G'], lo['G']
        hi, lo = nepo.both_equilibria('rel', m, kappa=k, nu=0.5)
        Gr[i, j] = hi['G'], lo['G']
print('楽観の釣り合いでの G(行は κ、列は m)。上: 市場型、下: 関係融資(ν = 0.5)')
sel = [0, 2, 4, 6, 8, 10]
print('  κ \\ m ' + ''.join(f'{ms[j]:7.2f}' for j in range(0, len(ms), 2)))
for i in range(0, len(ks), 2):
    print(f'  {ks[i]:4.2f}  ' + ''.join(f'{Gm[i, j, 0]:7.3f}' for j in range(0, len(ms), 2)))
print()
for i in range(0, len(ks), 2):
    print(f'  {ks[i]:4.2f}  ' + ''.join(f'{Gr[i, j, 0]:7.3f}' for j in range(0, len(ms), 2)))

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ext = [ms[0], ms[-1], ks[0], ks[-1]]
d = Gm[:, :, 0] - Gr[:, :, 0]
d = np.where((Gm[:, :, 0] < 1 + 1e-3) & (Gr[:, :, 0] < 1 + 1e-3), np.nan, d)   # どちらも誰も貸さない(G = 1)ところは白
im = ax[0].imshow(d, origin='lower', extent=ext, aspect='auto', cmap='RdBu_r', vmin=-0.3, vmax=0.3)
plt.colorbar(im, ax=ax[0], label='G(market) − G(relationship, ν = 0.5)')
ax[0].set_xlabel('environment m (larger = worse)'); ax[0].set_ylabel('externality κ')
ax[0].set_title('who wins (optimistic eq.; white = nobody lends)')
two = (Gm[:, :, 0] - Gm[:, :, 1] > 1e-3).astype(float) + 2 * (Gr[:, :, 0] - Gr[:, :, 1] > 1e-3)
im2 = ax[1].imshow(two, origin='lower', extent=ext, aspect='auto', cmap='Greys', vmin=0, vmax=3.5)
ax[1].set_xlabel('environment m'); ax[1].set_ylabel('externality κ')
ax[1].set_title('two equilibria (light: market, mid: relationship, dark: both)')
plt.tight_layout(); plt.show()
````

**読み方**

- 左の図の赤は市場型の勝ち、青は関係融資の勝ちである。**外部性 κ が強いほど、市場型の勝つ範囲が広がる**。κ = 0.4、m = 0.5 では、市場型の一期の成長は 1.26、関係融資(網の目が半分)は 1.04 で、差は外部性のないとき(1.09 対 1.06)の 8 倍ほどになる。広く貸すと投資の割合 x が上がり、それがすべての事業を成功しやすくする。作者の「投資が乗数的に広がり、経済が大きくなる」は、外部性があれば**出てくる**。
- 一方、環境が悪いほうでは、どちらも誰も貸さない釣り合い(G = 1、全員が現金を持つ)に落ちる。外部性は、良い環境での差を広げ、悪い環境での底を深くする。
- 右の図の色のついたところは、二つの釣り合いがある範囲である(薄い灰色は市場型だけ、中くらいは関係融資だけ、濃いのは両方)。同じ m と κ でも、楽観なら貸し合って成長し、悲観なら誰も貸さない。どちらになるかは、見込みと経路で決まる。

二つの釣り合いがあると、**履歴**が出る。良い環境 m = 0.5、外部性 κ = 0.3 で、市場型の貸し手の慎重さ h を少しずつ上げていき、そのあと下げて戻す(各段階で、一つ前の段階の x を見込みの出発点にする)。

````python
hs = np.round(np.r_[np.arange(-0.10, 0.301, 0.02), np.arange(0.28, -0.101, -0.02)], 3)
x = 1.0; path = []
for h in hs:
    r = nepo.lending('mkt', 0.5, h=h, kappa=0.3, x0=x)
    x = r['x']; path.append((h, r['x'], r['G']))
path = np.array(path); half = len(path) // 2 + 1
for h0 in [0.0, 0.1]:
    up = path[:half][np.isclose(path[:half, 0], h0)][0]
    dn = path[half:][np.isclose(path[half:, 0], h0)][0]
    print(f'h = {h0:.2f}: 上げていくとき x = {up[1]:.2f}, G = {up[2]:.3f} / 下げて戻すとき x = {dn[1]:.2f}, G = {dn[2]:.3f}')
fig, ax = plt.subplots(figsize=(6.5, 3.6))
ax.plot(path[:half, 0], path[:half, 2], 'o-', color=PAL[0], ms=3, label='caution h raised')
ax.plot(path[half:, 0], path[half:, 2], 's--', color=PAL[1], ms=3, label='caution h lowered back')
ax.set_xlabel('caution h of market lenders'); ax.set_ylabel('growth G per period')
ax.set_title('hysteresis (m = 0.5, κ = 0.3)'); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
````

慎重さを上げていくと、h = 0.14 で投資が崩れ、誰も貸さない釣り合いに落ちる。そのあと慎重さを元に戻しても、**投資は戻らない**(皆が「誰も貸さない」と見込んでいる限り、貸すと損だから)。h = 0(ふつうの損益の分かれ目)まで戻しても x = 0 のままで、h が負(自分の見込みでは少し損でも貸す)になって、ようやく戻る。戻るには、慎重さを崩れた値よりずっと下げる(リスクを積極的に取る)か、誰かが見込みそのものを変える(皆が貸すと信じさせる。公共の投資や、中央銀行の約束がその役をする)必要がある。

作者は、その 2 の左の枝を「安定志向が少子化を加速した」日本と読んだ。この模型の言葉では、それは山の左の枝ではなく、**崖の下の釣り合い**として書くほうが筋が通る。慎重さが一度ある値を超えると経済は崖から落ち、慎重さを元に戻しても上がってこない。ただし、これは模型の中の話で、日本の長い停滞がこの型かどうかは、模型からは言えない(少子化との関係も、この模型には入っていない)。

### 3.5 実証との釣り合い

作者の仮定「縁故主義は、リスクがまるわかりのため、小さいリスクで投資を引き上げる」(慎重さ h が大きい)は、実証とは逆向きの面がある。

- ボルトンら(Bolton, Freixas, Gambacorta and Mistrulli 2016)は、イタリアの銀行の貸出を調べ、関係融資の銀行は、平時には少し高い金利をとるが、**危機のときには貸出を続けた**(取引型の銀行より貸出を減らさなかった)ことを示した。付き合いのある借り手を知っているので、危機でも良い借り手を見分けて貸し続けられる。つまり、情報は、危機のときの慎重さを**下げる**方向に働く。
- 一方、縁故の害を示す実証もある。クワジャとミアン(Khwaja and Mian 2005)は、パキスタンで、政治家とつながりのある企業が 45% 多く借り、50% 多く返済できなくなることを示した。これは、情報ではなく**えこひいき**(見分けられるのに、見分けない)の害である。作者は「縁故主義に不正があるとかそういうこととは関係がなく」と、この害を問いから外している。

まとめると、関係融資と市場型融資の勝ち負けは、**見分ける効果、広さの効果、慎重さ、外部性**の釣り合いで決まり、作者の直感(非縁故のほうが経済を大きくする)が成り立つのは、外部性が強く、環境が良く、関係融資の網の目が狭いときである。「縁故は早く引き上げる」は仮定であって、実証はむしろ逆を示すことが多い。

## 第 4 章 投票は無意味か ── ダウンズへの反論

### 4.1 作者の式

ダウンズ(Downs 1957)の合理的選択の理論では、投票の期待効用は

$$B \cdot P - C$$

と書かれる。B は自分の推す候補が勝ったときの利益、P は**自分の一票**が結果を変える確率、C は投票の費用である。有権者が何万人もいれば P はとても小さいので、B·P は C に負け、「合理的な人は投票しない」ことになる(投票のパラドックス)。ライカーとオードシュック(Riker and Ordeshook 1968)は、これに「投票の義務を果たす満足」D を足して B·P − C + D としたが、これは「合理性では説明できない分を足した」とも読める。

作者は、2024年12月にこう書いた(元のノートの最初のセル)。

> 選挙の結果により政策A1と政策A2だけが異なるとする。自分が参加することで似た他の人も参加し A1 が実行される可能性が p1 になるとする。自分が参加しないことで似た他の人も参加しないことで A1 が実行される可能性が p2 になるとする。投票のコストを C とする。
>
> 選択(投票行動)の効用は、A1 * p1 + A2 * (1 - p1) - A1 * p2 - A2 * (1 - p2) - C になる。
>
> これは = (A1 - A2) * (p1 - p2) - C。

式の変形は正しい(下の Lean の `vote_utility`)。ダウンズの式と比べると、B = A1 − A2、P = p1 − p2 で、**形は同じ**である。違いは P の中身にある。ダウンズの P は「自分の一票だけ」が動いたときの差、作者の p1 − p2 は「自分と、自分に似た人たち」が動いたときの差である。作者は「p1 - p2 はとても大きいものであることが許されるはずだ」と書き、その理由を「相転移とかカオスセオリーとかありそうだが、そこはまだ私には理論化できない」とした。この章は、その理由を模型にする。

````python
%%writefile -a misc_models.lean
/-! ## 第 4 章 投票の効用と相関

作者の式: 投票するときの期待値から、棄権するときの期待値と費用を引くと、
(A1 − A2)(p1 − p2) − C になる。 -/

theorem vote_utility (A1 A2 p1 p2 C : ℝ) :
    (A1 * p1 + A2 * (1 - p1)) - (A1 * p2 + A2 * (1 - p2)) - C
      = (A1 - A2) * (p1 - p2) - C := by
  ring

/-- ベータ分布 Beta(a, b) の平均は a / (a + b)。一人が投票したと知ると Beta(a + 1, b)、
棄権したと知ると Beta(a, b + 1) になる。二つの平均の差は 1 / (a + b + 1)、
つまり級内相関 ρ そのものである。自分の選択は、仲間の投票の確率を ρ だけ動かす「しるし」になる。 -/
theorem beta_update_shift {a b : ℝ} (ha : 0 < a) (hb : 0 < b) :
    (a + 1) / (a + 1 + b) - a / (a + (b + 1)) = 1 / (a + b + 1) := by
  have h : a + b + 1 ≠ 0 := by linarith
  rw [show a + 1 + b = a + b + 1 by ring, show a + (b + 1) = a + b + 1 by ring]
  field_simp
  ring

/-- 級内相関 ρ の n 人の票の和の分散。一人の分散を v、二人の間の共分散を ρ v とすると、
和の分散は n v + n (n − 1) ρ v。ρ > 0 なら、分散は n² の速さで増え、票の数の揺れは人数に比例する。
独立(ρ = 0)なら n v で、揺れは √n に比例する。 -/
theorem equicorrelated_variance (n : ℕ) (v ρ : ℝ) :
    ∑ i ∈ range n, ∑ j ∈ range n, (if i = j then v else ρ * v)
      = n * v + n * (n - 1) * ρ * v := by
  have hrow : ∀ i ∈ range n,
      ∑ j ∈ range n, (if i = j then v else ρ * v) = v + (n - 1) * ρ * v := by
    intro i hi
    rw [Finset.sum_ite, Finset.sum_const, Finset.sum_const, Finset.filter_eq,
      Finset.filter_ne]
    simp only [hi, ite_true, Finset.card_singleton, nsmul_eq_mul, Nat.cast_one, one_mul]
    rw [Finset.card_erase_of_mem hi, Finset.card_range]
    rcases Nat.eq_zero_or_pos n with h0 | hpos
    · subst h0; simp at hi
    · rw [Nat.cast_sub (by omega)]; push_cast; ring
  rw [Finset.sum_congr rfl hrow, Finset.sum_const, Finset.card_range, nsmul_eq_mul]
  ring
````

### 4.3 「似た人も参加する」の二つの読み方

作者の文「自分が参加することで似た他の人も参加し」には、二つの読み方がある。

1. **因果の読み**: 自分が投票に行くと、それを見た友人も行く(自分が原因になる)。
2. **しるしの読み**: 自分が投票に行くと決めるとき、自分と同じ考え方(規則)で決める人たちも、同じように決めている。自分の決定は、仲間の決定の**しるし**になる。作者が別の箇所で書いた「自分が動くとき他が相転移的に動いていることを**見ている**確率」は、こちらに近い。

しるしの読みは、決定の理論ではよく知られた立場である。ハルサニ(Harsanyi 1980)は、投票を「一人の行為」ではなく「規則の選択」とみる規則功利主義で投票のパラドックスを解いた。フェダーセンとサンドローニ(Feddersen and Sandroni 2006)は、同じ規則に従う「倫理的な投票者」の群れの模型を作った。ホフスタッター(Hofstadter 1983)の「超合理性」(自分と同じように考える相手は、自分と同じ答えを出す)も同じ考えである。一方、因果だけを数える立場(因果的決定理論)からは、「自分の決定は他人の決定の原因ではないので、自分の一票の P だけを数えるべきだ」という反論がある(ニューカムの問題と同じ対立)。

どちらの読みでも、要になるのは**相関**である。自分と仲間の決定に相関がなければ(独立なら)、p1 − p2 はダウンズの P に戻る。相関があれば、大きくなりうる。以下では、まずしるしの読みで、相関の強さ ρ を振って p1 − p2 を計算する(4.4)。次に因果の読みで、友人の間の連鎖を見る(4.5)。

### 4.4 相関を振る ── しるしの読みの模型

有権者 N 人を、半分ずつ二つの陣営に分ける。自分は陣営 1 にいる。陣営 1 の一人ひとりは、確率 Θ で投票に行く。Θ はその選挙の「空気」のようなもので、陣営の全員に共通だが、前もってはわからない。Θ は平均 q、ばらつきのある分布(ベータ分布)に従うとする。すると、二人の決定の間の相関(級内相関)は ρ = 1/(a + b + 1) になる(a、b はベータ分布の母数)。ρ = 0 なら Θ = q と決まっていて、全員が独立に決める。陣営 2 も同じように決めるが、陣営 1 とは独立である。

自分が投票に行くと決めたとき、それは「Θ が高めだ」というしるしになる。ベータ分布の更新で、Θ の平均はちょうど ρ だけ上がる(Lean の `beta_update_shift`。投票したと知ったときの平均と、棄権したと知ったときの平均の差が 1/(a + b + 1) = ρ)。だから

- p1 = 自分が投票し、仲間の Θ の分布が「投票した」で更新されたときに、陣営 1 が勝つ確率、
- p2 = 自分が棄権し、仲間の Θ の分布が「棄権した」で更新されたときの確率

である。ρ = 0 なら、仲間の分布は変わらず、差は自分の一票だけになり、p1 − p2 はダウンズの P(一票で結果が変わる確率)になる。

票の数の分布は、ベータ二項分布という名前のついた分布なので、乱数を使わずに正確に計算できる。

````python
%%writefile vote.py
# vote.py: 第 4 章 投票は無意味か(ダウンズへの反論)の計算
# 元のノートのプログラムは使わない。相関のある投票の模型を、ここで新しく書く。
import numpy as np
from collections import deque
from scipy.stats import betabinom, binom
````

````python
%%writefile -a vote.py
def votes_pmf(n, q, rho, cond=0):
    """n 人の票の数の分布。一人ひとりの投票の確率 Θ が、平均 q、級内相関 ρ のベータ分布に従う(ベータ二項分布)。
    ρ = 0 なら二項分布(独立)。cond = +1 は「自分が投票した」という条件、-1 は「棄権した」という条件のもとでの、残りの人の分布。
    ベータ分布 Beta(a, b) は、一人が投票したと知ると Beta(a + 1, b)、棄権したと知ると Beta(a, b + 1) になる。"""
    k = np.arange(n + 1)
    if rho == 0:
        return binom.pmf(k, n, q)
    s = 1 / rho - 1
    a, b = q * s, (1 - q) * s
    if cond == 1:
        a += 1
    elif cond == -1:
        b += 1
    return betabinom.pmf(k, n, a, b)


def win_prob(v1, v2, add=0):
    """陣営 1 の票 v1(+ add)が陣営 2 の票 v2 を上回る確率。同数は半分の勝ち。v1, v2 は票の数の分布。"""
    c2 = np.concatenate([[0.0], np.cumsum(v2)])        # c2[k] = P(V2 < k)
    k = np.arange(len(v1)) + add
    lt = c2[np.clip(k, 0, len(v2))]
    eq = np.where(k < len(v2), v2[np.clip(k, 0, len(v2) - 1)], 0.0)
    return float(np.sum(v1 * (lt + 0.5 * eq)))


def p1_p2(N, rho, q1=0.5, q2=0.5, rho2=None):
    """有権者 N 人を半分ずつの二つの陣営に分ける。自分は陣営 1。
    p1 = 自分が投票したときに陣営 1 が勝つ確率、p2 = 棄権したときの確率。
    相関 ρ があると、自分の選択は、同じ規則で動く仲間の選択の「しるし」になる(ベータ分布の更新)。
    ρ = 0 なら、自分の一票の差だけが残り、p1 − p2 は一票で結果が変わる確率(ダウンズの P)になる。"""
    n1 = N // 2
    n2 = N - n1
    rho2 = rho if rho2 is None else rho2
    v2 = votes_pmf(n2, q2, rho2)
    p1 = win_prob(votes_pmf(n1 - 1, q1, rho, +1), v2, add=1)
    p2 = win_prob(votes_pmf(n1 - 1, q1, rho, -1), v2, add=0)
    return p1, p2


def scan_corr(Ns, rhos, **kw):
    """(N, ρ) の格子で p1 − p2 を計算する。"""
    return np.array([[np.subtract(*p1_p2(N, r, **kw)) for r in rhos] for N in Ns])
````

- **ρ > 0 では、人数がどれほど多くても、p1 − p2 は √(ρ/π) ≈ 0.56√ρ より小さくならない**(破線)。ρ = 0.001(千組に一組ほどの弱い相関)でも 0.018、ρ = 0.01 なら 0.056 である。百万人の選挙で、ダウンズの P の 70 倍になる。
- 右の図は、横軸を ρN、縦軸を (p1 − p2)√N にとったもので、どの N の線も**一本の曲線に重なる**。ρN が小さいところでは √(2/π) ≈ 0.8(ダウンズ)、大きいところでは √(ρN/π)(相関)で、二つが交わるのは ρN ≈ 2 である。**独立(ダウンズ)の世界と、相関の世界の境目は ρ ≈ 2/N にある**。人数が多いほど、境目は 0 に近づく。

**なぜ √(ρ/π) か**。人数が多いと、勝ち負けはほぼ「陣営 1 の Θ と陣営 2 の Θ のどちらが大きいか」で決まる。二つの Θ の差は、平均 0、分散 2ρq(1 − q) = ρ/2(q = 1/2 のとき)の山の形に広がる。自分の選択は、陣営 1 の Θ の平均を ρ だけずらす(Lean の `beta_update_shift`)。勝つ確率の変化は、ずらす幅 × 差が 0 のところの山の高さ = ρ × 1/√(πρ) = √(ρ/π) である。票の数の揺れは、相関があると人数に比例して大きくなる(Lean の `equicorrelated_variance`: 和の分散は Nv + N(N − 1)ρv)。だから、自分の選択がずらす幅(ρN 票)と、揺れの幅(√ρ·N 票ほど)が、**同じ速さで**大きくなり、比が人数によらなくなる。

**これは相転移か**。人数 N を無限に大きくした極限では、p1 − p2 は

$$\lim_{N\to\infty} (p_1 - p_2) = \begin{cases} 0 & (\rho = 0) \\ \sqrt{\rho/\pi} & (\rho > 0) \end{cases}$$

となる。ρ = 0 のところで、関数がなめらかでなくなる(√ρ は ρ = 0 で傾きが無限大)。物理の言葉では、ρ = 0 が臨界点で、p1 − p2 が秩序の量、指数が 1/2 の連続の相転移(平均場の型)と同じ形である。有限の N では、その角が幅 2/N ほどで丸まる(右の図の揃い方は、物理の「有限の大きさのスケーリング」と同じ見方である)。作者の「p1 − p2 は、とても大きいものであることが許される」は、**相関が少しでもあれば、大きな選挙では必ず成り立つ**。逆に言えば、ダウンズの P は、相関がちょうど 0 という特別な点での値である。

接戦でない場合も見ておく。陣営 2 の平均の投票率を 0.52 にする(陣営 1 は 0.5)。

````python
D2 = vote.scan_corr(Ns, [0, 1e-4, 1e-3, 1e-2, 0.1], q2=0.52)
print('陣営 2 が 0.52、陣営 1 が 0.5 のとき(行は N、列は ρ = 0, 1e-4, 1e-3, 1e-2, 0.1)')
for N, d in zip(Ns, D2):
    print(f'  N = {N:>9,}: ' + '  '.join(f'{x:.5f}' for x in d))
````

接戦でないと、独立の世界(ρ = 0)では、p1 − p2 は人数とともに**指数的に**小さくなる(百万人では、ほぼ 0)。一方、相関の世界(ρ = 0.01 なら 0.054)では、ほとんど変わらない。勝ち負けを決めるのは「空気」Θ の揺れであって、平均の差 0.02 は、Θ の揺れ(標準偏差は √(ρ/4) = 0.05 ほど)の中に埋もれるからである。

**費用との比べ方**。投票の効用は (A1 − A2)(p1 − p2) − C である。p1 − p2 ≈ 0.56√ρ なら、ρ = 0.001 のとき、二つの政策の差 A1 − A2 が投票の費用 C の 56 倍あれば、投票に行くのが得になる。ダウンズの P(百万人で 0.0008)なら、1250 倍が要る。

**注意**: これは「しるしの読み」での計算である。自分の決定が仲間の決定の原因でない限り、因果だけを数える人は、この p1 − p2 を認めない。しかし、自分が「投票に行く人間かどうか」を決めることが、自分と同じように考える人たちの決定と相関しているのは、事実としてある。どちらの決定の理論をとるかは、数学ではなく哲学の選択である。作者の立場(「見ている確率」)は、ハルサニやフェダーセンとサンドローニと同じ側にある。

### 4.5 因果の相関 ── 友人の連鎖

次に、因果の読みを見る。有権者は友人の網の目の上にいて、**友人のうち割合 φ 以上が投票に行くなら、自分も行く**とする(閾値の模型。Granovetter 1978、網の目の上ではワッツ Watts 2002)。初めは誰も行かない。一人(自分)が先に行くと決めたとき、連鎖で何人が動くかを数える。友人の数の平均 z を振る。友人の数が多いほど、決定の相関は強くなる。

````python
%%writefile -a vote.py
def random_graph(N, z, rng):
    """平均の次数 z のランダムなグラフ(エルデシュ–レーニィ)。隣接の表を CSR の形で返す。"""
    m = rng.poisson(z * N / 2)
    u = rng.integers(0, N, m)
    v = rng.integers(0, N, m)
    keep = u != v
    u, v = u[keep], v[keep]
    a = np.concatenate([u, v])
    b = np.concatenate([v, u])
    order = np.argsort(a, kind='stable')
    a, b = a[order], b[order]
    return np.searchsorted(a, np.arange(N + 1)), b


def cascade_size(ptr, nb, phi, seed):
    """閾値の連鎖(ワッツ 2002)。友人のうち割合 φ 以上が投票に行くと、自分も行く。
    一人(seed)が先に動いたとき、最後に何人が動くか。"""
    N = len(ptr) - 1
    deg = np.diff(ptr)
    active = np.zeros(N, bool)
    count = np.zeros(N, int)
    active[seed] = True
    queue = deque([seed])
    size = 1
    while queue:
        i = queue.popleft()
        for j in nb[ptr[i]:ptr[i + 1]]:
            if active[j]:
                continue
            count[j] += 1
            if count[j] >= phi * deg[j]:
                active[j] = True
                queue.append(j)
                size += 1
    return size


def cascade_scan(N, zs, phi=0.18, seeds=200, rng_seed=1, big=0.1):
    """平均の次数 z ごとに、一人が全体の連鎖を起こす確率と、そのときの大きさ。"""
    rng = np.random.default_rng(rng_seed)
    rows = []
    for z in zs:
        ptr, nb = random_graph(N, z, rng)
        sizes = np.array([cascade_size(ptr, nb, phi, rng.integers(N)) for _ in range(seeds)])
        hit = sizes > big * N
        rows.append((z, hit.mean(), sizes[hit].mean() / N if hit.any() else 0.0, np.median(sizes)))
    return rows
````

````python
importlib.reload(vote)   # 書き足した部分を読み直す
zs = [0.5, 1, 1.5, 2, 3, 4, 5, 6, 7, 8, 10, 12]
rows = vote.cascade_scan(20_000, zs, phi=0.18, seeds=300)
print(' 友人の平均 z | 一人が全体の連鎖を起こす確率 | 起きたときの大きさ(全体に対して) | 連鎖の大きさの中央値')
for z, p, s, med in rows:
    print(f'   {z:>5.1f}     |   {p:6.3f}   |   {s:5.3f}   |   {med:6.0f}')
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot([r[0] for r in rows], [r[1] for r in rows], 'o-', color=PAL[0], label='P(global cascade from one seed)')
ax.plot([r[0] for r in rows], [r[2] for r in rows], 's--', color=PAL[1], label='size if it happens')
ax.set_xlabel('mean number of friends z'); ax.set_ylabel('fraction')
ax.set_title('threshold φ = 0.18, N = 20,000')
ax.legend(fontsize=8); plt.tight_layout(); plt.show()
````

友人の数の平均 z が 1.5 から 5 ほど(1 と 6 の間)の**窓**の中でだけ、一人の決定が全体の連鎖(全体の 1 割以上)を起こす。窓の下では、網の目がつながっていないので広がらない。窓の上では、一人ひとりの友人が多すぎて、一人の友人が動いただけでは、閾値に届かない。窓の両端は、どちらも相転移である(ワッツ 2002 の結果と同じ)。窓の中では、一人の決定で、全体の 6 割から 9 割余りが動く。これは、作者の「自分が動くとき他が相転移的に動いている」の、因果の版である。

ただし、因果の読みには落とし穴がある。窓の中では、**誰が先に動いても**全体が動く。だから、自分以外の誰かが先に動くなら、自分が動いたかどうかで結果は変わらない。自分の決定が効くのは、「まだ誰も動いていない」ときに先に動く場合だけである。因果の p1 − p2 が大きいのは、窓の中で、しかも自分が最初の一人になれるときに限られる。しるしの読み(4.4)には、この制限はない。

### 4.6 もう一つの反論 ── 利益 B も人数とともに大きくなる

ダウンズの式には、もう一つ見落としがある。政策の差 A1 − A2 が社会全体に及ぶものなら、自分が他人の利益もいくらか気にかける(利他的である)限り、自分にとっての B は、人数 N に比例して大きくなる。ゲルマンらは、アメリカの大統領選挙で、一票で結果が変わる確率が平均で 6000 万分の 1 ほど(接戦の州では 1000 万分の 1 ほど)であることを推定し(Gelman, Silver and Edlin 2012)、利他的な有権者には B·P が人数によらない大きさになりうるとして、「合理的な選択としての投票」を論じた(Edlin, Gelman and Kaplan 2007)。作者の p1 − p2 の議論(相関で P が大きくなる)と、この議論(利他で B が大きくなる)は、別々の理由で、どちらもダウンズの結論を覆す。

### 4.7 経済の側への意味

このシリーズの経済の模型では、再分配の率、公共事業の量、金利などは、外から与える**パラメータ**だった。投票は、そのパラメータを社会の内側で決めるしくみである。この章の結果は、そこに二つのことを付け加える。

1. 投票率は、相関の強さで決まる。相関が 0 の世界(ダウンズ)では誰も投票せず、パラメータは少数の熱心な人(費用 C が負の人)が決める。相関があると、広い層が投票し、パラメータは多数の利害を映す。
2. 連鎖の窓(4.5)のように、つながり方が少し変わるだけで、投票の結果(したがってパラメータ)が飛ぶことがある。経済の模型のパラメータを投票で内生化すると、**臨界の近くで、政策が飛ぶ**ことがありうる。これは、ノート 3 の最後の「効くのは臨界の近く」と同じ型である。

ノート 4で、政策のパラメータを内側で決めるなら、まずこの章の模型(相関のある有権者の群れ)を、再分配の率を選ぶ選挙として使うのが、いちばん小さな一歩になる。

作者が元のノートの最初に書いていた見方、「合理的でない無党派層の投票行動が選挙を予測不能にしている」のではなく「本来の公正な社会に近付いている」は、この章の言葉では、**相関のある有権者の群れが、自分たちの規則で投票することは合理的であり、そのとき選挙は多数の利害を映す**、と言い換えられる。

## 第 5 章 AI と弾圧と再分配(アスモグル拡張 1・2)

材料: 《AI社会における再分配問題 - アスモグルらの議論から》(2026年9月)、《「注意労働」論とロボット軍の民主主義モデル》(2026年9月)

https://github.com/JRF-2018/misc_ipynb/blob/master/expansion_of_acemoglu_model_2.ipynb

二つのノートは、作者と複数の AI(Claude、Grok、ChatGPT、Gemini)の議論の記録で、模型が何度も作り直されている。後半の構図(AI 経済が人間経済の中で育ち、両者の接続口が問題になる)は、ノート O の第 7.5 節(AI 経済と人間経済)で使った。この章では、決まったとおり(JRF)、**議論が間違っていないかを確かめて、短く書く**。

### 5.1 模型の要点

**その 1 のモデル 1**(4 番目のセル `model.py`)。資本家の国家が、自動化 a、弾圧 r、再分配 τ を選ぶ。アセモグルらの論文(元のノートでは Acemoglu, Gitmez and Shadmehr "Automation and Repression"、NBER Working Paper 35336 とされている)の定性的な結論、「資本が蓄積すると、弾圧に収束する」を、単純な目的関数で再現している。作者の拡張は、**AI の出力が資本家の意図どおりかを確かめる監査(注意労働)には人間が要る**という制約で、要る監査 a に対して、弾圧で潰されたあとに残る監査の供給 1 − r が足りないと、収益が φ_max·max(0, a + r − 1) だけ目減りする。

**その 1 のモデル 2 以降**。要る監査を a√K(AI の規模とともに増える)にし、少数の忠実なエリートの監査者 E を雇えるようにし、エリートも社会の再生産に頼る(弾圧で供給が (1 − r)^δ に減る)とした。さらに、再分配が監査の供給を増やす道(1 + ξτ)を足した。

**その 2**。二段の模型である。段階 1 で、社会が国家ロボットの能力 R と注意労働 A(独立した司法・議会・報道・監査)を選び、国家の装置が乗っ取られる確率(捕獲の確率)Π が決まる。段階 2 で、資本家が弾圧・再分配・AI の成果の取り込みを選ぶ。段階 2 の監査の需要に、国家ロボットの「統制されていない残差」R·Π を足す。

### 5.2 確かめ

**(1) モデル 1 の「弾圧の頭打ち」は、仮定から素直に出る**。目減りの項 max(0, a + r − 1) は、a + r = 1 に折れ目を持つ。資本家にとって、r を 1 − a より上げると、収益が φ_max の速さで目減りするので、r* は 1 − a* のあたりで頭打ちになる。結果は仮定の言い換えだが、論理は正しい。元のノートの Claude さんの留保(「監査の担い手 = 民衆全体」という仮定を supply(r) = 1 − r に埋め込んでいる)も、そのとおりである。

**(2) モデル 2 の「巨大な AI では中位存続に近づく」は、Grok さんの指摘(11 番目のセル)のとおり、成り立っていない**。要る監査が √K で、生産が K^0.35 で伸びるので、要る監査と生産の比は K^0.15 で限りなく大きくなる。大きな K で三つの体制が揃うのは、監査を諦めて、目減り(φ ≈ 0.8)を飲んだうえで、反乱だけを再分配で抑えているからである。Grok さんは、この結論が「監査の需要の伸び(指数 0.5)が、生産の伸び(0.35)より大きい」という仮定に強く依存することも指摘している。この点は、ノートの中で正しく直されている(再分配が監査の供給を増やす道を足した)。

**(3) その 2 の「規模の上限は二重計上だった」は、向きは正しいが、上限が消えたとまでは言えない**。段階 1 で注意労働 A が捕獲の確率 Π を下げたなら、段階 2 で国家ロボット R をまるごと監査の需要に足すのは、同じ安全化を二度数えることになる。残差 R·Π を足すのが筋である。ここまでは正しい。ただし、二つの点が残る。

- **段階の間で、残差の測り方が違う**。段階 1 の捕獲の被害は κ·Π·R²(巨大な装置ほど暴走の被害が大きい、として二乗にした)、段階 2 の監査の需要は D·R·Π(一次)である。被害が R² で増えるなら、それを監査する手間も一次より速く増えると考えるのが自然で、そうすると上限は一部戻る。
- **注意の予算が共有されていない**。段階 1 の注意労働 A(司法・議会・報道・監査)は、お金の費用 c_A·A² として払われるだけで、段階 2 の市民の監査の能力 N_BAR(定数)からは引かれていない(26 番目のセル `model_v7_with_R_Pi.py` と、13 番目のセル `model_v3.py` の `audit_supply`)。しかし、独立した司法や報道で働くのも、AI の出力を監査するのも、**同じ人間の注意**である。

二つ目を、小さな模型で確かめる。捕獲を防ぐ注意 A と、残差の監査 R·Π(A) と、AI の監査 n0 を、同じ注意の量 N から払うとする。要る注意の合計は A + n0 + R·Π(A) で、A を最も良く選んだときの値が N 以下なら、規模 R を持てる。Π(A) = e^{−A} なら、最小値は n0 + 1 + log R になり(Lean の `attention_lower_bound` と `attention_at_optimum`)、持てる規模の上限は R ≤ e^{N − n0 − 1} である。

````python
%%writefile -a misc_models.lean
/-! ## 第 5 章 注意の予算と、国家ロボットの規模

国家ロボットの規模を R、捕獲を防ぐ注意労働を A、捕獲の確率を Π(A) = e^{−A} とする。
捕獲を防ぐ注意 A と、統制されていない残差 R·Π(A) の監査を、同じ人間の注意から払うなら、
要る注意は A + R e^{−A} で、その最小値は 1 + log R(A = log R のとき)。
注意の量が N なら、持てる規模は R ≤ e^{N − 1} どまりになる。上限はあるが、注意の量に対して指数的に大きい。 -/

theorem attention_lower_bound {R : ℝ} (hR : 0 < R) (A : ℝ) :
    1 + Real.log R ≤ A + R * Real.exp (-A) := by
  have h1 : R * Real.exp (-A) = Real.exp (Real.log R - A) := by
    rw [Real.exp_sub, Real.exp_log hR, div_eq_mul_inv, Real.exp_neg]
  have h2 := Real.add_one_le_exp (Real.log R - A)
  rw [h1]
  linarith

theorem attention_at_optimum {R : ℝ} (hR : 0 < R) :
    Real.log R + R * Real.exp (-Real.log R) = 1 + Real.log R := by
  rw [Real.exp_neg, Real.exp_log hR, mul_inv_cancel₀ hR.ne']
  ring
````

````python
%%writefile attn.py
# attn.py: 第 5 章 注意の予算を共有したときの、国家ロボットの規模の上限
# 元のノートのプログラムは使わない。小さな模型を、ここで新しく書く。
import numpy as np
from scipy.optimize import minimize_scalar


def capture(A, form='exp', k=2.0):
    """注意労働 A(独立した司法・議会・報道・監査)を積んだときの、捕獲の確率 Π(A)。
    'exp' は e^{−A}(一単位ごとに一定の割合で下がる)、'power' は (1 + A)^{−k}(下がり方がだんだん鈍る)。"""
    return np.exp(-A) if form == 'exp' else (1 + A) ** (-k)


def attention_needed(R, n0, form='exp', k=2.0):
    """国家ロボット R を持つ社会が、捕獲を防ぐ注意 A と、残差 R·Π(A) の監査と、AI の監査 n0 に要る、
    人間の注意の合計の最小値: min_A [A + n0 + R·Π(A)]。"""
    if R <= 0:
        return n0
    f = lambda A: A + n0 + R * capture(A, form, k)
    res = minimize_scalar(f, bounds=(0, 200), method='bounded')
    return min(res.fun, f(0.0))


def max_R(N, n0, form='exp', k=2.0):
    """注意の量 N の社会が持てる国家ロボットの規模の上限(注意が足りる最大の R)。"""
    if attention_needed(1e-12, n0, form, k) > N:
        return 0.0
    lo, hi = 0.0, 1.0
    while attention_needed(hi, n0, form, k) <= N and hi < 1e12:
        hi *= 2
    for _ in range(80):
        mid = 0.5 * (lo + hi)
        lo, hi = (mid, hi) if attention_needed(mid, n0, form, k) <= N else (lo, mid)
    return lo
````

**読み方**

- 注意の予算を共有すると、**規模の上限は戻ってくる**。ただし、その上限は、注意の量に対して**指数的に**大きい(Π = e^{−A} のとき e^{N − 2})。注意を一単位増やすごとに、持てる規模は e ≈ 2.7 倍になる。
- 捕獲の確率の下がり方が鈍い(Π = (1 + A)^{−2})と、上限は注意の量の**べき**でしか伸びない。上限の大きさを決めるのは、「注意を一単位積むと、捕獲の確率がどれだけ下がるか」の形である。
- だから、その 2 の結論は、次のように言い直すのが正確だと思う。「規模の上限が低く見えたのは、二重計上のせいだった。注意の予算を共有すれば上限はあるが、注意労働が捕獲を効率よく防げる(Π が速く下がる)限り、上限は非常に高い。逆に、注意労働が形骸化して Π が下がりにくくなると、上限は急に下がる」。元のノートの最後の文「弾圧コストを高く保てるかどうかは、装置の大きさではなく、それを監査する制度をどれだけ本気で作り込むかにかかっている」は、この言い直しのもとでも、そのまま成り立つ。

**まとめ**。アスモグル拡張の議論は、大筋で間違っていない。モデル 1 の結果は仮定の言い換え、モデル 2 の高い K の結論は(ノートの中で指摘されたとおり)監査の放棄、その 2 の二重計上の修正は正しい向きである。付け加えるべき留保は、段階 1 と段階 2 が**同じ人間の注意**を使うことで、それを入れると、規模の上限は消えずに、注意の量に対して指数的(または、べき)に高い位置に移る。

````python
# Lean の準備: .leandir に書いた lake のプロジェクト(Mathlib 入り)があれば、それを使う。
# なければ elan(Lean の版の管理)を入れ、Lean 4 v4.34.1 と Mathlib の入ったプロジェクト jrf_lean/ を作る(Google Colab でもそのまま通る)。
# 初めてのときは Mathlib の組み上がったものを取ってくるので、10 分ほどかかり、10GB ほどの場所を使う。同じディレクトリのほかのノートは、それを使い回す。
import os, shutil, subprocess

def lean_project_ok(d):
    return bool(d) and os.path.exists(os.path.join(d, 'lakefile.lean')) and os.path.isdir(os.path.join(d, '.lake', 'packages', 'mathlib'))

LEANDIR = open('.leandir').read().strip() if os.path.exists('.leandir') else None
if lean_project_ok(LEANDIR):
    print('Lean のプロジェクト: .leandir に書いた場所にあるものを使う')
else:
    elan_bin = os.path.expanduser('~/.elan/bin')
    os.environ['PATH'] = elan_bin + ':' + os.environ.get('PATH', '')
    if not shutil.which('lake'):
        subprocess.run('curl -sSfL https://raw.githubusercontent.com/leanprover/elan/master/elan-init.sh'
                       ' | sh -s -- -y --default-toolchain none', shell=True, check=True)
    LEANDIR = os.path.abspath('jrf_lean')
    os.makedirs(LEANDIR, exist_ok=True)
    with open(os.path.join(LEANDIR, 'lean-toolchain'), 'w') as f:
        f.write('leanprover/lean4:v4.34.1\n')
    with open(os.path.join(LEANDIR, 'lakefile.lean'), 'w') as f:
        f.write('import Lake\nopen Lake DSL\n\npackage jrf_lean\n\n'
                'require mathlib from git\n  "https://github.com/leanprover-community/mathlib4.git" @ "v4.34.1"\n')
    lake = shutil.which('lake') or os.path.join(elan_bin, 'lake')
    for args in (['update'], ['exe', 'cache', 'get']):
        r = subprocess.run([lake] + args, cwd=LEANDIR, capture_output=True, text=True)
        print(f'lake {" ".join(args)}: 終了コード {r.returncode}')
        if r.returncode != 0:
            print(r.stdout[-2000:], r.stderr[-2000:])
            raise RuntimeError('Lean のプロジェクトを作れなかった')
    with open('.leandir', 'w') as f:
        f.write(LEANDIR + '\n')
    print('Lean のプロジェクト: jrf_lean/ に作り、.leandir に書いた')
````

## 付録 B: 用語集

| 用語 | 意味 |
|---|---|
| 帳簿を閉じる | すべての支払いに、払う側と受け取る側を書き、支出の合計と所得の合計を一致させること |
| 部門の収支 | 家計・企業・政府などの部門ごとの、収入と支出の差。帳簿が閉じていれば、和は 0 |
| 漏れ | 所得のうち、使われも投資されもしない分(現金や国債で持つ)。需要が決まる模型では、漏れの大きさが所得の大きさを決める |
| カレツキの利潤の式 | 「資本家は自分が使った分を稼ぐ」。労働者が貯めないなら、資本家の所得は、資本家の支出と政府の赤字で決まる |
| ストーン–ギアリーの需要 | まず必要量を買い、残り(余裕)を一定の割合で分ける需要の型 |
| ゴーマンの集計 | 余裕の分け方の割合が皆同じなら、所得を移しても需要の合計は変わらない、という性質 |
| 多様性への愛好 | 同じ量の財でも、種類が多いほど満足が大きい、という好み(ディキシット–スティグリッツ) |
| 値上げ幅(マークアップ) | 価格と限界費用の比。差別化された財では 1 より大きく、差が利潤になる |
| レント | 供給が足りないときに、価格が費用を超えて上がった差額。持ち主の所得になる |
| 費用の経路 | 企業が運転資金を借りるので、金利が上がると費用と価格が上がる経路 |
| 関係融資・市場型融資 | 付き合いのある相手に、帳簿に出ない情報で貸す融資と、公の情報だけで誰にでも貸す融資 |
| 外部性 | ある人の行動が、市場を通さずに他人の得失に効くこと。ここでは、投資が他人の事業を成功しやすくすること |
| 二つの釣り合い | 皆が同じことを見込むと、その見込みどおりになる状態が二つあること(皆が貸す、誰も貸さない) |
| 履歴(ヒステリシス) | 条件を元に戻しても、状態が元に戻らないこと |
| 級内相関 ρ | 同じ群れの二人の決定の相関。ベータ分布で混ぜると 1/(a + b + 1) |
| ベータ二項分布 | 一人ひとりの確率 Θ がベータ分布に従うときの、n 人のうち当たる人数の分布 |
| しるしの読み・因果の読み | 自分の決定を、仲間の決定の「しるし」(相関)とみるか、仲間の決定の「原因」とみるか |
| 閾値の連鎖 | 友人のうち一定の割合が動くと自分も動く、という規則での連鎖(グラノヴェッター、ワッツ) |
| 注意労働 | AI や国家の装置が意図どおりに動いているかを、人間が確かめる仕事 |
| 捕獲の確率 Π | 国家の装置が、特定の者に乗っ取られる確率 |
