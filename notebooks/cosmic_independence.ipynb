### この文書の見立て

結論の芯を先に書いておく。

作者の議論の中心は、「0 × ∞ は評価できない、だから贅沢品の量は決められない」という部分である。この文書は、これを次のように読み替える。

> 決められないのは、期待値の計算ではない。**贅沢品が将来もたらすものについての信念が、一つに決まらない**のである。信念が一つに決まれば、最適な配分も一つに決まる。信念が区間でしか言えないなら、「どの信念のもとでも他の配分に負けない配分」は、ちょうどその区間全体になる。区間のどこを選ぶかは、データでは決まらず、曖昧さへの態度(悲観か楽観か)で決まる。

これは、経済学で**ナイト的不確実性**(確率そのものがわからない不確実性。Knight 1921)と呼ばれるものの、Bewley(2002)の扱い方である。そして、作者の「全体主義は必需品への偏り、抵抗権・革命権は贅沢品」は、**曖昧さに悲観的な社会と楽観的な社会**の違いとして、正確に読める。

### 答えの見通し

結論を先に書いておく(詳しくは各章と、第 9 章のまとめで)。

1. **絶滅の確率が 1 になるのは、破局の確率が毎期ある正の値より下がらないとき**である(第 2 章、Lean)。平均的に人口が増えていても、破局が一定の率で来れば、絶滅は確実になる。
2. **「確率 0 でも望みがある」は、破局の確率の総和が有限になりうること**と読める(第 3 章、Lean)。望みは、時間を引き伸ばすことではなく、発明が破局の確率を下げ続けることから来る。
3. **「0 × ∞ に値はない」は正しいが、それは仮定を置けば決まるということでもある**(第 4 章、Lean)。無限は値打ちを比で測るから現れ、差で測れば有限である。決まらないのは、信念のほうである。
4. **信念が区間でしか言えないとき、支配されない配分はちょうどその区間になる**(第 5 章、Lean)。これが「独立性」の正体である。悲観的な社会(全体主義)は区間の下端、楽観的な社会(抵抗権・革命権)は上端を選ぶ。どちらも区間の中なら咎められない。贅沢品を探索と読むと、探索の量についても同じ構造が現れる(第 6 章)。
5. **国が競うと、配分は相手に依存し、似てくる**(第 7 章、Lean)。時代ごとに勝つ配分が変わる環境では、競争のある世界はそろって時代を追い、多様性を失う。
6. **「経済 = 学習」の模型では、贅沢品(自己教師あり学習)を少しでもすれば、そこから先は平らな領域になる**(第 8 章)。全振りが選ばれうるのは、未知の危機を強く見込み、平時を軽く見るときだけである。析出の値打ちは、危機で少ない試行から学べることにある。危機の近傍では信念の区間が縮み、配分も一点に決まる(Lean)。独立性は平時の性質である。

**Claude Sonnet 4 の版**(「必需品と贅沢品の最適配分に関する数理的分析 ── 不確実性下での資源配分問題」)は、道具が多い。人口・資本・知識の確率微分方程式、ギルサノフの定理、裾の重い分布、ミニマックス、CVaR、ナッシュ均衡、モンテカルロの数値例。しかし、中身には、はっきりした誤りがある。

1. **定理 1**(生存確率の下界)の証明は「ギルサノフの定理を用いて、確率測度変換により導出される。詳細は付録に記載」とあるが、**付録はない**。ハザード関数 φ の積分で生存確率を書くのは、定義をそう置いただけで、ギルサノフの定理(確率過程の測度を取り替える定理)とは関係がない。
2. **定理 2**(配分最適化の非決定性)の証明は、「目的関数の L に関する期待値が発散するため、一階条件が定義されない。したがって、α ∈ (0, α_max) の任意の値が『最適』となりうる」である。これは**誤り**である。期待値がすべての α > 0 で +∞ に発散するなら、すべての α > 0 で目的関数が +∞ になる。そのとき言えるのは「どれも最適」ではなく、「**期待値では比べられない**」である。比べられないことと、どれも最適であることは違う(第 5 章で、「比べられない」の正しい扱い方を示す)。
3. **第 8 節の数値例**は、「Monte Carlo シミュレーション(10,000 回試行)の結果: 単独社会 α* ∈ [0.2, 0.8]、競争環境 α* ≈ 0.45 ± 0.05」とあるが、**計算の跡(コード)がない**。パラメータの式に「0.011{B<0.3N}」のような、意味の取れない項もある。この数字は、計算された結果ではなく、それらしく書かれたものと考えるほかない。

それでも、Claude の版の結論のうち、「単独の社会では価値観に基づく配分の選択が合理的、国際競争の下では戦略的な相互依存が生じる」と、「期待効用の最大化が困難な場合、ミニマックスなどのロバストな戦略が有効」は、方向として正しい。この文書は、その方向を、誤りのない形で示し直す。

作者は、2025 年に「ChatGPT さんにうかがったところ、共に(ほぼ)正しい議論だとのこと」と書いている。四つの AI が「(ほぼ)正しい」と認めたものに、証明のない定理と、誤った証明と、計算の跡のない数字が含まれていた。**AI が「できる」と言ったことを、別の AI に「正しい」と言わせても、確かめたことにはならない。** この文書で Lean を使うのは、そのためでもある。Lean は、証明が通らなければ通らないと言う。

### 1.2 第 1 段: 財の二分法

作者は、財を必需品・贅沢品・原料に分け、原料は「人々が求めるものではない」(中間財)として除く。そして、知的財産は贅沢品から生まれるとみなす。

ノート 2 の第 8.3 節では、作者が混同しがちだという「贅沢品・嗜好品・知財」を、余剰の三つの使い道として分けた。

| | 余剰をどうするか | 例 |
|---|---|---|
| 贅沢品 | 使い切る(消費) | 旅行、外食 |
| 嗜好品・文化資本 | 貯める(残る) | 美術品、技芸、蔵書 |
| 知財 | 探す(当たれば、新しい経済が育つ) | 研究、発明 |

この文書の「贅沢品」は、作者の二分法に従い、**必需品でないものすべて**を指す。そのうち、議論の中心になるのは三つ目、**探索**としての贅沢品である。作者の議論で贅沢品が値打ちを持つのは、それが「確率 0 を生き延びる発明」をもたらしうるからであり、それは探索の働きだからである。第 6 章で、贅沢品を探索として扱う。

### 1.3 第 2 段: 中位存続

作者は、人類・生命の目的を「ある程度の数を存続させること」(中位存続)とした。この文書では、これを**生き延びる確率**で測る。人口の大きさそのものではなく、0 にならないこと(絶滅しないこと)を目的とする。

ただし、作者は続けて、「中位存続がほぼできてはじめて『贅沢をする』ことが次の目的となるように思うかもしれない。しかし、そうだろうか」と問う。つまり、**必需品が先、贅沢品があと、という順番(辞書式の優先)を疑う**。この問いへの答えが、第 3〜5 章である。贅沢品は、生き延びる確率そのものを変えうる(第 3 章)。だから、必需品の下限を満たしたあとの贅沢品は、「余裕があればする」ものではなく、生き延びることそのものに関わる。ただ、その関わりの大きさが、一つに決まらない(第 5 章)。

## 第 2 章 絶滅の確率 1(第 3 段)

> 中位存続には不確実性下でという条件がある。このような不確実性下…つまりランダムな変動を許す中…で有限値が無限の時間、0 に達することなく維持されるということはほぼありえない。つまり、数学的には絶滅の確率は 1 である。

### 2.1 いつ確率 1 になるか

2025 年の Gemini の版は、これを「ギャンブラーの破産」(吸収状態のあるランダムウォークは、いずれ 0 に達する)として述べた。2026 年に Claude Sonnet 4.6 は、それには条件が要ると指摘した。人口の増え方の平均(増分の期待値)が正なら、ランダムウォークは確率 1 では 0 に達しない。

この指摘は正しい。ただし、作者の言う「不確実性」には、もう一つの読み方がある。**平均的な増え方とは別に、一定の確率で起きる破局**(戦争、疫病、天体の衝突)である。人口が平均的に増えていても、毎期、少なくとも確率 ε > 0 で全滅に近い破局が起きるなら、n 期生き延びる確率は (1 − ε)ⁿ 以下になり、0 に向かう。これは、平均の増え方が正でも成り立つ。作者の「ランダムな変動を許す中」は、こちらの読み方のほうが近いと思う。

これを Lean で書く。ここから、Lean のファイル `independence.lean` を書きはじめる。

`tendsto_pow_atTop_nhds_zero_of_lt_one` は Mathlib の定理で、「0 ≤ r < 1 なら rⁿ → 0」を言う。ここでは r = 1 − ε とした。仮定は二つ、0 < ε(破局の確率が正)と ε ≤ 1 である。**破局の確率が、どれだけ小さくても、正である限り**、生き延びる確率は 0 に向かう。

### 2.2 人口の模型で確かめる

人口の模型として、**分岐過程**(各人が平均 m 人の子を残す)に、破局を足したものを使う。人口には上限(環境の収容力)があるとする。2000 本の歴史を並べて回し、まだ生き延びている歴史の割合を見る。

````python
%%writefile extinction.py
# 人口の模型: 環境の揺らぎ(破局)のある分岐過程。
# 毎期、各人は平均 m 人の子を残す(ポアソン分布)。さらに、確率 h_t で破局が起き、そのときは人口が割合 s まで減る。
# 破局の確率 h_t が一定なら、平均の子の数が 1 を超えていても、いずれ絶滅する(人口は上限 cap で頭打ちになる)。
import numpy as np


def simulate(T=2000, n_paths=2000, N0=50, m=1.05, cap=1000, hazard=lambda t: 0.01, s=0.0, seed=0):
    """n_paths 本の歴史を同時に回し、各期の「まだ生き延びている割合」を返す。
    hazard(t): その期の破局の確率。s: 破局のときに残る割合(0 なら全滅)。"""
    rng = np.random.default_rng(seed)
    N = np.full(n_paths, N0)
    alive = np.zeros(T)
    for t in range(T):
        N = rng.poisson(m * np.minimum(N, cap))
        hit = rng.random(n_paths) < hazard(t)
        N = np.where(hit, rng.binomial(N, s) if s > 0 else 0, N)
        alive[t] = (N > 0).mean()
    return alive
````

````python
import extinction as ex
cases = [('m = 1.05, no catastrophe', dict(hazard=lambda t: 0.0), PAL[0]),
         ('m = 1.00, no catastrophe', dict(m=1.0, hazard=lambda t: 0.0), PAL[1]),
         ('m = 0.98, no catastrophe', dict(m=0.98, hazard=lambda t: 0.0), PAL[3]),
         ('m = 1.05, wipe-out with prob. 0.01', dict(hazard=lambda t: 0.01), PAL[2]),
         ('m = 1.05, 95% loss with prob. 0.02', dict(hazard=lambda t: 0.02, s=0.05), PAL[4])]
fig, ax = plt.subplots(figsize=(7.2, 3.8))
for lab, kw, c in cases:
    a = ex.simulate(**kw)
    ax.plot(a, color=c, label=lab)
    print(f'{lab:38s}  生き延びている割合: 100 期 {a[99]:.3f}   500 期 {a[499]:.3f}   2000 期 {a[-1]:.3f}')
ax.set(xlabel='period', ylabel='share of histories still alive', title='Extinction under uncertainty (2000 histories)')
ax.legend(fontsize=8); plt.show()
````

結果は、作者の主張と、2026 年の Claude の注意の両方を裏付ける。

- **平均の増え方が 1 を下回る(m = 0.98)か、ちょうど 1(m = 1.00)なら、破局がなくても、絶滅は確実**である。m = 1 では、ゆっくりだが確実に 0 に向かう(ランダムウォークの再帰性)。
- **平均の増え方が 1 を上回り(m = 1.05)、破局がなければ、ほとんどの歴史は生き延び続ける**。Claude の注意どおり、ギャンブラーの破産だけでは、絶滅の確率 1 は言えない。
- **しかし、平均の増え方が 1 を上回っていても、一定の確率で破局が来れば、絶滅は確実**になる。全滅でなくても、95% を失う破局が繰り返されれば、小さくなったときに偶然の揺らぎで消える。

作者の「数学的には絶滅の確率は 1」は、**破局の確率が、毎期、ある正の値より下がらない**という条件のもとで正しい。この条件は、人類が自分の力で変えられないもの(天体の衝突、宇宙線、太陽の寿命)を考えれば、もっともらしい。そして、次の章で見るとおり、この条件が崩れるところに、作者の「望み」がある。

### 2.3 終末論法との関係

作者の「絶滅の確率は 1」は、2.1 節のような、不確実性のもとでの人口の揺らぎ(ランダムウォーク)の議論から出ている。しかし、作者自身の回想によれば、その背後には、科学哲学の**終末論法**(Doomsday Argument)の影響も少しある。作者は、同じ『宗教学雑考集』の《終末論法を論駁する》で、これを次のように要約している。

> 「我々は多数派にいる」というコペルニクス原理がある。すると、我々が人間の総数の最後の 95% にいると仮定しても良さそうだ。すると、人類がこれまでに 600億人が生まれたと仮定すれば、95%の信頼度で、人類の総数は 1兆2000億以下と推定される。(…)人口が 100億人で安定し、寿命が 80年で安定するなら(…)この論は 95%の信頼度で、人類は 9120年以内に消滅すると予測する。

終末論法(Carter、Leslie 1996、Gott 1993)は、人口の動きの模型を使わない。「自分は、これまでに生まれた人間の中で、ありふれた順番に生まれたはずだ」という自己位置の推論(Bostrom 2002 の言う「自己標本化の仮定」SSA)だけから、人類の総数の上限を出す。2.1 節の議論(破局の確率が下がらなければ絶滅は確実)とは、**根拠がまったく違う**。

作者は、この論法を、三浦俊彦『多宇宙と輪廻転生』を参考に、次のように論駁した。

> もし、「人類」が、多宇宙の中に存在する場合、その数は 600億人ではなく、600兆人やもっと多く人数の選択から「現在」が選ばれることもありえる。そういう巨大な可能性に対して、600億という「少ない」数の仮説を選ぶのは元々、確率が小さいのだ。

これは、哲学の文献で**自己指示の仮定**(SIA: 自分が存在するという事実そのものが、観測者の多い世界の仮説を支持する。Dieks 1992、Bostrom 2002)と呼ばれる反論と同じ形である。SIA をとると、終末論法の結論は打ち消される。

そして作者は、こう続けた。

> ただ、逆に言えば、多宇宙を仮定できないなら、こういうことは言えない。むしろ、我々が 9120年を超えて生き延びるのは奇跡なのだ。…とすると、第1章(我思)《必需品と贅沢品の宇宙的独立関係》の、「無限の宇宙に比べたとき、確率を計算すれば 0 でも、有限な生存時間は無限にまぎれ、生き残る「可能性」はありうる。」という方向で論を組み立てるべきなのかもしれない。

#### SSA と SIA を Lean で確かめる

終末論法と、作者の論駁の芯を、二つの仮説の場合に Lean で証明する。仮説は「小さい人類」(総数 Ns、事前確率 p)と「大きい人類」(総数 Nl > Ns、事前確率 1 − p)。観測は「自分の順番は r で、r ≤ Ns」(どちらの仮説とも両立する)。

- **SSA**(自己標本化の仮定): 自分は、その世界の観測者から一様に選ばれた一人。順番 r の尤度は 1/N。
- **SIA**(自己指示の仮定): さらに、観測者の多い世界ほど自分がいる確率が高いとして、事前確率に N を掛けてから、SSA と同じ尤度を掛ける。

````python
%%writefile -a independence.lean
/-! ## 第 2.3 節 終末論法: 自己標本化の仮定(SSA)と自己指示の仮定(SIA)

仮説は二つ。「小さい人類」(観測者の総数 Ns)の事前確率 p と、「大きい人類」(総数 Nl > Ns)の事前確率 1 − p。
観測は「自分の生まれた順番が r で、r ≤ Ns」(どちらの仮説とも両立する)。

* SSA: 自分は、その世界の観測者から一様に選ばれた一人とみなす。順番 r の尤度は 1 / N。
* SIA: さらに、観測者の多い世界ほど、自分がいる確率が高いとみなす。事前確率に N を掛けてから、SSA と同じ尤度を掛ける。 -/

/-- SSA のもとでの、小さい人類の事後確率。 -/
noncomputable def ssaPost (p Ns Nl : ℝ) : ℝ := (p / Ns) / (p / Ns + (1 - p) / Nl)

/-- SIA のもとでの、小さい人類の事後確率。 -/
noncomputable def siaPost (p Ns Nl : ℝ) : ℝ :=
  (p * Ns * (1 / Ns)) / (p * Ns * (1 / Ns) + (1 - p) * Nl * (1 / Nl))

/-- 終末論法: SSA のもとでは、自分の順番を見ただけで、小さい人類(早い終末)の確率が事前より上がる。 -/
theorem ssa_doomsday_shift {p Ns Nl : ℝ} (hp0 : 0 < p) (hp1 : p < 1) (hs : 0 < Ns) (hsl : Ns < Nl) :
    p < ssaPost p Ns Nl := by
  have hl : 0 < Nl := lt_trans hs hsl
  have e : ssaPost p Ns Nl = p * Nl / (p * Nl + (1 - p) * Ns) := by
    unfold ssaPost; field_simp
  rw [e, lt_div_iff₀ (by nlinarith)]
  nlinarith [mul_pos hp0 (sub_pos.mpr hp1), mul_pos (mul_pos hp0 (sub_pos.mpr hp1)) (sub_pos.mpr hsl)]

/-- 大きい人類の総数が大きくなるほど(無限に近づくほど)、SSA のもとでは、小さい人類の事後確率は 1 に近づく。
第 3 章で、永遠に生き延びる(総数が無限になる)確率が正でも、SSA のもとでは終末論法が別の形で生き残る理由。 -/
theorem ssa_large_world_penalized {p Ns : ℝ} (hp0 : 0 < p) (hs : 0 < Ns) :
    Tendsto (fun Nl : ℝ => ssaPost p Ns Nl) atTop (𝓝 1) := by
  have h1 : Tendsto (fun Nl : ℝ => (1 - p) / Nl) atTop (𝓝 0) := by
    simpa using tendsto_const_nhds.div_atTop tendsto_id
  have h2 : Tendsto (fun Nl : ℝ => p / Ns + (1 - p) / Nl) atTop (𝓝 (p / Ns + 0)) :=
    tendsto_const_nhds.add h1
  have hne : p / Ns + 0 ≠ 0 := by rw [add_zero]; positivity
  have := (tendsto_const_nhds (x := p / Ns)).div h2 hne
  rw [add_zero, div_self (ne_of_gt (div_pos hp0 hs))] at this
  exact this

/-- 自己指示の仮定: SIA のもとでは、自分の順番を見ても、事後確率は事前確率のまま。終末論法は打ち消される。 -/
theorem sia_cancels_doomsday {p Ns Nl : ℝ} (hs : 0 < Ns) (hl : 0 < Nl) :
    siaPost p Ns Nl = p := by
  unfold siaPost
  have a : p * Ns * (1 / Ns) = p := by field_simp
  have b : (1 - p) * Nl * (1 / Nl) = 1 - p := by field_simp
  rw [a, b]; ring_nf

/-- 仮説がいくつあっても同じ: どの仮説とも両立する観測なら、SIA の事後確率は事前確率に一致する。 -/
theorem sia_cancels_general {ι : Type*} (S : Finset ι) (w N : ι → ℝ) (hN : ∀ i ∈ S, 0 < N i)
    (hw : ∑ j ∈ S, w j = 1) (i : ι) (hi : i ∈ S) :
    (w i * N i * (1 / N i)) / ∑ j ∈ S, w j * N j * (1 / N j) = w i := by
  have e : ∀ j ∈ S, w j * N j * (1 / N j) = w j := fun j hj => by
    have := (hN j hj).ne'; field_simp
  rw [Finset.sum_congr rfl e, hw, e i hi, div_one]

/-- 「僭越な哲学者」: SIA は、順番を見る前から、観測者の多い世界を強く支持する。
大きい人類の総数が大きくなるほど、その(順番を見る前の)確率は 1 に近づく。作者の多宇宙の論駁は、この性質を使っている。 -/
theorem sia_presumptuous {p Ns : ℝ} (hp1 : p < 1) (_hs : 0 < Ns) :
    Tendsto (fun Nl : ℝ => (1 - p) * Nl / (p * Ns + (1 - p) * Nl)) atTop (𝓝 1) := by
  have hq : 0 < 1 - p := sub_pos.mpr hp1
  have e : ∀ᶠ Nl : ℝ in atTop, (1 - p) * Nl / (p * Ns + (1 - p) * Nl) = 1 / (p * Ns / ((1 - p) * Nl) + 1) := by
    filter_upwards [eventually_gt_atTop 0] with Nl hNl
    have : (1 - p) * Nl ≠ 0 := by positivity
    field_simp
  rw [tendsto_congr' e]
  have h0 : Tendsto (fun Nl : ℝ => p * Ns / ((1 - p) * Nl)) atTop (𝓝 0) :=
    tendsto_const_nhds.div_atTop (tendsto_id.const_mul_atTop hq)
  have := (tendsto_const_nhds (x := (1 : ℝ))).div (h0.add tendsto_const_nhds) (by norm_num : (0 : ℝ) + 1 ≠ 0)
  rw [zero_add, div_one] at this
  exact this
````

- `ssa_doomsday_shift`: SSA のもとでは、自分の順番を見ただけで、小さい人類(早い終末)の事後確率が事前より上がる。終末論法そのものである。
- `sia_cancels_doomsday`、`sia_cancels_general`: SIA のもとでは、事後確率は事前確率のまま。総数 N を掛ける重みと、尤度 1/N が、ちょうど打ち消し合う。仮説がいくつあっても同じである。作者の論駁(「600 億という少ない数の仮説を選ぶのは元々、確率が小さい」)は、この打ち消しにあたる。
- `sia_presumptuous`: その代わり、SIA は、順番を見る前から、観測者の多い世界を強く支持する。大きい人類の総数が大きいほど、その確率は 1 に近づく。Bostrom はこれを「僭越な哲学者」の問題と呼び、SIA の弱点とした(観測者の数だけで宇宙論の仮説を決めてしまう)。作者が「多宇宙を仮定できないなら、こういうことは言えない」と書いたのは、この性質を自覚していたからだと読める。
- `ssa_large_world_penalized`: SSA のもとでは、大きい人類の総数が大きいほど(無限に近づくほど)、小さい人類の事後確率は 1 に近づく。第 3 章で「永遠に生き延びる確率が正」になっても、SSA をとるかぎり、終末論法は別の形で生き残る理由である。

作者の数字で計算してみる。これまでに生まれた人を 600 億人とし、小さい人類の総数を 1 兆 2000 億人、大きい人類を 600 兆人、事前確率を半々とする。

````python
p, Ns, Nl = 0.5, 1.2e12, 6.0e14
ssa = (p / Ns) / (p / Ns + (1 - p) / Nl)
sia = (p * Ns / Ns) / (p * Ns / Ns + (1 - p) * Nl / Nl)
pre_sia = (1 - p) * Nl / (p * Ns + (1 - p) * Nl)
print(f'SSA: 小さい人類(早い終末)の事後確率 {ssa:.4f}(事前 {p})')
print(f'SIA: 小さい人類の事後確率 {sia:.4f}(事前のまま)')
print(f'SIA: 順番を見る前の、大きい人類の確率 {pre_sia:.4f}(僭越な哲学者)')
````

つまり、作者の第 4 段(「確率 0 でも望みがある」)は、終末論法への、多宇宙を使わない答えとしても考えられていた。この文書の第 3 章の読み方 B(破局の確率の総和が有限になりうる)は、その答えの一つになる。終末論法は、人類の総数が有限であることを前提に、その中での自分の位置から推論する。破局の確率の総和が有限で、永遠に生き延びる確率が正なら、人類の総数は正の確率で無限になり、「総数の最後の 95% にいる」という推論の前提そのものが崩れる。ただし、そのとき SSA のもとでは、「総数が無限の世界にいるのに、こんなに早い順番に生まれた」ことが、ひどくありそうにないことになり、終末論法は、別の形で生き残る。この先は、自己位置の推論のどの仮定をとるかの問題で、この文書の範囲を超える。

この文書では、終末論法とは別の、作者がもともと拠った根拠(破局の確率)で議論を進める。

### 2.4 この章のまとめ

- 作者の「絶滅の確率は 1」は、**破局の確率が毎期 ε > 0 以上**という条件のもとで正しい。Lean の定理 `survival_const_tendsto_zero` が、それを示す。
- 平均の増え方が正でも、破局が一定の率で来れば、絶滅は確実になる。2025 年の Gemini の版の「ギャンブラーの破産」は、条件(増え方の平均が 0 以下)が要り、2026 年の Claude の注意は正しかった。作者の言う「ランダムな変動」を、破局として読むほうが、作者の主張をそのまま生かせる。
- 作者の議論の背後には、終末論法の影響もある。作者は、終末論法を多宇宙(自己指示の仮定)で論駁しつつ、多宇宙を仮定しない場合の答えとして、第 4 段の「確率 0 でも望み」を考えていた。

## 第 3 章 確率 0 でも望みがある(第 4 段)

> しかし、有限の集合に対するなら確率 1 は望みがないが、無限の集合ならば確率 1 でも望みがある。
>
> 人類は「有限」と言いつつリアルな存在だから、その生存(時間)の分割はいかようにでもできる。宇宙は、大きさは仮に有限であってもリアルにいかようにも分割できるという意味で、人類の生存時間に対して無限とみることができる(例えば宇宙の実時間を x 軸に概念時間を y 軸にし a を定数として y = tan(a × x) や y = exp(tan(a × x)) などとすればいい)。無限の宇宙に比べたとき、確率を計算すれば 0 でも、有限な生存時間は無限にまぎれ、生き残る「可能性」はありうる。

### 3.1 二つの読み方

**読み方 A(2025 年の Gemini の版): 確率 0 は不可能ではない。** 連続の確率空間では、確率 0 の事象も起こりうる(たとえば、[0, 1] から一様に数を選ぶとき、ちょうど 0.5 を選ぶ確率は 0 だが、選ばれうる)。だから、永遠に生き延びる確率が 0 でも、その「経路」は存在しうる。

これは数学として正しい。しかし、**行動の指針にはならない**。確率 0 の事象は、期待値の計算にも、どんな意思決定の理論にも、重み 0 でしか入らない。「起こりうるが重み 0」の望みは、何をすべきかについて何も言わない。

**読み方 B(この文書): 破局の確率の総和が有限なら、生き延びる確率は正になる。** 第 2 章の条件「破局の確率が毎期 ε 以上」が崩れ、破局の確率が時間とともに減っていくなら、どうなるか。毎期の破局の確率を a₀, a₁, a₂, … とすると、n 期生き延びる確率は ∏ₖ (1 − aₖ) である。そして、

$$ \prod_{k<n} (1 - a_k) \ge 1 - \sum_{k<n} a_k $$

が成り立つ(ワイエルシュトラスの積の不等式)。だから、**破局の確率の総和が 1 より小さく抑えられれば、どれだけ先まで行っても、生き延びる確率は 1 − Σaₖ 以上の正の値に残る**。

Lean で示す。

````python
%%writefile -a independence.lean
/-! ## 第 3 章 確率 0 でも望みがある: 危険が減っていけば、永遠に生き延びる確率は正になりうる

毎期の破局の確率を a k とすると、n 期生き延びる確率は ∏ (1 - a k) で、これは 1 - ∑ a k 以上である
(ワイエルシュトラスの不等式)。だから、危険の総和が 1 より小さく抑えられれば(発明が危険を減らし続ければ)、
生き延びる確率は、どれだけ先でも正のまま残る。 -/

theorem prod_one_sub_ge (a : ℕ → ℝ) (h0 : ∀ k, 0 ≤ a k) (h1 : ∀ k, a k ≤ 1) :
    ∀ n, 1 - ∑ k ∈ range n, a k ≤ ∏ k ∈ range n, (1 - a k) := by
  intro n
  induction n with
  | zero => simp
  | succ n ih =>
    rw [sum_range_succ, prod_range_succ]
    have hS : 0 ≤ ∑ k ∈ range n, a k := sum_nonneg (fun k _ => h0 k)
    have hP : 0 ≤ 1 - a n := by linarith [h1 n]
    nlinarith [mul_le_mul_of_nonneg_right ih hP, h0 n, mul_nonneg hS (h0 n)]

theorem survival_positive_of_summable (a : ℕ → ℝ) (h0 : ∀ k, 0 ≤ a k) (h1 : ∀ k, a k ≤ 1)
    (hs : Summable a) (hlt : ∑' k, a k < 1) :
    ∀ n, 0 < 1 - ∑' k, a k ∧ 1 - ∑' k, a k ≤ ∏ k ∈ range n, (1 - a k) := by
  intro n
  refine ⟨by linarith, ?_⟩
  have := prod_one_sub_ge a h0 h1 n
  have hle : ∑ k ∈ range n, a k ≤ ∑' k, a k := hs.sum_le_tsum _ (fun k _ => h0 k)
  linarith
````

`prod_one_sub_ge` は、n についての帰納法で示す。n 期の不等式 Pₙ ≥ 1 − Sₙ に、正の数 (1 − aₙ) を掛けると、Pₙ₊₁ ≥ (1 − Sₙ)(1 − aₙ) = 1 − Sₙ₊₁ + Sₙaₙ ≥ 1 − Sₙ₊₁ になる(Sₙaₙ ≥ 0)。`survival_positive_of_summable` は、それに「有限の和は全体の和以下」(`Summable.sum_le_tsum`)を組み合わせる。

この不等式は十分条件で、正確な条件はもっと広い。aₖ < 1 のとき、∏(1 − aₖ) が正の値に収束するのは、**Σaₖ が有限であるとき、かつそのときに限る**(無限積の標準的な定理)。総和が 1 を超えていても、有限なら、生き延びる確率は正に残る。

````python
%%writefile -a extinction.py
def survival_curve(hazards):
    """破局の確率の列 h_0, h_1, ... から、各期まで生き延びる確率 ∏ (1 − h_k)。"""
    return np.cumprod(1 - np.asarray(hazards))
````

### 3.2 作者の tan(ax) を、読み方 B で読む

作者は、宇宙の実時間 x を、概念の時間 y = tan(ax) に写した。x が有限の区間 [0, π/(2a)) を動くとき、y は [0, ∞) を動く。有限の宇宙の時間の中に、無限の概念の時間が入る。

読み方 B では、この写像を次のように読める。破局が**概念の時間の一単位ごと**に、確率 λ(y) dy で起きるとする。概念の時間とは、文明が経験する出来事の数、たとえば発明の世代の数である。すると、実時間の終わりまで生き延びる確率は、

$$ \exp\left(-\int_0^\infty \lambda(y)\, dy\right) $$

である。これが正になるのは、∫λ(y)dy が有限のとき、つまり、**概念の時間が進むにつれて、一単位あたりの破局の確率が十分速く下がる**ときである。発明の世代を重ねるごとに、文明が破局に強くなっていく(一世代ごとの破局の確率が、たとえば半分になる)なら、無限の世代を重ねても、破局の確率の総和は有限で、生き延びる確率は正に残る。

つまり、作者の「望み」は、写像そのもの(時間を引き伸ばすこと)から来るのではない。写像だけなら、破局の確率が概念の時間あたり一定のとき、無限の概念の時間のあいだに、確率 1 で破局が来る。**望みは、概念の時間が進むにつれて、破局の確率が下がっていくこと**から来る。そして、破局の確率を下げるのは、発明、つまり贅沢品から生まれる知財である。

ボレル–カンテリの補題も、同じことを別の形で言う。独立な事象 Aₖ(k 期目の破局)の確率の総和が有限なら、無限に多くの Aₖ が起きる確率は 0 である(破局は、確率 1 で有限回しか起きない)。総和が無限で、事象が独立なら、確率 1 で無限回起きる。

### 3.3 贅沢品の値打ちは、どこにあるか

この読み方から、作者の第 5 段(「確率 0 を生き延びる発明の価値は、0 の逆数で無限」)を、次のように言い直せる。

> 贅沢品(知財)の値打ちは、**破局の確率の列を、総和が発散する列から、総和が有限な列へ変えること**にある。それができれば、永遠に生き延びる確率は、0 から正の値に変わる。

これは、はっきりした命題で、無限も、0 の逆数も使わない。次の章では、作者の「0 × ∞」に戻り、この言い直しから何が見えるかを考える。

### 3.4 この章のまとめ

- 「確率 0 でも望みがある」を、測度論の「確率 0 ≠ 不可能」として読むと、正しいが、行動の指針にならない。
- **破局の確率の総和が有限なら、永遠に生き延びる確率は正**である。総和が 1 より小さければ、下界 1 − 総和が成り立つ(Lean の `prod_one_sub_ge`、`survival_positive_of_summable`)。正確な条件は、総和が有限であること。
- 作者の tan(ax) の写像は、時間を引き伸ばすだけでは望みを生まない。望みは、概念の時間(発明の世代)が進むにつれて、破局の確率が下がることから来る。
- 贅沢品の値打ちは、破局の確率の列を、総和が発散する列から有限の列へ変えることにある。

## 第 4 章 無限の価値と 0 × ∞(第 5 段)

> そのような確率 0 を生き延びるような発明を持たらす知的財産があるとすれば、その価値は 0 の逆数で無限になる。
>
> しかし、だからといって、贅沢品に全賭けするのがいいかというとそうではない。次の時間に無限の価値がある知財が得られる確率はまたほぼ 0 だろうからである。
>
> すると 0 と無限をかけて現在価値というのを考えると、その平均は決まらないということになろう。評価できないのである。

### 4.1 「0 × ∞」には値がない

2026 年の Claude Sonnet 4.6 は、2025 年の Gemini の版の 1/0 を、極限として書き直した。発明がないときの生存確率を ε、あるときを q > ε とし、発明の値打ちを V = W(q − ε)/ε(W は人類の全価値)とする。発明が得られる確率を p ∼ Cε^r とすると、期待値は

$$ p \cdot V \sim C W \varepsilon^{r-1} $$

で、ε → 0 のとき、r > 1 なら 0、r = 1 なら有限の値 CW、r < 1 なら無限大に向かう。r の値は原理的にわからない。これが「評価できない」の数学的な実体だ、というのが Claude Sonnet 4.6 の議論だった。

この「0 × ∞ に決まった値はない」を、Lean で示す。

````python
%%writefile -a independence.lean
/-! ## 第 4 章 「0 × ∞」には値がない

確率が 0 に向かい、価値が無限に向かうとき、その積(期待値)の行き先は、向かい方しだいで何にでもなる。
とくに、「0 × ∞ の値」として決まった一つの数は存在しない。 -/

theorem zero_mul_infty_no_value :
    ¬ ∃ c : ℝ, ∀ p V : ℕ → ℝ, Tendsto p atTop (𝓝 0) → Tendsto V atTop atTop →
      Tendsto (fun n => p n * V n) atTop (𝓝 c) := by
  rintro ⟨c, hc⟩
  have hV : Tendsto (fun n : ℕ => (n : ℝ) + 1) atTop atTop :=
    tendsto_atTop_add_const_right _ 1 tendsto_natCast_atTop_atTop
  have hp1 : Tendsto (fun n : ℕ => 1 / ((n : ℝ) + 1)) atTop (𝓝 0) :=
    tendsto_one_div_add_atTop_nhds_zero_nat
  have hp2 : Tendsto (fun n : ℕ => 2 * (1 / ((n : ℝ) + 1))) atTop (𝓝 0) := by
    simpa using hp1.const_mul 2
  have e1 : (fun n : ℕ => 1 / ((n : ℝ) + 1) * ((n : ℝ) + 1)) = fun _ => (1 : ℝ) := by
    funext n; field_simp
  have e2 : (fun n : ℕ => 2 * (1 / ((n : ℝ) + 1)) * ((n : ℝ) + 1)) = fun _ => (2 : ℝ) := by
    funext n; field_simp
  have l1 := hc _ _ hp1 hV
  have l2 := hc _ _ hp2 hV
  rw [e1] at l1
  rw [e2] at l2
  have : (1 : ℝ) = c := tendsto_nhds_unique tendsto_const_nhds l1
  have : (2 : ℝ) = c := tendsto_nhds_unique tendsto_const_nhds l2
  linarith

/-- 収束の速さを決める指数 r しだいで、期待値 ε^r · (1/ε) = ε^(r-1) は 0 にも、有限の値にも、無限にもなる。
ここでは r = 2(0 に行く)と r = 1(1 にとどまる)を示す。 -/
theorem expectation_depends_on_rate :
    Tendsto (fun n : ℕ => (1 / ((n : ℝ) + 1)) ^ 2 * ((n : ℝ) + 1)) atTop (𝓝 0) ∧
    Tendsto (fun n : ℕ => (1 / ((n : ℝ) + 1)) ^ 1 * ((n : ℝ) + 1)) atTop (𝓝 1) := by
  constructor
  · have e : (fun n : ℕ => (1 / ((n : ℝ) + 1)) ^ 2 * ((n : ℝ) + 1)) = fun n : ℕ => 1 / ((n : ℝ) + 1) := by
      funext n; field_simp
    rw [e]; exact tendsto_one_div_add_atTop_nhds_zero_nat
  · have e : (fun n : ℕ => (1 / ((n : ℝ) + 1)) ^ 1 * ((n : ℝ) + 1)) = fun _ => (1 : ℝ) := by
      funext n; field_simp
    rw [e]; exact tendsto_const_nhds
````

`zero_mul_infty_no_value` は、「0 に向かう列 p と、無限に向かう列 V の積が、どんな組でも同じ値 c に向かう」ような c は存在しない、と言う。証明は、二つの組を示すだけである。p = 1/(n+1)、V = n+1 なら積はいつも 1、p = 2/(n+1) なら 2。両方が c に向かうなら c = 1 かつ c = 2 で、矛盾する。`expectation_depends_on_rate` は、r = 2 なら 0 に、r = 1 なら 1 に向かうことを示す。

### 4.2 しかし、それは「仮定を一つ置けば決まる」ということでもある

この結果を、もう一度よく見る。積の行き先が決まらないのは、**p と V の向かい方を指定していない**からである。向かい方(r)を一つ指定すれば、行き先は一つに決まる。「0 × ∞ は評価できない」は、「r がわからない限り評価できない」ということで、言い換えれば、**評価できないのは r についての知識が足りないから**である。

これは、作者の議論の芯が、実は期待値の計算ではなく、**r(あるいは、贅沢品が将来もたらすもの)についての信念**にあることを示している。この点を、次の 4.3 節で、別の角度から確かめる。

### 4.3 無限は「比」で測るから現れる

作者は、発明の価値を「0 の逆数」とした。第 3 章の言葉で書き直すと、こうなる。発明がなければ、破局の確率の総和は発散し、永遠に生き延びる確率は S₀ = 0。発明があれば、総和は有限になり、S₁ > 0。

- 発明の値打ちを**比** S₁ / S₀ で測ると、S₀ = 0 なので無限になる。作者の「0 の逆数」は、これである。
- 発明の値打ちを**差** S₁ − S₀ = S₁ で測ると、有限である。

発明が確率 p で得られるなら、贅沢品の期待値は、差で測れば p × S₁ × W で、有限である。**0 × ∞ は現れない**。無限は、値打ちを「生き延びる確率が何倍になるか」で測ったときにだけ現れる。「生き延びる確率がどれだけ増えるか」で測れば、すべて有限である。

期待効用の理論では、値打ちは差(効用の増分 × 確率)で測る。比で測ると、元の確率がほぼ 0 のものほど、ほんの少しの改善が巨大な値打ちに見える。これは計算の誤りではないが、行動の指針としては危うい。たとえば、確率 10⁻¹⁰⁰ の生存を 10⁻⁹⁹ にする手段は、比で測れば「生存確率を 10 倍にする」大発明に見えるが、差で測ればほぼ 0 である。

````python
# 4.3 節の例: 比で測る値打ちと、差で測る値打ち
for S0, S1 in ((1e-100, 1e-99), (1e-6, 1e-3), (0.01, 0.5), (0.0, 0.3)):
    ratio = S1 / S0 if S0 > 0 else float('inf')
    print(f'発明なしの生存確率 {S0:8.0e}、ありのとき {S1:8.0e}:  比 {ratio:10.3g}   差 {S1 - S0:10.3g}')
````

### 4.4 それでも残るもの: 裾の重い不確実性

では、差で測れば、すべて決まるか。そうとは限らない。気候変動の経済学で、Weitzman(2009)は「陰鬱な定理」(dismal theorem)と呼ばれる結果を示した。破局の大きさの確率分布の裾が重く、しかもその裾の重さ自体が過去のデータから推定されなければならないとき、期待効用は(差で測っても)発散しうる。そのとき、どれだけ払って破局を避けるべきかは、期待効用の計算では決まらない。

2025 年の Claude Sonnet 4 の版の補題 1(価値の分布の裾が P(V > v) ∼ v^(−β)、β ≤ 1 なら期待値は発散する)は、この方向の正しい事実である。ただし、第 1 章で見たとおり、そこから「どの α も最適」を導いたのは誤りだった。期待値が発散するなら、言えるのは「期待値では比べられない」である。

次の章では、「比べられない」を正しく扱う方法を示す。それが、この文書の中心の結果になる。

### 4.5 この章のまとめ

- 「0 × ∞ に決まった値はない」は正しい(Lean の `zero_mul_infty_no_value`、`expectation_depends_on_rate`)。
- しかし、それは「0 と ∞ への向かい方(r)を指定すれば決まる」ということで、**評価できないのは、r(贅沢品が将来もたらすもの)についての信念が足りないから**である。
- 作者の「0 の逆数で無限の価値」は、値打ちを**比**で測るから現れる。**差**で測れば有限で、0 × ∞ は現れない。
- それでも、裾の重い不確実性(Weitzman 2009)のもとでは、差で測っても期待値が発散しうる。そのときに言えるのは「どれも最適」ではなく「期待値では比べられない」である。

## 第 5 章 独立性の正体: 信念の区間(第 6 段と第 7 段の前半)

> そこから、贅沢品を必需品と比べてどれぐらい作るべきかというのは、決し難い…独立であると考えるのだ。もちろん、ある程度、必需品がなければ生きられないから、必需品の必要量というのはありうる。しかし、それを超えてどれほど贅沢品を作るかというのは、決められることがらではないのだ。
>
> 軍の力をとても強くしがちな全体主義は必需品への偏りである。一方、言葉は悪いが、抵抗権・革命権は贅沢品にあたる。

この章が、この文書の中心である。第 4 章で見たとおり、「決められない」のは、期待値の計算ではなく、贅沢品が将来もたらすものについての信念である。この章では、それを、できるだけ簡単な模型で、Lean で証明する。

### 5.1 模型

社会は、資源のうち割合 a を贅沢品に、残り 1 − a を必需品に回す。贅沢品の値打ちについての信念を θ ≥ 0 で表す。θ は、「贅沢品(知財)が、将来、どれだけのものをもたらすか」の見込みで、第 3 章の言葉で言えば、贅沢品が破局の確率をどれだけ下げるかである。θ は、事前には確かめられない。

信念 θ のもとでの社会の値打ちを、

$$ F(\theta, a) = \theta a - \frac{a^2}{2} $$

とする。第一項は、贅沢品の見込みの値打ち(θ に比例する)。第二項は、必需品を削る痛みで、削るほど痛みは速く増える(必需品の限界的な値打ちは、残りが少ないほど大きい)。いちばん簡単な形だが、これで議論の構造は尽くせる。第 5.6 節で、生き延びる確率そのものを使う模型でも同じことを確かめる。

### 5.2 信念が決まれば、配分は一つに決まる

θ がわかっていれば、F(θ, a) を最大にする a はただ一つ、a = θ である。F(θ, θ) − F(θ, b) = (b − θ)²/2 > 0(b ≠ θ)だからである。

````python
%%writefile -a independence.lean
/-! ## 第 5 章(2) 独立性の正体: どの配分も、ある信念のもとでは最適である

贅沢品の割合を a、贅沢品の値打ちの「見込み」を θ ≥ 0 とする(θ は、贅沢品が将来もたらすものへの信念で、
事前には確かめられない)。社会の目的を、贅沢品の見込みの値打ち θ·a から、必需品を削る痛み a²/2 を引いたものとする。 -/

/-- 信念 θ のもとでの、贅沢品の割合 a の値打ち。 -/
noncomputable def F (θ a : ℝ) : ℝ := θ * a - a ^ 2 / 2

/-- 信念 θ がわかっていれば、最適な割合はただ一つ(a = θ)に決まる。 -/
theorem unique_optimum (θ b : ℝ) (hb : b ≠ θ) : F θ b < F θ θ := by
  unfold F
  have : 0 < (b - θ) ^ 2 := by positivity
  nlinarith

/-- どの割合 a ≥ 0 も、ある信念 θ ≥ 0 のもとでは最適である(どの配分も合理化できる)。 -/
theorem every_allocation_rationalizable (a : ℝ) (ha : 0 ≤ a) :
    ∃ θ, 0 ≤ θ ∧ ∀ b, F θ b ≤ F θ a :=
  ⟨a, ha, fun b => by unfold F; nlinarith [sq_nonneg (b - a)]⟩
````

`unique_optimum` は、信念 θ がわかっていれば、最適な配分は θ 以外にないことを言う。**信念が一つに決まれば、「決められない」ことは何もない。** これは、第 4 章の「r を指定すれば期待値は決まる」と同じことである。

`every_allocation_rationalizable` は、その裏返しで、**どの配分 a ≥ 0 も、ある信念(θ = a)のもとでは最適である**ことを言う。ある社会の配分を見て、「その配分は間違っている」と言うには、その社会の信念が間違っていると言わなければならない。そして、信念は事前には確かめられない。

### 5.3 比べられないことを、正しく扱う

では、信念が一つに決まらず、区間 [θlo, θhi] のどこかとしか言えないとき(**ナイト的不確実性**: 確率そのものがわからない不確実性)、社会はどうすればよいか。

2025 年の Claude Sonnet 4 の版は、ここで「どの α も最適」と言った。それは誤りだった(第 1 章)。正しい扱い方の一つは、Bewley(2002)の**不完備な選好**である。

> 配分 b が配分 a を**支配する**とは、区間のどの信念のもとでも、b のほうが a より厳密によいことである。

区間の信念のすべてが b を支持するなら、a を選ぶ理由はない。逆に、区間のある信念が a を支持するなら、a を退ける理由もない。そこで、「支配されない配分」を、合理的に選んでよい配分とする。どの配分が支配されないかは、次の定理で決まる。

````python
%%writefile -a independence.lean
/-! ## 第 5 章(3) ナイト的不確実性のもとでの独立性(Bewley の不完備な選好)

信念がひとつに決まらず、区間 [θlo, θhi] のどこにあるかしかわからないとする。
配分 a が「支配される」とは、別の配分 b が、区間のどの信念のもとでも、a より厳密によいことである。
支配されない配分は、ちょうど区間 [θlo, θhi] であり、そのどれを選んでも、データだけからは咎められない。
これが、「必需品の下限を超えて、贅沢品をどれだけ作るかは決められない」の、もっとも素直な数学的な意味である。 -/

def Dominated (θlo θhi a : ℝ) : Prop :=
  ∃ b, ∀ θ, θlo ≤ θ → θ ≤ θhi → F θ a < F θ b

theorem undominated_iff (θlo θhi a : ℝ) (_h : θlo ≤ θhi) :
    ¬ Dominated θlo θhi a ↔ θlo ≤ a ∧ a ≤ θhi := by
  unfold Dominated F
  constructor
  · intro hnd
    by_contra hout
    apply hnd
    rcases not_and_or.mp hout with hlo | hhi
    · push Not at hlo
      exact ⟨θlo, fun θ h1 _ => by nlinarith⟩
    · push Not at hhi
      exact ⟨θhi, fun θ _ h2 => by nlinarith⟩
  · rintro ⟨h1, h2⟩ ⟨b, hb⟩
    have := hb a h1 h2
    nlinarith [sq_nonneg (b - a)]

/-- 悲観的な社会(最悪の信念で評価する、Gilboa–Schmeidler の maxmin)は、区間の下端 θlo を選ぶ。 -/
theorem maxmin_picks_lower (θlo θhi a : ℝ) (ha : 0 ≤ a) :
    (∀ θ, θlo ≤ θ → θ ≤ θhi → F θlo a ≤ F θ a) ∧ F θlo a ≤ F θlo θlo := by
  refine ⟨fun θ h1 _ => by unfold F; nlinarith, ?_⟩
  unfold F; nlinarith [sq_nonneg (a - θlo)]

/-- 楽観的な社会(最良の信念で評価する)は、区間の上端 θhi を選ぶ。 -/
theorem maxmax_picks_upper (θlo θhi a : ℝ) (ha : 0 ≤ a) :
    (∀ θ, θlo ≤ θ → θ ≤ θhi → F θ a ≤ F θhi a) ∧ F θhi a ≤ F θhi θhi := by
  refine ⟨fun θ _ h2 => by unfold F; nlinarith, ?_⟩
  unfold F; nlinarith [sq_nonneg (a - θhi)]
````

`undominated_iff`: **支配されない配分は、ちょうど区間 [θlo, θhi] である**。

- 区間の中の a は、信念 θ = a のもとで最適なので、どの b にも、その信念のもとでは負けない(支配されない)。
- 区間より下の a(a < θlo)は、b = θlo に支配される。区間のどの信念 θ ≥ θlo のもとでも、F(θ, θlo) − F(θ, a) = (θlo − a)(θ − (θlo + a)/2) > 0 だからである。
- 区間より上の a(a > θhi)は、同じように b = θhi に支配される。

これが、作者の「決められない」の、もっとも素直な数学的な意味だと私は考える。

> **必需品の下限を確保したうえで、贅沢品をどれだけ作るかは、信念の区間の中なら、どれを選んでもデータでは咎められない。区間の外は咎められる。**

作者の「独立」は、「何でもよい」ではない。信念の区間の外の配分(贅沢品を作らなすぎる、作りすぎる)は、退けられる。区間の中では、データは何も言わない。

````python
%%writefile belief.py
# 信念と配分の模型(第 5 章の Lean の定理と同じ形)。
# 贅沢品の割合 a、贅沢品の値打ちの見込み θ のもとでの社会の値打ち F(θ, a) = θ a − a²/2。
# 必需品の下限: a ≤ amax = 1 − Bmin / R。
import numpy as np


def F(theta, a):
    return theta * a - a ** 2 / 2


def undominated(grid, theta_lo, theta_hi, n_theta=201):
    """格子 grid の配分のうち、区間 [θlo, θhi] のどの信念のもとでも厳密に勝つ別の配分がないもの。"""
    thetas = np.linspace(theta_lo, theta_hi, n_theta)
    V = F(thetas[:, None], grid[None, :])                  # 信念 × 配分
    keep = []
    for i in range(len(grid)):
        dom = np.all(V > V[:, [i]] + 1e-12, axis=0)       # どの信念でも厳密に勝つ配分
        keep.append(not dom.any())
    return np.array(keep)
````

````python
import belief as bf
lo, hi = 0.15, 0.35
grid = np.linspace(0, 0.6, 601)
keep = bf.undominated(grid, lo, hi)
print('数値で求めた支配されない配分: [%.3f, %.3f](定理では [θlo, θhi] = [%.2f, %.2f])' % (grid[keep].min(), grid[keep].max(), lo, hi))
fig, ax = plt.subplots(figsize=(7.2, 3.8))
for th, c in zip(np.linspace(lo, hi, 5), PAL):
    ax.plot(grid, bf.F(th, grid), color=c, lw=1.5, label=f'belief θ = {th:.2f}')
    ax.plot([th], [bf.F(th, th)], 'o', color=c)
ax.axvspan(lo, hi, color=PAL[2], alpha=0.12, label='undominated allocations')
ax.set(xlabel='luxury share a', ylabel='F(θ, a) = θa − a²/2', title='Every belief in the interval picks a different allocation', ylim=(-0.05, 0.08))
ax.legend(fontsize=8, loc='lower left'); plt.show()
````

### 5.4 悲観と楽観: 全体主義と抵抗権

区間の中のどれを選ぶかは、データでは決まらない。それを決めるのは、**曖昧さへの態度**である。

- **悲観的な社会**は、区間のうち最悪の信念で評価し、それを最大にする(Gilboa and Schmeidler 1989 の maxmin 期待効用)。F は θ について増えるので(a ≥ 0)、最悪の信念は区間の下端 θlo で、選ぶ配分は a = θlo になる。**贅沢品を区間のいちばん少なく作る。**
- **楽観的な社会**は、最良の信念で評価し、a = θhi を選ぶ。**贅沢品を区間のいちばん多く作る。**

Lean の `maxmin_picks_lower` と `maxmax_picks_upper` が、これを示す。

作者は、「軍の力をとても強くしがちな全体主義は必需品への偏りである。一方、言葉は悪いが、抵抗権・革命権は贅沢品にあたる」と書いた。この模型の言葉では、

> **全体主義は、曖昧さに悲観的な社会の選択**(贅沢品が将来もたらすものを、最悪の見込みで評価する)であり、**自由(抵抗権・革命権)は、曖昧さに楽観的な社会の選択**である。どちらも、区間の中にあるかぎり、データでは咎められない。

悲観と楽観のあいだには、その混ぜ合わせがある。最悪の評価に重み λ、最良の評価に重み 1 − λ を置く(α-maxmin、Ghirardato, Maccheroni and Marinacci 2004)と、選ぶ配分は a = λθlo + (1 − λ)θhi になり、区間のどの点も、ある λ で選ばれる。**λ は、その社会の曖昧さへの態度、つまり政治哲学の選択である。** 2025 年の Gemini の版が「社会の価値観や信念に基づく政治的な決定」と書き、Claude Sonnet 4 の版が「単独社会では価値観に基づく配分選択が合理的」と書いたのは、この意味で正しかった。

### 5.5 必需品の下限

作者は、「もちろん、ある程度、必需品がなければ生きられないから、必需品の必要量というのはありうる」と書いた。全資源を R、必需品の必要量を Bmin とすると、贅沢品の割合には上限 amax = 1 − Bmin / R がある。

````python
%%writefile -a independence.lean
/-! ## 第 5 章(1) 必需品の下限

全資源 R のうち、割合 α を贅沢品に回す。必需品の必要量 Bmin を確保するには α ≤ 1 - Bmin / R。 -/

theorem necessity_floor {R Bmin α : ℝ} (hR : 0 < R) :
    Bmin ≤ (1 - α) * R ↔ α ≤ 1 - Bmin / R := by
  rw [show α ≤ 1 - Bmin / R ↔ Bmin / R ≤ 1 - α by constructor <;> intro h <;> linarith,
    div_le_iff₀ hR]
````

合わせると、**選んでよい配分は、信念の区間 [θlo, θhi] と、下限から来る上限 [0, amax] の重なり**になる。信念の区間が amax を超えて広がっていても、amax より上は選べない。逆に、信念の区間の上端が amax より下なら、必需品の下限は「効かない」(必需品は、下限より多く作られる)。作者が言う「必需品の必要量」と「それを超えてどれほど贅沢品を作るか」の二つが、ここで、上限と区間という二つの別の制約として分かれる。

### 5.6 生き延びる確率の模型で確かめる

F(θ, a) は、いちばん簡単な形だった。目的を、作者の言う「中位存続」、つまり生き延びる確率そのものにしても、同じ構造が出ることを確かめる。毎期の破局の確率は、二つの部分からなるとする。

- **いまの危険**: 必需品の余裕(下限を超えた分)が多いほど下がる。
- **先の危険**: 贅沢品から生まれた知財の蓄積が多いほど下がる。下がり方の強さが θ で、事前にはわからない。

必需品を下限より削れば、その期に滅ぶ。400 期生き延びる確率を、配分 a と信念 θ ごとに計算する。

````python
%%writefile -a belief.py
# 生き延びる確率の模型で、同じことを確かめる。毎期の破局の確率は二つの部分からなる。
#   いまの危険: 必需品の余裕(下限を超えた分)が多いほど下がる   h_n · exp(−γ ((1 − a) R − Bmin))
#   先の危険:   贅沢品から生まれた知財の蓄積 K が多いほど下がる  h_k · exp(−θ K)、 K_{t+1} = K_t + a R
# 必需品が下限を割れば(a > amax)、その期に滅ぶ。θ(知財が先の危険をどれだけ下げるか)は、事前にはわからない。
def survival(a, theta, T=400, R=1.0, Bmin=0.6, h_n=0.004, gamma=8.0, h_k=0.01):
    if (1 - a) * R < Bmin - 1e-12:
        return 0.0
    K, logS = 0.0, 0.0
    for t in range(T):
        h = h_n * np.exp(-gamma * ((1 - a) * R - Bmin)) + h_k * np.exp(-theta * K)
        logS += np.log1p(-min(h, 1.0))
        K += a * R
    return float(np.exp(logS))


def survival_table(alphas, thetas, **kw):
    return np.array([[survival(a, th, **kw) for a in alphas] for th in thetas])
````

````python
importlib.reload(bf)
alphas = np.linspace(0, 0.4, 81)                 # 必需品の下限 0.6 から、amax = 0.4
thetas = np.geomspace(0.003, 0.03, 15)            # 信念の区間
tab = bf.survival_table(alphas, thetas)
best = alphas[tab.argmax(1)]
dom = [np.all(tab > tab[:, [i]] + 1e-15, axis=0).any() for i in range(len(alphas))]
und = alphas[~np.array(dom)]
i_mm, i_mx = tab.min(0).argmax(), tab.max(0).argmax()
print('信念ごとの最適な配分: θ = %.3f で %.3f … θ = %.3f で %.3f' % (thetas[0], best[0], thetas[-1], best[-1]))
print('支配されない配分: [%.3f, %.3f]' % (und.min(), und.max()))
print('悲観(maxmin)の選ぶ配分: %.3f   楽観(maxmax)の選ぶ配分: %.3f' % (alphas[i_mm], alphas[i_mx]))
print('贅沢品なし(a = 0)の生き延びる確率: %.4f   上限いっぱい(a = 0.4): 信念しだいで %.4f 〜 %.4f' % (tab[0, 0], tab[0, -1], tab[-1, -1]))
fig, ax = plt.subplots(figsize=(7.2, 3.8))
for k, c in zip((0, 5, 10, 14), PAL):
    ax.plot(alphas, tab[k], color=c, label=f'θ = {thetas[k]:.3f}')
ax.axvspan(und.min(), und.max(), color=PAL[2], alpha=0.12, label='undominated')
ax.set(xlabel='luxury share a (necessity floor: a ≤ 0.4)', ylabel='P(survive 400 periods)', title='Survival-based version')
ax.legend(fontsize=8); plt.show()
````

結果は、F の模型と同じ構造である。

- 信念 θ ごとに、最適な配分はただ一つに決まる(θ が小さければ少なめ、大きければ多め)。
- 信念が区間でしか言えないと、支配されない配分は区間になる。悲観的な社会はその下のほう、楽観的な社会は上のほうを選ぶ。
- 贅沢品を作らない(a = 0)のも、上限いっぱい作る(a = 0.4)のも、支配される。贅沢品を作らなければ、先の危険が下がらない。上限いっぱい作れば、いまの危険が上がる。

一つ、F の模型にない性質がある。θ をさらに大きくすると(知財が先の危険を非常によく下げるなら)、最適な配分は、かえって**下がっていく**。少しの贅沢品で先の危険を十分に下げられるなら、残りを必需品に回すほうがよいからである。この模型の信念の区間を [0.003, 0.03] に取ったのは、その手前の範囲にするためである。区間をもっと広げると、「楽観的な社会ほど贅沢品を多く作る」とは限らなくなる。それでも、支配されない配分が区間になることは変わらない。

### 5.7 二つの「独立」: 平らさと、比べられなさ

ここまでの「決められない」には、実は二つの違う意味が混ざりうる。区別しておく。

1. **平らさ**: 信念を一つに固定しても、いくつもの配分が**同じ値打ち**になる。「どの配分をとっても期待値が変わらない」は、この意味である。第 8 章の学習の模型(表現の容量が十分なとき)で現れる平らな領域が、これにあたる。作者が 2026 年に考えたプログラム 4(α を固定して比べ、差のない領域を見せる)も、この意味の独立を狙っていた。
2. **比べられなさ**: 信念ごとに最適な配分は一つに決まり、**値打ちは配分によって違う**。しかし、どの信念をとるかが決まらないので、信念の違う最適どうしを、一つの物差しで比べられない。この章の F の模型と、Bewley の定理が示したのは、こちらである。

この章の模型では、どの信念をとっても期待値が変わらない、ということは**言えていない**。信念 θ のもとでは、a = θ が他のどの配分より厳密によい(`unique_optimum`)。言えているのは、「信念 θ₁ のもとでは a₁ がよく、信念 θ₂ のもとでは a₂ がよい。そして、θ₁ と θ₂ のどちらをとるべきかは、データでは決まらない」である。

#### なぜ、最適どうしを比べられないのか

a₁(θ₁ のもとでの最適)と a₂(θ₂ のもとでの最適)を比べるには、θ₁ と θ₂ のどちらを、どれだけ重く見るかの**重み**が要る。重み w があれば、二つの信念の平均 wθ₁ + (1 − w)θ₂ のもとでの値打ちで比べればよい。F は θ について一次なので、平均した値打ちは、平均した信念のもとでの値打ちに一致する。そして、平均した信念のもとでは、最適はただ一つに決まる。

````python
%%writefile -a independence.lean
/-- 信念に重みを付けて平均すると(θ₁ に w、θ₂ に 1 − w)、値打ちは、平均した信念 wθ₁ + (1 − w)θ₂ のもとでの値打ちに一致する。
だから、信念どうしを比べる重み(信念についての信念)があれば、区間はただ一つの信念に縮み、最適も一つに決まる。 -/
theorem mixture_collapses (w θ₁ θ₂ a : ℝ) :
    w * F θ₁ a + (1 - w) * F θ₂ a = F (w * θ₁ + (1 - w) * θ₂) a := by
  unfold F; ring

/-- 区間の中のどの配分 a も、端の信念に適当な重みを付けた平均のもとでは最適になる。
支配されない配分の区間は、「信念に付けうるすべての重み」で選ばれる最適の集まりである。 -/
theorem interval_is_all_weightings (θlo θhi a : ℝ) (h1 : θlo ≤ a) (h2 : a ≤ θhi) (hlt : θlo < θhi) :
    ∃ w, 0 ≤ w ∧ w ≤ 1 ∧ ∀ b, F (w * θlo + (1 - w) * θhi) b ≤ F (w * θlo + (1 - w) * θhi) a := by
  refine ⟨(θhi - a) / (θhi - θlo), ?_, ?_, ?_⟩
  · apply div_nonneg <;> linarith
  · rw [div_le_one (by linarith)]; linarith
  · intro b
    have hne : θhi - θlo ≠ 0 := by linarith
    have hθ : (θhi - a) / (θhi - θlo) * θlo + (1 - (θhi - a) / (θhi - θlo)) * θhi = a := by
      field_simp [hne]; ring
    rw [hθ]; unfold F; nlinarith [sq_nonneg (b - a)]
````

`mixture_collapses` は、信念に重みを付けて平均すると、一つの信念(平均した信念)に縮むことを言う。`interval_is_all_weightings` は、区間の中のどの配分も、端の信念に適当な重みを付ければ最適になることを言う。つまり、**支配されない配分の区間は、「信念に付けうるすべての重み」で選ばれる最適の集まり**である。

だから、「最適どうしを比べられない」の理由は、ただ一つである。**信念にどう重みを付けるか(信念についての信念)が、決まっていない**。重みが決まれば、比べられる(そして、区間は一点に縮む)。ナイト的不確実性とは、この重みがない状態のことである。

では、「わからないなら、均等に重みを付ければよい」(理由不十分の原理)ではいけないのか。これには二つの答えがある。

- **重みの付け方は、言い換えに依存する**。θ について均等に重みを付けるのと、θ² について均等に重みを付けるのとでは、平均が違う。どの変数で均等にするかを決める根拠がない。
- **人は、実際にそうふるまわない**。エルスバーグの壺の実験(Ellsberg 1961)では、確率のわかっている賭けと、わからない賭けを、多くの人が区別し、わからないほうを避ける。それは、一つの重み(確率)で平均する人には起こりえない選び方である。悲観(maxmin)は、この「わからないものを避ける」ふるまいを、まっすぐ表したものである。

#### θ は、信念とも、値打ちの換算率とも読める

作者自身は、この「比べられなさ」を、信念の問題としてよりも、**値打ちの種類の違い**として考えていた節がある。『宗教学雑考集』の別の箇所で、作者は、最適化の価値判断について「本質的に比較可能でないようなものを比較していることもあるかもしれない」と書き、また、発明を「必需と比較可能と見せ、大き過ぎる価値を主張し、他を圧迫し過ぎる」ことを戒めている。さらに、神の視点の会計として、発明のような値打ちを「資産や負債」、細々とした値打ちを「損益」とし、両者を「次元」の違うものとする比喩も書いている。

F の模型の θ は、「贅沢品の値打ちを、必需品の単位に直すときの換算率」とも読める。換算率が一つに決まらない(生存への寄与と、日々の暮らしへの寄与を、同じ物差しで測れない)なら、それは数学的には、信念が一つに決まらないのとまったく同じ構造になる。経済学では、これを**多効用表現**(Dubra, Maccheroni and Ok 2004: 完備性を仮定しない期待効用。いくつもの効用関数のどれでも勝つときだけ、「よい」と言える)と呼ぶ。Bewley の定理(信念の集合)と、Dubra–Maccheroni–Ok の定理(効用の集合)は、同じ形をしている。だから、この章の結論は、**θ を信念と読んでも、換算率と読んでも成り立つ**。作者の直観(値打ちの次元の違い)は、後者の読み方のほうに近い。

### 5.8 2025 年の Claude の定理 2 を、正しく言い直す

2025 年の Claude Sonnet 4 の版の定理 2 は、「テール分布が重く、イノベーション実現確率が小さく、効用関数が分離可能なら、最適配分比率 α* は一意に決定されない」と述べ、その証明は「期待値が発散するので、任意の値が最適になりうる」だった。

正しく言い直すと、次のようになる。

> 贅沢品の値打ちについての信念が、一つの確率分布に決まらず、区間(確率分布の集合)でしか言えないなら、支配されない配分は一点ではなく区間になる。その区間のどこを選ぶかは、曖昧さへの態度による。

期待値の発散は要らない。要るのは、**信念が一つに決まらないこと**だけである。そして、言えるのは「どれも最適」ではなく、「区間の中のどれも支配されない、区間の外は支配される」である。

### 5.9 この章のまとめ

- 信念が一つに決まれば、最適な配分は一つに決まる(Lean の `unique_optimum`)。どの配分も、ある信念のもとでは最適である(`every_allocation_rationalizable`)。
- 信念が区間 [θlo, θhi] でしか言えないとき、**支配されない配分は、ちょうど区間 [θlo, θhi] である**(`undominated_iff`、Bewley 2002)。これが、作者の「贅沢品をどれだけ作るかは決められない」の数学的な正体である。「何でもよい」ではなく、区間の外は退けられる。
- 区間の中のどこを選ぶかは、曖昧さへの態度で決まる。**悲観的な社会(全体主義)は下端、楽観的な社会(抵抗権・革命権)は上端を選ぶ**(`maxmin_picks_lower`、`maxmax_picks_upper`)。
- 必需品の下限は、上限 amax = 1 − Bmin / R として、信念の区間とは別の制約になる(`necessity_floor`)。
- 生き延びる確率そのものを目的にしても、同じ構造が出る。
- この章で言えているのは「平らさ」(信念を固定しても配分によらず同じ値打ち)ではなく、「**比べられなさ**」(信念ごとの最適を、一つの物差しで比べられない)である。比べられないのは、信念に付ける重みが決まっていないからで、重みがあれば区間は一点に縮む(`mixture_collapses`、`interval_is_all_weightings`)。θ は、信念とも、贅沢品と必需品の値打ちの換算率とも読め、どちらでも結論は同じである。

## 第 6 章 探索としての贅沢品

### 6.1 贅沢品 = 探索、必需品 = 活用

第 1.2 節で、作者の議論の中心にある贅沢品は「探索」だと書いた。贅沢品が値打ちを持つのは、それが、まだ知られていない発明(新しい技術、新しい考え方)をもたらしうるからである。

経済学と統計学で、この構造を扱う標準の道具は、**多腕バンディット**の問題である。スロットマシンが何台もあり、それぞれ当たりの確率が違うが、事前にはわからない。わかっている台を引き続けるか(**活用**)、まだ知らない台を試すか(**探索**)。探索は、いまの得を削るが、よりよい台を見つける可能性を買う。Gittins(1979)は、腕ごとに「指数」を計算して、指数の最大の腕を引くのが最適であることを示した。その指数は、まだ試していない腕の値打ちについての**事前の信念**に依存する。

この章では、必需品 = わかっている腕の活用、贅沢品 = 新しい腕の探索、と読む。そして、最適な探索の量が、事前の信念(新しい腕の値打ちの分布、とくに裾の重さ)にどれだけ依存するかを見る。

### 6.2 模型

- 社会は毎期、資源の割合 1 − a を、わかっている最良の技術(値打ち b)での生産に回し、割合 a を探索に回す。
- 探索は、毎期、確率 a で新しい技術を一つ試す。新しい技術の値打ちは、未知の分布(パレート分布。裾の重さ α_tail が小さいほど、とても大きな値打ちの技術が出やすい)から引かれる。よりよい技術が見つかれば、以後はそれを使う。
- 社会の値打ちは、割り引いた生産の合計(割引率 0.97、200 期)。2000 本の歴史の平均で測る。

````python
%%writefile bandit.py
# 探索としての贅沢品: 多腕バンディットの模型。
# 社会は毎期、資源の割合 1 − a を「わかっている腕」(必需品の技術、値打ち 1)に、割合 a を探索(贅沢品)に使う。
# 探索は、毎期、確率 a で新しい腕(新しい技術)を一つ試す。新しい腕の値打ちは、未知の分布(裾の重さ)から引かれる。
# よい腕が見つかれば、以後はそれを「わかっている腕」として使う(技術の進歩)。
# 社会の値打ち = 割り引いた生産の合計。ただし、探索に回した分は、その期の生産から引かれる。
import numpy as np


def run(a, T=200, beta=0.97, draw=None, n_paths=2000, seed=0):
    """探索の割合 a で T 期回したときの、割り引いた生産の合計の平均。draw(rng, n) は新しい腕の値打ちを引く関数。"""
    rng = np.random.default_rng(seed)
    best = np.ones(n_paths)
    total = np.zeros(n_paths)
    for t in range(T):
        total += beta ** t * (1 - a) * best
        found = rng.random(n_paths) < a
        new = draw(rng, n_paths)
        best = np.where(found, np.maximum(best, new), best)
    return total.mean()


def pareto(alpha_tail, scale=0.5):
    """裾の重さ alpha_tail のパレート分布(小さいほど裾が重い)。最小値 scale。"""
    return lambda rng, n: scale * (1 + rng.pareto(alpha_tail, n))


def best_share(draw, grid=np.linspace(0, 0.6, 31), **kw):
    vals = np.array([run(a, draw=draw, **kw) for a in grid])
    return grid[vals.argmax()], vals
````

````python
import bandit as bd
grid = np.linspace(0, 0.6, 31)
fig, ax = plt.subplots(figsize=(7.2, 3.8))
print('新しい技術の値打ちの裾の重さ α_tail と、最適な探索の割合')
for at, c in zip((3.0, 2.0, 1.5, 1.2), PAL):
    a_best, vals = bd.best_share(bd.pareto(at), grid=grid)
    ax.plot(grid, vals / vals[0], color=c, label=f'tail index {at}')
    print(f'  α_tail = {at:3.1f}: 最適な探索の割合 {a_best:.2f}   値打ち(探索なしを 1 として){vals.max() / vals[0]:6.2f}')
ax.set(xlabel='exploration (luxury) share a', ylabel='value relative to a = 0', title='Optimal exploration depends on the believed tail', yscale='log')
ax.legend(fontsize=8); plt.show()
````

裾が軽い(α_tail = 3、とても大きな発明はまず出ない)と、最適な探索の割合は 1 割ほどである。裾が重くなるほど(α_tail = 1.2、とても大きな発明が出うる)、最適な探索の割合は 5 割まで上がる。

**新しい技術の値打ちの分布の裾の重さ**は、事前にはわからない。過去の発明の記録から推定するしかないが、裾の重さの推定は、まさに稀な大発明の数で決まるので、とても不安定である(Weitzman 2009 が気候の破局について指摘したのと同じ問題)。裾の重さについての信念が [1.2, 3] の区間でしか言えないなら、最適な探索の割合も [0.1, 0.5] の区間でしか言えない。第 5 章の「信念の区間 → 支配されない配分の区間」が、探索の量について、そのまま現れる。

### 6.3 裾が重すぎると、期待値そのものが当てにならない

裾の重さが 1 以下(α_tail ≤ 1)になると、新しい技術の値打ちの**平均が無限大**になる。第 4 章の「期待値が発散する」場合である。そのとき、2000 本の歴史の平均で最適な探索の割合を測ると、測るたびに(乱数の種を変えるたびに)答えが大きく変わる。

````python
for at in (3.0, 1.5, 1.0, 0.8):
    res = [bd.best_share(bd.pareto(at), grid=grid, seed=s) for s in range(6)]
    picks = [round(float(r[0]), 2) for r in res]
    v = np.array([r[1][-1] / r[1][0] for r in res])            # a = 0.6 の値打ち(探索なしを 1 として)
    print(f'α_tail = {at:3.1f}: 6 回で測った最適な探索の割合 {picks}   a = 0.6 の値打ちの、6 回での最小 {v.min():9.1f}、最大 {v.max():9.1f}')
````

裾が軽い(3)と、測るたびにほぼ同じ答えが出て、値打ちの測り方も安定している。裾が重くなるほど、最適な探索の割合の測り方がばらつき(1.5 で 0.32〜0.44、0.8 で 0.18〜0.5)、値打ちそのものの測り方は、回ごとに数倍から十倍近く違ってくる(裾 1.0 で 24〜74、0.8 で 159〜1362)。期待値が発散するので、標本の平均は、たまたま引いた最大の大発明に支配される。**期待値で比べること自体が当てにならない**。2025 年の Claude Sonnet 4 の版の補題 1(β ≤ 1 で期待値が発散する)が指していたのは、この現象である。ここから言えるのは、第 5 章のとおり、「期待値では比べられない」であって、「どれも最適」ではない。

### 6.4 贅沢品・嗜好品・知財

ノート 2 の第 8.3 節で、作者が混同しがちな三つを、余剰の三つの使い道として分けた(使い切る贅沢品、貯める嗜好品・文化資本、探す知財)。この章の模型は、そのうち三つ目、**探す**の働きだけを取り出したものである。

作者の二分法(必需品と贅沢品)では、三つはすべて「贅沢品」に入る。作者の議論が贅沢品に値打ちを認めるのは、探すの働きがあるからである。では、使い切る贅沢品(旅行、外食)や、貯める嗜好品(美術品、技芸)は、作者の議論ではどう位置づけられるか。

- 使い切る贅沢品は、それだけでは探索にならない。しかし、人がそれを求めることが、新しい財やサービスを作る誘因になる(ノート 1 で、作者の模型の贅沢品が「労働の受け皿」として働いたように)。間接的な探索である。
- 貯める嗜好品・文化資本は、探索の結果を保存し、のちの探索の材料になる(第 8 章の「析出」で使う共通の核にあたる)。

だから、三つは、探索という一つの働きの、違う段階として並べられる。**探す(知財)、探した結果を残す(嗜好品・文化資本)、探す誘因を作る(使い切る贅沢品)**。作者の「贅沢品」は、その全体を指していたと読める。最終監査で、この区別を、ほかのノートにも広げる。

(贅沢品・嗜好品・知財の区別は、シリーズ全体にかかわるので、ノート 2 の 8.3 節の表にまとめた。この節の三つの読みは、その表の三つの行にあたる)

### 6.5 贅沢品が複数あるとき: 相対的な値打ちは決まるか

ここまでは、贅沢品を一つの量(割合 a)として扱った。贅沢品が何種類もあるとき、信念が決まれば、それぞれの相対的な値打ちも決まってしまうのか。

**模型の中で、信念を一つに固定し、贅沢品どうしが互いに影響しないなら、決まる**。贅沢品 j の値打ちの見込みを θⱼ とすると、それぞれの最適な量は θⱼ で決まり、相対的な値打ちは θⱼ の比で固定される。

しかし、それが崩れる理由が三つある。

1. **贅沢品どうしが影響しあう**。二つの探索が同じ因子を探しているなら(重なりがあるなら)、一方を進めると、もう一方の値打ちは下がる(代わりになる)。逆に、一方が他方の道具になるなら、値打ちは上がる(補いあう)。そのとき、贅沢品 j の値打ちは、θⱼ だけでなく、ほかの贅沢品をどれだけ作るかで変わる。下の計算で示す。
2. **探索すると、信念そのものが変わる**。多腕バンディットでは、腕を引くたびに、その腕についての信念が更新され、ギッティンズ指数(その腕の値打ち)が変わる。相対的な値打ちは、固定ではなく、探索とともに動く。第 8 章の「析出」は、その極端な形である(重なりから共通の核が取り出されると、多くの贅沢品の値打ちがまとめて変わる)。
3. **人によって信念が違う**。ある人は θ₁ を高く、別の人は θ₂ を高く見込む。二人の信念を一つにまとめる仕方は、決まらない。Mongin(1995)は、信念も好みも違う人々について、全員が一致して「よい」と言うものを社会も「よい」とする(パレート原則)規則で、信念と好みをまとめようとすると、一人の独裁者の信念と好みに従うしかないことを示した。人々の信念が違うかぎり、**社会全体の、贅沢品どうしの相対的な値打ちは、一つに決まらない**。

三つ目は、第 5 章の「信念の区間」の、社会の側からの現れである。一人の中では信念が一つでも、社会の中には、信念の違う人々がいる。それぞれが自分の信念で探索するなら、社会全体の探索は、いくつもの信念の混ぜ合わせになる。第 7 章で見た「多様性」(競争がないときの、国のばらつき)の値打ちは、ここにある。作者の「抵抗権・革命権は贅沢品にあたる」は、**多数派の信念に従わずに、自分の信念で探索する権利**とも読める。

````python
# 二つの贅沢品の量 a1, a2 の値打ち: θ1 a1 + θ2 a2 − a1²/2 − a2²/2 − ρ a1 a2(ρ > 0 なら代わりになる、ρ < 0 なら補いあう、ρ = 0 なら影響しあわない)
def best_pair(t1, t2, rho, grid=np.linspace(0, 1, 201)):
    A1, A2 = np.meshgrid(grid, grid, indexing='ij')
    V = t1 * A1 + t2 * A2 - A1 ** 2 / 2 - A2 ** 2 / 2 - rho * A1 * A2
    i, j = np.unravel_index(V.argmax(), V.shape)
    return grid[i], grid[j]
print('信念 (θ1, θ2)   重なり ρ    最適な (a1, a2)')
for t1, t2 in ((0.4, 0.3), (0.3, 0.4)):
    for rho in (-0.5, 0.0, 0.5, 0.9):
        a1, a2 = best_pair(t1, t2, rho)
        print(f'  ({t1}, {t2})      {rho:5.1f}      ({a1:.3f}, {a2:.3f})')
````

影響しあわない(ρ = 0)なら、それぞれの量は見込みのとおり(0.4 と 0.3)で、相対的な値打ちは信念の比で決まる。しかし、互いに代わりになるほど(ρ が大きいほど)、最適な組み合わせは見込みの高いほうに寄り、見込みの低いほうは作られなくなっていく。補いあうなら(ρ < 0)、両方を見込み以上に作る。そして、二人の信念が入れ替われば(θ1 と θ2 のどちらを高く見るか)、最適な組み合わせも入れ替わる。代わりになる度合いが強いほど、その入れ替わりは極端になる。

信念が決まっても、相対的な値打ちは、贅沢品どうしの関係で変わる。そして、信念は、人によって違い、探索とともに動く。だから、**贅沢品の相対的な値打ちは、固定されない**。

### 6.6 この章のまとめ

- 贅沢品を探索、必需品を活用と読むと、作者の問いは、多腕バンディットの「どれだけ探索するか」の問題になる。その答えは、まだ試していない技術の値打ちについての事前の信念に依存する(Gittins 1979)。
- 模型では、新しい技術の値打ちの分布の裾が重いほど、最適な探索の割合は大きくなる(裾の重さ 3 で 1 割、1.2 で 5 割)。裾の重さは事前にはわからないので、**最適な探索の割合も区間でしか言えない**。第 5 章の構造が、探索について現れる。
- 裾の重さが 1 以下では、期待値が発散し、標本の平均で比べること自体が当てにならない(2025 年の Claude の補題 1 の正しい中身)。
- 贅沢品・嗜好品・知財は、探索の三つの段階(探す誘因を作る、探す、探した結果を残す)として並べられる。
- 贅沢品が複数あるとき、信念を固定し、贅沢品どうしが影響しあわないなら、相対的な値打ちは信念で決まる。しかし、贅沢品どうしの重なり(代わりになる、補いあう)、探索による信念の更新、人による信念の違い(Mongin 1995: 信念も好みも違う人々をパレート原則でまとめると独裁になる)のために、社会全体の相対的な値打ちは固定されない。抵抗権・革命権は、自分の信念で探索する権利とも読める。

## 第 7 章 国が複数あると(第 7 段の後半)

> この必需品と贅沢品の宇宙的独立関係は、地球全体の話である。地球全体ではなく、国が複数あるときその国でどういう選択をするかというのはそれほど独立ではないだろう。
>
> その時代の全体の状況によって、必需品(軍事)と贅沢品(自由)のどちらにウェイトを置くかが勝利につながるか違うというのはありそうで、どの国も競争をしている以上、ある程度は似た配分にしないといけないという面はあるように思う。地球としてはどういうウェイトでも許容しうるが、個々の国にとっては独立に決められるものではないだろう。
>
> どちらが勝つかは、その時代の、贅沢品=知的財産のアイデアの降りて来ぐあいにもよるだろう。ときには全体主義が優勢となることもありうる。

### 7.1 競争があると、配分は相手に依存する

第 5 章の模型に、国どうしの競争を足す。国 i の値打ちに、相手の国 j と配分が違うことの不利 κ(aᵢ − aⱼ)²/2 を加える。軍事で劣れば攻められ、経済で劣れば市場を取られる。その不利は、相手との差が大きいほど大きい。

$$ G(a_i) = \theta a_i - \frac{a_i^2}{2} - \kappa \frac{(a_i - a_j)^2}{2} $$

このとき、国 i の最もよい配分(最適反応)は、

$$ a_i^* = \frac{\theta + \kappa a_j}{1 + \kappa} $$

で、相手の配分 aⱼ に依存する。そして、相手との差は

$$ a_i^* - a_j = \frac{\theta - a_j}{1 + \kappa} $$

で、自分の信念と相手の配分の差の 1/(1 + κ) に縮む。競争が強いほど(κ が大きいほど)、各国は相手に似た配分を強いられる。

````python
%%writefile -a independence.lean
/-! ## 第 7 章 国が複数あると、独立ではなくなる

国 i の目的に、相手の国 j と配分が違うことの不利 κ (a_i - a_j)²/2 を加える(軍事や経済の競争)。
すると、国 i の最もよい配分は相手の配分に依存し、(θ + κ a_j) / (1 + κ) になる。
相手との差は、自分の信念と相手の配分の差の 1 / (1 + κ) に縮む。 -/

noncomputable def G (κ θ aj ai : ℝ) : ℝ := θ * ai - ai ^ 2 / 2 - κ * (ai - aj) ^ 2 / 2

theorem best_response (κ θ aj b : ℝ) (hκ : 0 ≤ κ) :
    G κ θ aj b ≤ G κ θ aj ((θ + κ * aj) / (1 + κ)) := by
  unfold G
  have h1 : (0 : ℝ) < 1 + κ := by linarith
  have key : G κ θ aj ((θ + κ * aj) / (1 + κ)) - G κ θ aj b
      = (1 + κ) * (b - (θ + κ * aj) / (1 + κ)) ^ 2 / 2 := by
    unfold G; field_simp; ring
  unfold G at key
  nlinarith [sq_nonneg (b - (θ + κ * aj) / (1 + κ))]

theorem best_response_gap (κ θ aj : ℝ) (hκ : 0 ≤ κ) :
    (θ + κ * aj) / (1 + κ) - aj = (θ - aj) / (1 + κ) := by
  have h1 : (1 : ℝ) + κ ≠ 0 := by linarith
  field_simp; ring
````

`best_response` は、G(b) ≤ G(aᵢ*) をすべての b について示す。差が (1 + κ)(b − aᵢ*)²/2 になることを `field_simp` と `ring` で確かめ、それが 0 以上であることを使う。`best_response_gap` は、相手との差の式である。

作者の言う「どの国も競争をしている以上、ある程度は似た配分にしないといけない」は、この 1/(1 + κ) である。そして、第 5 章の言葉で言えば、**信念の区間の中のどこを選ぶかを、各国はもう自由に決められない**。曖昧さへの態度(政治哲学)で選べる幅は、競争の強さに応じて狭まる。

### 7.2 時代ごとに勝つ配分が変わる環境での淘汰

作者は、「どちらが勝つかは、その時代の知財のアイデアの降りて来ぐあいにもよる。ときには全体主義が優勢となることもありうる」と書いた。これを、淘汰の模型で確かめる。

- 200 か国が、それぞれ贅沢品の割合 aᵢ を持つ。
- 時代ごとに「その時代に勝つ配分」a* がある(知財の降りて来ぐあいで変わる)。時代は平均 200 期ごとに入れ替わり、新しい a* は [0.1, 0.5] から引かれる。
- **競争がある場合**: 毎期、1 割の国が、強い国(その時代の a* に近い国)の配分をまねる(または併合される)。まねるときに少しずれる。
- **競争がない場合**(地球全体を一つの社会とみる): 各国は、ゆっくり、ばらばらに配分を変えるだけ。

````python
%%writefile nations.py
# 国が複数あるときの淘汰の模型。
# N か国がそれぞれ贅沢品の割合 a_i を持つ。時代ごとに「その時代に勝つ配分」a*_e が変わる
# (作者の「その時代の、贅沢品=知的財産のアイデアの降りて来ぐあい」)。
# 毎期、国の強さは、その時代の勝つ配分からの離れ方で決まり、強い国ほど生き残って広がる(弱い国は、強い国の配分をまねるか、併合される)。
# 時代は、平均 L 期ごとに入れ替わり、新しい時代の勝つ配分は [lo, hi] から一様に引かれる。
import numpy as np


def simulate(N=200, T=3000, L=200, lo=0.1, hi=0.5, sel=20.0, mut=0.01, comp=True, seed=0):
    """comp=True: 国どうしが競い、毎期 1 割の国が、強さに比例して選ばれた国の配分をまねる(と少しずれる)。
    comp=False: 地球全体を一つの社会とみる(各国は独立に、ゆっくりランダムに配分を変えるだけ)。
    返す値: 各期の配分の平均・標準偏差、時代の勝つ配分、その時代の勝つ配分にいちばん近い国の離れ方。"""
    rng = np.random.default_rng(seed)
    a = rng.uniform(lo, hi, N)
    star = rng.uniform(lo, hi)
    mean, sd, stars, near = np.zeros(T), np.zeros(T), np.zeros(T), np.zeros(T)
    for t in range(T):
        if rng.random() < 1 / L:
            star = rng.uniform(lo, hi)
        if comp:
            fit = np.exp(-sel * (a - star) ** 2 / (hi - lo) ** 2 * 4)
            k = rng.random(N) < 0.1
            src = rng.choice(N, k.sum(), p=fit / fit.sum())
            a[k] = a[src] + rng.normal(0, mut, k.sum())
        else:
            a = a + rng.normal(0, mut / np.sqrt(10), N)
        a = np.clip(a, 0, 1)
        mean[t], sd[t], stars[t], near[t] = a.mean(), a.std(), star, np.abs(a - star).min()
    return mean, sd, stars, near
````

````python
import nations as nt
fig, axs = plt.subplots(1, 2, figsize=(11, 3.6), sharey=True)
for ax, comp, title in ((axs[0], True, 'competing nations'), (axs[1], False, 'no competition (the whole earth)')):
    m, s, st, _ = nt.simulate(comp=comp)
    ax.plot(st, color=INK2, lw=1, label="era's winning allocation a*")
    ax.plot(m, color=PAL[0], lw=1.5, label='mean allocation')
    ax.fill_between(np.arange(len(m)), m - s, m + s, color=PAL[0], alpha=0.2, label='± 1 s.d. across nations')
    ax.set(xlabel='period', title=title); ax.legend(fontsize=7, loc='upper right')
    print(f'{"競争あり" if comp else "競争なし"}: 国のあいだの配分のばらつき(標準偏差){s[500:].mean():.3f}   '
          f'平均と時代の勝つ配分との差 {np.abs(m - st)[500:].mean():.3f}   平均の動いた幅 {m[500:].min():.2f}〜{m[500:].max():.2f}')
axs[0].set_ylabel('luxury share a'); plt.show()
````

- **競争があると、各国の配分は互いに近づき(ばらつき 0.02 ほど)、その時代の勝つ配分を追う**。時代が変わると、全体がそろって新しい配分へ移る。勝つ配分が必需品寄り(a* が小さい)の時代には、全体が必需品寄りになる。作者の「ときには全体主義が優勢となることもありうる」である。
- **競争がないと、各国の配分はばらばらのまま(ばらつき 0.16 ほど)で、時代の勝つ配分とは関係なく漂う**。地球全体で見れば、どんな配分も許容される。作者の「地球としてはどういうウェイトでも許容しうる」である。

### 7.3 競争の代償: 多様性が失われる

競争は、各国を時代に合わせるが、同時に、**国のあいだの多様性**を奪う。時代が急に変わったとき、どの国も前の時代の配分に寄っているので、新しい時代に合う国がない。

````python
rows = {True: [], False: []}
for seed in range(8):
    for comp in (True, False):
        m, s, st, near = nt.simulate(comp=comp, seed=seed, L=400, T=4000)
        for t in np.flatnonzero(np.diff(st) != 0) + 1:
            back = np.flatnonzero(np.abs(m[t:] - st[t]) < 0.03)
            rows[comp].append((near[t], back[0] if len(back) and t + back[0] < len(st) and st[t + back[0]] == st[t] else np.nan))
for comp in (True, False):
    r = np.array(rows[comp])
    print(f'{"競争あり" if comp else "競争なし"}: 時代が変わった直後、新しい勝つ配分にいちばん近い国の離れ方 {np.mean(r[:, 0]):.3f}   '
          f'平均が新しい勝つ配分の ±0.03 に入るまでの期数 {np.nanmean(r[:, 1]):.0f}(入れた割合 {np.mean(~np.isnan(r[:, 1])):.0%})')
````

時代が変わった直後、競争のある世界では、新しい勝つ配分にいちばん近い国でも、かなり離れている(どの国も前の時代の配分に寄っているから)。競争のない世界では、ばらつきが大きいので、新しい時代にほぼ合う国が、たいていどこかにある。一方、全体の平均が新しい勝つ配分に追いつくのは、競争のある世界だけで、競争のない世界の平均は、ほとんど追いつかない。

**競争のある世界は、時代に素早く合わせる力を持つが、多様性を失う。競争のない世界は、多様性を保つが、時代に合わせない。** 作者の「地球全体の独立」と「国ごとの非独立」は、この二つの面に対応する。第 8 章で見る「平時の多様な探索が、未知の危機への備えになる」は、この多様性の値打ちを、学習の言葉で言い直したものになる。

### 7.4 この章のまとめ

- 国どうしが競うと、各国の最もよい配分は相手の配分に依存し、相手との差は 1/(1 + κ) に縮む(Lean の `best_response`、`best_response_gap`)。信念の区間の中で自由に選べる幅は、競争の強さに応じて狭まる。
- 時代ごとに勝つ配分が変わる環境で、競争のある世界は、そろって時代の勝つ配分を追う(ときに全体が必需品寄りになる = 全体主義が優勢になる)。競争のない世界(地球全体)は、どの配分も許したまま、ばらばらに漂う。作者の第 7 段の後半は、この対比でそのまま確かめられる。
- 競争は、時代に合わせる力と引き換えに、国のあいだの多様性を奪う。

## 第 8 章 経済 = 学習: 析出と、危機(2026 年の再考)

### 8.1 2026 年の疑問と議論

2026 年 4 月、作者は、この議論に一つの疑問を投げた。

> 贅沢品に投資することが、絶滅を避けるのに「有効」だという議論にするとします。すると(ホルムズ海峡封鎖のような)危機となったとき、それまでの贅沢品の厚みが必需品の危機をやわらげるようなことになるでしょう。すると、贅沢品にはできるだけ投資しておいたほうが、よいとなりかねません。これでは贅沢品は全然「独立」ではなくなります。

Claude Sonnet 4.6 と Gemini 3.1 Pro との議論は、次のように進んだ。

- **どの贅沢品が効くかは、危機が来るまでわからない**(危機の種類は事前に列挙できない)。だから、贅沢品の「危機への備え」としての値打ちも、事前には評価できない(Claude)。
- 危機が近づくと、どの贅沢品が効くかが確定していき、それは必需品に「転嫁」する。危機までの距離を d とすると、どの贅沢品が効くかのエントロピーが d → 0 で下がる。平時の独立性は、危機の近傍で失われる(Claude)。
- 作者は、贅沢品の重なり(同じ要素を含む贅沢品)から共通の要素を「析出」させると、計算ができるようになる、と考えた。そして、**析出 = 事前学習、危機 = 強化学習**ではないか、と問うた。Claude と Gemini は、これを、平時に目的から独立した表現を学んでおくこと(事前学習)が、未知の危機という新しい課題への素早い適応(少ない試行での学習)を可能にする、と読んだ。作者は、ここから「社会全体が学習装置である」という見方を得た。

Gemini と Claude は、これを PyTorch のシミュレーションにした。作者は、それを何度も作り直させ、自分でも批判した。

- 作者は、α をランダムに揺らすより、**α を固定した複数の値で比べ、学習の出来に差がない(相関がない)領域があることを示すほうが、独立性そのものを見せられる**と考えた(プログラム 4)。そして、「危機時に差がないこと(平時に差があっても)を示すほうが、私の議論に合っている」と書いた。
- Gemini は、ノイズの多い危機の初期(霧の中)でこそ多様な社会が有利だと説明した。作者は、これを**誤謬あるいは誇張**ではないかと指摘した。もしそうなら、既知の危機でも、霧が晴れないあいだは多様な社会が有利になるはずだが、シミュレーションはそうなっていなかったからである。

この章では、まず作者のシミュレーションを再現し(8.2)、そのうえで、それを、構造のある世界での線形の模型として組み直す(8.3〜8.6)。作者の二つの指摘(相関のない領域、霧の中の誇張)には、どちらも、この模型で答えが出る。

### 8.2 作者のシミュレーションの再現

#### プログラム 1(社会 A・B・C)

100 次元の入力から、平時には「既知の課題」(最初の 20 次元に依存する。必需品)を学ぶ。社会は、20 次元の表現(背骨のネットワーク)を持つ。

- 社会 A: α = 0(既知の課題だけを学ぶ)
- 社会 B: α = 1(入力の再構成だけを学ぶ。自己教師あり、贅沢品)
- 社会 C: α を 0.2〜0.8 でランダムに揺らす

危機では、表現を固定し、未知の課題(残りの 80 次元に依存する)に、線形の出力だけを 80 歩学ばせ、最後の損失を比べる。作者のプログラムは乱数の種 42 の一回だけだったので、ここでは種を 6 通りにし、比べる相手として「学習していない表現」と「表現を使わず入力をそのまま使う」を足す。

````python
%%writefile orig2026.py
# 作者の 2026 年の IPYNB(Gemini 3.1 Pro と Claude Sonnet 4.6 が書いたもの)のプログラム 1 を、そのまま関数にしたもの。
# 変えたのは、乱数の種を引数にしたことと、比べる相手(学習していない表現、表現を使わない入力)を足したことだけ。
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
torch.set_num_threads(1)
INPUT_DIM, HIDDEN_DIM, NUM_SAMPLES = 100, 20, 2000


def environment(seed):
    torch.manual_seed(seed); np.random.seed(seed)
    X = torch.randn(NUM_SAMPLES, INPUT_DIM)
    W_known = torch.zeros(INPUT_DIM, 1); W_known[:20, :] = torch.randn(20, 1) * 2.0
    y_known = torch.sin(X @ W_known)
    W_unknown = torch.zeros(INPUT_DIM, 1); W_unknown[20:, :] = torch.randn(80, 1) * 2.0
    y_unknown = torch.relu(X @ W_unknown)
    return X[:1500], X[1500:], y_known[:1500], y_unknown[1500:]


def backbone():
    return nn.Sequential(nn.Linear(INPUT_DIM, 64), nn.ReLU(), nn.Linear(64, HIDDEN_DIM), nn.ReLU())


def decoder():
    return nn.Sequential(nn.Linear(HIDDEN_DIM, 64), nn.ReLU(), nn.Linear(64, INPUT_DIM))


def peacetime(X_train, y_known_train, alpha_of_epoch, epochs=300):
    """平時: 損失 (1 − α)·必需品 + α·再構成(贅沢品)で、背骨を学ぶ。α は期ごとに与える(固定でも揺らぎでもよい)。"""
    bb, head, dec = backbone(), nn.Linear(HIDDEN_DIM, 1), decoder()
    params = list(bb.parameters()) + list(head.parameters()) + list(dec.parameters())
    opt = optim.Adam(params, lr=0.01, weight_decay=1e-4)
    mse = nn.MSELoss()
    for e in range(epochs):
        a = alpha_of_epoch(e)
        loss = (1 - a) * mse(head(bb(X_train)), y_known_train) + a * mse(dec(bb(X_train)), X_train)
        opt.zero_grad(); loss.backward(); opt.step()
    with torch.no_grad():
        known_loss = mse(head(bb(X_train)), y_known_train).item()
    return bb, known_loss


def crisis(features, y, epochs=80, lr=0.02):
    """危機: 表現を固定し、線形の出力だけを学ぶ。最後の損失を返す。"""
    head = nn.Linear(features.shape[1], 1); opt = optim.Adam(head.parameters(), lr=lr)
    mse = nn.MSELoss()
    for e in range(epochs):
        loss = mse(head(features), y); opt.zero_grad(); loss.backward(); opt.step()
    return loss.item()


def program1(seed):
    """社会 A(α = 0)、B(α = 1)、C(揺らぐ α)、学習していない表現、表現を使わない入力の、危機の損失。"""
    Xtr, Xc, yk, yu = environment(seed)
    rng = np.random.default_rng(seed); alphas, a = [], 0.5
    for i in range(300):
        a = float(np.clip(a + rng.normal(0, 0.05), 0.2, 0.8)); alphas.append(a)
    out = {}
    for name, f in (('A', lambda e: 0.0), ('B', lambda e: 1.0), ('C', lambda e: alphas[e])):
        bb, _ = peacetime(Xtr, yk, f)
        with torch.no_grad():
            F = bb(Xc)
        out[name] = crisis(F, yu)
    with torch.no_grad():
        F = backbone()(Xc)
    out['random'] = crisis(F, yu)
    out['raw'] = crisis(Xc, yu)
    out['var'] = yu.var().item()
    return out
````

````python
%%writefile -a orig2026.py
def fixed_alpha(alpha, seed):
    """作者のプログラム 4 の考え: α を固定して平時を過ごし、危機の損失と平時の必需品の損失を測る。"""
    Xtr, Xc, yk, yu = environment(seed)
    bb, known_loss = peacetime(Xtr, yk, lambda e: alpha)
    with torch.no_grad():
        F = bb(Xc)
    return crisis(F, yu), known_loss
````

### 8.3 構造のある世界で組み直す

作者のシミュレーションの弱点(世界に構造がない)を直す。世界に、**少数の潜在因子**を入れる。

- 10 個の潜在因子 z が、100 次元の観測 x を生む(x = zA + 雑音)。
- 必需品の課題(既知)は、最初の 3 個の因子に依存する。
- 危機の課題は、「既知型」なら同じ 3 個の因子に、「未知型」なら残りの 7 個の因子に依存する、新しい課題である。
- 社会は、平時に、観測を k 次元に圧縮する表現を学ぶ。必需品の学習(α = 0)は既知の課題に役立つ方向だけを残し、贅沢品の学習(α = 1)は観測そのものの構造(主成分。因子が「析出」される)を残す。その間は、二つを α で混ぜた行列の上位 k 個の固有ベクトルを取る。
- 危機では、表現を固定し、**30 個の標本**(少ない試行)から、新しい課題への線形の出力だけを学ぶ。出来は、決定係数 R²(1 なら完全に当てる、0 なら平均を当てるのと同じ)で測る。

模型を線形にしたのは、表現が固有ベクトルの計算で決まり、学習の乱数に左右されず、結果が見通せるからである。

````python
%%writefile learning.py
# 経済 = 学習 の線形の模型。
# 世界: 少数(m 個)の潜在因子 z が、d 次元の観測 x を生む。x = z A + 雑音。
# 必需品の課題(既知): 最初の n_known 個の因子に依存する。y = z · w + 雑音(w は最初の n_known 個だけ非零)。
# 危機の課題(未知): 「既知型」なら同じ因子群、「未知型」なら残りの因子に依存する、新しい w。
# 社会は、平時に、観測を k 次元に圧縮する表現 P を学ぶ(社会の制度・技術・知識の「型」)。
#   必需品の学習(α = 0): 既知の課題を当てるのに役立つ方向だけを残す。
#   贅沢品の学習(α = 1): 観測そのものの構造(主成分 = 析出された因子)を残す(自己教師あり)。
#   その間: M(α) = (1 − α) · 既知の課題の方向 + α · 観測の共分散 の上位 k 個の固有ベクトル。
# 危機では、表現 P を固定し、少ない標本 n_c から、新しい課題への線形の出力だけを学ぶ(リッジ回帰)。
import numpy as np


class World:
    def __init__(self, m=10, d=100, n_known=3, noise=0.5, seed=0):
        rng = np.random.default_rng(seed)
        self.m, self.d, self.n_known, self.noise = m, d, n_known, noise
        self.A = rng.normal(0, 1, (m, d)) / np.sqrt(m)
        self.w_known = np.zeros(m); self.w_known[:n_known] = rng.normal(0, 1, n_known)
        self.rng = rng

    def sample(self, n):
        z = self.rng.normal(0, 1, (n, self.m))
        return z, z @ self.A + self.noise * self.rng.normal(0, 1, (n, self.d))

    def crisis_task(self, kind, overlap=0.0):
        """kind='known': 既知の因子群に依存。kind='unknown': 残りの因子に依存。
        overlap: 未知型の危機が、既知の因子にも少し依存する度合い(重なり)。"""
        w = np.zeros(self.m)
        if kind == 'known':
            w[:self.n_known] = self.rng.normal(0, 1, self.n_known)
        else:
            w[self.n_known:] = self.rng.normal(0, 1, self.m - self.n_known)
            w[:self.n_known] += overlap * self.rng.normal(0, 1, self.n_known)
        return w


def representation(world, alpha, k, n_peace=4000):
    """平時に学ぶ表現(k × d の射影)と、平時の必需品の出来(既知の課題の決定係数 R²)。"""
    z, x = world.sample(n_peace)
    y = z @ world.w_known + 0.1 * world.rng.normal(0, 1, n_peace)
    xc = x - x.mean(0)
    Sxx = xc.T @ xc / n_peace
    sxy = xc.T @ (y - y.mean()) / n_peace
    Mk = np.outer(sxy, sxy); Mk /= np.trace(Mk)
    Mx = Sxx / np.trace(Sxx)
    vals, vecs = np.linalg.eigh((1 - alpha) * Mk + alpha * Mx)
    P = vecs[:, ::-1][:, :k]
    return P, r2_fit(x @ P, y, x @ P, y)


def r2_fit(Ftr, ytr, Fte, yte, lam=1e-3):
    F1 = np.c_[Ftr, np.ones(len(Ftr))]
    beta = np.linalg.solve(F1.T @ F1 + lam * np.eye(F1.shape[1]), F1.T @ ytr)
    pred = np.c_[Fte, np.ones(len(Fte))] @ beta
    return 1 - np.mean((yte - pred) ** 2) / np.var(yte)


def crisis_r2(world, P, w, n_c=30, n_test=2000, label_noise=0.3):
    """危機: 表現 P を固定し、n_c 個の(雑音の乗った)標本から出力だけを学び、新しい課題をどれだけ当てられるか(R²)。"""
    z, x = world.sample(n_c)
    y = z @ w + label_noise * np.std(z @ w) * world.rng.normal(0, 1, n_c)
    zt, xt = world.sample(n_test)
    return r2_fit(x @ P, y, xt @ P, zt @ w, lam=1.0)
````

````python
%%writefile -a learning.py
def sweep(alphas, k=5, kinds=('known', 'unknown'), n_c=30, n_tasks=40, worlds=4, overlap=0.0, label_noise=0.3):
    """α ごとの、平時の必需品の出来と、危機の種類ごとの危機の出来(R²)の平均。"""
    out = {'peace': np.zeros(len(alphas))}
    for kind in kinds:
        out[kind] = np.zeros(len(alphas))
    for s in range(worlds):
        for i, a in enumerate(alphas):
            W = World(seed=s)
            P, r2k = representation(W, a, k)
            out['peace'][i] += r2k / worlds
            for kind in kinds:
                Wt = World(seed=s)                       # 同じ世界で、危機の課題と標本を引く
                Wt.rng = np.random.default_rng(1000 + s)
                out[kind][i] += np.mean([crisis_r2(Wt, P, Wt.crisis_task(kind, overlap), n_c=n_c, label_noise=label_noise)
                                         for _ in range(n_tasks)]) / worlds
    return out


def evaluate(alpha, k, kind, n_c=30, label_noise=0.3, rep='learned', overlap=0.0, worlds=4, n_tasks=40):
    """一つの α での危機の出来(R²)。rep='random' なら、学習していない(でたらめな)k 次元の表現を使う。"""
    r = []
    for s in range(worlds):
        W = World(seed=s)
        if rep == 'random':
            P = np.linalg.qr(np.random.default_rng(500 + s).normal(0, 1, (W.d, k)))[0]
        else:
            P = representation(W, alpha, k)[0]
        Wt = World(seed=s); Wt.rng = np.random.default_rng(1000 + s)
        r += [crisis_r2(Wt, P, Wt.crisis_task(kind, overlap), n_c=n_c, label_noise=label_noise) for _ in range(n_tasks)]
    return float(np.mean(r))
````

````python
import learning as lrn
al = np.array([0.0, 0.01, 0.05, 0.2, 0.5, 0.8, 0.95, 1.0])
out = {k: lrn.sweep(al, k=k, n_tasks=30) for k in (5, 10)}
fig, axs = plt.subplots(1, 2, figsize=(11, 3.6), sharey=True)
for ax, k in zip(axs, (5, 10)):
    o = out[k]
    ax.plot(al, o['peace'], 'o-', color=PAL[0], label='peacetime: necessity task')
    ax.plot(al, o['known'], 's-', color=PAL[3], label='crisis: known type')
    ax.plot(al, o['unknown'], '^-', color=PAL[1], label='crisis: unknown type')
    ax.set(xlabel='luxury (self-supervised) share α', title=f'capacity k = {k} (10 latent factors)')
    print(f'k = {k}:  α        ' + ' '.join(f'{a:6.2f}' for a in al))
    for key, lab in (('peace', '平時の必需品'), ('known', '危機・既知型'), ('unknown', '危機・未知型')):
        print(f'        {lab:10s} ' + ' '.join(f'{x:6.3f}' for x in o[key]))
axs[0].set_ylabel('R²'); axs[0].legend(fontsize=8); plt.show()
````

二つの場合で、様子が違う。

**表現の容量が十分なとき(k = 10、因子の数と同じ)**:

- α = 0(必需品だけ)の社会は、平時の課題はこなすが、危機には弱い。未知型の危機だけでなく、**既知型の危機にも弱い**。必需品の学習は、いまの課題の方向だけを残し、同じ因子に依存する別の課題には備えていないからである。
- **α を少しでも正にすると(0.01 でも)、危機の出来は跳ね上がり、そこから α = 1 まで、平時の出来も危機の出来も、まったく変わらない**。表現が因子の全体を析出しているので、どの α でも同じ表現になる。

これが、作者の言う「相関がない領域」の、もっとも純粋な形である。**α = 0 は退けられるが、それより上は、どれを選んでも同じ**。第 5 章の「支配されない配分の区間」が、ここでは (0, 1] という広い平らな領域として現れる。

**表現の容量が足りないとき(k = 5、因子の数の半分)**:

- α が小さいうちは、表現は必需品の方向と、主成分のいくつかを残す。平時の出来はよく、危機の出来は中くらいで、α = 0.05〜0.8 ではほぼ平らである。
- α を 1 に近づけると、必需品の方向が表現から押し出され、**平時の必需品の出来が落ちる**(必需品の下限を割る)。その代わり、未知型の危機の出来は少し上がる。

容量が足りないと、贅沢品の学習と必需品の学習は、限られた表現の取り合いになる。それでも、真ん中に平らな領域は残る。作者のプログラム 4 の結果(α = 0.1〜0.9 で横ばい、α = 1 で必需品が崩れる)は、この k = 5 の形とよく合う。

### 8.4 危機の種類の確率と、最適な α

危機が既知型か未知型かは、事前にはわからない。未知型である確率を θ とし、社会の値打ちを、平時の必需品の出来(重み w)と、危機の出来の期待値の和とする。

$$ V(\alpha; \theta) = w \cdot \text{平時}(\alpha) + (1 - \theta)\,\text{既知型}(\alpha) + \theta\,\text{未知型}(\alpha) $$

θ が区間でしか言えないとき、支配されない α を求める(容量 k = 5)。平らな領域の中の違いは計算の細かな揺れ(0.005 未満)でしかないので、それより小さい差は「同じ」とみなす(b が a を支配するのは、区間のどの θ でも 0.005 より大きく勝つとき)。

````python
o = out[5]
thetas = np.linspace(0.0, 1.0, 11)
def undominated_alphas(w, lo_, hi_, tol=0.005):
    th = thetas[(thetas >= lo_ - 1e-9) & (thetas <= hi_ + 1e-9)]
    V = np.array([w * o['peace'] + (1 - t) * o['known'] + t * o['unknown'] for t in th])
    dom = [np.all(V > V[:, [i]] + tol, axis=0).any() for i in range(len(al))]
    return [float(x) for x in al[~np.array(dom)]]
for w in (1.0, 0.1):
    print(f'平時の重み w = {w}')
    for lo_, hi_ in ((0.0, 0.5), (0.2, 0.9), (0.8, 1.0)):
        print(f'   θ ∈ [{lo_}, {hi_}] のとき、支配されない α: {undominated_alphas(w, lo_, hi_)}')
````

- 平時の必需品を重く見る(w = 1)なら、θ がどこにあっても、**支配されない α は真ん中の平らな領域(0.01〜0.5)**で、α = 0(必需品だけ)と、α = 0.8 以上(贅沢品に寄りすぎ)は退けられる。平らな領域のどこを選ぶかは、データでは決まらない。
- 平時の必需品を軽く見る(w = 0.1)と、平らな領域は広がり、θ の区間が未知型に寄っていれば、α = 1(贅沢品への全振り)も支配されなくなる。ただし、全振りがほかを支配する(唯一の答えになる)わけではない。

**危機の種類についての信念の区間と、平時の重みが、支配されない α の集合を決める**。第 5 章の定理の構造が、学習の模型の中で、そのまま現れる。作者の 2026 年の疑問「危機に効くなら、贅沢品に全振りすべきでは」への答えは、こうなる。

> 全振りが選ばれうるのは、未知型の危機を強く見込み、かつ、平時の必需品を軽く見るときだけである。それ以外では、全振りは真ん中の平らな領域に支配される。そして、平らな領域の中では、どの α でも(ほぼ)同じである。**贅沢品は、少しは要るが、多ければよいわけではない。**

### 8.5 析出は、少ない試行での学習を可能にする

事前学習(析出)の値打ちは、危機で**少ない標本から学べる**ことにある。危機で得られる標本の数 n_c を変えて、未知型の危機の出来を比べる(容量 k = 10)。比べる相手は、α = 0 の社会と、学習していない(でたらめな)10 次元の表現を持つ社会である。

````python
print('危機の標本の数 n_c と、未知型の危機の出来(R²)')
print('  n_c    α = 0(必需品だけ)   α = 0.3(析出あり)   でたらめな表現')
for n_c in (10, 30, 100):
    r = [lrn.evaluate(0.0, 10, 'unknown', n_c=n_c, n_tasks=30), lrn.evaluate(0.3, 10, 'unknown', n_c=n_c, n_tasks=30),
         lrn.evaluate(0.0, 10, 'unknown', n_c=n_c, rep='random', n_tasks=30)]
    print(f'  {n_c:4d}       {r[0]:7.3f}             {r[1]:7.3f}           {r[2]:7.3f}')
````

析出した表現を持つ社会は、**10 個の標本でも、未知の危機の 4 割ほどを当てられる**。必需品だけの社会と、でたらめな表現の社会は、10 個の標本では何も当てられない(R² が負 = 平均を当てるより悪い)。標本が増えれば差は縮むが、析出した社会は、30 個で 9 割に達する。

危機では、何度も失敗して学ぶ余裕がない(作者の議論での「何度も絶滅して学ぶことは許されない」)。**事前学習の厚みが、危機での学習の速さを決める**。2026 年の議論での「析出 = 事前学習、危機 = 強化学習(少ない試行での適応)」は、この形で正確に言える。

### 8.6 霧の中では、多様な社会が有利か

作者は、Gemini の「霧の中(ノイズの多い危機の初期)でこそ多様な社会が有利」を、誤謬あるいは誇張ではないかと指摘した。危機の標本の雑音(霧の濃さ)を変えて確かめる(容量 k = 5、標本 30 個)。

````python
print('霧の濃さ(標本の雑音の大きさ)と、危機の出来(R²)')
print('  霧     既知型: α = 0   α = 0.3      未知型: α = 0   α = 0.3')
for fog in (0.3, 1.0, 3.0):
    r = [lrn.evaluate(a, 5, kind, label_noise=fog, n_tasks=30) for kind in ('known', 'unknown') for a in (0.0, 0.3)]
    print(f'  {fog:3.1f}          {r[0]:7.3f}  {r[1]:7.3f}            {r[2]:7.3f}  {r[3]:7.3f}')
````

- **既知型の危機では、霧が濃くても薄くても、必需品だけの社会(α = 0)と析出のある社会(α = 0.3)は、ほぼ同じ**である。霧そのものが、多様な社会を有利にするわけではない。
- **未知型の危機では、析出のある社会がいつも有利**で、霧が濃くなると、その差は縮む(霧が濃すぎると、どちらも何も学べない)。

作者の指摘は正しかった。多様な社会の有利さは、「霧の中だから」ではなく、「**危機の構造が、析出された表現の中にあるから**」である。既知型の危機の構造は、必需品の表現の中にもあるので、多様性は助けにならない。未知型の危機の構造は、析出された表現の中にしかないので、多様性が効く。そして、霧はその差を、増やすのではなく、薄める。Gemini の説明は、二つの違う効果(構造があるかどうか、霧の濃さ)を混ぜていた。

### 8.7 危機の近傍: 信念のエントロピーと、配分の収束

最後に、2026 年の議論の「危機の近傍でエントロピーが下がり、独立性が失われる」を、信念の側で書く。2026 年の Claude のシミュレーションでは、「エントロピー」を、台本どおりに動かした α の履歴から計算していて、信念の不確かさを測っていなかった。ここでは、危機の種類(既知型か未知型か)についての信念を、ベイズで更新する。

- 事前の信念(未知型である確率)は、一つに決まらず、区間 [0.2, 0.8] でしか言えない(ナイト的不確実性)。
- 危機が近づくと、毎期、危機の種類について少しだけ情報を含む兆しが届く。
- 区間の両端から、それぞれベイズで更新すると、事後の区間が得られる。兆しが積もるほど、区間は縮む。

````python
%%writefile -a learning.py
# 危機の近傍での信念の更新。危機の種類は二つ(既知型、未知型)。
# 事前の信念(未知型である確率)は一つに決まらず、区間 [πlo, πhi] でしか言えない。
# 危機が近づくと、毎期、兆し s ~ N(+μ, 1)(未知型)または N(−μ, 1)(既知型)が届く。
# 事前の区間のそれぞれの端からベイズで更新すると、事後の区間 [πlo_t, πhi_t] が得られ、兆しが積もるほど縮む。
def posterior_interval(signals, pi_lo, pi_hi, mu=0.5):
    llr = np.cumsum(2 * mu * np.asarray(signals))          # 対数尤度比(未知型 / 既知型)の累積
    post = lambda p: 1 / (1 + (1 - p) / p * np.exp(-llr))
    return post(pi_lo), post(pi_hi)


def binary_entropy(p):
    p = np.clip(p, 1e-12, 1 - 1e-12)
    return -(p * np.log2(p) + (1 - p) * np.log2(1 - p))
````

````python
importlib.reload(lrn)
rng = np.random.default_rng(3)
signals = rng.normal(+0.2, 1.0, 120)                       # 本当は未知型の危機
plo, phi = lrn.posterior_interval(signals, 0.2, 0.8, mu=0.2)
mid = (plo + phi) / 2
fig, axs = plt.subplots(1, 2, figsize=(11, 3.4))
axs[0].fill_between(np.arange(1, 121), plo, phi, color=PAL[1], alpha=0.3, label='posterior interval (prior [0.2, 0.8])')
axs[0].set(xlabel='signals received (approaching the crisis)', ylabel='P(unknown-type crisis)', title='The belief interval shrinks')
axs[0].legend(fontsize=8)
axs[1].plot(np.arange(1, 121), lrn.binary_entropy(mid), color=PAL[6])
axs[1].set(xlabel='signals received', ylabel='entropy (bits)', title='Entropy of the (mid) belief')
plt.show()
for t in (0, 10, 40, 119):
    print(f'兆し {t + 1:3d} 個: 事後の区間 [{plo[t]:.3f}, {phi[t]:.3f}]  幅 {phi[t] - plo[t]:.3f}   エントロピー {lrn.binary_entropy(mid[t]):.3f} ビット')
````

兆しが積もるにつれ、信念の区間は(兆しの偶然で一時的に広がることもあるが)縮み、エントロピーは下がる。区間が一点に縮めば、第 5 章の定理から、支配されない配分も一点になる。Lean で示す。

````python
%%writefile -a independence.lean
/-! ## 第 8 章 危機の近傍: 信念の区間が縮めば、配分は一点に決まる

危機が近づき、どの贅沢品が効くかがわかってくると、信念の区間 [θlo, θhi] が縮む。
区間の幅が 0 になれば、支配されない配分はただ一つになる(贅沢品から必需品への「転嫁」)。 -/

theorem crisis_collapse (θ a : ℝ) : ¬ Dominated θ θ a ↔ a = θ := by
  rw [undominated_iff θ θ a le_rfl]
  constructor
  · rintro ⟨h1, h2⟩; linarith
  · rintro rfl; exact ⟨le_rfl, le_rfl⟩
````

`crisis_collapse`: 信念の区間の両端が同じ θ なら、支配されない配分は a = θ だけである。**独立性は、平時(信念の区間が広いとき)の性質で、危機の近傍(区間が縮むとき)では失われる**。2026 年の Claude の「相転移」の描写は、この信念の区間の縮みとして、正確に言える。ただし、区間は、兆しが積もるにつれて連続的に縮むので、どこかで急に変わる「相転移」というより、**滑らかな収束**である。

### 8.8 経済 = 学習

この章の模型から、作者の「社会全体が学習装置である」という見方を、次のように言い直せる。

| 経済の言葉 | 学習の言葉 |
|---|---|
| 制度、技術、知識の蓄積(社会の「型」) | 表現(観測を圧縮する仕方) |
| 必需品の生産 | 既知の課題への教師あり学習 |
| 贅沢品(探索、文化、知財) | 目的から独立した自己教師あり学習(構造の析出) |
| 危機 | 少ない試行で新しい課題を解くこと |
| 必需品の下限 | 既知の課題の出来の下限 |
| 社会の容量(人口、時間) | 表現の次元 k |

この対応のもとで、作者の議論の三つの部分が、一つの模型の中に収まる。

1. **独立性**: 表現の容量が十分なら、贅沢品の学習を少しでもすれば、そこから先はどの α でも同じ(平らな領域)。容量が足りなくても、真ん中に平らな領域が残る。
2. **必需品の下限**: 容量が足りないとき、贅沢品に振りすぎると、必需品の課題が表現から押し出される。
3. **危機の近傍**: 危機の種類についての信念が縮むと、支配されない配分も縮む。

そして、析出(事前学習)の値打ちは、危機で少ない試行から学べることにある。これは、ノート O の「格上げ」(経済の中で別の経済が育ち、全体に格上げされる)の、学習の側からの説明にもなる。探索が当たったとき(未知型の危機が、析出された構造の中にあったとき)、それまで「贅沢品」だった表現は、いっきに「必需品」の中心になる。作者が 2026 年に書いた「贅沢品から必需品への転嫁」である。

### 8.9 この章のまとめ

- 作者の 2026 年のシミュレーション(プログラム 1)は、6 シードで向き(B < C < A)が再現される。ただし、効果はとても小さく、原因は世界に構造がないことにある。プログラム 4 の「相関がない領域」も再現される。
- 構造(潜在因子)のある世界の線形の模型では、**α = 0 は退けられ、それより上には平らな領域がある**。容量が足りないと、α = 1 付近で必需品の下限を割る。作者のプログラム 4 の結果と、第 5 章の定理の両方に合う。

### 9.2 作者の議論のうち、そのまま残るもの、言い直したもの

**そのまま残るもの**: 結論。「必需品の必要量を超えて、贅沢品をどれだけ作るかは決められない。それは地球全体の話で、国が複数あると独立ではない。全体主義は必需品への偏り、抵抗権・革命権は贅沢品」。この結論は、Lean で証明できる形で、ほぼそのまま残る。

**言い直したもの**: 理由。作者は、「決められない」理由を、「0 × ∞ の期待値が評価できない」に置いた。この文書は、それを「**贅沢品が将来もたらすものについての信念が、一つに決まらない**」に置き換えた。

- 「0 × ∞」は、仮定(向かい方)を一つ置けば決まってしまう。そして、無限は比で測るから現れる。だから、それだけでは「決められない」の理由として弱い。
- 「信念が一つに決まらない」(ナイト的不確実性)は、仮定を置いても消えない。どの仮定(信念)を置くかが、まさに決まらないからである。そして、信念が区間なら、支配されない配分は区間になる、という定理が、作者の結論をそのまま導く。

作者の議論のもう一つの芯、「確率 0 でも望みがある」は、測度論の「確率 0 ≠ 不可能」ではなく、「**破局の確率の総和が有限になりうる**」として言い直した。こうすると、贅沢品(知財)の値打ちが、はっきりした形(破局の確率の列を、総和が有限な列に変えること)で言える。

### A.2 Lean の準備

次のセル(「Lean の準備」)が、Lean を用意する。このノートと同じディレクトリの `.leandir` に、Lean 4(v4.34.1)と Mathlib の入った lake のプロジェクトの場所(lakefile のあるディレクトリ)が書いてあれば、それを使う。なければ、elan(Lean の版の管理)を入れ、そのプロジェクト `jrf_lean/` を作って、`.leandir` に書く。Google Colab でも、そのまま通る。初めてのときは Mathlib の組み上がったものを取ってくるので 10 分ほどかかり、10GB ほどの場所を使う。同じディレクトリでほかのノートを動かすときは、それを使い回す。

手元の Lean を使うなら、たとえば次のようにして、`.leandir` を先に書いておく。

```bash
lake new leanproj math        # Mathlib を使うプロジェクトを作る
cd leanproj && lake exe cache get
pwd > /path/to/notebook/.leandir
```

````python
# Lean の準備: .leandir に書いた lake のプロジェクト(Mathlib 入り)があれば、それを使う。
# なければ elan(Lean の版の管理)を入れ、Lean 4 v4.34.1 と Mathlib の入ったプロジェクト jrf_lean/ を作る(Google Colab でもそのまま通る)。
# 初めてのときは Mathlib の組み上がったものを取ってくるので、10 分ほどかかり、10GB ほどの場所を使う。同じディレクトリのほかのノートは、それを使い回す。
import os, shutil, subprocess

def lean_project_ok(d):
    return bool(d) and os.path.exists(os.path.join(d, 'lakefile.lean')) and os.path.isdir(os.path.join(d, '.lake', 'packages', 'mathlib'))

LEANDIR = open('.leandir').read().strip() if os.path.exists('.leandir') else None
if lean_project_ok(LEANDIR):
    print('Lean のプロジェクト: .leandir に書いた場所にあるものを使う')
else:
    elan_bin = os.path.expanduser('~/.elan/bin')
    os.environ['PATH'] = elan_bin + ':' + os.environ.get('PATH', '')
    if not shutil.which('lake'):
        subprocess.run('curl -sSfL https://raw.githubusercontent.com/leanprover/elan/master/elan-init.sh'
                       ' | sh -s -- -y --default-toolchain none', shell=True, check=True)
    LEANDIR = os.path.abspath('jrf_lean')
    os.makedirs(LEANDIR, exist_ok=True)
    with open(os.path.join(LEANDIR, 'lean-toolchain'), 'w') as f:
        f.write('leanprover/lean4:v4.34.1\n')
    with open(os.path.join(LEANDIR, 'lakefile.lean'), 'w') as f:
        f.write('import Lake\nopen Lake DSL\n\npackage jrf_lean\n\n'
                'require mathlib from git\n  "https://github.com/leanprover-community/mathlib4.git" @ "v4.34.1"\n')
    lake = shutil.which('lake') or os.path.join(elan_bin, 'lake')
    for args in (['update'], ['exe', 'cache', 'get']):
        r = subprocess.run([lake] + args, cwd=LEANDIR, capture_output=True, text=True)
        print(f'lake {" ".join(args)}: 終了コード {r.returncode}')
        if r.returncode != 0:
            print(r.stdout[-2000:], r.stderr[-2000:])
            raise RuntimeError('Lean のプロジェクトを作れなかった')
    with open('.leandir', 'w') as f:
        f.write(LEANDIR + '\n')
    print('Lean のプロジェクト: jrf_lean/ に作り、.leandir に書いた')
````
