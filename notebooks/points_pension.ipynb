### この文書について

この文書は、同じ作者の古いモデルを読み直すシリーズの一冊である。

ここまでの三冊が扱ったのは、経済の**シミュレーション**だった。この文書が扱うのは、それとは種類の違うものである。作者が 2006 年から 2018 年にかけて考え、一部は実際に作って動かした、**ネットの上の小さな制度**の設計である。

- **投げ銭予約**(2006)。はてなブックマークのポイント(投げ銭)を、まだ払えないページにも「予約」できるようにする。作者は、予約を紙幣、実際の投げ銭を本貨にたとえ、その比を公開して予約の信用を支えようとした。
- **空スター**(2009)。読んで欲しい人を名指しして、その人が読んでスターを付けたらポイントが渡るよう、予約しておく。読む側ではなく、**読んでもらう側が払う**。
- **はてなポイントの分配**(2011)。作者は、匿名で受け取ったポイントを、四半期ごとに分配する規則を作り、Perl のプログラムにして公開し、実際に運用した。その期に作者がポイントを送る書き手には**ボーナス**を上乗せし、過去に送った書き手には**ペンション**(年金)を払う。
- **BI が否認されたときの返し方**(2011)。受け取ったポイントが、もし不正に得たベーシックインカム(BI)から送られたものだと国が主張したら、どう返すべきか。作者は、この計算も Perl で書いた。
- **No Ad URL**(2013)。リンクを張る人が、リンク先の広告の非表示を、リンクを受け取る人のために「買う」。
- **時間記録デバイス**(2009 の構想)、**Loaded Magic**(2011)、**Tweet Sig**(2018)。文書がいつ、誰によって書かれたかを、どう示すか。
- **プライバシープロバイダ API**(2006 の構想)と **PP Authorizer**(2017)。サイトには身元を渡さず、信用の高い第三者が保証するログインと支払い。

作者は、2026 年の初めに、これらを振り返って次のように書いている。P2P ファイル共有が流行ったころ、著作物の対価を集める道は寄付しかないと思った。「情報を囲って儲ける」のではない道を探していた。しかし、寄付だけでは経済は回らない。そういう書き手が投げ銭だけでは生きていけないから、BI のような仕組みが要る、という直感もあった。

この文書では、これらを一つにまとめて読む。ばらばらに見える構想は、組み合わせると、**ネットの上の小さな通貨の制度**を、ひと揃い形づくっている、というのがこの文書の見立てである。そして、それぞれが 2026 年の今、どういうものとして実現しているか(あるいは、していないか)を調べ、これからどう使えるかを考える。

### シミュレーションではなく、制度の設計

ノート 1・O・2 のモデルは、「こういう規則で動く経済は、どうなるか」を調べるものだった。この文書の材料は、「こういう結果になって欲しいから、規則をこう決める」というものである。経済学では、こちらを**メカニズムデザイン**(制度設計)と呼ぶ。

作者は、規則の細部を決めるとき、参加者がどう振る舞うかを読んでいた。たとえば、分配のボーナスを送った額の平方根に比例させると、一人が ID を二つに分ければ得をする。そこで、ペンションの上限には 1.05 乗をかけ、ID を分けるとわずかに損になるようにした。今の言葉で言えば、一人が多くの ID を作って得をすること(**シビル攻撃**)への対策を、2011 年に、自分の制度の中で考えていたことになる。

だから、この文書のシミュレーションは、制度を試す実験である。人工の参加者を作り、その参加者が規則の中で得な振る舞いを選んだとき、作者の狙った結果になるかを確かめる。実在の人のデータは使わない(はてなブックマークの記録には、実在の人の ID が含まれるため)。すべて人工のデータを使う。

### 三つの層

作者の構想は、次の三つの層に分けられる。この文書の章立ても、これに沿っている。

| 層 | 問い | 作者の構想 | 章 |
|---|---|---|---|
| **価値の層** | 誰が誰に、何の対価として払うか | 投げ銭予約、空スター、No Ad URL、分配とペンション、BI 否認時の返し方 | 第 1・2 章、第 5〜8 章 |
| **証明の層** | それが、いつ、誰によって起きたかを、どう示すか | 時間記録デバイス、Loaded Magic、Tweet Sig | 第 3 章 |
| **経路の層** | 誰が仲立ちし、身元を誰に渡すか | 日銀カード、PP ログイン・PP 振込、印栞、PP Authorizer | 第 4 章 |

三つの層は、お金の制度に必要なものに対応している。お金には、値打ち(誰が誰に払うか)、記録(払ったことの証明)、決済の経路(誰が仲立ちするか)が要る。紙幣で言えば、券面の額、偽造を防ぐ印刷、銀行の窓口である。作者は、これをネットの上の小さな規模で、一つずつ考えていた。

そして、三つの層に共通する作者の態度が二つある。

1. **サイトに対しては匿名で、どこかには責任がある。** 作者は、すべての取引がサーバーに記録される「1984 的な」お金を嫌った。一方で、まったくの無責任も認めなかった。身元を知るのは信託会社や認証プロバイダ(作者の言う「正義の味方」)だけにして、サイトには渡さない、という形を繰り返し考えた。
2. **仕様は小さく、商売っ気なく公開する。** 作者は、No Ad URL について「この部分だけは商売っけなくできることが何より重要」と書き、凝った部分は他の人に任せるとした。作ったものは、ほとんどがパブリックドメインで公開されている。

### この文書で考える問い

1. 投げ銭予約は、どういう通貨の制度か。予約(紙幣)を本貨で支える仕組みは、どこまで信用を保てるか。
2. 空スターのような「読んでもらう側が払う」支払いは、成り立つか。読まずにスターを付ける人がいても、評判で抑えられるか。
3. 文書がいつ書かれたかの証明は、どう作れるか。作者の三つの構想は、今の技術とどう対応するか。
4. 分配の規則は、ID を分けたりまとめたりする誘因を、作者の狙いどおり「ほどほど」に抑えられるか。今の書き手と過去の書き手の取り分は、公平か。
5. BI が否認されたときの返し方は、何を保証していたか。マイナス金利で、何が壊れるか。
6. 書き手は、どう食べていけばよいのか。作者の構想は、2026 年の今、どういうものとして実現しているか。AI の時代に、どう使えるか。

### 答えの見通し

結論を先に書いておく(詳しくは各章と、第 11 章のまとめで)。

1. **投げ銭予約は、部分準備の通貨である**。作者が「要証明」と書いた信頼性フィルタは、証明できない。受けられないページへの予約は開示に現れないので、少し払っておけば、最も信頼できる利用者に見える。偽りに強いのは、開示ではなく、本貨の裏付け(予約証拠金)である(第 1 章)。
2. **空スターは条件付きの支払い、No Ad URL は手形の裏書きである**。「読まない読み手は、だんだん指定されなくなる」という作者の見込みは、評判が私的なら半分しか当たらず、公開すれば当たる。早い者勝ちは、読まない人を呼び込む(第 2 章)。
3. **時間記録デバイス、Loaded Magic、Tweet Sig は、今の時刻証明・監査の記録・投稿による証明に対応する**。時間記録デバイスは、二重の譲渡を見抜けるが、複製は防げない。それは P2P の電子マネーの根本の問題である(第 3 章)。経路の層の構想(PP ログイン、PP Authorizer)は、パスキーと FedCM として、ブラウザに標準で入った(第 4 章)。
4. **作者の分配のプログラムは、今の Perl でそのまま動き、書き直しと一人ずつの支払いまで一致する**(第 5 章)。ペンションは、過去の書き手の上乗せの倍率を、今期の倍率に追いつかせる仕組み(過去の拠出の再評価)である。ID を分けた得は、ペンションの上限で取り返され、統合を 10 年ほど待たせれば、ほぼ中立になる。作者の「ほどほどの ID の分散と統合」は実現している(第 6 章)。
5. **BI 否認時の返し方では、新しい債権者の負担は利率一期分を超えない**。作者の閉じた式の誤りが、ありもしない発散を見せていた。マイナス金利では、作者のプログラムはゼロでの割り算で止まり、貸倒れの判定は成り立たない(第 7・8 章)。
6. **作者の目的は「補償金で基礎、投げ銭で上乗せ」(2007)にあった**。集めたお金は、使われ方に比例して配る限り、基礎の収入にならない。2011 年の分配のプログラムは、AI の利用料を引用された書き手に配る仕組みの雛形になる(第 9・10 章)。

# 第 I 部 価値の層: 誰が誰に払うか

## 第 1 章 投げ銭予約: 紙幣と本貨(2006)

### 1.1 作者の構想

2006 年 4 月、作者は「はてなブックマーク『投げ銭予約』のアイデアまとめ ── 金本位制のメタファより」を書いた。当時のはてなブックマークには、ブックマークしたページの書き手に、はてなポイント(1 ポイント = 1 円)を送る「投げ銭」の機能があった。ただし、送れるのは、受け取る宣言をしたはてなのブログだけだった。

作者の提案は、次のようなものだった。

- **投げ銭を、「ほんの少し責任のあるレーティング」として使う。** 普通のレーティング(星いくつ)は、高く付けても低く付けても同じコストしかかからない。だから、怒りにまかせた低い評価や、不自然な高い評価が出る。投げ銭は、高く評価するほど多くのコストがかかる。
- **まだ投げ銭を受けられないページにも、「予約」できるようにする。** 予約は取り消せない。支払いは、本人の責任で猶予できる。「以前とは違うページになっていても、あの頃の感動をもとに投げ銭する」のが、予約の本質だという。
- **金本位制のたとえ。** 予約を紙幣、実際の投げ銭を本貨とみる。紙幣の信用は、求められればいつでも本貨に換えられる(兌換できる)ことで保たれる。そこで、利用者ごとに次の三つを公開する。
  - 送ポイント総額(実際の本貨)
  - 予約総額(紙幣の流通量)
  - 予約総額のうち送ポイント可能額(受けられるようになったのに、まだ払っていない額。兌換の猶予の状況)
- **収益力のある関心の指標。** タグやページごとに、この三つを集計して公開する。どのページとどのページの累積ポイントを比べても意味は薄いが、タグごとの集計は、社会の中で「お金を払ってでも見たい関心」がどこにあるかを示す。
- **信頼性フィルタ。** 集計には、信用の違う利用者が混ざっている(「通貨の信用がまったく違う国のデータをゴッチャにして兌換可能性が示されるようなもの」)。そこで、「予約総額のうち送ポイント可能額」を「送ポイント総額」で割った値がごく小さい利用者だけで集計できるようにする。作者は、ここに「(←要証明。不都合あるかも。)」と書き添えた。
- **予約証拠金。** 投げ銭できないページにタグを乱発して、そのタグへの関心が高いように装う人が出るかもしれない。そこで、予約にはある程度の積み立て(証拠金)を求める。ただし、払えない若者のために、証拠金の額と割合を公開すれば積み立てなくてもよい、とする。
- そのほか、衝動的に付けたポイントを見直すための「レジ」(買い物かご)、詐欺への備えとしての「入金遅延」、そして**ポイントバック**(違法かもしれない利用への「投げ銭」を、正規に買うまでの積立金とみなし、買ったときに精算する)。

最後のポイントバックについて、作者は、クレジット会社を通さずに「著作者がユーザーを直接信用して金融する構図」と書き、寄付で支えられるモデルから、所有権を売るモデルへ、連続的に移っていけると考えた。

この章では、作者が「要証明」と書いた信頼性フィルタを、模型で確かめる。

### 1.2 今の目で読む

**部分準備の通貨。** 予約(紙幣)を、実際の投げ銭(本貨)で支え、その比を公開する、という設計は、銀行の**部分準備制度**と同じ形をしている。銀行は、預金(求められれば払う約束)の一部だけを現金で持つ。予約と投げ銭の比は、預金と準備の比にあたる。作者の予約証拠金は、**準備率**(預金の一定割合を準備として置かせる規則)である。

**準備の証明。** 「送ポイント総額」と「予約総額」を公開して、紙幣の信用を保とうとする発想は、2022 年に暗号資産の取引所 FTX が破綻したあと、取引所に求められるようになった**準備の証明**(proof of reserves)と同じである。そこでも問題になったのは、持っている資産(準備)は示せても、負っている債務(払う約束)の全体を示すのは難しい、ということだった。投げ銭予約で言えば、送ポイント総額は確かめられても、予約がどれだけ本気かは確かめにくい。この章の模型は、その難しさを具体的に見せる。

**コストのかかる評価。** 高く評価するほどコストがかかるレーティングは、経済学では**コストのかかるシグナル**(Spence 1973)と呼ばれる。安く出せる信号は、誰でも出せるので信用されない。コストがかかるから、強い選好を表せる。近年の**二次投票**(票を、その数の 2 乗のコストで買う。Lalley and Weyl 2018)も、同じ問題意識から出ている。投げ銭は、評価にお金というコストを付けることで、レーティングに「責任」を持たせる。

**ポイントバックは、売り手による直接の与信。** 「著作者がユーザーを直接信用して金融する」は、売り手が買い手に信用を与える**企業間信用**(売り手信用)の形である。今の「後払い」(BNPL)や、クラウドファンディングの支援者に製品版を割り引く仕組みに近い。

### 1.3 模型

人工の利用者と、人工のページで、投げ銭予約の一年ほど(40 期)を作る。

- ページは 5 つのタグに 60 ずつ、計 300。3 割は最初から投げ銭を受けられ、3 割は永久に受けられず、残りはそのうち(一期に 3% の確率で)受けられるようになる。
- 正直な利用者 300 人。それぞれ、
  - 個人相場(予約の額の大きさ。対数正規分布)、
  - タグの好み、
  - 兌換の速さ(受けられるページへの予約を、一期に払う確率。平均 0.5)、
  - 誠実さ(予約のうち、いずれ本当に払うものの割合。平均 0.73。残りは忘れたり踏み倒したりする)、
  を持つ。1 割は「パッション」型で、大きく予約し、ゆっくり払う(一期に 2%)。作者が「100 万ポイントを予約して、折にふれ、少しずつ充足していく」使い方も許されるべきだとした人たちである。
- **スパムの利用者** 10 人。タグ 0 の関心を高く見せるため、決して投げ銭を受けられないページに、大きな額(40)を毎期予約する。払う気はない。
  - 「素朴な」スパムは、それだけをする。
  - 「装う」スパムは、さらに 4 期に一度、受けられるページに 1 ポイントを予約してすぐ払い、送ポイント総額を作る。40 期で 10 ポイントの出費である。

40 期の時点の開示(利用者ごとの送ポイント総額 S、予約総額 R、予約総額のうち送ポイント可能額 D)から、各タグの「収益力のある関心」を推定する。**真の値**は、そのあと「すべてのページが投げ銭を受けられるようになった」として、新しい予約なしに 200 期走らせ、実際に払われた額とする。受けられないページのせいで払われない分は、関心がないのではないので、真の値に含める。

指標はタグの間で関心を比べるためのものなので、**タグの取り分**(そのタグの推定 / 全タグの推定の和)が、真の取り分にどれだけ近いかで評価する。比が 1 なら正確、2 なら倍に見せられている。

推定のしかたは六つ。

| 名前 | 推定 |
|---|---|
| 予約総額 | R をそのまま集計する |
| 送ポイント総額 | S だけを集計する(本貨だけを見る) |
| フィルタ A | 作者の信頼性フィルタ。D / S が小さい(0.3 未満)人の R だけを集計。S = 0 の人は D / S が定まらないので、通す |
| フィルタ B | 同じだが、S = 0 の人は除く |
| 重み付け | 利用者ごとの兌換率 S / (S + D) を推定し(予約の少ない人は 0.7 に寄せる)、R に掛ける |
| 過去の実現度 | 予約し、受けられるようになってから 20 期以上たった予約のうち、払った額の割合で R に重みを付ける(作者の「送ポイント総額のうち長期間予約状態だったもの」) |

```python
%%writefile nagesen.py
# 投げ銭予約(2006)の模型。
# 利用者はページに投げ銭を「予約」し、ページが投げ銭を受けられるようになったら、自分の速さで払う(兌換)。
# 開示する三つの量(作者の言葉):
#   S = 送ポイント総額(実際に払った額、本貨)
#   R = 予約総額(紙幣の流通量。払ったものも含む予約の総額)
#   D = 予約総額のうち送ポイント可能額(受けられるのに、まだ払っていない額。兌換の猶予)
import numpy as np


def simulate(T=40, H=200, n_honest=300, n_spam=10, n_tags=5, pages_per_tag=60, spam_tag=0,
             spam_amount=40.0, spam_style='plain', margin=0.0, spam_budget=200.0, closed_tag=None, seed=0):
    """T 期のあいだ予約を集め、その時点の開示から各タグの「収益力のある関心」を推定する。
    真の値は、そのあと H 期、新しい予約なしで走らせて実際に払われた額。

    spam_style: 'plain' = 受けられないページに予約するだけ、'mimic' = 受けられるページに少し払って信用を装う。
    closed_tag: このタグのページは 9 割が永久に投げ銭を受けられない(企業や NPO のページなど)。
    margin: 予約証拠金の率。未払いの予約額の margin 倍を積み立てる必要があり、
            スパムの利用者は予算 spam_budget の範囲でしか予約できない(正直な利用者は予算に余裕があるとする)。
    """
    rng = np.random.default_rng(seed)
    n_pages = n_tags * pages_per_tag
    tag_of = np.repeat(np.arange(n_tags), pages_per_tag)
    # ページが投げ銭を受けられるようになる時刻。3 割は最初から、3 割は永久に受けられない、残りはそのうち。
    u = rng.random(n_pages)
    open_at = np.where(u < 0.3, 0, np.where(u < 0.6, 10 ** 9, rng.geometric(0.03, n_pages)))
    if closed_tag is not None:
        c = tag_of == closed_tag
        open_at[c] = np.where(rng.random(c.sum()) < 0.9, 10 ** 9, open_at[c])
    n = n_honest + n_spam
    spam = np.arange(n) >= n_honest
    scale = rng.lognormal(0.0, 0.8, n)             # 個人相場
    pay_rate = rng.beta(2.0, 2.0, n)                # 受けられるページへの予約を、一期に払う確率(兌換の速さ)
    passion = (rng.random(n) < 0.1) & ~spam         # 1 割は「パッション」型: 大きく予約し、ゆっくり払う
    pay_rate[passion] = 0.02
    honor = rng.beta(4.0, 1.5, n)                   # 予約のうち、いずれ本当に払うものの割合(残りは忘れる・踏み倒す)
    honor[spam] = 0.0
    pref = rng.dirichlet(np.ones(n_tags), n)       # タグの好み
    never = np.flatnonzero((tag_of == spam_tag) & (open_at > 10 ** 8))
    # 予約の記録: 利用者, ページ, 額, 予約した期, 払った期(未払いは -1)
    U, P, A, TT, PD, W = [], [], [], [], [], []      # W: その予約をいずれ払うか
    owed = np.zeros(n)                              # 未払いの予約額(証拠金の計算用)

    def add(i, p, a, t, paid=-1):
        U.append(i); P.append(p); A.append(a); TT.append(t); PD.append(paid)
        W.append(paid >= 0 or rng.random() < honor[i])
        if paid < 0:
            owed[i] += a

    def pay_step(t):
        if not U:
            return
        pg = np.array(P); paid = np.array(PD); us = np.array(U)
        due = np.flatnonzero((paid < 0) & (open_at[pg] <= t) & np.array(W))
        hit = due[rng.random(len(due)) < pay_rate[us[due]]]
        for k in hit:
            PD[k] = t
            owed[U[k]] -= A[k]

    for t in range(T):
        pay_step(t)
        payable = np.flatnonzero(open_at <= t)
        for i in range(n):
            if spam[i]:
                # スパム: 狙ったタグの、決して受けられないページに大きく予約する。払う気はない
                if margin == 0 or margin * (owed[i] + spam_amount) <= spam_budget:
                    add(i, rng.choice(never), spam_amount, t)
                if spam_style == 'mimic' and t % 4 == 0:
                    # 受けられるページに少額を予約してすぐ払い、送ポイント総額 S を作る
                    add(i, rng.choice(payable), 1.0, t, paid=t)
                continue
            for _ in range(rng.poisson(1.0)):
                tag = rng.choice(n_tags, p=pref[i])
                p = rng.choice(np.flatnonzero(tag_of == tag))
                add(i, p, scale[i] * rng.lognormal(0.0, 0.5) * (5.0 if passion[i] else 1.0), t)
    open_T = open_at.copy()
    snap = dict(user=np.array(U), page=np.array(P), amt=np.array(A), t=np.array(TT), paid=np.array(PD))
    open_at[:] = np.minimum(open_at, T)             # 真の値: すべてのページを受けられるようにして、新しい予約なしで走らせる
    for t in range(T, T + H):
        pay_step(t)
    truth_paid = np.array(PD) >= 0
    return dict(snap=snap, open_at=open_T, tag_of=tag_of, spam=spam, passion=passion, n=n, n_tags=n_tags, T=T,
                truth=np.bincount(tag_of[snap['page']], weights=snap['amt'] * truth_paid, minlength=n_tags))
```

```python
%%writefile -a nagesen.py
def disclose(res):
    """T 期の時点の開示: 利用者 × タグごとの S、R、D。"""
    s = res['snap']; T = res['T']
    paid = s['paid'] >= 0
    payable = res['open_at'][s['page']] < T
    n, nt = res['n'], res['n_tags']
    tg = res['tag_of'][s['page']]
    by = lambda w: np.bincount(s['user'] * nt + tg, weights=w, minlength=n * nt).reshape(n, nt)
    return dict(S=by(s['amt'] * paid), R=by(s['amt']), D=by(s['amt'] * (~paid & payable)))


def aged_rate(res, L=20, prior=0.7, prior_w=3.0):
    """利用者ごとの兌換率の推定: 受けられるようになって L 期以上たった予約のうち、払った額の割合。
    作者の「送ポイント総額のうち長期間予約状態だったもの」(過去の兌換実現度)にあたる。予約が少ない人は prior に寄せる。"""
    s = res['snap']; T = res['T']
    old = np.maximum(res['open_at'][s['page']], s['t']) <= T - L     # 予約して、かつ受けられるようになってから L 期以上
    paid = s['paid'] >= 0
    n = res['n']
    num = np.bincount(s['user'], weights=s['amt'] * old * paid, minlength=n)
    den = np.bincount(s['user'], weights=s['amt'] * old, minlength=n)
    return (num + prior * prior_w) / (den + prior_w)


def estimates(res, filt=0.3, prior=0.7, prior_w=3.0):
    """各タグの「収益力のある関心」の推定。
    naive    : 予約総額 R
    paid     : 送ポイント総額 S(いま払われている分だけ)
    filter_a : 作者の信頼性フィルタ。D_i / S_i < filt の利用者の R だけを集計する
               (S_i = 0 の人は D_i / S_i が定まらないので、0 として通す)
    filter_b : 同じだが、S_i = 0 の人は除く
    weighted : 利用者ごとの兌換率の推定 c_i = (S_i + prior·w) / (S_i + D_i + w) で R_i を重み付け
    aged     : 過去の兌換実現度(aged_rate)で R_i を重み付け
    """
    d = disclose(res)
    S, R, D = d['S'].sum(1), d['R'].sum(1), d['D'].sum(1)
    ratio = np.divide(D, S, out=np.zeros_like(D), where=S > 0)
    ok_a = ratio < filt
    ok_b = ok_a & (S > 0)
    c = (S + prior * prior_w) / (S + D + prior_w)
    return dict(naive=d['R'].sum(0), paid=d['S'].sum(0), filter_a=d['R'][ok_a].sum(0), filter_b=d['R'][ok_b].sum(0),
                weighted=(d['R'] * c[:, None]).sum(0), aged=(d['R'] * aged_rate(res, prior=prior, prior_w=prior_w)[:, None]).sum(0),
                pass_a=ok_a, pass_b=ok_b)
```

```python
%%writefile -a nagesen.py
def error_table(seeds=range(8), filt=0.3, which=0, **kw):
    """タグの間の取り分(各タグの推定 / 全タグの推定の和)を、真の取り分と比べる。
    返すのは (タグ which の、推定の取り分 / 真の取り分, ほかのタグの、同じ比の平均)(シードの平均)。
    指標は、タグの間で関心を比べるためのものなので、総額の水準ではなく取り分で見る。"""
    keys = ('naive', 'paid', 'filter_a', 'filter_b', 'weighted', 'aged')
    out = {k: [] for k in keys}
    passed = {'a': [], 'b': []}
    for s in seeds:
        res = simulate(seed=s, **kw)
        e = estimates(res, filt=filt)
        share_true = res['truth'] / res['truth'].sum()
        for k in keys:
            r = (e[k] / e[k].sum()) / share_true
            out[k].append((r[which], np.delete(r, which).mean()))
        sp = res['spam']
        m = lambda x: x.mean() if len(x) else np.nan
        passed['a'].append((m(e['pass_a'][sp]), m(e['pass_a'][~sp])))
        passed['b'].append((m(e['pass_b'][sp]), m(e['pass_b'][~sp])))
    return {k: np.mean(v, axis=0) for k, v in out.items()}, {k: np.mean(v, axis=0) for k, v in passed.items()}
```

### 1.4 一回走らせて、開示を見る

```python
import nagesen
res = nagesen.simulate(seed=0)
d = nagesen.disclose(res)
sp = res['spam']
print('全体   送ポイント総額 S = %.0f   予約総額 R = %.0f   予約総額のうち送ポイント可能額 D = %.0f   R / S = %.1f'
      % (d['S'].sum(), d['R'].sum(), d['D'].sum(), d['R'].sum() / d['S'].sum()))
print('正直な人の R = %.0f、スパムの R = %.0f' % (d['R'][~sp].sum(), d['R'][sp].sum()))
print('\nタグ        ' + '  '.join(f'{k:>6d}' for k in range(5)))
for lab, v in (('S', d['S'].sum(0)), ('R', d['R'].sum(0)), ('D', d['D'].sum(0)), ('真の値', res['truth'])):
    print(f'{lab:10s}' + '  '.join(f'{x:6.0f}' for x in v))
```

紙幣(予約)は、本貨(払った額)の 5 倍以上流通している。タグ 0 の予約総額は、スパムのせいで、ほかのタグの数倍に膨らんでいる。一方、払った額 S と、受けられるのに払っていない額 D には、スパムはほとんど現れない。スパムは、決して受けられないページに予約しているからである。

これが、この章の結果の核心になる。**D は「受けられるページ」への予約しか数えない。受けられないページへの予約は、開示のどの欄にも「怪しさ」として現れない。**

### 1.5 信頼性フィルタは「証明」できるか

```python
labels = {'naive': '予約総額', 'paid': '送ポイント総額', 'filter_a': 'フィルタ A', 'filter_b': 'フィルタ B',
          'weighted': '重み付け', 'aged': '過去の実現度'}
scen = [('スパムなし', dict(n_spam=0)), ('素朴なスパム', dict()), ('装うスパム', dict(spam_style='mimic'))]
print('タグ 0 の「推定の取り分 / 真の取り分」(8 シードの平均。1 なら正確)')
print(f'{"":14s}' + ''.join(f'{labels[k]:>12s}' for k in labels))
for lab, kw in scen:
    t, p = nagesen.error_table(**kw)
    print(f'{lab:14s}' + ''.join(f'{t[k][0]:12.2f}' for k in labels))
t, p = nagesen.error_table()
print('\n素朴なスパムで、フィルタを通る割合: A はスパム %.0f%%・正直な人 %.0f%%、B はスパム %.0f%%・正直な人 %.0f%%'
      % (100 * p['a'][0], 100 * p['a'][1], 100 * p['b'][0], 100 * p['b'][1]))
```

結果は次のとおりである(数字は上の出力)。

1. **スパムがなければ、どの推定も正確である**(比はほぼ 1)。利用者の誠実さや速さはばらばらだが、それはタグの間で平均され、取り分には効かない。
2. **素朴なスパムに対して、フィルタ A(S = 0 の人を通す)は、かえって悪くする**。スパムの利用者は、受けられるページに予約していないので D = 0 であり、S = 0 なので D / S は定まらない。0 として通すと、スパムは全員通り、正直な人は 3 割ほどしか通らない。フィルタが正直な人を落とし、スパムを残すので、タグ 0 の取り分は、何もしないときより大きく見せられる。
3. **フィルタ B(S = 0 の人を除く)は、素朴なスパムを完全に除く**。
4. **しかし、装うスパムには、フィルタ B も破られる**。受けられるページに 4 期に一度、1 ポイントを払うだけで、S > 0、D = 0 となり、D / S = 0 の「最も信頼できる利用者」になる。一人 40 期で 10 ポイントの出費で、10 人 × 40 期 × 40 ポイント = 1 万 6000 ポイント分の予約を「信頼できる紙幣」として流通させられる。
5. **利用者ごとの兌換率で重みを付けても(重み付け、過去の実現度)、ほとんど効かない**。スパムの予約は、受けられないページへのものなので、兌換率を測る材料がない。材料のない人には、ふつうの人と同じ値(0.7)を仮に置くしかなく、スパムの予約がそのまま数えられる。
6. **送ポイント総額 S(本貨)だけを見れば、スパムに強い**。スパムは払わないからである。

つまり、作者の「要証明」への答えは、**「証明できない。不都合がある」**である。理由は一つにまとめられる。

> D / S は、**受けられるページ**への予約が払われているかを見る。決して受けられないページへの予約は、D にも S にも現れない。だから、受けられないページに予約し、受けられるページで少し払っておけば、開示の上では、最も信頼できる利用者になれる。

これは、準備の証明が「持っている資産は示せても、負っている債務の全体は示せない」と言われるのと同じ構造である。予約という債務のうち、検査できる部分(受けられるページへの予約)だけが検査され、検査できない部分(受けられないページへの予約)は、検査をすり抜ける。

なお、フィルタ B には、もう一つの代償がある。S = 0 の人、つまり、まだ一度も払っていない新しい利用者を、すべて除くことである。作者が「払えないけど、他のユーザーの参考になるレーティングができる若者」を気にかけたのと、逆の方向に働く。

### 1.6 本貨だけを見ればよいか

送ポイント総額 S はスパムに強かった。では、本貨だけを見ればよいか。そうはいかない。投げ銭予約は、そもそも**投げ銭を受けられないページへの関心**を表すために考えられた(企業や NPO のページ、ブログ以外のページ)。

タグ 4 のページの 9 割が、決して投げ銭を受けられない(企業のページが多いタグ)とする。スパムはいないとする。

```python
t, p = nagesen.error_table(n_spam=0, closed_tag=4, which=4)
print('タグ 4(9 割が受けられない)の「推定の取り分 / 真の取り分」')
for k in labels:
    print(f'  {labels[k]:12s} {t[k][0]:.2f}')
```

本貨だけを見ると、受けられないページの多いタグへの関心は、本当の 1 割ほどにしか見えない。予約(紙幣)を数えれば、正確に見える。

**本貨は、偽れないが、受けられるところにしか流れない。紙幣は、どこにでも流れるが、偽れる。** これは、金本位制の時代に、金貨と紙幣の両方が要った理由そのものである。作者は、まさにこの二つを組み合わせて使おうとした。問題は、紙幣を偽れないようにする方法だった。

### 1.7 予約証拠金: 準備率としての証拠金

作者のもう一つの提案、予約証拠金を試す。スパムの利用者は、未払いの予約額の m 倍を、証拠金として積み立てなければならない。スパムの予算は 200 ポイントとする(1.5 節の装うスパムが払った 10 ポイントの 20 倍)。正直な利用者には、予算の余裕があるとする。

```python
ms = [0.0, 0.2, 0.5, 1.0, 2.0]
rows = [nagesen.error_table(spam_style='mimic', margin=m)[0] for m in ms]
fig, ax = plt.subplots(figsize=(6.4, 3.6))
for k, c in (('naive', PAL[0]), ('filter_b', PAL[1]), ('aged', PAL[2]), ('paid', PAL[3])):
    ax.plot(ms, [r[k][0] for r in rows], 'o-', color=c, label={'naive': 'total reservations R', 'filter_b': 'filter B',
            'aged': 'past realization', 'paid': 'paid only S'}[k])
ax.axhline(1, color=INK2, lw=1)
ax.set(xlabel='margin rate m', ylabel='estimated share / true share (tag 0)', title='Margin requirement vs. disguised spam')
ax.legend(fontsize=8); plt.show()
for m, r in zip(ms, rows):
    print(f'm = {m:3.1f}:  予約総額 {r["naive"][0]:.2f}   フィルタ B {r["filter_b"][0]:.2f}   過去の実現度 {r["aged"][0]:.2f}')
```

証拠金の率 m を上げると、スパムが出せる予約の総額は、予算 / m に抑えられる。m = 2 で、予約総額の取り分の誇張は 1 割ほどになる。m = 0.2 では、一人 200 / 0.2 = 1000 ポイント、つまり 25 回分(40 期のうち 25 期)まで予約を出せるので、効果は小さい。

これは、銀行の準備率と**信用乗数**の関係そのものである。準備(証拠金)の額が決まっていれば、出せる預金(予約)の総額は、準備 / 準備率までに限られる。作者の予約証拠金は、紙幣の発行量を、本人の持つ本貨に結び付ける規則だった。そして、この模型では、**開示とフィルタ(情報で信用を測る)よりも、証拠金(本貨の裏付けを求める)のほうが、偽りに強い**。

ただし、証拠金は、正直な利用者、とくにお金のない若者にも重くかかる。作者が「証拠金の額と割合を公開すれば、積み立てなくてもよい」とした例外は、その手当てだった。しかし、この例外を認めると、スパムも「公開して積み立てない」を選べる。作者は、その場合、証拠金を公開している利用者を集計から外せるようにしておけばよい、と書いている。結局、**払えない人の評価は、集計の中では、払える人の評価と同じ重みを持てない**。お金で責任を持たせる仕組みの、避けられない性質である。

### 1.8 遅い払い手と、払わない人は、見分けられない

最後に、「パッション」型の利用者を見る。大きく予約して、ゆっくり払う人たちである。作者は、この使い方も許されるべきだとした。

```python
res = nagesen.simulate(seed=0, n_spam=0)
rate = nagesen.aged_rate(res)
pas = res['passion']
print('過去の実現度(20 期以上たった予約のうち払った割合)の平均: パッション型 %.2f、それ以外 %.2f'
      % (rate[pas].mean(), rate[~pas].mean()))
```

パッション型の人は、誠実さ(いずれ払う割合)はほかの人と同じだが、過去の実現度では、ずっと低く見える。限られた期間の記録からは、**「ゆっくり払う人」と「払わない人」を見分けられない**。作者が支払いの猶予を本人の責任で認めたのは、自由のためだったが、その自由は、開示の上では信用の低さとして現れる。作者もそれを承知で、「レーティングの信頼性より自分のパッションを大事にしたいなら」と書いていた。

### 1.9 今の形

| 作者の構想 | 今の形 |
|---|---|
| 予約(受けられないページへの投げ銭を、受けられるまで預かる) | ブラウザ Brave の投げ銭(BAT)では、まだ受け取りの登録をしていない書き手への投げ銭は、ブラウザの中に 90 日間預かられ、その間に書き手が登録すれば渡り、しなければ送り手に戻る。予約に期限を付けた形である |
| 送ポイント総額・予約総額・兌換の猶予の開示 | 暗号資産の取引所の準備の証明(2022 年の FTX の破綻のあと)。利用者が自分の残高が負債の集計に含まれていることを、マークル木で確かめられる。ただし、資産の側は示せても、負債の全体を示すのは難しいとされる |
| 高く評価するほどコストがかかるレーティング | 二次投票、二次資金調達(第 6 章) |
| ポイントバック(寄付から所有権の購入へ連続的に移る) | 「好きな額を払う」販売と、あとからの格上げ。クラウドファンディングの支援者特典 |
| 予約証拠金 | 準備率、証拠金取引の証拠金 |

### 1.10 この章のまとめ

- 投げ銭予約は、予約を紙幣、投げ銭を本貨とする、**部分準備の通貨**である。作者は、紙幣の信用を開示(準備の証明)で保とうとし、偽りへの備えとして、信頼性フィルタと予約証拠金を考えた。
- 作者が「要証明」と書いた**信頼性フィルタは、証明できない**。D / S は受けられるページへの予約しか見ないので、受けられないページに予約し、受けられるページで少し払っておけば(この模型では 40 期で 10 ポイント)、最も信頼できる利用者に見える。S = 0 の人を通すと、フィルタはかえって偽りを助ける。
- **本貨(送ポイント総額)は偽れないが、受けられるページにしか流れない。紙幣(予約総額)はどこにでも流れるが、偽れる。** 両方が要る。
- 偽りに強いのは、情報(開示とフィルタ)よりも、**本貨の裏付けを求めること(予約証拠金)**である。出せる予約は、証拠金 / 率までに限られる(信用乗数)。
- ゆっくり払う人と払わない人は、限られた記録からは見分けられない。

次の章では、お金の向きを逆にする。読む人が払うのではなく、**読んでもらいたい人が払う**。

## 第 2 章 空スターと No Ad URL: 条件付きの支払いと、裏書き(2009、2013)

### 2.1 空スター(2009)

はてなスターは、記事に☆を付けて気持ちを伝える機能である(2007 年に始まった)。色の付いた「カラースター」は、はてなポイントで買えた。2009 年 6 月、作者は、はてなアイデアに次の機能を提案した。

> ポイントバック予約付き空スター。読んで欲しい人が読んでスターを付けてくれたらポイントを渡すよう、はてなに予約する機能。(…)社会人の弟子が師匠の指導を受けるとき、契約等にもとづきこの機能「空[から]スター」を使うことなどを想定しています。

手順は次のとおりである。

1. 書き手が、スターを選ぶところに新しく置かれた「空白のスター」を選び、読んで欲しい人とポイントを指定する。
2. 記事には、普通より大きい空白のスターが現れる。マウスを重ねると、読んで欲しい人の ID と、予約されたポイントが見える。
3. 読んで欲しい人が記事を訪れ、空白のスターをクリックしてスターを付けると、その人にポイントが送られる。

作者は、さらに次のような細部を書いている。

- 読んで欲しい人は、複数の空スターで複数指定できる。それとは別に、グループや OR 指定で「その内の誰か早い者勝ち」も指定できる。
- 書き手でない「紹介者」が、読んで欲しい人を指定してもよい。
- 予約の期間を定めたり、長く確定されない予約には、証拠金を予約額より少なくできるようにしたりしてもよい。
- 師匠が、いつか弟子に読んで欲しいものを、それとなく指定しておく使い方もある(その場合、通知は要らない)。

スターは、読まなくても付けられる。作者は、これを承知していた。

> スターを付けるのは読んでなくても付けられるわけであるが、「プロ読者」がそういうマナー違反をしてると、だんだんポイントバック付き読者に指定されなくなるだろうと想定している。

2010 年には、空スターを「名宛人が評価しないと意味がない」ものとして、「労働債権のような、師匠にいつまでも先取特権が残るもの」が望ましい、とも書いている。同じころ、作者は、はてなスターに「閲覧時間」(ページを開いてからスターを付けるまでの時間)を表示する機能も提案していた。読んだかどうかの、弱い手がかりである。

### 2.2 No Ad URL(2013)

2013 年、作者は、広告をクリックした先の募金サイトが、画面全体を覆う広告(オーバーレイ)を出してきたことに憤り、URL の指定で広告を消す Greasemonkey のスクリプト `no_ad_url.user.js` を書いた。

- URL に `no_ad=…` を付けると、指定した要素を消して、そこに大きな赤い×を描く(何かが消されたことがわかるように)。
- 版 0.2 で、`linker`(誰がリンクしたか)を足した。信用する linker の指定なら黙って従い、信用しない linker なら無視し、それ以外は確認を求める。
- のちに `linkee`(リンクを使うと想定している人)と `mature`(有効期限)も足し、リンクをコピーするための補助のスクリプトで、linker の後ろに自分を足していけるようにした。作者は、これを「no_ad 指定に対して『手形の裏書き』をするような感じ」と書いた。

そして 2018 年、作者は、EU の「リンク税」の議論の中で、二つをまとめて次のように書いた。

> 読ませたい記事がある linker が金を払って、linkee に読んでもらうというモデルである。(…)読む者が読むために金を払うのではなく、読んでもらう側が金を払うという「パラダイムシフト」なわけだが、理系の論文誌への投稿は読んでもらうために金を払うとも言えるといったように、これまでにないモデルとも言いがたい。

さらに、空スターでは、ポイントが新聞社の前を素通りして読み手に渡るので、新聞社が「No」と言えるよう、空スターの設置にサイトごとの課金を認めるべきかもしれない、とも書いている。

### 2.3 今の目で読む: 条件付きの支払い

**お金の向きが逆。** 普通は、読む人が読むために払う(購読料、投げ銭)。空スターでは、読んでもらいたい人が、読む人に払う。これは、作者の言うとおり珍しくない。論文の投稿料、広告(広告主は、見てもらうために払う)、試供品、そして**注意の市場**(人の注意そのものに値段が付く)。空スターの特徴は、**名宛人がいる**ことである。不特定の誰かではなく、特定の人に読んでもらうために払う。

**条件付きの支払いの予約。** 空スターは、

- 名宛人(一人、または「その内の誰か早い者勝ち」の組)、
- 条件(名宛人がスターを付ける)、
- 期限、
- 証拠金(予約額より少なくてよい)

を持つ支払いの約束である。これは、今の言葉で言えば、**エスクロー**(第三者が代金を預かり、条件が満たされたら渡す)であり、暗号資産のライトニングネットワークで使われる**ハッシュ時間ロック契約**(HTLC)と同じ形をしている。HTLC は、「ある秘密を示した人に払う。期限が過ぎたら払い手に戻す」という約束である。

第 1 章の言葉で並べると、作者の道具は次のようになる。

| 道具 | 金融の道具 |
|---|---|
| 投げ銭予約 | 紙幣(いつでも本貨に換える約束) |
| 空スター | 条件付きの手形(名宛人が条件を満たしたら払う約束) |
| No Ad URL(linker を足していく) | 裏書きできる手形(受け取った人が、次の人に回せる) |
| 証拠金を予約額より少なくする | 部分準備 |

**裏書き。** 手形の裏書きは、受け取った手形を次の人に渡すとき、裏に自分の名前を書くことである。裏書きした人は、振り出した人が払えなかったときの責任を負う。No Ad URL で linker を足していくのは、「このリンクを勧めたのは誰か」を連ねていくことで、責任の連鎖を作る。作者は、linker の名乗りに嘘を書いてはならないと法で決まっていれば、署名は要らない、とも書いている。

**「読んだ」は証明できない。** 空スターの弱点は、条件が「スターを付けた」であって、「読んだ」ではないことである。次の節では、空スターを HTLC の形で書いてみる。そうすると、何が証明でき、何ができないかが、はっきりする。

### 2.4 空スターを、ハッシュ時間ロックで書く

書き手は、記事の中に秘密の文字列を埋め込む(たとえば、本文の途中に置いた合言葉)。そして、そのハッシュ(第 3 章で詳しく扱う。文字列から作る指紋のようなもので、指紋から元の文字列は戻せない)と、名宛人と、期限を付けて、ポイントを預ける。名宛人が期限までに合言葉を示せば、ポイントはその人に渡る。期限を過ぎれば、書き手に戻る。

```python
%%writefile karastar.py
# 空スターを、ハッシュ時間ロック(HTLC)の形で書く。
# 書き手は、記事の中に秘密の文字列 s を埋め込み、そのハッシュ H(s) と名宛人と期限を付けてポイントを預ける。
# 名宛人が期限までに s を示せば、ポイントはその人に渡る。期限を過ぎれば、書き手に戻る。
# s を知るには記事を開く必要がある。示せるのは「開いた」ことで、「読んだ」ことではない。
import hashlib


def H(s):
    return hashlib.sha256(s.encode()).hexdigest()


class KaraStar:
    def __init__(self, writer, payees, amount, lock, expiry, deposit=None):
        """payees: 名宛人の集合(一人なら名指し、複数なら「その内の誰か早い者勝ち」)。
        deposit: 実際に預ける額(作者の「長期の予約は証拠金を予約額より少なくできる」)。既定は amount。"""
        self.writer, self.payees, self.amount, self.lock, self.expiry = writer, set(payees), amount, lock, expiry
        self.deposit = amount if deposit is None else deposit
        self.state = 'open'

    def claim(self, who, secret, t):
        if self.state != 'open' or t > self.expiry or who not in self.payees or H(secret) != self.lock:
            return 0.0
        self.state = 'paid:' + who
        return self.amount          # 証拠金が予約額より少なければ、差額は書き手がこのとき払う(払えなければ不渡り)

    def refund(self, t):
        if self.state == 'open' and t > self.expiry:
            self.state = 'refunded'
            return self.deposit
        return 0.0
```

```python
import karastar
secret = '本文の 3 段落目に置いた合言葉: しだれ桜'
s1 = karastar.KaraStar('writer', ['disciple'], amount=100, lock=karastar.H(secret), expiry=10)
print('名宛人でない人が合言葉を示す:', s1.claim('someone', secret, t=3))
print('名宛人が違う言葉を示す:      ', s1.claim('disciple', 'あてずっぽう', t=3))
print('名宛人が合言葉を示す:        ', s1.claim('disciple', secret, t=5), s1.state)
s2 = karastar.KaraStar('writer', ['a', 'b', 'c'], amount=100, lock=karastar.H(secret), expiry=10, deposit=30)
print('\nOR 指定、期限切れ:', s2.claim('b', secret, t=12), '/ 払い戻し(証拠金):', s2.refund(t=12), s2.state)
```

この形で示せるのは、**名宛人が記事を開いた(合言葉のところまで目を通した)**ことである。**読んで理解したこと**は示せない。合言葉は、読まなくても探せる。そして、一度示された合言葉は、ほかの人にも伝わる(OR 指定では、最初の一人が合言葉を言えば、ほかの人にも知られてしまう。HTLC で、秘密が一度明かされると、同じ秘密で鍵をかけた支払いがすべて解けるのと同じである)。

作者の「スターを付ける」も、HTLC の「合言葉を示す」も、**条件の確かめは、行為のごく一部しか捉えられない**。残りは、評判に任せるしかない。作者は、それを「プロ読者は、マナー違反をしていると、だんだん指定されなくなるだろう」と見込んだ。次の模型で、この見込みを確かめる。

### 2.5 模型: 評判で、読まない読み手は指定されなくなるか

- 書き手 65 人、読み手 100 人。読み手の約 3 割は、**読まずにスターだけ付ける**(不誠実な読み手)。
- 書き手は毎期、一人を名指しして空スターを置く。読み手が一期に応じられるのは一つまで。
- 書き手は、名指しした人が読まずにスターを付けたことを、確率 `detect`(既定 0.3)でしか見抜けない。見抜けなかった場合は、「よい読み」に見える(ちゃんと読んだ人は、いつも「よい読み」)。
- 書き手は、読み手ごとの「よい読み」と「悪い読み」の回数から見込みを作り、試しながら学ぶ(トンプソン抽出: 見込みの分布から一回引いて、いちばん高い人を選ぶ)。
- 評判は二通り。**私的**(書き手ごとの経験だけ)と、**公開**(全員の経験を合わせる。作者が求めた、スターの履歴からの「仕事ぶり」の分析にあたる)。
- 指定は二通り。**名指し**と、**OR 指定**(評判の上位 5 人を指定し、早い者勝ち)。読まずに付ける人は、読む人より速い。候補に一人でもいれば、9 割はその人が先にスターを付ける。
- 誠実な読み手は、一期に 0.5% の確率で入れ替わる(新顔は経験なしから始まる)。

測るのは、後半 100 期に、不誠実な読み手に渡ったポイントの割合である。不誠実な読み手は約 28% いるので、評判がまったく効かなければ 28% 前後になる。

```python
%%writefile -a karastar.py
# 空スター(2009)の模型。
# 書き手は毎期、読んで欲しい人を一人(または候補の組)指定して、ポイントつきの空スターを置く。
# 指定された人が読んでスターを付けると、ポイントが渡る。
# 読み手には、ちゃんと読む人と、読まずにスターだけ付ける人(不誠実な読み手)がいる。
# 書き手は、不誠実を見抜ける確率 detect でしか見抜けず、見抜けなかった不誠実は「よい読み」に見える。
import numpy as np


def simulate(T=300, n_writers=65, n_readers=100, frac_cheat=0.3, detect=0.3, share="private", mode="named",
             group=5, capacity=1, whitewash=False, prior=(1.0, 1.0), honest_turnover=0.005, seed=0):
    """share: 'private' = 評判は書き手ごとの経験だけ、'public' = 全員の経験を合わせた公開の評判。
    mode: 'named' = 一人を名指し、'or' = 評判の上位 group 人を指定し、早い者勝ち。
    whitewash: 公開の評判が悪くなった不誠実な読み手は、新しい ID で登録し直す。
    prior: 経験のない読み手への見込み(よい, 悪い)の仮の回数。(1, 3) なら新顔を疑ってかかる。
    honest_turnover: 誠実な読み手が一期に入れ替わる確率(新顔は経験なしから始まる)。
    返す値: 期ごとの、不誠実な読み手に渡ったポイントの割合、誠実な新顔(経験 5 回未満)の受け取り。
    """
    rng = np.random.default_rng(seed)
    cheat = rng.random(n_readers) < frac_cheat
    good = np.zeros((n_writers, n_readers)); bad = np.zeros((n_writers, n_readers))
    a0, b0 = prior
    cheat_share = np.zeros(T); newcomer_pay = np.zeros(T); n_jobs = np.zeros((T, n_readers))
    for t in range(T):
        # 誠実な読み手の入れ替わり
        out = (~cheat) & (rng.random(n_readers) < honest_turnover)
        good[:, out] = 0; bad[:, out] = 0
        G, B = (good.sum(0, keepdims=True).repeat(n_writers, 0), bad.sum(0, keepdims=True).repeat(n_writers, 0)) \
            if share == 'public' else (good, bad)
        load = np.zeros(n_readers, int)
        paid_cheat = paid = new_pay = 0.0
        for w in rng.permutation(n_writers):
            # トンプソン抽出: 見込みの分布から一回引いて、高い順に候補にする(試しながら学ぶ)
            score = rng.beta(G[w] + a0, B[w] + b0)
            score[load >= capacity] = -1
            order = np.argsort(-score)
            if mode == 'named':
                r = order[0]
            else:
                cand = order[:group]
                ch = cand[cheat[cand]]
                # 読まずに付ける人は、読む人より速い。候補に一人でもいれば、9 割はその人が先にスターを付ける
                r = rng.choice(ch) if len(ch) and rng.random() < 0.9 else rng.choice(cand[~cheat[cand]] if (~cheat[cand]).any() else cand)
            load[r] += 1; n_jobs[t, r] += 1
            paid += 1
            if cheat[r]:
                paid_cheat += 1
                if rng.random() < detect:
                    bad[w, r] += 1
                else:
                    good[w, r] += 1
            else:
                if good[:, r].sum() + bad[:, r].sum() < 5:
                    new_pay += 1
                good[w, r] += 1
        if whitewash:
            gs, bs = good.sum(0), bad.sum(0)
            redo = cheat & (gs + bs >= 5) & ((gs + a0) / (gs + bs + a0 + b0) < 0.6)
            good[:, redo] = 0; bad[:, redo] = 0
        cheat_share[t] = paid_cheat / paid
        newcomer_pay[t] = new_pay
    return dict(cheat_share=cheat_share, newcomer_pay=newcomer_pay, frac_cheat=cheat.mean())
```

```python
%%writefile -a karastar.py
def summary(seeds=range(6), last=100, **kw):
    """後半 last 期の、不誠実な読み手に渡ったポイントの割合(シードの平均)と、誠実な新顔への支払いの数(一期あたり)。"""
    rs = [simulate(seed=s, **kw) for s in seeds]
    return (float(np.mean([r['cheat_share'][-last:].mean() for r in rs])),
            float(np.mean([r['newcomer_pay'][-last:].mean() for r in rs])),
            float(np.mean([r['frac_cheat'] for r in rs])))
```

```python
import importlib
importlib.reload(karastar)            # 2.4 節で読み込んだあとに、模型の部分を書き足したので読み直す
rows = [('私的・名指し', dict()), ('公開・名指し', dict(share='public')),
        ('私的・OR 指定', dict(mode='or')), ('公開・OR 指定', dict(share='public', mode='or'))]
print('不誠実な読み手に渡ったポイントの割合(後半 100 期、6 シードの平均。不誠実な読み手は約 28%)')
for lab, kw in rows:
    c, nw, fc = karastar.summary(**kw)
    print(f'  {lab:14s} {100 * c:5.1f}%')
print('\n見抜く確率 detect を変える')
for sh in ('private', 'public'):
    print('  ' + {'private': '私的', 'public': '公開'}[sh] + ':  ' + '   '.join(
        f'detect {d}: {100 * karastar.summary(detect=d, share=sh)[0]:4.1f}%' for d in (0.0, 0.1, 0.3, 0.6)))
```

結果(数字は上の出力):

1. **評判が私的なら、作者の見込みは半分しか当たらない。** 不誠実な読み手への支払いは、28% から 2 割ほどに下がるが、なくならない。書き手一人の経験は少なく、見抜ける確率も低いので、不誠実な読み手は、新しい書き手のところで、また指定される。
2. **評判を公開すれば、見込みどおりになる。** 不誠実な読み手は、ほとんど指定されなくなる。見抜く確率が 0.1 でも、全員の経験を合わせれば十分である。作者が同じころ「スターを付けた履歴から、その『仕事ぶり』が分析されるべき」と書いたのは、この意味で正しかった。
3. **OR 指定(早い者勝ち)は、不誠実な読み手を呼び込む。** 私的な評判では、不誠実な読み手への支払いは、何もしないとき(28%)より**多く**なる。読まない人のほうが速いので、候補の組に一人でも混ざれば、その人が取る。早い者勝ちは、速さを報い、速さは読まないことと結び付いている。これは**逆選択**の一つの形である。
4. **見抜く確率が 0 なら、評判は何も選べない。** そのうえ、誠実な読み手の入れ替わりのせいで、長くいる不誠実な読み手のほうが、経験の多い「古参」として選ばれやすくなり、取り分は 28% を超える。

### 2.6 新しい ID で出直す読み手と、新顔の負担

評判の仕組みにつきものの抜け道は、評判が悪くなったら、新しい ID で出直すこと(ホワイトウォッシュ)である。Friedman and Resnick(2001)は、ID を安く作り直せるところでは、新顔を疑ってかかる(新顔に不利な初期値を置く)しかなく、それは誠実な新顔にも負担をかける、と論じた。

```python
print('公開の評判、名指し。不誠実な読み手への支払いの割合 / 誠実な新顔(経験 5 回未満)への一期あたりの支払いの数')
for lab, kw in (('基準', dict()), ('ID で出直す', dict(whitewash=True)),
                ('ID で出直す + 新顔を疑う', dict(whitewash=True, prior=(1.0, 3.0))), ('新顔を疑う', dict(prior=(1.0, 3.0)))):
    c, nw, fc = karastar.summary(share='public', **kw)
    print(f'  {lab:24s} {100 * c:5.1f}%   {nw:.2f}')
```

この模型では、ID で出直しても、不誠実な読み手はほとんど得をしない。評判のよい誠実な読み手が十分にいれば、経験のない新顔は、そもそも選ばれにくいからである。一方、新顔を疑ってかかる初期値を置くと、**誠実な新顔への支払いが、1 割から 3 割ほど減る**。不誠実への備えの代償は、正直な新しい人が払う。しかも、この模型では、不誠実な読み手への支払いも、かえって増える(0.5% から 5% ほどへ)。新顔を疑うと、経験の少しある古参(見抜かれずに残った不誠実な読み手を含む)が、相対的に有利になるからである。第 1 章で、信頼性フィルタが「まだ一度も払っていない新しい利用者」を締め出したのと同じ構造である。

作者が、空スターの最初の用途として**師弟**を挙げたのは、この意味で筋がよかった。師弟は、長く続く、相手の決まった関係である。読んだかどうかは、次に会ったときにわかる(見抜く確率が高い)。そして、相手を替えにくいので、ID を替えて出直すことができない。**証明できない条件は、関係の長さで補う**しかない。

### 2.7 近い構想: 貸出アフィリエイト、グループクレジット

作者は、同じころ、似た形の構想をいくつか出している。

- **貸出アフィリエイト**(2009 年ごろから)。貸し出された物(本やゲーム)を再生したあとに、購入を促す宣伝を表示し、それが購入につながったら、利益の一部を貸主にキックバックする。貸すことを、売ることの敵ではなく、売ることの入口にする。空スターと同じく、「誰のおかげで読んだ(買った)か」に支払う仕組みである。作者は、のちに、図書館がゲームを所蔵数の分だけ貸し出す形にも広げた(2021 年)。
- **グループクレジット**(2012 年)。グループが一つのクレジットカードを共有し、誰がいくら使ったかを、グループの皆が SNS で即座に見られるようにする。使いすぎた人を、グループが外したり、通報したりできる。作者は、これを、無尽講などの P2P の金融ができなかったことを、SNS の安い監視で可能にするもの、と書いた。2.5 節の言葉で言えば、**評判を公開することで、見抜く確率を上げる**仕組みである。ノート O で扱った無尽(ROSCA)とも、ここでつながる。

### 2.9 この章のまとめ

- 空スターは、名宛人・条件・期限・証拠金を持つ、**条件付きの支払いの予約**であり、今のエスクローやハッシュ時間ロック契約と同じ形をしている。No Ad URL の linker の連なりは、**手形の裏書き**である。作者は、紙幣(第 1 章)に続いて、手形と裏書きまでを作っていた。
- 条件として確かめられるのは「スターを付けた」「合言葉を示した」までで、「読んだ」ではない。残りは評判に任せるしかない。
- 作者の見込み「読まない読み手は、だんだん指定されなくなる」は、**評判が私的なら半分しか当たらず、公開すれば当たる**。
- **早い者勝ちは、読まない読み手を呼び込む**(速さは読まないことと結び付く)。
- 新顔を疑うと、正直な新顔が負担を払う。**証明できない条件は、関係の長さで補うしかない**。作者が最初に挙げた師弟という用途は、その点で筋がよかった。

# 第 II 部 証明と経路の層

## 第 3 章 時刻の証明: 時間記録デバイス、Loaded Magic、Tweet Sig(2009〜2018)

第 1・2 章の道具(予約、空スター、裏書き)は、どれも「いつ、誰が、何を約束したか」の記録の上に立っている。記録が書き換えられるなら、約束は意味を持たない。作者は、この記録の層についても、三つの構想を出していた。

### 3.1 三つの構想

**時間記録デバイス(2009 年 7 月)。** きっかけは、作者が「レジローン」と名付けた話だった。店員が、他の客の返品を装ってレジの金を友人に貸し、あとで手数料を上乗せして返させる。その証拠になるはずの店内の映像も、日付だけ書き換えられたら見抜くのは難しい。作者は「タイムスタンプの偽造防止というのはなかなか難しい」と書き、次のようなデバイスを考えた(2012 年の書き直しによる)。

- USB に差すようなデバイスで、読み出しと追記だけができ、時計を内蔵し、偽造の防止がなされ、PC 側の計算なしに電子署名ができる。
- 契約の文書を作ったら、このデバイスに書き込む。文書には、デバイスのメーカーによる署名が付いて保存される。
- ネットを介して、たとえばコンテンツを譲渡するとき、その契約に基づく譲渡が以前に行われていないことを示すため、譲渡の日付に署名したデバイスの記録目録を渡す。
- 受け取る側は、メーカーから取り寄せた公開鍵で記録を確かめ、契約を結ぶ。

作者は、「これができるということは、デジタル著作物の『物々交換』も可能で、つまりそれは **P2P 型の『電子マネー』**もできるということになる」と書いている。サーバーに記録を集める「サーバー型の電子金融」とは違う道として、である。

**Loaded Magic(2011 年 5 月)。** 作者は、「ひとこと」を書くためのウェブのエディタを作っていて、コメントの一部を削除するときの取り違えを防ぐために、コメントに 4 桁の「マジックナンバー」を付けていた。数字は乱数でよかった。作者は、ここで次のことを思い付いた。

> 「悪者」が、ページの送信途中に人知れず偽造を行える環境なら、「正義の味方」も、ページの送信途中に人知れず細工できるだろう。その可能性があることが、「悪者」を牽制できるのではないか?

具体的には、乱数を返すだけの関数 `allot_magic` を、わざわざ別のファイルから読み込んで使う。「正義の味方」は、送信の途中で、この関数を、秘密の塩(SALT)とコメントのハッシュの先頭 4 桁を返し、さらにコメントとハッシュを自分のサーバーに記録する関数に差し替える。すると、悪者が途中でコメントを書き換えても、正義の味方は、それが利用者の意図ではないという証拠をネット上に持っている。そして、悪者は、どのページで関数が差し替えられているかを知らない。

**Tweet Sig(2018 年 1 月)。** 時間記録デバイスがいつまでも現れないので、その代わりとして作った、Firefox と Chrome のアドオンである。ページの `#tweet-sig-in` の中身(たとえば取引の内容のハッシュ)を Twitter に投稿し、投稿の URL を `#tweet-sig-out` に書き戻す。作者は、これを「三文判程度の認証」と呼んだ。

> あるサイトが話している本人のものとわかっていれば、取引内容の符牒をそこで公開させれば、第三者はその符牒に沿った取引がいつあったかを知ることができる。

作者は弱点も書いている。ツイートは削除でき、非公開にもできる。ただ、詐欺のような犯罪になれば、削除したデータもサーバーには残っているだろうし、公証人のような第三者がフォローしておけばよい、と。

この章では、三つの構想を、今の暗号の道具で書き直す。使うのは、Python の標準の `hashlib`(ハッシュ)と `hmac`(共有鍵の署名)だけである。

### 3.2 ハッシュと、連鎖する時刻証明

**ハッシュ**は、どんな長さの文書からも、決まった長さ(SHA-256 なら 64 桁の 16 進数)の値を作る関数である。同じ文書からは同じ値が出て、文書が 1 文字でも違えば、まったく違う値になる。値から元の文書を戻すことも、同じ値になる別の文書を作ることも、事実上できない。文書の「指紋」と考えればよい。

1991 年、Haber と Stornetta は、ハッシュを鎖のようにつなぐ時刻証明を考えた。記録を一つ足すたびに、**前の記録のハッシュも含めて**ハッシュを取る。すると、最新のハッシュ(鎖の頭)は、それまでのすべての記録に依存する。この頭を、改ざんできない公開の場に載せれば、それより前のすべての記録が、その時点で存在し、その後書き換えられていないことが示せる。

```python
%%writefile timestamp.py
# 連鎖する時刻証明(Haber and Stornetta 1991 の線形の連鎖)。
# 記録を一つ足すたびに、前の記録のハッシュを含めてハッシュを取る。最新のハッシュ(頭)を公開の場に載せれば、
# それより前のすべての記録が、その時点で存在し、その後書き換えられていないことが示せる。
import hashlib
import hmac


def H(*parts):
    h = hashlib.sha256()
    for p in parts:
        h.update(p.encode() if isinstance(p, str) else p)
        h.update(b'\x00')
    return h.hexdigest()


class Chain:
    def __init__(self):
        self.entries = []           # (時刻, 文書のハッシュ, 前のハッシュ, この記録のハッシュ)

    def head(self):
        return self.entries[-1][3] if self.entries else H('genesis')

    def append(self, t, doc):
        prev = self.head()
        d = H(doc)
        self.entries.append((t, d, prev, H(str(t), d, prev)))
        return self.head()

    def verify(self, published_head):
        """記録を最初からたどり直し、公開された頭と一致するか。書き換えがあれば、最初に食い違った位置を返す。"""
        prev = H('genesis')
        for k, (t, d, p, h) in enumerate(self.entries):
            if p != prev or H(str(t), d, p) != h:
                return False, k
            prev = h
        return prev == published_head, None
```

```python
import timestamp as ts
c = ts.Chain()
for t, doc in enumerate(['予約: ページ A に 100 ポイント', '予約: ページ B に 30 ポイント',
                         '支払い: ページ A に 100 ポイント', '空スター: 弟子に 50 ポイント、期限 10']):
    head = c.append(t, doc)
print('公開する頭:', head[:16], '...')
print('そのまま検証:', c.verify(head))
# 二番目の記録を書き換える(30 → 300)。記録のハッシュはそのままにしておく
t, d, p, h = c.entries[1]
c.entries[1] = (t, ts.H('予約: ページ B に 300 ポイント'), p, h)
print('書き換えたあと:', c.verify(head), '← 2 番目(位置 1)で食い違う')
```

書き換えた記録のハッシュを計算し直しても、次の記録が「前のハッシュ」として古い値を持っているので、食い違いが出る。それも書き換えれば、その次で食い違う。最後まで書き換えると、頭が変わり、公開した頭と合わなくなる。**過去の一つを偽るには、それ以後のすべてと、公開した頭を偽らなければならない。**

### 3.3 マークル木: たくさんの文書を、一つの値にまとめる

Haber と Stornetta の会社 Surety は、1995 年から、利用者から集めたハッシュをまとめた一つの値を、毎週、New York Times の広告欄に載せた。新聞は、図書館に何部も残るので、あとから書き換えられない。たくさんのハッシュを一つにまとめるのに使うのが、**マークル木**である。二つずつ組にしてハッシュを取ることを繰り返し、最後に残る一つ(根)を公開する。ある文書が根に含まれていることは、木の高さ(文書が n 個なら log₂ n 個)のハッシュだけで示せる。

```python
%%writefile -a timestamp.py
# マークル木: たくさんの文書のハッシュを一つの根にまとめる(Surety が新聞に載せたのは、この根)。
# ある文書が根に含まれていることは、木の高さ(log2 n)個のハッシュで示せる。
def merkle_levels(leaves):
    lv = [[H(x) for x in leaves]]
    while len(lv[-1]) > 1:
        cur = lv[-1]
        if len(cur) % 2:
            cur = cur + [cur[-1]]
        lv.append([H(cur[i], cur[i + 1]) for i in range(0, len(cur), 2)])
    return lv


def merkle_proof(levels, i):
    proof = []
    for cur in levels[:-1]:
        if len(cur) % 2:
            cur = cur + [cur[-1]]
        j = i ^ 1
        proof.append((cur[j], j < i))      # (隣のハッシュ, 隣が左か)
        i //= 2
    return proof


def merkle_verify(doc, proof, root):
    h = H(doc)
    for sib, left in proof:
        h = H(sib, h) if left else H(h, sib)
    return h == root


def sum_tree(leaves):
    """マークル和木: 葉は (名前, 額)。節は (ハッシュ, 額の和)。根の和が、負債(予約)の総額になる。
    各人は、自分の葉から根までの証明で、自分の額が総額に含まれていることを確かめられる(準備の証明の負債の側)。"""
    lv = [[(H(n, str(a)), a) for n, a in leaves]]
    while len(lv[-1]) > 1:
        cur = lv[-1]
        if len(cur) % 2:
            cur = cur + [(H('pad'), 0)]
        lv.append([(H(cur[i][0], cur[i + 1][0], str(cur[i][1] + cur[i + 1][1])), cur[i][1] + cur[i + 1][1])
                   for i in range(0, len(cur), 2)])
    return lv


def sum_proof(levels, i):
    proof = []
    for cur in levels[:-1]:
        if len(cur) % 2:
            cur = cur + [(H('pad'), 0)]
        j = i ^ 1
        proof.append((cur[j], j < i))
        i //= 2
    return proof


def sum_verify(name, amount, proof, root):
    h, a = H(name, str(amount)), amount
    for (sh, sa), left in proof:
        if sa < 0:
            return False                     # 負の額で総額をごまかすのを防ぐ
        h, a = (H(sh, h, str(sa + a)) if left else H(h, sh, str(a + sa))), a + sa
    return (h, a) == root
```

```python
import importlib; importlib.reload(ts)    # 書き足したモジュールを読み直す
docs = [f'文書 {i}' for i in range(1000)]
lv = ts.merkle_levels(docs)
root = lv[-1][0]
proof = ts.merkle_proof(lv, 123)
print('1000 個の文書の根:', root[:16], '...   文書 123 の証明に要るハッシュの数:', len(proof))
print('文書 123 は含まれるか:', ts.merkle_verify('文書 123', proof, root), '  別の文書で同じ証明を使うと:', ts.merkle_verify('文書 124', proof, root))
```

### 3.4 マークル和木: 負債の側の証明

第 1 章で、準備の証明の弱点は、資産(本貨)は示せても負債(払う約束)の全体を示すのが難しいことだった、と書いた。2022 年の FTX の破綻のあと、取引所は、この負債の側を示すために**マークル和木**を使うようになった。葉に各人の残高を置き、節にはハッシュと残高の和を持たせる。根の和が負債の総額になり、各人は、自分の残高がその総額に含まれていることを確かめられる。

投げ銭予約で言えば、各人の予約額を葉にして根を公開すれば、「予約総額」を、利用者一人一人が確かめられる数にできる。

```python
reservations = [('利用者 %d' % i, a) for i, a in enumerate([120, 30, 5, 250, 60, 15, 80])]
lv = ts.sum_tree(reservations)
root = lv[-1][0]
print('予約総額(根の和):', root[1])
pr = ts.sum_proof(lv, 3)
print('利用者 3 が「私の 250 は含まれているか」を確かめる:', ts.sum_verify('利用者 3', 250, pr, root))
print('運営が利用者 3 の予約を 25 と少なく数えていたら:  ', ts.sum_verify('利用者 3', 25, pr, root))
```

ただし、これで示せるのは、**含まれている予約が少なく数えられていない**ことまでである。第 1 章のスパムの予約(払う気のない予約)は、和木に正しく含まれる。和木は、運営が負債を隠すことは防ぐが、利用者が払う気のない約束をすることは防がない。第 1 章の結論(偽りに強いのは証拠金)は、ここでも変わらない。

### 3.5 時間記録デバイスと、二重の譲渡

作者の時間記録デバイスを模型にする。デバイスは、追記だけができる連鎖の記録を持ち、内蔵の時計(利用者は動かせない)で時刻を付け、記録の目録にメーカーの鍵で署名する。ここでは、署名の代わりに HMAC(共有の鍵による署名)を使う。本物は公開鍵の署名にして、誰でもメーカーの公開鍵で確かめられるようにする。

```python
%%writefile -a timestamp.py
# 時間記録デバイス(2009 の構想)の模型。
# 追記しかできず、時計を内蔵し、記録にメーカーの鍵で署名する。ここでは署名の代わりに HMAC(共有鍵の署名)を使う
# (本物は公開鍵の署名にして、誰でもメーカーの公開鍵で確かめられるようにする)。
class Device:
    def __init__(self, maker_key, serial):
        self.key, self.serial, self.log, self.clock = maker_key, serial, Chain(), 0

    def record(self, text):
        self.clock += 1                                  # 内蔵の時計(利用者は動かせない)
        head = self.log.append(self.clock, text)
        return self.clock, head, hmac.new(self.key, f'{self.serial}|{self.clock}|{head}'.encode(), 'sha256').hexdigest()

    def catalog(self, nonce=''):
        """署名つきの記録目録: 全記録と、頭への署名。nonce は、相手がその場で出す合言葉(古い目録の使い回しを防ぐ)。"""
        head = self.log.head()
        return dict(serial=self.serial, entries=list(self.log.entries), head=head, clock=self.clock, nonce=nonce,
                    sig=hmac.new(self.key, f'{self.serial}|{self.clock}|{head}|{nonce}'.encode(), 'sha256').hexdigest())


def check_catalog(cat, maker_key, nonce=''):
    """目録がメーカーの鍵で署名され、記録が連鎖として正しく、こちらの出した合言葉 nonce を含むか。"""
    c = Chain(); c.entries = list(cat['entries'])
    ok, _ = c.verify(cat['head'])
    good_sig = hmac.compare_digest(cat['sig'], hmac.new(maker_key, f"{cat['serial']}|{cat['clock']}|{cat['head']}|{nonce}".encode(), 'sha256').hexdigest())
    return ok and good_sig


def already_transferred(cat, item):
    """目録の中に、その品物の譲渡の記録がすでにあるか(二重の譲渡の検出)。"""
    return any(d == H('transfer:' + item) for (_, d, _, _) in cat['entries'])
```

```python
import importlib; importlib.reload(ts)    # 書き足したモジュールを読み直す
maker = b'maker-secret-key'
dev = ts.Device(maker, serial='JRF-0001')
dev.record('契約: ゲーム X を購入')
dev.record('transfer:ゲーム X')            # アリスに譲った
# ボブに同じゲームを売ろうとする。ボブは、その場で合言葉を出して、目録を求める
cat = dev.catalog(nonce='bob-1234')
print('ボブ: 目録は本物か', ts.check_catalog(cat, maker, nonce='bob-1234'), ' / すでに譲渡済みか', ts.already_transferred(cat, 'ゲーム X'))

# 攻撃 1: 譲渡の前に取っておいた古い目録を見せる
dev2 = ts.Device(maker, serial='JRF-0002'); dev2.record('契約: ゲーム Y を購入')
old = dev2.catalog(nonce='')                # 前もって作っておいた目録
dev2.record('transfer:ゲーム Y')
print('古い目録(合言葉なし)を見せる: 本物と認められるか', ts.check_catalog(old, maker, nonce='bob-5678'))

# 攻撃 2: デバイスを複製する(鍵を抜き出せれば、二台目で別の歴史を作れる)
clone = ts.Device(maker, serial='JRF-0002'); clone.record('契約: ゲーム Y を購入')
cat2 = clone.catalog(nonce='bob-5678')
print('複製したデバイスの目録: 本物と認められるか', ts.check_catalog(cat2, maker, nonce='bob-5678'),
      ' / 譲渡済みと見えるか', ts.already_transferred(cat2, 'ゲーム Y'))
```

三つのことがわかる。

1. **正直に使えば、二重の譲渡は見抜ける。** 目録の中に、すでに譲渡の記録がある。
2. **古い目録の使い回しは、その場の合言葉で防げる。** 受け取る側が、その場で考えた合言葉(nonce)を含めて署名させれば、前もって作っておいた目録は使えない。Tweet Sig の説明にある「今ここで考えた文字列をサイトのどこかに表示させる」と同じ考えである。
3. **デバイスそのものを複製されると、防げない。** 鍵を抜き出して二台目を作れば、譲渡の記録のない「もう一つの歴史」を持てる。

3 は、P2P の電子マネーの根本の問題、**二重支払い**の問題である。解き方は、二つしか知られていない。

- **複製できないハードウェアに頼る。** 鍵を、取り出せないチップ(セキュアエレメント)の中に閉じ込める。作者の時間記録デバイスは、この道である。2020 年代、欧州中央銀行は、デジタルユーロの**オフライン払い**を、まさにこの形で設計している。スマートフォンなどのセキュアエレメントに価値を置き、二つの端末のあいだだけで決済し、銀行にも中央銀行にも取引の中身を見せない。現金に近い匿名性を目指している。そして、チップが破られたときの不正が、この設計の弱点だと指摘されている(作者も「時間記録デバイスはある程度クラックされる」と書いていた)。
- **全員で一つの記録を持つ。** 歴史を一つにするために、誰もが同じ連鎖の記録を見られるようにし、どの歴史を正とするかを、計算量の証明(次の節)で決める。2008 年の Bitcoin の道である。Bitcoin の論文は、参考文献に Haber と Stornetta の時刻証明を引いている。

作者は 2009 年に、サーバー型ではない P2P 型の電子マネーとして、前者の道を考えていた。Bitcoin の論文が出た翌年である。

前者の道を本気でとるなら、デバイスの**発行者が、造幣局のような責任を負う**必要がある(JRF の 2026 年 10 月の指摘)。紙幣の偽造を防ぐのは、紙幣そのものの技術だけではなく、造幣局が偽札を見つけ、回収し、その損失を負う仕組みである。時間記録デバイスで言えば、

- 発行者が、デバイスの鍵の抜き出しにくさを保証し(今のセキュアエレメントは、第三者の評価の認証を受けている)、
- 複製が見つかったら(同じ品物の譲渡が、二つの目録に現れたら)、そのデバイスの鍵を失効させ、
- 複製で生じた損失を、発行者が負う。

損失を負うなら、発行者には、複製の被害を小さくする誘因が生まれる。一番効くのは、**一台のデバイスが持てる値打ちに上限を置く**ことである。複製されても、被害は「複製の数 × 上限」までで止まる。デジタルユーロでも、一人が持てる額に上限(3000 ユーロ前後が議論されている)を置き、オフラインで受け取った値打ちは、いずれオンラインに戻して照合する設計が検討されている。上限と、あとからの照合で、複製を「防ぐ」のではなく「被害を限り、見つけて止める」。作者が「時間記録デバイスはある程度クラックされる。しかし、クレジット IC カードがクラックされてないことになっているように、何らかの方法で、クラックへの弥縫策があるとできるはず」と書いたのは、この方向である。

### 3.6 Loaded Magic: 見えない監査による牽制

Loaded Magic の論理は、暗号よりも、**監査の経済学**に近い。悪者は、ページを途中で書き換えるとき、そのページの `allot_magic` が乱数のままか、「正義の味方」に差し替えられているかを知らない。差し替えられているページの割合を q とする。差し替えられたページを書き換えると、表示される 4 桁の数字が、記録されたハッシュと合わなくなり、証拠が残る(悪者が秘密の塩を知らずに、偶然 4 桁を合わせられる確率は 1 万分の 1)。悪者が N ページを書き換えたとき、一度でも証拠を残す確率は 1 − (1 − q)^N である。

```python
print('書き換えたページの数 N と、一度でも証拠を残す確率(差し替えられたページの割合 q ごと)')
print('   N   ' + ''.join(f'q={q:<8}' for q in (0.001, 0.01, 0.05)))
for N in (1, 10, 100, 1000):
    print(f'{N:5d}  ' + ''.join(f'{1 - (1 - q) ** N:9.3f} ' for q in (0.001, 0.01, 0.05)))
```

差し替えが 1% のページにしかなくても、100 ページを書き換えれば、6 割以上の確率で証拠が残る。**少数のランダムな監査が、大規模な不正を抑える。** これは、Becker(1968)の犯罪の経済学(罰の重さと、見つかる確率の積が抑止を決める)と同じ考えであり、税務調査や、抜き打ちの検査の理屈である。

今の形で一番近いのは、**証明書の透明性**(Certificate Transparency、2013 年〜)である。ウェブサイトの暗号の証明書を発行するたびに、追記しかできない公開の記録に載せさせ、独立の監視者が記録を見張る。偽の証明書を発行すれば、どこかの監視者に見つかる。作者の「正義の味方」は、この監視者にあたる。作者は「国家や信託銀行など合法的な『正義の味方』はブログのプロバイダを味方に付けられるのだから、最終的に有利なのは『正義の味方』だろう」と書いたが、今の監視者は、国家ではなく、ブラウザの会社や研究者である。

### 3.6.1 ランダムことわざ(2012): 暗号を信用できなくても働く目印

作者は、2012 年 2 月の「ユーザーによるコメント『削除』をココログプロで実現する」で、コメントを JavaScript で暗号化する仕組みを作った。そのとき、暗号化の前に、**本文の最後に、ランダムに選んだことわざやニュースを付け足す**ようにした。

> もし暗号の鍵がバラまかれるようなことが起きているなら、そこに同時に表示されることわざやニュースが人々の口にのぼるようになる、そこまでの影響力がないとしても、はっきりとは言えない立場のものが、盗聴があるのをそれとなくほのめかすのに使えるようになる…と夢想した。

暗号が破られて中身が読まれても、読んだ人は、付け足されたことわざも読む。本来その文を読んでいるはずのない人(作者の言う「正義の味方」)が、「盗聴がある」と口に出して言えない立場でも、そのことわざを何気なく口にすることで、それとなく知らせられる。

この仕掛けの強さは、**暗号やハッシュ(や、それを計算するブラウザ)そのものを信用できなくても働く**ことにある。Loaded Magic は、ハッシュを記録するサーバーと、それを差し込む経路を信用する必要があった。ランダムことわざは、漏れた中身が人の目に触れるという、社会の側の経路だけで働く。作者の「システムを造る際にソーシャルな要素を考慮に入れる」の、具体的な形である。

情報の世界では、似た仕掛けが二つ知られている。

- **カナリアの罠**(canary trap): 同じ文書を、少しずつ違う言い回しで何人かに渡し、漏れた文書の言い回しから、誰が漏らしたかを突き止める。ことわざが暗号化ごとに違えば、漏れた中身のことわざから、どの暗号文(どの鍵)が漏れたかがわかる。
- **令状のカナリア**(warrant canary): 「当局から秘密の命令を受けたことはない」と定期的に公表しておき、それが消えることで、口外を禁じられた命令があったことを知らせる。作者の「はっきりとは言えない立場のものが、それとなくほのめかす」と同じ問題を解いている。

小さなコードで、カナリアの罠として働くことを示す(暗号の代わりに、ハッシュから作った簡単な鍵の流れで、本文を隠す)。

```python
import hashlib, random
PROVERBS = ['身から出た錆。', '猿も木から落ちる。', '鶴は千年、亀は萬年。', '温故知新。', '七転び八起き。',
            '急がば回れ。', '石の上にも三年。', '情けは人の為ならず。']

def keystream(key, n):                                  # 暗号の代わり(説明用。実用の強さはない)
    out, c = b'', 0
    while len(out) < n:
        out += hashlib.sha256(f'{key}|{c}'.encode()).digest(); c += 1
    return out[:n]

def seal(text, key, rng):
    salt = rng.choice(PROVERBS)                         # 暗号化の前に、ランダムなことわざを付け足す
    data = (text + '\nランダムことわざ: ' + salt).encode()
    return bytes(a ^ b for a, b in zip(data, keystream(key, len(data)))), salt

rng = random.Random(0)
box = {}
for who in ('A さん', 'B さん', 'C さん', 'D さん'):
    ct, salt = seal('ここだけの話です。', key=f'{who} の鍵', rng=rng)
    box[who] = (ct, salt)
    print(f'{who} のコメント: 暗号文 {ct[:12].hex()}...  付け足したことわざ(書いた本人と管理者だけが知る) {salt}')
# ある日、どこかで「ここだけの話です。…七転び八起き。」という文が出回った
leaked_salt = box['C さん'][1]
print('\n出回った文の終わり: ランダムことわざ:', leaked_salt)
print('このことわざを付けて暗号化したのは:', [w for w, (_, s_) in box.items() if s_ == leaked_salt])
```

ことわざの種類が少ないと、同じことわざが重なり、漏れたものを一つに絞れないことがある(上の例でも重なりうる)。作者のことわざ集(`proverbs.js`)には、本などから選んだ七百ほどのことわざや名言があり、最初の版では、その日のはてなブックマークの人気の記事の題も付け足していた(その部分は、使っていた Google Feed API が 2016 年 12 月に終わったため、のちに外された)。ニュースは毎日変わるので、**いつ暗号化したものが漏れたか**の目印にもなる。時刻の証明(3.2 節)の、社会の側の版である。

### 3.7 計算量の証明

作者は、のちに(2020 年と 2025 年)、Loaded Magic のプログラムの中で「ハッシュの先頭何文字かだけ一致させて計算量を調整する」処理を使っていたこと、それが、自分よりずっと前にサトシ・ナカモトが考えていたことだと知った、と書いている。

Loaded Magic では、ハッシュの先頭 4 桁を「マジックナンバー」にした。逆向きに、**ハッシュの先頭 k 桁が 0 になるような数を探させる**と、それには平均 16^k 回の試行がかかる。探し当てた数は、それだけの計算をしたことの証明になる。これが**計算量の証明**(Proof of Work)である。もとは、迷惑メールを防ぐために、メール一通ごとに少しの計算を求める案として出された(Dwork and Naor 1992、Back の Hashcash 1997)。Bitcoin は、これを、どの歴史を正とするかの決め方に使った。

```python
%%writefile -a timestamp.py
# 計算量の証明(PoW): ハッシュの先頭 k 桁が 0 になる数(nonce)を探す。平均 16^k 回の試行がかかる。
# Loaded Magic(2011)の「マジックナンバー = ハッシュの先頭 4 桁」は、逆向きに読むと、同じ道具になる。
def pow_search(text, k, limit=10 ** 7):
    for nonce in range(limit):
        if H(text, str(nonce)).startswith('0' * k):
            return nonce
    return None
```

```python
import importlib; importlib.reload(ts)    # 書き足したモジュールを読み直す
import time
for k in (1, 2, 3, 4):
    n_txt = 200 if k < 4 else 20
    t0 = time.time()
    tries = [ts.pow_search(f'記録 {i}', k) + 1 for i in range(n_txt)]
    print(f'先頭 {k} 桁が 0: 平均の試行 {np.mean(tries):8.0f} 回(理論 16^{k} = {16 ** k:6d})   一つあたり {1000 * (time.time() - t0) / n_txt:6.2f} ms')
```

一桁増やすごとに、かかる計算は 16 倍になる。計算の難しさを、桁の数で調整できる。

ノート O の第 3 章では、ペニーオークションと入札講(無尽)の「レント散逸」(賞品の値打ちが、取り合いのコストで食いつぶされる)を扱い、Bitcoin の採掘も同じ構造だと書いた。計算量の証明は、**偽造を防ぐための費用を、わざと払わせる**仕組みである。第 1 章の予約証拠金(偽りを防ぐために本貨を積ませる)と、同じ種類の解き方だと言える。証拠金はお金を積ませ、計算量の証明は電気を燃やさせる。

### 3.8 Tweet Sig: 公開の場としての SNS

Tweet Sig は、3.2 節の「頭を公開の場に載せる」を、Twitter で行うものと読める。Surety が新聞を使ったところを、SNS で置き換えた。作者が挙げた弱点(削除できる、非公開にできる)は、新聞にはない弱点である。新聞は、何部も刷られて図書館に残る。SNS の投稿は、会社のサーバーに一つしかない。

ほぼ同じ形の仕組みは、2014 年から **Keybase** が使っていた。Keybase は、利用者が暗号の鍵を持っていることを、Twitter などに署名つきの投稿を載せることで示させた。誰でも、その投稿を見て確かめられる(Keybase は 2020 年に Zoom に買収され、開発は止まった)。今の SNS では、Mastodon の、プロフィールに載せた自分のサイトへのリンクが相互に張られていることで本人と確かめる仕組みや、Bluesky の、自分のドメインを名前にする仕組みが、同じ役割をしている。

### 3.9 投げ銭予約の開示を、公開の記録で満たす

ここまでの道具を、第 1 章の投げ銭予約に戻してみる。作者が 2006 年に求めた開示(送ポイント総額、予約総額、兌換の猶予)は、運営(はてな)が集計して見せるものだった。運営を信じるしかない。

この章の道具を使えば、次のようにできる。

1. 予約と支払いを、一つずつ連鎖の記録に足していく(3.2 節)。
2. 定期的に、その頭を公開の場(SNS、新聞、ほかの会社の記録)に載せる(3.3 節、Tweet Sig)。
3. 各人の予約額を和木にして、その根も公開する(3.4 節)。各人は、自分の予約が総額に含まれていることを確かめられる。

こうすれば、運営を信じなくても、開示が正しいことを、誰でも確かめられる。これは、Bitcoin が実際に通った道(Haber と Stornetta の時刻証明の上に通貨を乗せた)と同じである。ただし、確かめられるのは、記録が書き換えられていないことまでである。**記録された約束が本気かどうか**は、第 1・2 章で見たとおり、別の仕組み(証拠金、評判)に頼るしかない。

### 3.10 今の形

| 作者の構想 | 今の形 |
|---|---|
| 時刻を記録の連鎖で証明する | Haber と Stornetta の連鎖する時刻証明(1991)、OpenTimestamps |
| 公開の場に頭を載せる(Tweet Sig) | Surety の New York Times の広告欄(1995〜)、Keybase の投稿による証明(2014〜) |
| 時間記録デバイス(追記のみ、時計、署名、P2P の電子マネー) | スマートフォンのセキュアエレメント、デジタルユーロのオフライン払いの設計 |
| Loaded Magic(第三者がハッシュを記録して、偽造を牽制する) | 証明書の透明性(2013〜)と、その監視者。画像の来歴の証明(C2PA) |
| ハッシュの先頭を合わせて計算量を調整する | 計算量の証明(Hashcash 1997、Bitcoin 2008) |
| 予約総額の開示 | マークル和木による準備の証明(2022〜) |

### 3.11 この章のまとめ

- 作者の三つの構想は、**記録の層**を作ろうとするものだった。時間記録デバイスは、記録をハードウェアで守る。Loaded Magic は、見えない第三者の記録で牽制する。Tweet Sig は、公開の場に符牒を載せる。
- 今の言葉では、連鎖する時刻証明、マークル木、監査の記録(証明書の透明性)、そして計算量の証明である。作者は、これらを、Bitcoin とほぼ同じ時期に、独立に考えていた。
- 時間記録デバイスは、正直に使えば二重の譲渡を見抜け、その場の合言葉で古い目録の使い回しも防げる。しかし、**デバイスの複製は防げない**。P2P の電子マネーには、複製できないハードウェア(作者の道、デジタルユーロのオフライン払い)か、全員で一つの記録を持つこと(Bitcoin の道)が要る。
- 少数のランダムな監査は、大規模な不正を抑える(1% の監査で、100 件の不正の 6 割以上に証拠が残る)。
- 記録の層で確かめられるのは、**記録が書き換えられていないこと**までである。約束が本気かどうかは、価値の層の仕組み(証拠金、評判)に頼るしかない。

## 第 4 章 誰が仲立ちするか: 日銀カード、PP ログイン・PP 振込、印栞、PP Authorizer(2001〜2017)

価値の層(第 1・2 章)と記録の層(第 3 章)の下に、もう一つ層がある。お金を払うとき、ログインするとき、**誰が仲立ちし、身元を誰に渡すか**である。作者は、この層について、2001 年から 2017 年まで、少しずつ形を変えながら、同じ問題を考え続けた。この章には、シミュレーションはない。作者の構想を順に読み、今の形と並べる。最後に、小さなコードで、作者が気にした問題の今の解き方を一つだけ示す。

### 4.1 「日銀カード(仮称)」構想(2001〜2002)

作者は、2001 年ごろ、金融の仕組みの勉強のまとめと、**支払い専用の匿名口座**の提案を書き、2002 年に本名で公開した。支払いのためだけの口座を、身元と切り離して持てるようにする、という案である。2006 年に再録したとき、作者は、この案を自分で退けている。

> 現在では、支払い専用匿名口座は善意の第三者が偽造カードとパスワードを取得したとき、救済手段がないため、あまり良いものではないと考えています。残り度数などがカード内にあれば偽造カードによるものが自分の支払いでないことを示せますが、それなら今普及し始めている Edy などの IC カード型電子マネーを使えば済むことでしょうし。

匿名にすると、盗まれたときに「自分ではない」と示す手段もなくなる。**匿名と救済は、ぶつかる**。作者のこのあとの構想は、すべて、この衝突をどう解くかの試みとして読める。

### 4.2 PP ログイン・PP 振込(2006)

2006 年 12 月、作者は「保証付き匿名ログインと即時銀行振込を可能にして! ── プライバシープロバイダ API(仮称)」を書いた。PP は privacy-protected の略である。

登場するのは三者である。

- **閲覧する人**(利用者)
- **PP クライアントサイト**(利用者がいま見ているサイト。掲示板やネットの店)
- **PP サーバーサイト**(利用者の身元を知っている、信用の高いサイト。銀行やプロバイダ)

仕組みは次のとおり。

1. 利用者が、クライアントサイトのログインや支払いの準備画面で、ブックマークレット(ブラウザのお気に入りに登録した小さなスクリプト)をクリックする。
2. ブックマークレットは、利用者をサーバーサイトに移し、クライアントサイトの URL を伝える。
3. サーバーサイトは、クライアントサイトから、何を求めているか(ログインか、支払いの額と、あいまいな内訳か)の XML を取ってくる。
4. サーバーサイトで、利用者がログインし、確認する。
5. サーバーサイトは、終わったことだけをクライアントサイトに知らせ、利用者を元のページに戻す。

肝心なのは、**サーバーサイトの持つ身元の情報は、クライアントサイトに渡らない**ことである。クライアントサイトが知るのは、「このサーバーサイトが保証する誰か」であることと、支払いが済んだことだけである。利用者がクライアントサイトで問題を起こしたら、クライアントサイトはサーバーサイトに照会し、両方の管理者が問題を認めたときだけ、身元が明かされる。

作者は、応用として次の三つを挙げた。

- **匿名的な ID の取得とログイン**: サイトごとに別の ID とパスワードを、サーバーサイトが作って管理する。
- **保証付きの匿名ログイン**: ID を名乗らずに、「保証された誰か」としてログインする。
- **即時の銀行振込(PP 振込)**: 支払いのときだけ、銀行が主導権を握る。

動機として、作者は二つを書いている。一つは、**ID の相続**である。持ち主が亡くなったとき、第三者が関わって ID を譲るには、パスワードを利用者の側で管理していると難しい。もう一つは、作者がクレジットカードを使うことに強い抵抗があり、ネット経由の銀行振込が安全なまま、もっと簡単になって欲しかったことである。

そして、欠点も三つ書いている。二社以上がやらなければ意味がないこと。単純で目新しい技術を使っていないので、すでに誰かが試して失敗しているかもしれないこと。ネットから本物の匿名の文化が失われる危険があること。

4.1 節の衝突への答えは、ここではっきりしている。**サイトに対しては匿名、信用の高い第三者には身元がある**。盗まれたときの救済は、第三者が引き受ける。

### 4.3 印栞(いんかん)と OAuth(2011〜2012)

2011 年 5 月、作者は、OAuth(あるサイトが、利用者の許可を得て、別のサイトの利用者の情報や機能を使うための仕組み)を使う CGI を自分で書いていて、「OAuth は PP ログインに使えそうだ」と気づいた。同時に、問題も見つけた。

OAuth では、利用者が使っているサイトが、利用者のブラウザを、許可を出すサイト(はてなや Twitter)に**自動で移す**(リダイレクトする)。作者は、こう書いた。

> 簡単にリダイレクトしてしまうような常態になると、知らない間に偽装サイトが表示されていることが起きうる。

信用の低いサイトが、信用の高いサイトへ利用者を自動で移すのが当たり前になると、信用の低いサイトは、本物そっくりの偽のサイトへ移すこともできる。利用者は、移った先で、本物のつもりでパスワードを入れてしまう。**フィッシング**である。

作者の対策は、自動で移さないことだった。サイトは、移るための URL を表示するだけにする(しかも、普通のリンクにしてはいけない)。利用者は、自分のお気に入りに登録したブックマークレットをクリックして、そこへ移る。作者は、このブックマークレットを「**印栞**」と名付けた。印鑑のように、自分の手で押すものである。作者は、これを家のインターフォンにたとえている。どんな訪問者にも、いったん門の前で待ってもらい、インターフォンを押してもらう。

作者は、さらに、銀行協会がブラウザのアドオンとしてボタンを配り、ページに支払いの合図があり、それが利用者の登録した銀行と合えば、ボタンが光ってクリックを促す、という形を考えた(**OAuth 振込**)。「クレジットカードや PayPal でももちろんいいが、私は分散的に銀行振込でやって欲しい」。

2012 年には、OpenID(大きなサイトが、利用者が誰かを他のサイトに保証する仕組み)と OAuth を比べ、OpenID は「オレ様のほうがデカイんだからここは通せよ」というプロトコルで、発想が「セレブすぎる」と書いた。そして、セレブにふさわしい役割は、アジアでは**グループクレジット**(第 2 章)のようなものを仕切る「親分」の ID になることだろう、とも書いている。

### 4.4 PP Authorizer と PP Interrupter(2017)

2017 年 11 月、作者は、印栞をブラウザのアドオンにした **PP Authorizer** を、Firefox と Chrome 向けに公開した。

- ページの中の、`pp-authorizer` という印の付いたリンクを探し、アドレスバーにアイコンを出す。
- 利用者がアイコンをクリックすると、登録しておいた認証プロバイダ(初めから Twitter とはてなが入っている)のうち、そのページで使えるものが並び、選ぶとそこへ移る。
- 認証プロバイダは、設定画面か、ネットに置かれた `.ppauth.js` というファイルを読んで登録する。

作者の夢は、この機能が**ブラウザに標準で搭載され**、すべてのサイトが自動のリダイレクトをやめることだった。初めから入れる認証プロバイダを増やしたい組織(たとえば銀行協会)から寄付を募り、ブラウザの開発資金が潤沢になれば、とも書いた。

しかし、PP Authorizer を使うには、サイトの側が自動のリダイレクトをやめなければならない。そこで作者は、もう一つのアドオン **PP Interrupter** を作った。特定の URL(信用できる認証プロバイダ)へのアクセスを、いったん止める。逆転の発想である。

> 本当にやりたいのは、悪意のあるサイトへの自動的なリダイレクトを禁止することである。しかし、悪意のあるサイトへ導かれるかどうかは判定が難しい。そこで、逆転の発想で、判定が容易な健全なサイト(認証局)へのアクセスを一旦ブロックすることを習慣化し、ブロックがあるはずなのにないことによって、詐欺サイトに導かれたことを検知するようにするのだ。

本物に行くときは必ず止められる。止められなかったら、そこは本物ではない。作者は「PP Interrupter は現実への対応だが、PP Authorizer は未来への投資である」と書いた。

### 4.5 今の目で読む

作者の構想を通して読むと、問いは一貫している。

1. **サイトに身元を渡さない**(PP ログイン)。
2. **身元を知る第三者は、信用の高いところに限る**(PP サーバーサイト、認証プロバイダ、銀行)。
3. **利用者を第三者へ移す経路を、偽れないようにする**(印栞、PP Authorizer、PP Interrupter)。

3 は、セキュリティの言葉では**信頼できる経路**(trusted path)の問題である。利用者が「いま本物の銀行と話している」と確かめるには、偽のサイトが真似できない何かが要る。作者は、それを**利用者自身の手の動き**(自分のお気に入りに登録したブックマークレットを押す)に求めた。偽のサイトは、利用者のお気に入りの中身を書き換えられないからである。

今の解き方は、それを**ブラウザ**に求めた。ブラウザは、いまどのサイト(出所、origin)を表示しているかを知っている。偽のサイトは、ブラウザに出所を偽れない。そこで、ログインの鍵を出所に結び付けておけば、偽のサイトでは、鍵がそもそも使えない。これが**パスキー**(WebAuthn)の考え方である。

小さなコードで、違いを示す。パスワードは、どのサイトで入力しても同じ値なので、偽のサイトに入力すれば、そのまま本物で使われる。出所に結び付けた鍵は、ブラウザが「いまのサイトの出所」と組み合わせて使うので、偽のサイトで作った答えは、本物では通らない(本物のパスキーは公開鍵の署名を使うが、ここでは考え方だけを、ハッシュで示す)。

```python
import hashlib, secrets

def H(*xs):
    return hashlib.sha256('|'.join(xs).encode()).hexdigest()

user_secret = secrets.token_hex(16)                     # 利用者の端末の中だけにある秘密
REAL, FAKE = 'https://bank.example', 'https://bank-example.login.example'

# パスワード: どこで入力しても同じ値
password = 'correct horse'
stolen = password                                       # 偽のサイトで入力させて盗む
print('パスワード: 盗んだ値で本物に入れるか ->', stolen == password)

# 出所に結び付けた鍵: ブラウザが、いま表示しているサイトの出所を必ず混ぜる(サイトには偽れない)
def browser_answer(shown_origin, challenge):
    return H(user_secret, shown_origin, challenge)

challenge = secrets.token_hex(8)                        # 本物の銀行が出した、その場の問い
answer_on_fake = browser_answer(FAKE, challenge)        # 利用者は偽のサイトにいる(問いは本物から中継された)
print('出所に結び付けた鍵: 偽のサイトで作らせた答えで本物に入れるか ->', answer_on_fake == browser_answer(REAL, challenge))
```

偽のサイトは、本物の銀行の問いを中継して、利用者に答えさせることはできる。しかし、ブラウザは答えに「いま表示しているのは偽のサイト」という事実を混ぜるので、その答えは本物では通らない。**利用者が見分けなくても、ブラウザが見分ける。**

作者の印栞は、利用者に見分けさせる(自分のお気に入りから押させる)方式だった。PP Interrupter は、利用者の習慣で見分けさせる(本物なら必ず止まる)方式だった。どちらも、利用者の注意に頼る。今の方式は、利用者の注意に頼らない。作者が「ブラウザに標準で搭載されること」を望んだのは、まさにこの方向だった。

### 4.6 今の形

| 作者の構想 | 今の形 |
|---|---|
| サイトには身元を渡さない(PP ログイン) | OpenID Connect の、サイトごとに別の利用者番号を返す仕組み(pairwise ID)。Apple の「メールを非公開」(サイトごとに別の転送用アドレスを作る) |
| 自動のリダイレクトをやめ、ブラウザが仲立ちする(PP Authorizer、「ブラウザへの標準搭載」) | **FedCM**(2022 年に Chrome に搭載。リダイレクトもサードパーティの Cookie も使わず、ブラウザ自身が認証プロバイダとのやりとりを仲立ちする)。**パスキー**(2022 年に Apple・Google・Microsoft が共同で推進を発表。鍵が出所に結び付くので、偽のサイトでは使えない) |
| 支払いのときだけ銀行が主導権を握る(PP 振込、OAuth 振込) | オープンバンキングの支払い指図(EU の PSD2、2018 年〜。日本では 2017 年の改正銀行法で、電子決済等代行業が制度になった)。ブラウザの Payment Request API と Secure Payment Confirmation(支払いの確認を、ブラウザと銀行の認証で行う) |
| 銀行協会がアドオンでボタンを配る | 銀行のアプリへの遷移、スマートフォンの決済の画面(OS が仲立ちする) |
| ID の相続 | Apple の故人アカウント管理連絡先(2021 年)、Google のアカウント無効化管理ツール(2013 年)、Facebook の追悼アカウント |
| 支払い専用の匿名口座(日銀カード) | 使い捨てのカード番号(バーチャルカード)、プリペイドの電子マネー |
| グループクレジット(親分の ID) | 家族の共有の支払い(親が承認する子の買い物)、法人カードの利用の共有 |

### 4.7 なぜ作者の版は広まらず、今の版は広まったか

作者は 2006 年に、自分の案の欠点の第一に「二社以上がやらねば意味がない」を挙げた。PP Authorizer でも、「サイトの側が自動のリダイレクトをやめる必要がある」ことを、導入の難しさとして書いた。

これは、経済学で**両面市場**(two-sided market)と呼ばれる問題である。利用者は、対応するサイトがなければアドオンを入れない。サイトは、アドオンを入れた利用者がいなければ対応しない。どちらも、相手が先に動くのを待つ。

今の版が広まったのは、両側を一度に動かせる者が動いたからである。パスキーは、OS とブラウザを持つ Apple・Google・Microsoft が、同時に、標準として入れた。FedCM は、ブラウザの最大手が、サードパーティの Cookie をやめる流れの中で入れた。**両面市場の立ち上がりは、個人の小さなアドオンには、ほとんど不可能**だった。作者の構想の中身が誤っていたのではなく、それを動かせる位置にいなかった、と私は考える。作者自身、PP Authorizer の説明に「こんなどこの誰とも知らない私が作ったソフトをセキュリティ向上のために使う気になるかというとそれは難しいだろう」と書いている。

一方で、作者の構想のうち、今もまだ実現していないものがある。**サイトに対しては匿名で、問題が起きたときだけ、両方の管理者が認めて身元を明かす**、という部分である。今の仕組みは、サイトごとに別の番号を返すところまではするが、「問題が起きたときの照会の手続き」は、仕組みの外(法と、各社の規約)に置かれている。作者が繰り返し考えた「信託会社」や「正義の味方」の役割は、まだ制度になっていない。

### 4.7.1 ブラウザに「息がかかる」こと

4.5 節で、今の解き方は、経路の確かめを**ブラウザ**に任せた、と書いた。すると、ブラウザそのものを信用できるかが問題になる。作者は、これにも答えを用意していた。

- 2013 年(No Ad URL の記事): ブラウザは、広く広告を募るより、信託会社や、印栞の仕組みを使う組合(シンジケート)をスポンサーに付けるべきだ。
- 2017 年(PP Authorizer の説明): 初めから入れる認証プロバイダを増やしたい組織(たとえば銀行協会)から寄付を募り、ブラウザの開発資金が潤沢になれば。

第 3 章の Loaded Magic の論理(「悪者」が途中で細工できるなら、「正義の味方」も細工できる)からすると、正義の味方が有利になるには、**ブラウザに正義の味方の「息がかかって」いる**必要がある。作者は、それを、オープンソースのブラウザの財務を良くする道として考えていた(JRF の 2026 年 10 月の回想)。信用の高い組織が、ブラウザの開発を支える。その代わりに、ブラウザは、利用者と信用の高い組織とのあいだの経路を守る。

今の現実は、半分だけこの形になっている。Firefox を作る Mozilla の収入は、8 割以上が、Google を既定の検索エンジンにする契約から来ている。ブラウザは、広告の会社に支えられている。2025 年 9 月、アメリカの Google の独占禁止の裁判で、裁判所は、Google が既定の検索の位置にお金を払うこと自体は認め(払いを止めれば、Mozilla のような小さなブラウザに深刻な害が出るとして)、独占的な契約だけを禁じた。作者の言う「スポンサー」は、銀行協会や信託会社ではなく、検索と広告の会社になった。一方、国がオープンソースの基盤を直接支える仕組み(ドイツの Sovereign Tech Fund、2022 年〜)も現れている。

作者の構想の要点は、**ブラウザに息をかける者は、ブラウザを通る経路の信用を引き受ける者であるべきだ**、ということだったと思う。広告の会社が支えるブラウザは、広告の経路を守る誘因を持つ。銀行や信託が支えるブラウザは、支払いと身元の経路を守る誘因を持つ。ただし、どちらも、ブラウザを支える者が、利用者を見張る側に回る危険がある。それを抑えるのが、ブラウザのソースが公開され、誰でも「細工」を見つけられること(オープンソース)である。Loaded Magic の牽制の論理は、ここで、ブラウザそのものに向けられる。

### 4.8 この章のまとめ

- 作者は、2001 年から 2017 年まで、**誰が仲立ちし、身元を誰に渡すか**を考え続けた。出発点は、匿名にすると盗まれたときの救済ができない、という衝突だった。
- 答えは、**サイトに対しては匿名、信用の高い第三者には身元がある**(PP ログイン・PP 振込)。そして、利用者を第三者へ移す経路を偽れないように、自動のリダイレクトをやめ、利用者の手で移る(印栞、PP Authorizer)、本物なら必ず止まる習慣を作る(PP Interrupter)。
- 今の解き方は、経路の確かめを利用者の注意ではなく**ブラウザ**に任せた(パスキー、FedCM)。作者が望んだ「ブラウザへの標準搭載」は、実現した。
- 作者の版が広まらなかったのは、**両面市場**の立ち上がりを個人では動かせなかったからで、構想の中身の誤りではない。
- まだ実現していないのは、**問題が起きたときだけ身元を明かす手続き**を、仕組みの中に置くことである。

第 II 部はここまでである。第 III 部では、作者が実際に動かした制度、はてなポイントの分配に戻る。第 I 部の価値の層を、本当に回した実験である。

# 第 III 部 はてなポイントの分配(2011)

## 第 5 章 分配プログラムを動かす

### 5.1 作者の記事

2011 年 3 月 1 日、作者は、匿名で、ポイントのやりとりとしては高額の、はてなポイントを受け取った。それまでに受け取ったポイントの累計は、分配するには少なかった。そこで作者は、翌日、「重要な更新情報: 受け取ったはてなポイントの分配を行います」を公開した。分配の方法を決めるにあたって、作者は二点に留意したという。

1. 公平・公正であること
2. 私の事務負担が減ること

作者は、もともと 2006 年のポリシー(「はてなブックマークについての私のポリシー」)で、送られたポイントは商業的に使わないと決めていた。記事に送られたことがはっきりしているポイントは、別枠にして使わず、作者がほかの人に送るポイントの総額が、それを上回るように運用する。どの記事に送られたかがはっきりしないポイント(匿名のものなど)は、四半期ごとに、次のように分配する。

- 前の期からの繰り越しと、その期に受け取ったポイントを合わせたもの(**原資**)の 1/3 を、**ボーナス**として、その期に作者がポイントを送る書き手に上乗せする。ボーナスは、送るポイント(**基礎ポイント**)の**平方根**に比例する。
- 残りの 2/3 から、過去に作者がポイントを送った書き手に、**ペンション**(年金)を払う。ペンションは、過去の基礎ポイントの累計に比例する。ただし、過去に受け取ったボーナスとペンションの累計が、基礎ポイントの累計の **1.05 乗** × その期のボーナス倍率の最小、を超えないように上限を付ける。
- ペンションを続けて受け取るには、登録が要り、毎期 5 ポイントの手数料を取る。
- 受け取り手が亡くなり、合法的な相続人もいない場合などは、「ペンション除外」のタグを付けて、累計から除く。
- はてなの手数料(送るポイントの 5%)は、その期の原資から引く。原資が 1000 ポイントに満たない期は、すべて翌期に繰り越す。

作者は、この分配のプログラム(Perl)をソースで公開し、記事の中で「逐条解説」をした。プログラムは、3 月中に三つの版が出た。

| 版 | 日付 | 変更 |
|---|---|---|
| 0.01 | 2011-03-02 | 最初の版 |
| 0.02 | 2011-03-29 | ID 統合の申請と、受け取りの申請(はてな以外のブログの持ち主が、自分の ID を名乗る)に対応 |
| 0.03 | 2011-03-31 | 「IDポイント一部未送」「一部不可」に対応。処理用の別名 |

この章では、作者のプログラムを、人工のデータで、そのまま走らせる。そして、Python に書き直して、同じ答えが出ることを確かめる。ノート O で 2011 年の外作用モデル(同じ月に同じ Perl で書かれた)を書き直したのと、同じ方法である。

なお、作者のプログラムには、はてな以外のブログの URL と、その持ち主の はてな ID の対応表が書き込まれている。実在の人の ID なので、この文書では使わない。人工の帳簿は、架空の ID(`writer001` など)で作る。

### 5.2 入力: ブックマークのタグ

作者のプログラムが読むのは、はてなブックマークを Atom フィードの形でエクスポートしたファイルである。一つのブックマーク(エントリ)には、ブックマークした URL、日付、コメント、タグが入っている。作者は、タグで、そのブックマークに関わるポイントの状態を記録していた。

| タグ | 意味 |
|---|---|
| `100p` | このブックマークで送る(送った)基礎ポイント |
| `b20p` | このとき送ったボーナス |
| `p10p` | このとき送ったペンション |
| `IDポイント未送` | 今期これから送る |
| `IDポイント済`、`ポイント済` | 送った |
| `IDポイント一部未送`(と `u20p`) | 一部をまだ送っていない(版 0.03 から) |
| `IDポイント一部不可`(と `n20p`) | 一部は送れなかった(版 0.03 から) |
| `ペンション登録` | ペンションを続けて受け取る登録をした(手数料 5 ポイント) |
| `ペンション除外` | ペンションの累計から除く |

受け取り手の ID は、URL がはてなダイアリーなら URL から読み、そうでなければ、コメントの「(はてなでは ○○ 氏)」から読む。

```python
%%writefile hatena_orig.py
# 作者の分配プログラム(Perl)を、そのまま走らせるための道具。
# プログラムは三つの版(0.01: 2011-03-02、0.02: 2011-03-29、0.03: 2011-03-31)。手元(refs/hatena/)になければ作者のサイトから取ってくる。
# 入力は、はてなブックマークを Atom フィード形式でエクスポートしたファイル。ここでは人工のフィードを作る。
import hashlib
import os
import re
import subprocess
from pathlib import Path
from xml.sax.saxutils import escape, quoteattr

VERSIONS = {'0.01': 'calc_bonus_bookmark_point-20110302.pl',
            '0.02': 'calc_bonus_bookmark_point-20110329.pl',
            '0.03': 'calc_bonus_bookmark_point-20110331.pl'}
SHA256 = {}
ARCHIVE = 'http://jrf.cocolog-nifty.com/archive/perl/'
EXPECTED = {   # 作者のサイトに置かれているファイルの sha256(2026-10-26 に確かめた値)
    'calc_bonus_bookmark_point-20110302.pl': 'a20d9b4bc8471e39e46229c8a37fdf569c021035857e0debdde68bcfee3fa60e',
    'calc_bonus_bookmark_point-20110329.pl': 'dffc676923c21444a7ab8829dd684591febc7df7910d7771218b5852942d3d76',
    'calc_bonus_bookmark_point-20110331.pl': 'cf5ef0c0409215588a0661193470d0e82136d5af537ae9a1e30f9ea925f1cb5b',
    'point_back_with_rejected_bi.pl': '6b19f1cd6728c336bf3ddaab65bfb57b6edb0e1eedf4b5ad31a8944c096dab7e',
    'point_back_with_rejected_bi_int.pl': '1f3ce89d9f3713ded218ab000f7f7a5226696011841e041783634bdc0fb03442',
}


def perl_file(name, dirs=('.', '../refs/hatena', 'refs/hatena')):
    """作者の Perl のファイルを探す。なければ作者のサイトから取ってくる。どちらでも sha256 を確かめる。"""
    for d in dirs:
        p = Path(d) / name
        if p.exists():
            break
    else:
        import urllib.request
        p = Path(name)
        p.write_bytes(urllib.request.urlopen(ARCHIVE + name, timeout=30).read())
    h = hashlib.sha256(p.read_bytes()).hexdigest()
    if h != EXPECTED[name]:
        raise RuntimeError(f'{name} の sha256 が違う: {h}')
    return p


def find_perl_files():
    """分配のプログラムの三つの版を探し(なければ取ってきて)、sha256 を記録する。"""
    out = {}
    for v, f in VERSIONS.items():
        out[v] = perl_file(f)
        SHA256[v] = EXPECTED[f]
    return out
```

```python
%%writefile -a hatena_orig.py
# プログラムが使う CPAN のモジュールのうち、URI、XML::Parser、HTML::Entities は、Perl の標準には入っていない。
# 入っていない環境のために、プログラムが使う機能だけを持つ、小さな代わりを書く(本物があれば本物を使う)。
# 代わりは、このノートが作る人工のフィード(改行や実体参照を含まない一行の文字列)を読むのに十分なだけのもの。
SHIM = {
    'URI.pm': r'''package URI;
use strict; use warnings;
sub new { my ($c, $s) = @_; return bless { s => $s }, $c; }
sub host { my ($self) = @_; return ($self->{s} =~ m{^[a-z]+://([^/:?#]+)}i) ? lc($1) : undef; }
1;
''',
    'HTML/Entities.pm': r'''package HTML::Entities;
use strict; use warnings;
use Exporter 'import';
our @EXPORT = qw(decode_entities);
my %E = (amp => '&', lt => '<', gt => '>', quot => '"', apos => "'");
sub decode_entities {
  my @r = map { my $s = $_;
    $s =~ s/&#x([0-9a-fA-F]+);/chr(hex($1))/ge; $s =~ s/&#(\d+);/chr($1)/ge;
    $s =~ s/&(amp|lt|gt|quot|apos);/$E{$1}/g; $s } @_;
  return wantarray ? @r : $r[0];
}
1;
''',
    'XML/Parser.pm': r'''package XML::Parser;
use strict; use warnings;
use HTML::Entities ();
sub new { my ($c) = @_; return bless { h => {} }, $c; }
sub setHandlers { my ($self, %h) = @_; $self->{h} = { %{$self->{h}}, %h }; }
sub parse {
  my ($self, $s) = @_;
  my $h = $self->{h};
  $s =~ s/<\?.*?\?>//gs; $s =~ s/<!--.*?-->//gs;
  while ($s =~ /\G(?:<\/([^\s>]+)\s*>|<([^\s\/>]+)((?:\s+[^\s=]+\s*=\s*(?:"[^"]*"|'[^']*'))*)\s*(\/?)>|([^<]+))/gc) {
    if (defined $1) { $h->{End}->($self, $1) if $h->{End}; }
    elsif (defined $2) {
      my ($tag, $a, $empty) = ($2, $3, $4);
      my @attr;
      while ($a =~ /([^\s=]+)\s*=\s*(?:"([^"]*)"|'([^']*)')/g) {
        push @attr, $1, HTML::Entities::decode_entities(defined $2 ? $2 : $3);
      }
      $h->{Start}->($self, $tag, @attr) if $h->{Start};
      $h->{End}->($self, $tag) if $empty && $h->{End};
    } else { $h->{Char}->($self, HTML::Entities::decode_entities($5)) if $h->{Char}; }
  }
}
1;
''',
}


def prepare_shim(root='perl_shim'):
    """足りないモジュールだけ、代わりを書き出す。返す値は -I に渡すディレクトリ(要らなければ None)と、代わりを使ったモジュール。"""
    used = []
    for rel, src in SHIM.items():
        mod = rel[:-3].replace('/', '::')
        ok = subprocess.run(['perl', f'-M{mod}', '-e', '1'], capture_output=True).returncode == 0
        if not ok:
            p = Path(root) / rel
            p.parent.mkdir(parents=True, exist_ok=True)
            p.write_text(src)
            used.append(mod)
    return (root if used else None), used
```

```python
%%writefile -a hatena_orig.py
def make_feed(entries):
    """人工のブックマークの Atom フィード。entries は dict のリスト:
    url, date(ISO の文字列), id(はてなの URL でないときに、コメントで名前を示す), tags(タグのリスト)。"""
    out = ['<?xml version="1.0" encoding="utf-8"?>',
           '<feed version="0.3" xmlns="http://purl.org/atom/ns#" xmlns:dc="http://purl.org/dc/elements/1.1/">']
    for e in entries:
        summary = f'(はてなでは {e["id"]} 氏) ' if e.get('id') else ''
        summary += e.get('comment', 'artificial bookmark')
        out.append('<entry>')
        out.append(f'<title>{escape(e.get("title", "page"))}</title>')
        out.append(f'<link rel="related" type="text/html" href={quoteattr(e["url"])}/>')
        out.append(f'<issued>{e["date"]}</issued>')
        out.append(f'<summary>{escape(summary)}</summary>')
        for t in e['tags']:
            out.append(f'<dc:subject>{escape(t)}</dc:subject>')
        out.append('</entry>')
    out.append('</feed>')
    return '\n'.join(out) + '\n'


RESULT_LINE = re.compile(r'^ (\S+)  (-?\d+)  \((.*) \)$')


def run_perl(version, feed_text, fund, pension=(), tax=None, workdir='.'):
    """作者のプログラムをそのまま走らせ、出力を読む。
    返す値: {'pay': {id: 払う額}, 'parts': {id: 'Np bNp pNp'}, 'BONUS', 'PENSION', 'TAX', 'REST', 'TOTAL', 'raw'}"""
    files = find_perl_files()
    feed = Path(workdir) / 'artificial.atom.xml'
    feed.write_text(feed_text, encoding='utf-8')
    inc, _ = prepare_shim(str(Path(workdir) / 'perl_shim'))
    cmd = ['perl'] + (['-I', inc] if inc else []) + [str(files[version]), '-f', str(feed)]
    if pension:
        cmd.append('--pension=' + ','.join(pension))
    if tax is not None:
        cmd += ['-t', str(tax)]
    cmd.append(str(fund))
    env = dict(os.environ, PERL_HASH_SEED='0', PERL_PERTURB_KEYS='0')
    r = subprocess.run(cmd, capture_output=True, text=True, env=env)
    if r.returncode != 0:
        raise RuntimeError(r.stderr)
    out = {'pay': {}, 'parts': {}, 'raw': r.stdout, 'warn': r.stderr}
    for line in r.stdout.splitlines():
        m = RESULT_LINE.match(line)
        if m:
            out['pay'][m.group(1)] = int(m.group(2))
            out['parts'][m.group(1)] = m.group(3).strip()
        m = re.match(r'^  (BONUS|PENSION|TAX|REST|TOTAL)  (-?[\d.]+)', line)
        if m:
            out[m.group(1)] = float(m.group(2))
    return out
```

### 5.3 人工の帳簿

架空の書き手 40 人に、2006〜2010 年に 200 回ポイントを送り(その 3 割には、過去の期のボーナスが付いていたとする)、今期(2011 年 1〜3 月)に 15 回送る、という帳簿を作る。書き手の 2 割はペンションに登録していて、少数は除外されている。登録していない書き手の一部は、今期、ペンションを求めてきた(プログラムの `--pension` で渡す)。ブックマークされやすさは、書き手によって大きく違う(ディリクレ分布で作る)。

```python
%%writefile bonus.py
# 分配プログラム(calc_bonus_bookmark_point.pl、版 0.03)の計算の部分を、Python に書き直したもの。
# 入力は同じ Atom フィード。ID の統合申請と受取申請(作者も「十分にテストしていない」と書いた部分)は書き直していない。
import math
import xml.etree.ElementTree as ET
import re

NS = {'a': 'http://purl.org/atom/ns#', 'dc': 'http://purl.org/dc/elements/1.1/'}
ID_PAT = r'[-a-zA-Z01-9_]+'


def parse_feed(text):
    """フィードから、作者のプログラムと同じ項目を取り出す。"""
    entries = []
    for e in ET.fromstring(text).findall('a:entry', NS):
        cur = {}
        for ln in e.findall('a:link', NS):
            if ln.get('rel') == 'related':
                cur['url'] = ln.get('href')
        cur['date'] = e.findtext('a:issued', namespaces=NS)
        for s in e.findall('dc:subject', NS):
            t = s.text or ''
            for pat, key in ((r'^(\d+)p$', 'point'), (r'^b(\d+)p$', 'bonus'), (r'^p(\d+)p$', 'pension'),
                             (r'^u(\d+)p$', 'unpaid_partially'), (r'^n(\d+)p$', 'not_payable_partially')):
                m = re.match(pat, t)
                if m:
                    cur[key] = int(m.group(1))
                    break
            else:
                if 'ポイント' in t:
                    cur['state'] = t
                elif 'ペンション' in t:
                    cur['pension_plan'] = t
        summ = (e.findtext('a:summary', namespaces=NS) or '').lstrip()
        m = re.match(r'^\(?(?:はてなでは\s*)?(' + ID_PAT + r')\s*氏\)?', summ)
        if m:
            cur['id'] = m.group(1)
        if 'id' not in cur:
            m = re.match(r'^http://[a-z\.]+\.hatena\.ne\.jp/(' + ID_PAT + r')/', cur.get('url', ''))
            if m:
                cur['id'] = m.group(1)
        entries.append(cur)
    return entries


def tally(entries):
    """ID ごとの、今期送る基礎ポイント BASE、過去に送った基礎ポイント PAID_BASE、過去のボーナスとペンション PAID_BONUS、
    ペンション登録した基礎ポイント PENSION_PLAN。"""
    BASE, PAID_BASE, PAID_BONUS, PLAN = {}, {}, {}, {}
    for e in entries:
        st = e.get('state')
        if st is None or 'id' not in e:
            continue
        unpaid = paid = None
        if st == 'IDポイント未送':
            unpaid = e['point']
        elif st == 'IDポイント一部未送':
            unpaid = e['unpaid_partially']; paid = e['point'] - unpaid
        elif st in ('IDポイント済', 'ポイント済'):
            paid = e['point']
        elif st in ('IDポイント一部不可', 'ポイント一部不可'):
            paid = e['point'] - e['not_payable_partially']
        i = e['id']
        if unpaid is not None:
            BASE[i] = BASE.get(i, 0) + unpaid
        if paid is not None:
            if e.get('pension_plan') != 'ペンション除外':
                PAID_BASE[i] = PAID_BASE.get(i, 0) + paid
            PAID_BONUS[i] = PAID_BONUS.get(i, 0) + e.get('bonus', 0) + e.get('pension', 0)
            if e.get('pension_plan') == 'ペンション登録':
                PLAN[i] = PLAN.get(i, 0) + paid
    return BASE, PAID_BASE, PAID_BONUS, PLAN


def distribute(BASE, PAID_BASE, PAID_BONUS, PLAN, fund, pension=(), tax=0.05, ratio=0.5, fee=5, condenser=0.05):
    """作者の計算そのもの。返す値は、作者のプログラムの出力と同じ項目。"""
    PAID_BONUS = dict(PAID_BONUS); PLAN = dict(PLAN)
    bonus_fund = (ratio / (1 + ratio)) * fund / (1 + tax)
    sqrt_total = sum(math.sqrt(b) for b in BASE.values())
    BONUS, bonus_total, tax_total, min_rate = {}, 0, 0, None
    for i, base in BASE.items():
        b = bonus_fund * math.sqrt(base) / sqrt_total
        b = int(b / 10) * 10                            # 下一桁を切り捨て
        if b > 0:
            BONUS[i] = b
        bonus_total += b
        if min_rate is None or min_rate > b / base:
            min_rate = b / base                         # その期のボーナス倍率の最小
        if b > 0:
            tax_total += int((b + base) * tax) - int(base * tax)
    for i, b in BONUS.items():
        PAID_BONUS[i] = PAID_BONUS.get(i, 0) + b
    pension_fund = (fund - bonus_total - tax_total) / (1 + tax)
    req = list(pension)
    for i in req:
        if i in PLAN:
            PLAN[i] = PAID_BASE[i]
    req = list(dict.fromkeys(req + list(PLAN)))
    paid_base_total = sum(PAID_BASE.values())
    PENSION = {}
    for i in req:
        if i in PAID_BASE:
            b = PLAN.get(i, PAID_BASE[i])
            p = pension_fund * b / paid_base_total       # 基礎ポイントの累計に比例
            limit = (b ** (1 + condenser)) * min_rate      # 上限: 累計の 1.05 乗 × ボーナス倍率の最小
            if p + PAID_BONUS[i] > limit:
                p = limit - PAID_BONUS[i]
            p = int(p)
            f = fee if i in PLAN else 0
            if p > f:
                PENSION[i] = p
    pension_total = 0
    for i, p in PENSION.items():
        f = fee if i in PLAN else 0
        pension_total += p - f
        o = BASE.get(i, 0) + BONUS.get(i, 0)
        tax_total += int((o + p - f) * tax) - int(o * tax)
    pay, parts = {}, {}
    for i in set(BASE) | set(BONUS) | set(PENSION):
        pay[i] = BASE.get(i, 0) + BONUS.get(i, 0) + (PENSION[i] - (fee if i in PLAN else 0) if i in PENSION else 0)
        parts[i] = ' '.join(([f'{BASE[i]}p'] if i in BASE else []) + ([f'b{BONUS[i]}p'] if i in BONUS else [])
                            + ([f'p{PENSION[i]}p'] if i in PENSION else []))
    base_total = sum(BASE.values())
    return dict(pay=pay, parts=parts, BONUS=bonus_total, PENSION=pension_total, TAX=tax_total,
                REST=fund - bonus_total - tax_total - pension_total,
                TOTAL=base_total + int(base_total * tax) + bonus_total + pension_total + tax_total,
                bonus=BONUS, pension=PENSION, min_rate=min_rate)


def run(feed_text, fund, pension=(), **kw):
    return distribute(*tally(parse_feed(feed_text)), fund, pension, **kw)
```

```python
%%writefile -a bonus.py
# 人工のブックマークの帳簿。書き手は架空の ID(writer001 など)で、はてなダイアリーの形の URL を持つ。
import numpy as np


def artificial_book(seed=0, n_writers=40, n_past=200, n_now=15, plan_frac=0.2, excl_frac=0.05, prior_bonus=True):
    """過去(2006〜2010 年)に送ったポイントと、今期(2011 年 1〜3 月)に送るポイントのブックマーク。
    返す値: エントリのリスト、ペンションを求める ID のリスト。"""
    rng = np.random.default_rng(seed)
    ids = [f'writer{k:03d}' for k in range(n_writers)]
    pop = rng.dirichlet(np.ones(n_writers) * 0.5)           # 書き手ごとの、ブックマークされやすさ
    entries = []
    plan = set(rng.choice(ids, int(plan_frac * n_writers), replace=False))
    excl = set(rng.choice(ids, max(1, int(excl_frac * n_writers)), replace=False)) - plan
    for k in range(n_past):
        w = ids[rng.choice(n_writers, p=pop)]
        pt = int(rng.choice([10, 20, 30, 50, 100], p=[0.3, 0.25, 0.2, 0.15, 0.1]))
        tags = [f'{pt}p', 'IDポイント済']
        if prior_bonus and rng.random() < 0.3:
            tags.append(f'b{int(rng.integers(1, 4)) * 10}p')   # 過去の期に受け取ったボーナス
        if w in plan:
            tags.append('ペンション登録')
        elif w in excl:
            tags.append('ペンション除外')
        y, m, d = int(rng.integers(2006, 2011)), int(rng.integers(1, 13)), int(rng.integers(1, 29))
        entries.append(dict(url=f'http://d.hatena.ne.jp/{w}/{y}{m:02d}{d:02d}/p{k}', date=f'{y}-{m:02d}-{d:02d}T12:00:00+09:00',
                            tags=tags))
    for k in range(n_now):
        w = ids[rng.choice(n_writers, p=pop)]
        pt = int(rng.choice([10, 20, 30, 50, 100], p=[0.3, 0.25, 0.2, 0.15, 0.1]))
        d = int(rng.integers(1, 29))
        entries.append(dict(url=f'http://d.hatena.ne.jp/{w}/201102{d:02d}/n{k}', date=f'2011-02-{d:02d}T12:00:00+09:00',
                            tags=[f'{pt}p', 'IDポイント未送']))
    past_ids = sorted({e['url'].split('/')[3] for e in entries[:n_past]})
    pension_req = [i for i in past_ids if i not in plan and rng.random() < 0.3]
    return entries, pension_req
```

```python
import hatena_orig as ho, bonus as bo
ents, req = bo.artificial_book(seed=0)
feed = ho.make_feed(ents)
print(feed[:1100])
print('...\nエントリ', len(ents), '件、ペンションを求める ID:', req)
```

### 5.4 そのまま走らせる

作者のプログラムは、CPAN のモジュール URI、XML::Parser、HTML::Entities を使う。これらは Perl の標準には入っていない。入っていない環境では、プログラムが使う機能だけを持つ小さな代わり(上の `SHIM`)を書き出して使う。**作者のプログラムには一文字も手を入れない。** Perl のハッシュの順番(ID を並べる順番)は実行ごとに変わるので、環境変数 `PERL_HASH_SEED=0` で固定する。

```python
files = ho.find_perl_files()
inc, used = ho.prepare_shim()
print('代わりを使ったモジュール:', used if used else 'なし(本物を使う)')
for v in files:
    print(f'版 {v}: {files[v].name}  sha256 {ho.SHA256[v][:16]}...')
print()
for v in files:
    r = ho.run_perl(v, feed, fund=3000, pension=req)
    print(f'版 {v}:  BONUS {r["BONUS"]:.0f}   PENSION {r["PENSION"]:.0f}   TAX {r["TAX"]:.0f}   REST {r["REST"]:.0f}')
print('\n版 0.03 の出力:\n' + r['raw'])
```

原資を 3000 ポイントとした。出力の読み方は次のとおり。`writer006  352  ( 70p b130p p157p )` は、writer006 に、今期の基礎ポイント 70、ボーナス 130、ペンション 157、合わせて 352 ポイントを送る、という意味である。登録した書き手のペンションからは、手数料 5 が引かれる(`writer026  28  ( p33p )`)。

最後の行の BONUS がボーナスの合計、PENSION がペンションの合計(手数料を引いたもの)、TAX がはてなの手数料、REST が翌期への繰り越しである。

### 5.5 書き直しと、一致の確認

```python
py = bo.run(feed, 3000, req)
print('Python:  BONUS', py['BONUS'], '  PENSION', py['PENSION'], '  TAX', py['TAX'], '  REST', py['REST'], '  TOTAL', py['TOTAL'])
print('一人ずつの支払いと内訳が一致するか:', r['pay'] == py['pay'] and r['parts'] == py['parts'])

# 帳簿の大きさと原資を変えた 100 通りで確かめる
ok = 0
for s in range(100):
    ents_s, req_s = bo.artificial_book(seed=s, n_writers=20 + s % 40, n_now=5 + s % 25)
    feed_s = ho.make_feed(ents_s); fund = 500 + 97 * s
    rp, pp = ho.run_perl('0.03', feed_s, fund=fund, pension=req_s), bo.run(feed_s, fund, req_s)
    ok += rp['pay'] == pp['pay'] and all(rp[k] == pp[k] for k in ('BONUS', 'PENSION', 'TAX', 'REST', 'TOTAL'))
print(f'100 通りのうち、すべての数が一致したもの: {ok}')
```

Python の書き直しは、作者のプログラムと、一人ずつの支払いの額まで一致する。

一つ注意がある。ボーナスの計算では、√(基礎ポイント) の和を、ID の順に足していく。浮動小数点の足し算は順番で最後の桁が変わりうるので、ボーナスの額が 10 の倍数の境目にぴったり来ると、切り捨ての結果が変わる可能性がある。Perl のハッシュの順番と Python の辞書の順番は違うので、原理的には食い違いうる。100 通りでは一度も起きなかった。

### 5.6 版の違い

三つの版で、計算の部分は同じである。違うのは、入力の読み方である。たとえば、版 0.03 で足された「IDポイント一部未送」(基礎ポイント 50 のうち 20 をまだ送っていない)のエントリを一つ加えると、版 0.01・0.02 は、このエントリを読み飛ばす。

```python
ents1, req1 = bo.artificial_book(seed=1)
ents1.append(dict(url='http://d.hatena.ne.jp/writer005/20110215/x', date='2011-02-15T12:00:00+09:00',
                  tags=['50p', 'u20p', 'IDポイント一部未送']))
feed1 = ho.make_feed(ents1)
for v in files:
    r1 = ho.run_perl(v, feed1, 2000, req1)
    print(f'版 {v}: writer005 の内訳 ( {r1["parts"].get("writer005")} )   PENSION の合計 {r1["PENSION"]:.0f}')
print(f'Python(版 0.03 の書き直し): ( {bo.run(feed1, 2000, req1)["parts"].get("writer005")} )')
```

版 0.03 では、未送の 20 が今期の基礎ポイントに、送った 30 が過去の累計に入る。そのぶん、writer005 のボーナスとペンションが変わり、ペンションの合計も変わる。

### 5.7 計算を式で読む

プログラムの計算を、式に書いておく。はてなの手数料の率を τ(= 0.05)、原資を F とする。

**ボーナス。** 今期、書き手 i に送る基礎ポイントを Bᵢ とする。

$$ b_i = 10 \left\lfloor \frac{1}{10} \cdot \frac{F}{3(1+\tau)} \cdot \frac{\sqrt{B_i}}{\sum_k \sqrt{B_k}} \right\rfloor $$

ボーナスに回すのは、原資の 1/3(から手数料のぶんを除いたもの)である(プログラムでは、ボーナスとペンションの比 1/2 から 1/3 を作っている)。平方根なので、小口の書き手ほど、基礎ポイントに対するボーナスの倍率 bᵢ / Bᵢ が高い。その期のボーナス倍率の最小を r_min = minᵢ bᵢ / Bᵢ とする。平方根の配分では、倍率は 1/√Bᵢ に比例するので、**r_min は、その期にいちばん多くのポイントを送る書き手で決まる**。

**ペンション。** 過去に書き手 j に送った基礎ポイントの累計を Aⱼ、過去に受け取ったボーナスとペンションの累計(今期のボーナスを含む)を Cⱼ とする。ペンションに回せるのは、原資からボーナスと手数料を引いた残り P = (F − Σbᵢ − 手数料) / (1 + τ) である。

$$ p_j = \left\lfloor \min\left( P \cdot \frac{A_j}{\sum_k A_k},\; A_j^{1.05}\, r_\text{min} - C_j \right) \right\rfloor $$

ペンションを受け取るのは、登録した書き手と、今期求めてきた書き手だけである。分母の ΣAₖ は、**求めていない書き手も含めた**過去の累計の全体なので、求めなかった人のぶんは、払われずに繰り越しに残る。登録した書き手からは、手数料 5 を引く。pⱼ が手数料以下なら、払わない。

**繰り越し。** 残り(REST)= F − ボーナス − ペンション − 手数料が、翌期の原資に入る。

作者が記事に書いた「1/3 をボーナス、2/3 を繰り越してそこからペンション」は、プログラムでは、「1/3 をボーナス、残り全部をペンションの財源にするが、上限と、求めない人のぶんで、実際には多くが繰り越される」という形になっている。5.4 節の例では、原資 3000 のうち、ボーナス 910、ペンション 816、手数料 86、繰り越し 1188 だった。

### 5.8 この章のまとめ

- 作者の分配プログラムは、今の Perl で、人工のブックマークを入力にして、そのまま動く(足りない CPAN のモジュールは、小さな代わりで補った。作者のプログラムには手を入れていない)。
- Python に書き直したものは、100 通りの帳簿で、一人ずつの支払いまで一致する。
- 三つの版の違いは入力の読み方にあり、計算は同じである。
- 計算を式にすると、二つのことが見える。ペンションの上限は、**その期にいちばん多く送られる書き手**のボーナス倍率で決まる。そして、ペンションを**求めない人のぶんは繰り越しに残る**。

次の章では、この規則を、人工の参加者で、何期も回す。

---

## 第 6 章 基金と、ボーナスとペンション

### 6.1 作者の狙い

作者は、分配の規則の細部を、受け取る人がどう振る舞うかを読んで決めていた。記事と、そのあとの「ひとこと」から、狙いを拾う。

- **平方根のボーナス。** 小口の書き手を重んじる。ただし、ID を分ければ得をする(第 0 章)。
- **1.05 乗の上限。** 「ID を複数に分けるのが不利になるようにするため」。作者は、これが「多く稼いだ人がさらにもらえるという『逆進性』のある不公平な制度かもしれない」と承知のうえで、「私などの事務負担を考えるとぜひ欲しいものなので、ごく微量(0.05)、その効果を取り入れています」と書いた。
- **登録の手数料(毎期 5 ポイント)。** ID を分けるほど多くかかる。そして、「ペンション登録していない者については、ID が受け取り不能になっていることが要求がないことでわかることに対する逆の対価という側面があります」。手数料は、受け取る側が生きていることを知らせる合図の値段でもある。
- **ID 統合の待ち期間**(2011 年 3 月 22 日の「ひとこと」)。ID を統合すれば、手数料が減り、上限が増える。だから、何でも統合するのが得になり、それでは「記事の『個性』が失なわれる方向に私が加担することになる」。そこで、統合を認めるには一定の期間(本格的にやるなら 15 年ほど)の経過を必要とし、「ID が別のほうが得するボーナス」と釣り合わせて、**「ほどほどの ID の分散と統合」**を制御する。

> 要するに、支払う「私」達を統合しようとする意志に従いながら、ペンションまで期間のあるボーナスによってそれに少し対抗するのである。

この章では、第 5 章で作者のプログラムと一致させた計算(`bonus.distribute`)を、そのまま何期も回して、これらの狙いが実際に働くかを確かめる。

### 6.2 模型

- 一期は四半期。200 期(50 年)回す。
- 書き手は、一期に平均 1 人入ってくる。入ってから 12 期(3 年)のあいだは活動していて、作者から基礎ポイントを送られる(一期に 15 回、書き手のブックマークされやすさに比例して選ぶ。額は 10〜100)。そのあとは引退して、ペンションに登録する。
- 作者が受け取るポイントは、一期に平均 1500(指数分布)。さらに、5% の確率で、1 万ポイントの匿名の大口の寄付がある(2011 年 3 月 1 日のような)。
- 原資が 1000 に満たない期は、分配せずに繰り越す(作者の規則)。

```python
%%writefile sybil.py
# 分配の規則を、何期(四半期)も回す模型。計算は bonus.distribute(作者のプログラムの書き直し)をそのまま使う。
# 書き手は入ってきて、活動する L_a 期のあいだ作者からポイントを送られ(ボーナスの対象)、そのあと引退してペンションの対象になる。
import numpy as np
import bonus


def run(T=80, inflow=1500.0, big_gift=(0.05, 10000.0), n_now=15, entry=1.0, L_a=12, pts=(10, 20, 30, 50, 100),
        focal_k=0, focal_wait=None, focal_pop=1.0, fee=5, condenser=0.05, min_fund=1000, extra_big=None, seed=0):
    """返す値: 期ごとの原資・ボーナス・ペンション・手数料・繰り越し・ペンションが 0 だった期、注目する書き手の受け取りの合計。

    inflow: 一期の受け取りの平均(指数分布)。big_gift: (確率, 額) の匿名の大口の寄付。
    focal_k: 0 なら注目する書き手はいない。k ≥ 1 なら、第 0 期に入る注目の書き手が、ID を k 個に分けて受け取る
             (ブックマークする人は、k 個の ID のどれかを等しい確率で選ぶ)。
    focal_wait: 統合の申請から、統合が認められるまでの期数。注目の書き手は引退した時点で統合を申請する(None なら統合しない)。
    extra_big: (期, 額) その期だけ、一人の書き手に大きな基礎ポイントを送る(r_min への影響を見る)。
    """
    rng = np.random.default_rng(seed)
    A, C, plan = {}, {}, set()          # 過去の基礎ポイントの累計、ボーナスとペンションの累計、ペンション登録
    writers = []                         # (ID, 入った期, ブックマークされやすさ)
    fid = [f'focal{j}' for j in range(focal_k)]
    merged = None
    fund = 0.0
    rec = {k: np.zeros(T) for k in ('fund', 'bonus', 'pension', 'tax', 'rest', 'zero_pension', 'min_rate', 'focal')}
    for t in range(T):
        for _ in range(rng.poisson(entry)):
            writers.append((f'w{len(writers):04d}', t, rng.lognormal(0, 1)))
        if t == 0 and focal_k:
            writers.append(('focal', 0, focal_pop * np.exp(0.5)))   # 平均的なブックマークされやすさ
        fund += rng.exponential(inflow) + (big_gift[1] if rng.random() < big_gift[0] else 0.0)
        rec['fund'][t] = fund
        active = [w for w in writers if t - w[1] < L_a]
        BASE = {}
        if active:
            wts = np.array([w[2] for w in active]); wts /= wts.sum()
            for j in rng.choice(len(active), n_now, p=wts):
                wid = active[j][0]
                if wid == 'focal':
                    wid = fid[int(rng.random() * focal_k)]          # 乱数の使い方を k によらず同じにする
                BASE[wid] = BASE.get(wid, 0) + int(rng.choice(pts))
        if extra_big and t == extra_big[0]:
            BASE['whale'] = BASE.get('whale', 0) + extra_big[1]
        # 引退した書き手は、ペンションに登録する
        for w in writers:
            if t - w[1] >= L_a:
                ids = fid if w[0] == 'focal' else [w[0]]
                if w[0] == 'focal' and merged is not None and t >= merged:
                    ids = ['focal_m']
                plan.update(i for i in ids if i in A)
        if focal_k and focal_wait is not None and merged is None and t >= L_a:
            merged = t + focal_wait                                  # 引退したときに統合を申請
        if merged is not None and t == merged:
            A['focal_m'] = sum(A.pop(i, 0) for i in fid); C['focal_m'] = sum(C.pop(i, 0) for i in fid)
            plan.difference_update(fid); plan.add('focal_m')
        if fund < min_fund or not BASE:
            rec['rest'][t] = fund
            continue
        r = bonus.distribute(BASE, dict(A), dict(C), {i: A[i] for i in plan if i in A}, fund, (), fee=fee, condenser=condenser)
        for i, b in BASE.items():
            A[i] = A.get(i, 0) + b
            C.setdefault(i, 0)              # 作者のプログラムでは、送ったことのある ID には必ずボーナスの累計(0 でも)がある
        for i, b in r['bonus'].items():
            C[i] = C.get(i, 0) + b
        for i, p in r['pension'].items():
            C[i] = C.get(i, 0) + p
        rec['bonus'][t], rec['pension'][t], rec['tax'][t], rec['rest'][t] = r['BONUS'], r['PENSION'], r['TAX'], r['REST']
        rec['zero_pension'][t] = r['PENSION'] == 0
        rec['min_rate'][t] = r['min_rate']
        rec['focal'][t] = sum(r['pay'].get(i, 0) - BASE.get(i, 0) for i in fid + ['focal_m'])   # 基礎ポイント以外の受け取り
        fund = r['REST']
    return rec
```

```python
%%writefile -a sybil.py
def focal_value(k, wait=None, seeds=range(8), T=200, beta=1.0, **kw):
    """ID を k 個に分けた注目の書き手が、T 期のあいだに受け取るボーナスとペンションの合計(シードの平均)と、
    その内訳(活動中 = 主にボーナス、引退後 = ペンション)。beta は一期あたりの割引(1 なら割り引かない)。"""
    out = []
    for s in seeds:
        r = run(T=T, focal_k=k, focal_wait=wait, seed=s, **kw)
        L_a = kw.get('L_a', 12)
        d = r['focal'] * beta ** np.arange(T)
        out.append((d[:L_a].sum(), d[L_a:].sum()))
    out = np.array(out)
    return out.sum(1).mean(), out[:, 0].mean(), out[:, 1].mean()
```

### 6.3 基金の動き

```python
import sybil
r = sybil.run(T=200, seed=0)
t = np.arange(200) / 4 + 2011
fig, ax = plt.subplots(figsize=(7.2, 3.6))
ax.plot(t, r['fund'], color=PAL[0], lw=1.2, label='fund (start of quarter)')
ax.plot(t, r['bonus'], color=PAL[1], lw=1.2, label='bonus')
ax.plot(t, r['pension'], color=PAL[2], lw=1.2, label='pension')
ax.set(xlabel='year', ylabel='points', title='The fund under the 2011 rules (artificial inflow)'); ax.legend(fontsize=8); plt.show()
ex = r['fund'] >= 1000
print('分配した期の平均:  原資 %.0f   ボーナス %.0f   ペンション %.0f   はてなの手数料 %.0f   繰り越し %.0f'
      % tuple(r[k][ex].mean() for k in ('fund', 'bonus', 'pension', 'tax', 'rest')))
print('ボーナス / 原資 = %.3f(規則では 1/3.15 = %.3f)' % (r['bonus'][ex].sum() / r['fund'][ex].sum(), 1 / 3.15))
spend = 1 - r['rest'][ex].sum() / r['fund'][ex].sum()
print('一期に使われる割合(ボーナス + ペンション + 手数料)/ 原資 = %.2f' % spend)
```

ボーナスは、原資のほぼ 1/3(手数料を除くと 1/3.15)に、10 の倍数への切り捨てのぶんだけ少なく、きちんと決まる。ペンションは、それよりずっと少なく、残りの多くは繰り越される。大口の寄付があった期のあと、原資は跳ね上がり、数年かけて、ボーナスとして少しずつ吐き出される。

これは、大学の基金(endowment)の**支出の規則**と同じ働きである。毎期、基金の一定割合だけを使えば、受け取りの波(大口の寄付)は、ならされて使われる。作者の規則では、ボーナスとペンションと手数料を合わせて、毎期、原資の 5 割近くが使われるので、大口の寄付の効果は、半分がほぼ 1 期で、9 割が 3〜4 期(1 年ほど)で使われる。大学の基金(毎年、数年分の平均の 5% ほどを使う)よりは、ずっと速い。

### 6.4 ペンションは何をしているか

ペンションの式(第 5.7 節)をもう一度見る。上限は、

$$ C_j + p_j \le A_j^{1.05}\, r_\text{min} $$

である。Cⱼ は、書き手 j が過去に受け取ったボーナスとペンションの累計、Aⱼ は基礎ポイントの累計、r_min はその期のボーナス倍率の最小である。両辺を Aⱼ で割ると、

$$ \frac{C_j + p_j}{A_j} \le A_j^{0.05}\, r_\text{min} $$

となる。左辺は、**書き手 j がこれまでに受け取った「上乗せの倍率」**(基礎ポイント 1 あたり、どれだけ上乗せをもらったか)である。右辺は、ほぼ、**今期の書き手がもらうボーナスの倍率の最小**である。

つまり、ペンションは、**過去の書き手の上乗せの倍率が、今期の書き手の倍率に追いつくまで、差を埋める**ものである。過去に、基金が乏しくてボーナスの倍率が低かった期に送られた書き手は、今の倍率に追いつくまでペンションを受け取る。基金が豊かな期に高い倍率のボーナスをもらった書き手は、ペンションをほとんど受け取らない。作者の言う「過去にポイントを送った方との公平を期すため」は、この意味だった。

年金の言葉で言えば、これは、**過去の拠出(ここでは、送られた基礎ポイント)を、今の水準で評価し直す**仕組みにあたる。公的年金で、過去の賃金を今の賃金の水準に直して年金額を計算する(再評価、賃金スライド)のと同じ考え方である。

```python
zs = []
for s in range(8):
    rr = sybil.run(T=200, seed=s); ex = rr['fund'] >= 1000
    zs.append((rr['zero_pension'][ex].mean(), rr['pension'][ex].sum() / (rr['bonus'][ex].sum() + rr['pension'][ex].sum())))
print('分配した期のうち、ペンションの合計が 0 だった期の割合: %.0f%%(8 シードの平均)' % (100 * np.mean([z[0] for z in zs])))
print('分配のうち、ペンションの占める割合: %.0f%%' % (100 * np.mean([z[1] for z in zs])))
dec = lambda k, j: r[k][40 * j:40 * j + 40].sum()
print('ペンションの割合の推移(10 年ごと、シード 0): ' + '  '.join(
    f'{2011 + 10 * j}〜{2020 + 10 * j}年 {100 * dec("pension", j) / (dec("pension", j) + dec("bonus", j)):.0f}%' for j in range(5)))
```

分配した期の半分ほどで、ペンションは誰にも払われない。過去の書き手の倍率が、すでに今期の最小の倍率に追いついているからである。ペンションが払われるのは、今期の倍率の最小が、過去の水準より高い期(基金が豊かで、大口の書き手がいない期)である。分配のうちペンションの占める割合は、引退した書き手が増えるにつれて、最初の 10 年の 1 割ほどから、4 割近くへ増え、そこで落ち着く。

### 6.5 いちばん多く送られる書き手が、ペンションを決める

第 5.7 節で、r_min は、その期にいちばん多くのポイントを送られる書き手で決まる、と書いた。ある期(40 期目)に、一人の書き手に 2000 ポイントを送る(ふだんの一回の送りは 10〜100)とどうなるか。

```python
base_, whale = [], []
for s in range(8):
    r0 = sybil.run(T=60, seed=s); r1 = sybil.run(T=60, seed=s, extra_big=(40, 2000))
    base_.append((r0['pension'][40], r0['min_rate'][40])); whale.append((r1['pension'][40], r1['min_rate'][40]))
print('40 期目のペンションの合計 / ボーナス倍率の最小(8 シードの平均)')
print('  ふだん:             %6.0f / %.2f' % tuple(np.mean(base_, 0)))
print('  一人に 2000 送る期: %6.0f / %.2f' % tuple(np.mean(whale, 0)))
```

一人に大きく送る期には、その人のボーナスの倍率が低くなり(平方根なので、2000 送られても、ボーナスは √2000 に比例するだけ)、r_min が下がる。すると、すべての過去の書き手の上限が下がり、**その期のペンションは、誰にも払われない**。

作者は、「今期の書き手と過去の書き手の公平」を、今期の倍率の**最小**で測った。最小をとるのは、過去の書き手が、今期のどの書き手よりも有利にならないようにするためだったと思われる。しかし、その結果、**過去の書き手全員の年金が、今期のいちばん大口の一人に左右される**。平均や中央値で測れば、この結び付きは弱まる。作者が意図した設計かどうかは、記事からはわからない。

### 6.6 切り捨ての罠

もう一つ、規則の組み合わせから出てくる、意図していなかったと思われる性質がある。ボーナスは 10 の倍数に切り捨てられる。基金が乏しい期に、小口の書き手のボーナスが 10 に届かず 0 になると、その人の倍率は 0 で、r_min = 0 になる。すると上限は 0 になり、**全員のペンションが消える**。

作者の「原資が 1000 に満たない期は、すべて繰り越す」という規則は、この罠を防いでいる。受け取りの平均を一期 200 に下げ、大口の寄付をなくして、この規則がある場合とない場合を比べる。

```python
for mf, lab in ((0, '1000 未満でも分配する'), (1000, '作者の規則(1000 未満は繰り越す)')):
    rows = []
    for s in range(6):
        rr = sybil.run(seed=s, inflow=200, big_gift=(0, 0), min_fund=mf); ex = rr['bonus'] > 0
        rows.append(((rr['min_rate'][ex] == 0).mean(), ex.mean()))
    zr, exr = np.mean(rows, 0)
    print(f'{lab:22s}: 分配した期の割合 {100 * exr:3.0f}%、そのうち r_min = 0(ペンションが消える)期 {100 * zr:3.0f}%')
```

1000 未満でも分配すると、分配した期の 4 割ほどで、r_min = 0 になり、ペンションが消える。作者の規則では、原資が貯まるまで待つので、この罠は起きない。作者が 1000 という下限を置いたのは、事務の負担(少ない額を何度も送らない)のためだったかもしれないが、結果として、ペンションの計算を守っている。

### 6.7 ID を分けるか、まとめるか

作者の狙いの中心を確かめる。第 0 期に入ってくる注目の書き手が、受け取りの ID を k 個に分ける(ブックマークする人は、k 個のどれかを等しい確率で選ぶ)。引退したとき(12 期目)に統合を申請し、待ち期間のあとで統合が認められる(統合しない場合も見る)。50 年間に受け取るボーナスとペンションの合計を比べる。乱数は、k によらず同じ列を使うので、違いは k だけから来る。

```python
ks, waits = (1, 2, 4, 8), (None, 0, 20, 40, 80)
wl = {None: '統合しない', 0: 'すぐ統合', 20: '5 年待つ', 40: '10 年待つ', 80: '20 年待つ'}
for beta, lab in ((1.0, '割り引かない'), (0.97, '一期 3% 割り引く(年 12% ほど)')):
    print(f'50 年間に受け取るボーナス + ペンション({lab}、8 シードの平均。k = 1 を 100 とする)')
    print('  ID の数   ' + ''.join(f'{wl[w]:>9s}' for w in waits) + '    内訳(統合しない): ボーナス / ペンション')
    v1 = sybil.focal_value(1, beta=beta)[0]
    for k in ks:
        vals = [sybil.focal_value(k, wait=w, beta=beta)[0] for w in waits]
        tot, bon, pen = sybil.focal_value(k, beta=beta)
        print(f'  k = {k}     ' + ''.join(f'{100 * v / v1:9.1f}' for v in vals) + f'    {bon:6.0f} / {pen:6.0f}')
    print()
```

結果(数字は上の出力)は、作者の狙いをよく裏付ける。

1. **ID を分けると、ボーナスは増える**(平方根のため。k = 8 で 3 割近く)。
2. **しかし、ペンションがそのぶん減る。** ペンションの上限は、受け取った上乗せの累計 Cⱼ で決まるので、ボーナスで多くもらったぶんは、あとのペンションで差し引かれる。そのうえ、k 個の ID の上限の合計は、1.05 乗のために k^(−0.05) 倍になる(k = 8 で 0.90 倍)。**統合しなければ、分けるほど、生涯の合計はかえって減る。**
3. **すぐに統合できれば、分けるほうが少し得**になる(k = 8 で 3% ほど)。上限が元に戻るうえ、ボーナスを早く受け取れるからである。
4. **10 年ほど待たせると、分けても分けなくても、ほとんど同じ**になる(20 年待たせると、分けるほうがわずかに損になる)。作者の言う「統合する期間を調整すれば、ほどほどの ID の分散と統合」は、この模型では、待ち期間を 10 年ほどにすると、ID の数を変える誘因がほぼ消える、という形で実現している。誘因がなければ、書き手は、記事の性格に合わせて(「個性」で)、ID を分けたり分けなかったりするだろう。
5. **割り引くと(早くもらうほうがうれしいと)、分けるほうが得になる**。ボーナスは早く、ペンションの差し引きは遅く来るからである。将来を強く割り引く人(お金に困っている人)ほど、ID を分ける誘因がある。

作者が恐れた「何でも統合するのが得」も、「何でも分けるのが得」も、この規則では起きない。**上限が、受け取った上乗せの累計に効くので、ボーナスで得たぶんを、長い目で見て取り返す**。それが、平方根の配分の弱点(シビル攻撃)を抑える、作者の規則の芯である。

### 6.8 二次資金調達との比較

平方根で小口の支持を重んじる配分は、2019 年に Buterin、Hitzig、Weyl が提案した**二次資金調達**(quadratic funding)と同じ形をしている。公共財(オープンソースのソフトなど)への寄付に、基金から上乗せするとき、上乗せを「寄付の平方根の和の 2 乗」に比例させる。多くの人から少しずつ集めた事業ほど、大きく上乗せされる。そして、作者の規則と同じく、**一人が多くの ID を作れば上乗せを稼げる**という弱点を持つ。Gitcoin などの実際の運用では、ID の本人確認(パスポートのような仕組み)や、つながりの近い寄付を割り引く工夫で対抗している。

作者の対抗のしかたは、それとは違う。本人を確かめるのではなく、**あとで取り返す**(ペンションの上限が累計に効く)。そして、取り返すまでの時間(統合の待ち期間)で、誘因の強さを調整する。本人確認の手間をかけずに済む代わりに、長い時間が要る。作者が「本格的にやるなら 15 年ぐらい」と書いたのは、その長さである。

もう一つ、近いものがある。過去に役立った貢献に、あとから報いる**遡及的な公共財の資金**(Optimism の RetroPGF、「インパクト = 利益」)である。「将来何が役立つかより、過去に何が役立ったかのほうが合意しやすい」という考えで、作者のペンションと同じ方向を向いている。

### 6.9 この章のまとめ

- 作者の規則で、ボーナスは毎期、原資の約 1/3 になり、大口の寄付は数年にならされて使われる(基金の支出の規則)。
- **ペンションは、過去の書き手の「上乗せの倍率」を、今期の書き手の倍率に追いつかせる**仕組みである(過去の拠出の再評価)。分配した期の半分ほどでは、誰にも払われない。分配に占める割合は、引退した書き手が増えるにつれ、最初の 10 年の 1 割ほどから 4 割近くへ増えて落ち着く。
- 上限は今期の倍率の**最小**で決まるので、**一人に大きく送る期には、過去の書き手全員のペンションが止まる**。
- ボーナスの 10 の倍数への切り捨てと、最小の倍率の組み合わせは、基金が乏しいとき、全員のペンションを消す。作者の「原資が 1000 に満たない期は繰り越す」規則が、この罠を防いでいる。
- **ID を分けると、ボーナスは増えるが、ペンションで取り返される**。統合しなければ生涯では損、すぐ統合できれば少し得、10 年ほど待たせればほぼ中立。作者の「ほどほどの ID の分散と統合」は、この模型では実現している。ただし、将来を強く割り引く人には、分ける誘因が残る。

### 6.10 Lean: ID を分けても、取り分は変わらない

6.7 節で、ID を分けると切り捨ての分だけ得をするが、それは取り返されることを見た。比例の配り方そのものは、ID の分割で得をしない。貢献 c を二つの ID に c₁ + c₂ と分けても、二つの取り分の和は分ける前と同じである。

## 第 7 章 BI が否認されたとき: 受け取ったポイントの返し方

### 7.1 作者のメモ

2011 年 3 月 17 日、東日本大震災の 6 日後に、作者は、大震災の前から考えていたことのメモを「ひとこと」に書き出した(aboutme:137725)。

問題は、こうである。作者が受け取ったポイントのうち、どの記事への対価かがはっきりしているものは、受け取って、記事を書くのに役立てるべきである。しかし、もし、送ってくれた人が、そのポイントを、不正に得た BI(ベーシックインカム)から送っていたと、国(元の債権者)が主張したら、どうなるか。

作者の言う BI は、生活のできる額の給付ではなく、**消費税の定額の還付**を念頭に置いた、生活保護と並んで受け取れる少額の、全国民への給付(「負の所得税」に近いもの)である。

作者は、次の原則を立てた。

- ポイントをすぐに返せというのは不当である。少なくとも、そのポイントは**貸付**とみなされるべきである。
- その利子は、BI の受け取りの資格が決まるまでの貸付の利子を超えるべきではない。
- 作者(新しい債権者)は、善意で受け取ったのだから、元の債権者(国)より地位が劣ることはない。だから、**元の債権者が債務者から受け取れないものを、新しい債権者が受け取れるはずがない**。つまり、
  1. 債務者が多く返しているなら、そのうちの一定の割合は、新しい債権者に支払われた分のものでなければならず、そのぶん新しい債権者の利子の負担は減るべきである。
  2. 債務者が返していないなら、元の債権者が受け取れないことを受け入れているのだから、やはり新しい債権者の利子の負担は減るべきである。
- 新しい債権者が、利子の負担を嫌って**代位弁済**(債務者の代わりに払う)する方法も考えたい。払ったあと、債務者が元の債権者に返したものがあれば、そこから分配を受けたい。

作者は、これを Perl のプログラム `point_back_with_rejected_bi.pl`(版 0.03、2011-03-18)にした。その後、ポイントを整数で扱う版(版 0.04、2011-04-04)も書いた。どちらも、入力を取らず、数値の例を計算する。

### 7.2 プログラムの仕組み

記号は作者のものを使う。

| 記号 | 意味 | 例の値 |
|---|---|---|
| R | BI を失ったときの貸付の利率(利息制限法を考えた上限から) | 0.15 |
| P1 | 毎期、債務者から元の債権者に払われる額 | 160 |
| X1 | 元の債権者から債務者への貸付の総額(残高 x1) | 1000 |
| X2 | 債務者から新しい債権者に送られたポイントの総額(残り x2) | 100 |
| Q | 新しい債権者が、最初に供託として積み立てる額(代位弁済に充てる) | 計算で決める |

毎期の動きは次のとおりである。

- 債務者は P1 を払う。そのうち R·x1 が利子で、残りの P1 − R·x1 が元本の返済になる。返済の割合は r3 = (P1 − R·x1) / x1。
- 新しい債権者が負う残り x2 は、債務者の返済の割合と同じだけ減る: x2' = x2 (1 − r3)。
- 新しい債権者は、減ったぶんに利子 R を付けて、元の債権者に払う: p2 = R · x2 · r3。
- 新しい債権者は、最初に供託金 Q を積み、それは債務者の元本に充てられる(x1 は X1 − Q から始まる)。毎期、その時点での必要な供託金を計算し直し、余った分を払い戻す。
- 債務者が利子すら払えない(R·x1 > P1)なら、債務者は返済できないとみて、新しい債権者は何も払わない(原則の 2)。

```python
%%writefile pointback.py
# point_back_with_rejected_bi.pl(版 0.03、2011-03-18)と、整数版(版 0.04、2011-04-04)の書き直し。
# 出力の文字列まで同じになるように書く(Perl の数の表示は %.15g)。
#
# 登場するのは三者。
#   元債権者(国など): 債務者に X1 を貸している(BI を返させる請求)。利率 R。
#   債務者: 毎期 P1 を元債権者に払う。そのうち R·x1 は利子、残りが元本の返済。
#   新債権者(作者): 債務者から X2 のポイントを受け取っていた。それは貸付とみなされる。
# 新債権者は、債務者が元本を返した割合(r3)に応じて、受け取ったポイントの残り x2 を減らし、そのとき利子 R をつけて元債権者に払う。
# さらに、新債権者は最初に供託金 Q を積み、それは債務者の元本の返済(代位弁済)に充てられ、のちに少しずつ払い戻される。
import math


def g(x):
    """Perl の数の表示(%.15g)。"""
    if isinstance(x, int):
        return str(x)
    s = f'{x:.15g}'
    return s


class PointBack:
    def __init__(self, R=0.15, P1=160.0, X1_0=1000.0, X2_0=100.0, integer=False):
        self.R, self.P1, self.X1_0, self.X2_0, self.integer = R, P1, X1_0, X2_0, integer
        q = self.calc_q(X1_0, X2_0)
        self.Q = math.ceil(q) if integer else q

    def x1next(self, x1):
        return x1 - (self.P1 - self.R * x1)

    def r3(self, x1):
        return (self.P1 - self.R * x1) / x1

    def x2next(self, x1, x2):
        return x2 * (1 - self.r3(x1))

    def p2(self, x1, x2):
        return self.R * x2 * self.r3(x1)

    def x1b(self, n):
        R, P1 = self.R, self.P1
        return ((1 + R) ** n) * (self.X1_0 - self.Q - P1 / R) + P1 / R

    def p2next(self, x1, x2, p2):
        R = self.R
        return R * (self.X2_0 / (self.X1_0 - self.Q)) * self.P1 - R * R * x2 + R * p2

    def _ka(self):
        R = self.R
        k = R * (self.X2_0 / (self.X1_0 - self.Q)) * self.P1
        return R, -k / (1 - R), self.p2(self.X1_0 - self.Q, self.X2_0)

    def q2(self, n):
        r, k1, a0 = self._ka()
        return (r ** n) * (a0 + k1) - k1

    def sum_q2(self, n):
        r, k1, a0 = self._ka()
        return (1 - (r ** n)) / (1 - r) * (a0 + k1) - n * k1

    def nlast(self):
        return math.log(self.P1 / (self.P1 - self.R * (self.X1_0 - self.Q))) / math.log(1 + self.R)

    def calc_q(self, x1_0, x2_0):
        R = self.R

        def sum_q2b(q):
            p1 = R * x1_0
            Tq = math.log(p1 / (p1 - R * (x1_0 - q))) / math.log(1 + R)
            k = R * (x2_0 / (x1_0 - q)) * p1
            k1 = -k / (1 - R)
            a0 = self.p2(x1_0 - q, x2_0)
            return (1 - (R ** Tq)) / (1 - R) * (a0 + k1) - (1 + Tq) * k1
        step, q = x1_0 / 4, x1_0 / 2
        while True:
            qb = sum_q2b(q)
            if abs(qb - q) < 0.1:
                break
            q = q - step if qb < q else q + step
            step /= 2
        return q

    def run(self):
        """作者のプログラムと同じ出力の文字列と、結果(Sum p2、Sum payback、期数)を返す。"""
        R, P1, out = self.R, self.P1, []
        pr = out.append
        if R * self.X1_0 > P1:
            return 'No need of P2\n', None
        pr('init:\n\n')
        ret = 0
        x1, x2, n = self.X1_0 - self.Q, self.X2_0, 0
        sq2, sum_payback = 0, 0
        qcur = math.ceil(self.calc_q(self.X1_0, self.X2_0)) if self.integer else self.calc_q(self.X1_0, self.X2_0)
        pr(f'q = {g(qcur)}\n')
        pr('\nloop:\n\n')
        while x1 > 0 and x2 > 0:
            pr(f'x1 = {g(x1)} {g(self.x1b(n))}\n')
            pr(f'x2 = {g(x2)}\n')
            if R * x1 > P1:
                pr('No need of P2. Ended for uncertain condition.\n\n')
                x1 = self.x1next(x1); x2 = self.x2next(x1, x2); n += 1
                break
            elif x1 < P1:
                pr('No need of P2. Ended normally.\n\n')
                x1 -= x1; n += 1
                continue
            elif not self.integer:
                p2 = self.p2(x1, x2); ret += p2
                pr(f'p2 ={g(p2)}\n'); pr(f'p2next = {g(self.p2next(x1, x2, p2))}\n'); pr(f'q2 = {g(self.q2(n))}\n')
                sq2 += self.q2(n)
            if self.integer:
                p2 = self.p2(x1, x2); ret += p2
                pr(f'p2 ={g(p2)}\n'); pr(f'p2next = {g(self.p2next(x1, x2, p2))}\n'); pr(f'q2 = {g(self.q2(n))}\n')
                sq2 += self.q2(n)
            pr(f'x2/x1 = {g(x2 / x1)}\n')
            x1n, x2n = self.x1next(x1), self.x2next(x1, x2)
            pr(f'x2n/x2cur = {g(x2n / x2)}\n'); pr(f'x1n/x1cur = {g(x1n / x1)}\n')
            if x1n > 0 and x2n > 0:
                qn = self.calc_q(x1n + qcur, x2n)
                if self.integer:
                    payback = math.floor(qcur - qn - p2)
                    pr(f'calc_q = {g(qn)}\n'); pr(f'payback = {g(payback)}\n')
                    qcur = qn - (p2 - (qcur - qn - payback))
                    pr(f'next qcur = {g(qcur)}\n')
                else:
                    payback = qcur - qn - self.p2(x1, x2)
                    pr(f'calc_q = {g(qn)}\n'); pr(f'payback = {g(payback)}\n')
                    qcur = qn
                sum_payback += payback
            else:
                payback = qcur - self.p2(x1, x2)
                pr(f'payback = {g(payback)}\n')
                sum_payback += payback
                qcur = 0
            x1, x2 = x1n, x2n
            n += 1
            pr('\n')
        if qcur:
            if self.integer:
                payback = math.floor(qcur)
                pr(f'last payback = {g(payback)}\n\n')
                sum_payback += payback
            else:
                sum_payback += qcur
        pr('result : \n\n')
        pr(f'Sum p2 = {g(ret)}\n'); pr(f'Sum payback = {g(sum_payback)}\n')
        pr(f'n = {n} < {g(self.nlast())} + 1\n'); pr(f'sum_q2 = {g(sq2)} {g(self.sum_q2(n))}\n')
        return ''.join(out), dict(sum_p2=ret, sum_payback=sum_payback, n=n, Q=self.Q)
```

### 7.4 閉じた式を解き直す

作者は、プログラムの中で、毎期の p2 と、その合計の閉じた式(q2、sum_q2)を導いている。出力の最後の行 `sum_q2 = 41.755... 44.750...` が、それである。ところが、実際の p2 の合計は 14.09 で、閉じた式とは大きく違う。どこが違うのかを確かめる。

x1 の更新は x1' = (1 + R) x1 − P1 である。両辺に R を掛けて P1 から引くと、

$$ P_1 - R\,x_1' = (1+R)(P_1 - R\,x_1) $$

なので、P1 − R x1ₙ = (1 + R)ⁿ (P1 − R x1₀)。一方、x2 と x1 は同じ割合 r3 で減るので、**x2 / x1 は一定**(c = X2 / x1₀)である(出力の `x2/x1 = 0.106...` が毎期同じ)。よって、

$$ p_{2,n} = R\,c\,(P_1 - R\,x_{1,n}) = R\,c\,(P_1 - R\,x_{1,0})\,(1+R)^n $$

p2 は、**比 (1 + R) の等比数列**である。作者の漸化式 `p2next = R c P1 − R² x2 + R p2` は、x2 = c·x1 を入れると (1 + R) p2 に等しく、正しい(出力の `p2next` は、次の期の `p2` とぴったり同じ)。間違っていたのは、それを解いた閉じた式 q2 で、作者は −R² x2 の項を定数のように扱い、比を R として解いていた。

```python
%%writefile -a pointback.py
# 正しく解き直した式。
# x1 の更新 x1' = (1+R) x1 − P1 から、P1 − R x1ₙ = (1+R)ⁿ (P1 − R x1₀)。
# x2 / x1 は一定(= c = X2 / x1₀)なので、p2ₙ = R c (P1 − R x1ₙ) = R c (P1 − R x1₀)(1+R)ⁿ: 比 (1+R) の等比数列。
# 作者の漸化式 p2' = R c P1 − R² x2 + R p2 は、これと同じ(= (1+R) p2)。作者の閉じた式 q2 は、−R² x2 を定数とみなして解いていた。
def p2_exact(n, R, P1, x1_0, X2):
    c = X2 / x1_0
    return R * c * (P1 - R * x1_0) * (1 + R) ** n


def sum_p2_exact(N, R, P1, x1_0, X2):
    """N 期ぶんの p2 の和 = R c (x1₀ − x1_N) = R × (x2 の減った分)。"""
    c = X2 / x1_0
    return c * (P1 - R * x1_0) * ((1 + R) ** N - 1)


def total_burden(R, P1, X1, X2, q=0.0):
    """供託金 q を代位弁済に充てたとき、新債権者が元債権者に払う合計(作者の終わり方: x1 < P1 の期は p2 なし)。"""
    x1, x2, tot, n = X1 - q, X2, 0.0, 0
    if R * x1 >= P1:
        return 0.0, math.inf                     # 債務者は元本を返せない(貸倒れ)。p2 も払わない
    while x1 >= P1:
        tot += R * x2 * (P1 - R * x1) / x1
        x2 *= 1 - (P1 - R * x1) / x1
        x1 = (1 + R) * x1 - P1
        n += 1
    return tot, n + 1


def fixed_point_q(R, X1, X2, tol=1e-10):
    """作者の供託金の考え方(最悪の場合 P1 = R·X1 でも、供託金 q を充てれば返済が進み、
    その間の p2 の合計が q に等しくなる)を、正しい合計で解く。"""
    lo, hi = 1e-9, X1 * (1 - R) * 0.999
    f = lambda q: total_burden(R, R * X1, X1, X2, q)[0] - q
    for _ in range(200):
        mid = (lo + hi) / 2
        if f(mid) > 0:
            lo = mid
        else:
            hi = mid
    return (lo + hi) / 2
```

```python
import importlib; importlib.reload(pb)    # 書き足したモジュールを読み直す
o = pb.PointBack()
x1_0 = o.X1_0 - o.Q
print(' n   p2(プログラムの値)   正しい閉じた式   作者の閉じた式 q2')
x1, x2 = x1_0, o.X2_0
for n in range(6):
    print(f'{n:2d}   {o.p2(x1, x2):16.6f}   {pb.p2_exact(n, o.R, o.P1, x1_0, o.X2_0):14.6f}   {o.q2(n):14.6f}')
    x1, x2 = o.x1next(x1), o.x2next(x1, x2)
N = res[False]['n'] - 1          # p2 を払った期の数(最後の期は残りを払うだけ)
print(f'\n{N} 期の p2 の合計:  プログラム {res[False]["sum_p2"]:.6f}   正しい閉じた式 {pb.sum_p2_exact(N, o.R, o.P1, x1_0, o.X2_0):.6f}   作者の式 {o.sum_q2(N):.6f}')
```

### 7.5 負担の合計は、利率一期分を超えない

正しい式から、もう一つのことがわかる。p2 は、毎期の x2 の減り方 x2 − x2' = x2·r3 に R を掛けたものだから、その合計は

$$ \sum_n p_{2,n} = R\,(X_2 - x_{2,\text{last}}) \le R\,X_2 $$

である。**新しい債権者の負担の合計は、受け取ったポイントの、利率一期分(R·X2 = 15)を決して超えない。** 債務者がすべて返せば、ほぼ R·X2 になり、返せなければ 0 になる。

作者は、プログラムの中に、次のように書いていた。

> sum_q2 は P1 → R·X1₀ で発散しうる。これを防ぐには、Sum P2 の中に元本返済に相当する分が少しでもあればいい。そこで Q を導入した。

つまり、供託金 Q は、債務者の払う額が利子ぎりぎり(P1 → R·X1)のときに、負担の合計が発散するのを防ぐために入れられた。しかし、正しい式では、合計は R·X2 で抑えられていて、発散しない。**発散するのは、合計ではなく、返し終わるまでの期間である。** 作者の閉じた式の誤りが、ありもしない発散を見せていた。

```python
print('債務者の払う額 P1 を変える(R = 0.15、X1 = 1000、X2 = 100。利子ぎりぎりは P1 = 150)')
print('   P1    供託金なし: 負担の合計  期数    作者の供託金 Q   Q を充てたとき: 負担の合計  期数')
for P1 in (150.5, 151, 155, 160, 200, 300, 500):
    t0, n0 = pb.total_burden(0.15, P1, 1000, 100, 0.0)
    Qa = pb.PointBack(P1=P1).Q
    t1, n1 = pb.total_burden(0.15, P1, 1000, 100, Qa)
    print(f'{P1:6.1f}           {t0:7.3f}      {n0:4d}          {Qa:7.2f}              {t1:7.3f}        {n1:4d}')
print('上限 R·X2 =', 0.15 * 100)
```

債務者の払う額が利子ぎりぎりに近づくと、返し終わるまでの期数は 20、25、36、41 と延びていくが、負担の合計は 15 を超えない。供託金を代位弁済に充てると、債務者の元本がそのぶん減るので、返済が進み、期数は短くなる(P1 = 150.5 で 41 期 → 20 期)。供託金の本当の役目は、**負担の上限を作ることではなく、返済を前に進めること**だった。そして、作者の供託金 Q ≈ 57 は、負担の上限 15 の 4 倍ほどの額で、多すぎる。ただし、毎期計算し直して払い戻すので、仕組みとしては壊れていない。新しい債権者は、多めに積んで、多めに戻される。

### 7.6 原則は守られているか

作者の原則を、結果で確かめる。

- **「元の債権者が債務者から受け取れないものを、新しい債権者が受け取れるはずがない」**: 債務者が利子すら払えない(R·X1 > P1)とき、プログラムは `No need of P2` と出して終わる。新しい債権者は何も払わない。返済が進めば進むほど(r3 が大きいほど)、新しい債権者の払う p2 は大きい。債務者が返したぶんにだけ、新しい債権者が利子を払う形である。
- **負担の上限**: 合計は R·X2 以下。つまり、新しい債権者は、受け取ったポイントを、**一期だけ利率 R で借りていた**のと同じ以上には払わない。作者の「その利子は、BI の受け取りの資格が決まるまでの貸付の利子を超えるべきではない」に合う。
- **代位弁済の払い戻し**: 供託金は、債務者の返済に応じて払い戻される(作者の「債務者が元債権者に返済したものがあるなら、そこから分配を受けたい」)。

作者の原則は、正しい式で見ても守られている。誤りは、供託金の額を決める閉じた式だけにあった。

### 7.7 マイナス金利で、何が壊れるか

作者は、2026 年 1 月の振り返りで、「マイナス金利だと、貸倒れの判断ができなくなり、論理が破綻する。その場合は、借金を積み増していけば得をする」と書いた。プログラムで確かめる。作者のプログラムの、定数 R だけを書き換えた写しを走らせる。

```python
import tempfile, re
src = ppath[False].read_text()
for R in (0.2, 0.0, -0.01):
    with tempfile.NamedTemporaryFile('w', suffix='.pl', delete=False) as f:
        f.write(re.sub(r'^my \$R = 0\.15;', f'my $R = {R};', src, flags=re.M))
    out = subprocess.run(['perl', f.name], capture_output=True, text=True)
    print(f'R = {R:5}: ' + (out.stdout.strip().splitlines() or [''])[0] + ' ' + out.stderr.strip().replace(f.name, '(写し)'))
print()
print('正しい計算で、負担の合計(供託金なし、P1 = 160):')
for R in (0.15, 0.05, 0.01, 0.0, -0.01, -0.05):
    t, n = pb.total_burden(R, 160, 1000, 100, 0.0)
    print(f'  R = {R:5}:  新しい債権者の払う額の合計 {t:7.3f}   期数 {n}')
```

- R = 0.2 では、R·X1 = 200 > P1 = 160 で、債務者は利子も払えない。プログラムは「P2 は要らない」として終わる(貸倒れの判定)。
- **R = 0 と R < 0 では、プログラムは、ゼロでの割り算で止まる**。供託金を決める式に log(1 + R) で割る部分や、二分法が x1₀ − q = 0 にたどり着く部分があるからである。
- 正しい計算で見ると、R = 0 では新しい債権者の負担は 0、R < 0 では**マイナス**、つまり、新しい債権者が元の債権者から受け取ることになる。
- そして、R ≤ 0 では、貸倒れの判定 R·X1 > P1 は、決して成り立たない。債務者がどれだけ少ししか払わなくても、「返せない」とは判定されない。

作者の 2026 年の言葉どおり、**マイナス金利では、貸倒れの判断ができなくなり、論理が破綻する**。プログラムは文字どおり止まり、式は「借りているほうが受け取る」を示す。次の章で、これを、もう少し広い文脈(借入の上限と、ポンジ・ゲーム)で考える。

### 7.8 この章のまとめ

- 作者の二つのプログラムは、今の Perl でそのまま動き、Python の書き直しは、出力の文字列が一字一句一致する。

### 7.9 Lean: 返し方の閉じた式と、負担の上限

7.4 節で、作者の漸化式は正しく、閉じた式の比が誤っていた(R ではなく 1 + R)ことを見た。漸化式 p_{n+1} = (1 + R) p_n の解は p_n = p_0 (1 + R)^n で、その和は ((1 + R)^n − 1) / R になる。7.5 節の「負担の合計は、利率一期分を超えない」も、式で確かめる。

```python
%%writefile -a points_pension.lean
/-! ## 第 7 章 BI が否認されたときの返し方

作者の漸化式 p_{n+1} = (1 + R) p_n の解は、比 (1 + R) の等比数列 p_n = p_0 (1 + R)^n(閉じた式の比は R ではない)。 -/
theorem pointback_geometric (p₀ R : ℝ) (n : ℕ) : p₀ * (1 + R) ^ (n + 1) = (1 + R) * (p₀ * (1 + R) ^ n) := by ring

/-- 等比数列の和: Σ_{k<n} (1 + R)^k = ((1 + R)^n − 1) / R(R ≠ 0)。 -/
theorem pointback_sum (R : ℝ) (hR : R ≠ 0) (n : ℕ) : ∑ k ∈ range n, (1 + R) ^ k = ((1 + R) ^ n - 1) / R := by
  rw [geom_sum_eq (by intro h; apply hR; linarith)]
  congr 1; ring

/-- 負担の合計 R (X₂ − x) は、最後に残る分 x が 0 以上で R ≥ 0 なら、利率一期分 R X₂ を超えない。 -/
theorem burden_bound {R X₂ x : ℝ} (hR : 0 ≤ R) (hx : 0 ≤ x) : R * (X₂ - x) ≤ R * X₂ := by
  nlinarith
```

## 第 8 章 自然な借入制約と、ポンジ・ゲーム

### 8.1 BI を担保にした貸付

第 7 章のメモ(aboutme:137725)には、もう一つの考察があった。BI の受け取りの資格がはっきりしないとき、BI の給付を留め置いて、それを担保に貸し付ける場合である。

> BI 担保貸付利率を r1、無担保貸付利率を r2、一期の BI を I1、総貸付額を X とすると、少なくとも r1 * X < I1 でなければ担保あり貸付は成り立たない。
> n 期の貸付けで X = n * I1 とすると、n * r1 * I1 < I1 ⟺ n < 1/r1。これが少なくとも成り立つ。
> さらに複利などの問題があるので、これより短い期間しか有担保貸付けは成り立たない。この先は無担保貸付けとなるわけだが、そうなるとこの BI は成立しない公算が強くなり、最初から無担保貸付であったとして、割り戻し再計算がされなければならないだろう。すると、X < Y なる Y が貸付額となり、増えた貸付け額を加味して直近の利率を計算すると、r2 よりさえ大きくなる。(ひょっとして、これが高利貸しの必然性なのか?)

そして、利子が担保を食いつぶして 0 になれば、「貸倒れ」と判断せざるを得ない、と続く。

### 8.2 自然な借入制約

毎期 I の収入が永久に続くとき、利率 r で借りられる上限は、**利子を毎期の収入で払いきれる額**、X = I / r である。これより多く借りると、利子だけで収入を超え、借金はいつまでも減らない。経済学では、これを**自然な借入制約**と呼ぶ(Aiyagari 1994 では、最低の所得を利率で割った値)。作者の不等式 r1 X < I1 は、これそのものである。

作者は、「複利などの問題があるので、これより短い期間しか」成り立たない、と書いた。どれだけ短いかを計算しておく。毎期 I を借り、利子が複利で付くと、n 期後の借金は X_n = I((1 + r)ⁿ − 1)/r になる。条件 r X_n < I は、(1 + r)ⁿ < 2、つまり

$$ n < \frac{\ln 2}{\ln(1+r)} $$

となる。右辺は、**借金が倍になるまでの期数**(いわゆる「72 の法則」で、72 / 利率(%) と近似される値)である。単利の上限 1/r との比は、r が小さいとき ln 2 ≈ 0.69 になる。

```python
import math
print('  利率 r    単利の上限 1/r    複利の上限 ln2/ln(1+r)    比')
for r in (0.01, 0.03, 0.05, 0.10, 0.15, 0.30):
    a, b = 1 / r, math.log(2) / math.log(1 + r)
    print(f'  {r:5.2f}     {a:10.1f}           {b:10.1f}             {b / a:.2f}')
```

### 8.3 「高利貸しの必然性」: 上限を超えたときの利率の跳ね上がり

上限を超えたら、最初から無担保の貸付(利率 r2)だったとして計算し直す、というのが作者の考えだった。担保つきの利率を r1 = 0.05、無担保の利率を r2 = 0.15 とし、毎期 I = 1 を借りるとする。担保つきとして計算した借金を X_n、無担保として計算し直した借金を Y_n とする。作者の言う「直近の利率」を、計算し直した利子 r2 Y_n を、借り手が担保つきだと思っていた借金 X_n で割った値として計算する。

```python
r1, r2 = 0.05, 0.15
lim = math.log(2) / math.log(1 + r1)
print(f'担保つきの上限: {lim:.1f} 期(r1 = {r1})')
print('  期数 n   担保つきの借金 X   無担保で計算し直した借金 Y   直近の利率 r2·Y/X')
for n in (5, 10, 14, 15, 20, 25):
    X = ((1 + r1) ** n - 1) / r1
    Y = ((1 + r2) ** n - 1) / r2
    print(f'  {n:4d}      {X:10.2f}                {Y:10.2f}                 {r2 * Y / X:.3f}' + ('   ← 上限を超えた' if n > lim else ''))
```

上限(この例では 14.2 期)を超えた 15 期目に計算し直すと、直近の利率は、無担保の利率 0.15 をさらに上回る 0.33 になり、20 期目なら 0.47 になる。借り手は、それまで 0.05 で借りているつもりだった。**担保が尽きた瞬間に、利率は 0.05 から 0.33 へ、連続的にではなく跳ね上がる。** 作者は、これを「高利貸しの必然性」ではないかと書いた。

経済学では、Stiglitz and Weiss(1981)が、貸し手は借り手の返す力を完全には知らないので、利率を上げると、返す気の薄い借り手ばかりが残る(逆選択)、だから貸し手は利率を上げる代わりに貸す量を絞る(**信用割当**)、と論じた。作者の計算は、別の角度から同じ現象を見ている。担保の範囲の内と外とで、借り手は別の種類の借り手になり、利率は不連続に変わる。現実の消費者金融で、担保や保証のある貸付と、ない貸付のあいだに、大きな利率の差があるのはこのためである。日本の利息制限法(上限 15〜20%)は、作者の言うとおり、この跳ね上がりに天井を付けるものと読める。

### 8.4 貸倒れの判定と、マイナス金利

作者の貸倒れの判定は、「利子が担保を食いつぶすとき」、つまり r X ≥ I である。この判定が、利率によってどう働くかを見る。借り手は、毎期 I = 1 を借り、決して返さない(借り換え続ける)。

```python
print('毎期 1 を借りて返さない借り手。貸倒れと判定される(r·D ≥ 1)のは何期目か、200 期後の借金はいくらか')
for r in (0.10, 0.05, 0.01, 0.0, -0.01, -0.05):
    D, t_def = 0.0, None
    for t in range(1, 201):
        D = (1 + r) * D + 1
        if t_def is None and r * D >= 1:
            t_def = t
    lim = (f'{-1 / r:.0f} に近づく' if r < 0 else '際限なく増える')
    print(f'  r = {r:5.2f}:  判定 {str(t_def) + " 期目" if t_def else "されない":>8s}   200 期後の借金 {D:12.1f}  ({lim})')
```

- r > 0 なら、借金はいずれ利子だけで収入を超え、貸倒れと判定される(r = 0.05 で 15 期目。8.2 節の複利の上限と同じ)。
- **r = 0 なら、判定は決してされない。** 利子は 0 なので、借金がいくら増えても、「利子が担保を食いつぶす」ことは起きない。借金は際限なく増える。
- **r < 0 なら、判定されないうえに、借金は一定の値(−1/r)に近づく。** 毎期 1 を借りて使い続けても、借金はマイナスの利子で目減りして、増えない。借り続けることが、ただで使い続けることになる。

作者の 2026 年の言葉、「その場合は、わざわざ複雑に計算して自分にポイントを確定させるようなまだるっこしいことをせずとも、単に借金を積み増していけばいいだけになる。そうすれば得をするのだ」は、この r < 0 の行そのものである。

### 8.5 ポンジ・ゲームの禁止

借金を、返さずに借り換え続けることを、経済学では**ポンジ・ゲーム**と呼ぶ。ふつうの経済のモデルでは、これを禁じる条件(**ポンジ・ゲームの禁止**、横断性の条件)を置く。借金の現在価値が、遠い将来に 0 に収束しなければならない、という条件である。

$$ \lim_{t \to \infty} \frac{D_t}{(1+r)^t} \le 0 $$

r > 0 なら、借り換え続ける借金は (1 + r)ᵗ より速く増えるので、この条件は破れ、貸し手は貸さない。ところが、**r が経済の成長率 g 以下なら、借り換え続けても、借金の所得に対する比は発散しない**。Blanchard(2019)は、アメリカの国債の金利が長く成長率を下回ってきたことを指摘し、そのような国の借金の財政的なコストは小さい、と論じた。日本では、2016 年から 2024 年まで、日本銀行がマイナス金利の政策をとっていた。

作者の貸倒れの判定 r X ≥ I は、成長のない経済で、ポンジ・ゲームを見分けるための条件である。r ≤ 0(成長がある経済なら r ≤ g)では、この条件が働かなくなる。作者は、自分の小さな制度の中で、**r ≤ g の世界では、借金の返済を強制する論理そのものがなくなる**ことに、2026 年に気づいたのだった。

### 8.6 ノート 2 の 2019 年のモデル: ゼロ金利で、上限のない経済

ノート 2 で解剖した 2019 年の資産市場のモデルは、**利子がなく、借りる額に上限もない**経済だった。作者自身、BI 論の影響の名残として、借金がかなり自由にできると書いていた。8.4 節の言葉で言えば、r = 0 で、貸倒れの判定がされない経済である。

それでも、ノート 2 では、全員の借金の合計は際限なく増えなかった。その理由は、利子でも担保でもなく、**お金の流れがストックに応じて変わる負のフィードバック**(借金が多いほど返済が増え、取り崩しが減る)だった(ノート 2 の第 3.4・3.5 節)。そして、ノート 2 の第 3.6 節で利子を付けると、毎期返せる上限(5)を利率で割った値 5 / r を超えた人は、いつまでも返しきれなくなった。これは、この章の自然な借入制約と同じものである。

ゼロ金利の経済で借金を抑えるのは、利子の重みではなく、**流れの仕組み**である。そういう仕組みがなければ、借金は増え続ける(8.4 節)。

### 8.7 ペンションを担保にする

作者は、2026 年の振り返りで、もう一つのことを書いている。「ペンションプランは、支払いがありうるので、それを受けずに借金できる。それは、株式のような富の創造ではなく、銀行の信用創造に近い」。

第 6 章で見たように、作者のペンションは、毎期決まった額を払う年金ではない。過去の書き手の上乗せの倍率を、今期の倍率に追いつかせる仕組みで、分配した期の半分ほどでは払われず、今期のいちばん大口の書き手に左右される。そして、上乗せの累計には上限(A^1.05 × r_min)があるので、一人が生涯に受け取れるペンションの総額には、上限がある。

これを担保に借りるとすると、担保の値打ちは、

- **永久に続く流れではなく、上限のある残高**(まだ受け取っていない「追いつきの余地」)であり、
- **不確か**(いつ払われるかは、基金に入る匿名の寄付と、その期の大口の書き手しだい)

である。だから、自然な借入制約(I / r)をそのまま当てはめることはできず、担保の掛け目(ヘアカット)を大きくとる必要がある。そして、ペンションを担保に貸すことは、将来の寄付を当てにして今お金を作ることなので、作者の言うとおり、銀行の信用創造(将来の返済を当てにして、今預金を作る)に近い。信用循環モデル(credit_simulation_v6)の言葉で言えば、ペンションの基金は、準備に乏しい銀行のようなものになる。

現実の制度では、日本の公的年金を担保にした貸付(独立行政法人福祉医療機構の年金担保貸付制度)は、2010 年の閣議決定で廃止が決まり、2022 年 3 月末で新しい申し込みを終えた。2022 年 4 月以降、公的年金の受給権を担保にした融資を認める制度はなく、そうした融資を持ちかける業者は違法とされている。年金という「将来の確かな流れ」を担保にした貸付ですら、生活を守る観点から閉じられたのである。

### 8.8 この章のまとめ

- 作者の BI 担保貸付の不等式 r1 X < I1(n < 1/r1)は、**自然な借入制約**(収入 / 利率)である。複利では、上限は **ln 2 / ln(1 + r)**(借金が倍になる期数)に縮む。作者の「これより短い期間しか」は、この値だった。
- 上限を超えて無担保として計算し直すと、**利率は、無担保の利率をさらに上回って跳ね上がる**(作者の「高利貸しの必然性」)。信用割当(Stiglitz and Weiss 1981)と同じ、担保の内と外での不連続である。
- 作者の貸倒れの判定 r X ≥ I は、**r ≤ 0 では決して成り立たない**。r = 0 では借金が際限なく増え、r < 0 では借り続けても借金が増えない(借り続けることが、ただで使い続けることになる)。これは、**ポンジ・ゲームの禁止が r ≤ g で破れる**ことの、小さな制度の中での現れである。
- ノート 2 の 2019 年のモデルは、まさにこのゼロ金利・上限なしの経済で、借金を抑えていたのは、流れの負のフィードバックだった。
- ペンションを担保にした借金は、上限のある、不確かな残高を担保にした、信用創造に近いものになる。

第 III 部はここまでである。第 IV 部では、作者の構想全体の目的に戻る。**書き手は、どう食べていけばよいのか。**

### 8.9 Lean: 複利の上限と、ポンジ・ゲーム

8.2 節の「複利の上限 ln 2 / ln(1 + r) は単利の上限 1/r より小さい」は、log の凹さ(1 と 2 を結ぶ弦の上にある: log(1 + r) ≥ r log 2)から出る。8.4 節と 8.5 節の、毎期 1 を借りて返さない借り手の借金 D_{t+1} = (1 + r) D_t + 1 は、r < 0 なら −1/r に落ち着き(貸倒れと判定されない)、r = 0 なら毎期 1 ずつ際限なく増える。

```python
%%writefile -a points_pension.lean
/-! ## 第 8 章 自然な借入制約と、ポンジ・ゲーム

複利で倍になるまでの期数 log 2 / log(1 + r) は、単利で倍になるまでの期数 1 / r 以下(0 < r ≤ 1)。
log の凹さ(1 と 2 を結ぶ弦の上にある)から、log(1 + r) ≥ r log 2。 -/
theorem log_one_add_ge {r : ℝ} (h0 : 0 ≤ r) (h1 : r ≤ 1) : r * Real.log 2 ≤ Real.log (1 + r) := by
  have hc := strictConcaveOn_log_Ioi.concaveOn
  have := hc.2 (show (1:ℝ) ∈ Set.Ioi 0 by norm_num) (show (2:ℝ) ∈ Set.Ioi 0 by norm_num)
    (show 0 ≤ 1 - r by linarith) h0 (by ring)
  simp only [smul_eq_mul, Real.log_one, mul_zero, zero_add] at this
  have e : (1 - r) * 1 + r * 2 = 1 + r := by ring
  rw [e] at this; linarith

theorem compound_le_simple {r : ℝ} (h0 : 0 < r) (h1 : r ≤ 1) : Real.log 2 / Real.log (1 + r) ≤ 1 / r := by
  have hl : 0 < Real.log (1 + r) := Real.log_pos (by linarith)
  rw [div_le_div_iff₀ hl h0]
  have := log_one_add_ge h0.le h1
  linarith

/-- 毎期 1 を借りて返さない借り手の借金 D_{t+1} = (1 + r) D_t + 1。r < 0 なら、借金は −1/r に落ち着く(固定点)。 -/
theorem ponzi_fixed_point {r : ℝ} (hr : r ≠ 0) : (1 + r) * (-1 / r) + 1 = -1 / r := by
  field_simp; ring

/-- r = 0 なら、借金は毎期 1 ずつ増え、際限なく増える(貸倒れの判定は働かない)。 -/
theorem ponzi_zero_rate (D : ℕ → ℝ) (h0 : D 0 = 0) (h : ∀ t, D (t + 1) = (1 + 0) * D t + 1) : ∀ t, D t = t := by
  intro t; induction t with
  | zero => simp [h0]
  | succ n ih => rw [h, ih]; push_cast; ring
```

# 第 IV 部 著者はどう食べていくか

## 第 9 章 報酬請求権化と、著者の BI

### 9.1 「デジタル著作権の報酬請求権化に向けて」(2007)

作者の一連の構想の目的は、2007 年 11 月の記事「デジタル著作権の報酬請求権化に向けて」に、いちばんはっきり書かれている。

作者は、まず、なぜ著作権が独占権として守られてきたのかを問う。

> 他の独占権がなくなっていく中、なぜ、著作権は独占権としてずっと守られてきたのだろうか? 私はその大きな理由が紙幣制度にあるのではないかと思っている。

紙幣の制度を守るには、印刷の版を押さえなければならない。偽の版で刷られたものが出回ったとき、「報酬を請求できる」などと悠長なことは言っていられない。だから独占権が必要で、民間にもそれを認めて、お互いに見張らせるほうがよい。作者は、紙幣に載せられたデータ(数値)そのものには価値がなく、**メディアの製造を押さえられることが価値を決めている**と書く。

そして、データだけで流通できる時代には、メディアの製造を押さえるための独占権は意味を失い、著作者の収入は**「パトロン + 報酬請求権」**(独占して許諾するのではなく、使われたら報酬を請求できる、弱い権利)に近づいていく、と見立てた。最後に、こうまとめている。

> こう考えてみると、「間接税」である補償金からの報酬請求により基礎的収入を支え、薄く広い「直接税」によるパトロネージュで儲けを出すという昔とは逆転したような形が、これからの著作権周辺の在り方なのかもしれない。

ここで「間接税」とは、私的録音録画補償金のような、機器や媒体の値段に上乗せして集め、権利者に配るお金である。「直接税」とは、読者が書き手に直接払うお金(投げ銭)である。**基礎の収入を、集めて配るお金で支え、上乗せを、直接の支払いで得る。** これは、BI(基礎)と投げ銭(上乗せ)の形そのものである。

作者は、この記事で「著作権に関する話題からは意識して離れていこうと思う」と書いた。そして、そのあとの構想(2009 年の空スター、2011 年の分配)は、著作権の議論ではなく、**ポイントの小さな制度**として現れた。

### 9.2 分配の基金を、一人の規模の補償金制度として読む

この見立てで、第 5・6 章の分配の規則を読み直すと、次のように対応する。

| 2007 年の見立て | 2011 年の分配 |
|---|---|
| 間接税(補償金)で集めるお金 | どの記事への対価かがはっきりしない、匿名のポイント(基金) |
| 薄く広い直接税(パトロネージュ) | 作者が書き手に送る基礎ポイント(投げ銭) |
| 補償金からの基礎的な収入 | ペンション(過去の書き手への、追いつきの給付) |
| 直接の支払いへの上乗せ | ボーナス(今期の投げ銭に、平方根で上乗せ) |

作者の分配は、**一人の読者が運営する、小さな補償金制度**だった。集めたお金(匿名のポイント)を、使われ方(作者がブックマークして投げ銭した記事)に応じて配る。配り方に、平方根(小口を重んじる)と、ペンション(過去の書き手を今の水準に追いつかせる)の工夫がある。

### 9.3 配り方を比べる

集めたお金の配り方で、書き手の収入はどう変わるか。書き手 1000 人の「使われ方」(読まれる量)が、上位ほど極端に多い分布(ジップの法則、順位の 1.1 乗に反比例)に従うとする。集めたお金を、次の四通りで配る。

- **比例**: 使われ方に比例して配る(今の多くの補償金や、音楽配信の「プロラタ」の配り方)。
- **平方根**: 使われ方の平方根に比例して配る(作者のボーナス)。
- **均等**: 全員に同じ額を配る(BI)。
- **下限つき比例**: 使われ方が下位の半分の書き手には配らず、残りに比例して配る(音楽配信の Spotify が 2024 年から、年に 1000 回未満しか再生されない曲には払わないとしたのに近い)。

```python
N = 1000
rank = np.arange(1, N + 1)
use = 1 / rank ** 1.1
use /= use.sum()
cut = use >= np.median(use)
rules = {'比例': use, '平方根': np.sqrt(use) / np.sqrt(use).sum(), '均等': np.full(N, 1 / N),
         '下限つき比例': np.where(cut, use, 0) / use[cut].sum()}

def gini(x):
    x = np.sort(x); n = len(x)
    return (2 * np.arange(1, n + 1) - n - 1) @ x / (n * x.sum())

print('集めたお金の配り方と、書き手 1000 人への分かれ方')
print('  配り方          上位 1% の取り分   下位半分の取り分   ジニ係数   平均の 1/2 以上を受け取る人')
for k, s in rules.items():
    print(f'  {k:12s}      {100 * s[:10].sum():6.1f}%          {100 * s[N // 2:].sum():6.1f}%        {gini(s):.2f}        {100 * (s >= 0.5 / N).mean():5.1f}%')
```

比例で配ると、上位 1%(10 人)が半分近くを取り、下位の半分には 6% ほどしか届かない。均等に配ると、誰もが同じ額を受け取る。平方根は、そのあいだにあり、上位 1% の取り分は 1 割ほどに下がり、下位半分に 3 割近くが届き、9 割近くの書き手が平均の半分以上を受け取る。下限つき比例は、下位半分を切り捨てるので、比例よりもさらに偏る。

作者の分配は、ボーナス(平方根)とペンション(過去の書き手への追いつき)の組み合わせだった。そして、2007 年の見立ては、基礎を集めて配るお金、上乗せを直接の支払い、とした。この二つを合わせて、書き手の収入を、**直接の支払い(使われ方に比例する投げ銭)**と、**集めて配るお金**の和として考える。全体の額を同じにして、集めて配るお金の割合と配り方を変えると、書き手が「生活できる額」(ここでは仮に、全員に均等に配ったときの額の 3 割)に届く人の割合は、次のようになる。

```python
total = 1.0
living = 0.3 * total / N
print('直接の支払い(使われ方に比例)と、集めて配るお金を合わせた収入で、「生活できる額」に届く書き手の割合')
print('  集めて配る割合    ' + ''.join(f'{k:>10s}' for k in ('比例', '平方根', '均等')))
for share in (0.0, 0.1, 0.3, 0.5):
    row = []
    for k in ('比例', '平方根', '均等'):
        inc = (1 - share) * total * use + share * total * rules[k]
        row.append(100 * (inc >= living).mean())
    print(f'      {100 * share:3.0f}%          ' + ''.join(f'{v:9.1f}%' for v in row))
```

直接の支払いだけ(集めて配る割合 0%)では、生活できる額に届く書き手は、上位の 3 分の 1 しかいない。投げ銭だけでは、多くの書き手は食べていけない。作者が、P2P ファイル共有の時代に「寄付モデルしかない」と思い、同時に「それでは経済が回らない」と感じたのは、この形のことだったと思われる。

集めて配るお金を足すと、届く人は増える。ただし、**比例で配る限り、まったく増えない**(集めたお金も、直接の支払いと同じ割合で分かれるので、全体の額が同じなら、一人一人の収入は変わらない)。均等に配れば、集めて配る割合が 3 割で、全員が届く。平方根は、そのあいだで、3 割で 6 割近く、5 割で 9 割近くの書き手が届く。2007 年の作者の見立て(基礎は集めて配るお金、上乗せは直接の支払い)が意味を持つのは、**集めたお金を、使われ方に比例させずに配るとき**だけである。

作者のペンションは、「過去の上乗せの倍率を、今期の倍率に追いつかせる」という形で、過去に一度でも投げ銭を受けた書き手に、基礎の給付を届ける仕組みだった(第 6 章)。それは、均等(BI)ではないが、比例でもない。作者は、2011 年に、自分一人の規模で、「著者の BI」の小さな模型を動かしていたことになる。

### 9.4 定期休眠ポイント(2010): 失効の代わりに保管料を

作者は、2010 年 2 月、はてなアイデアに「はてな定期休眠ポイント」を提案した。はてなポイントは、最終の利用日から 1 年を過ぎると無効になる規約だった。作者は、これを、はてなが金融機関にならないための規制だと理解していた。そこで、**保管の手数料を取る代わりに、解約の手続きをしなければ普通の支払いには使えないが、1 年を過ぎても無効にしない**ポイントを提案した。そのとき、作者は「かつてデフレに対応するための減価する貨幣制度が必要でないかという議論も思い出しました」と書いている。

作者は、失効の仕組みの問題も指摘した。

> 失効を望むということは、究極的には、ポイント所有者が DEAD になることを望むことでもある。

ポイントの発行者は、使われずに失効したポイントのぶんだけ得をする(業界では breakage と呼ぶ)。すると、発行者には、持ち主がポイントを忘れたり、亡くなったりすることを望む誘因が生まれる。保管料なら、発行者は、持ち主がポイントを持ち続けるほど得をするので、誘因の向きが逆になる。

小さな計算で比べる。持ち主 1 万人が、それぞれ 1000 ポイントを持ち、平均 2 年(指数分布)使わずに放置するとする。放置している間に亡くなる人もいる(年 2%)。1 年で失効する規則と、年 5% の保管料で失効しない規則を比べる。

```python
rng = np.random.default_rng(0)
n, B = 10000, 1000.0
idle = rng.exponential(2.0, n)                  # 使わずに放置する年数
death = rng.exponential(50.0, n)                # 亡くなるまでの年数(年 2%)
dies = death < idle
t_end = np.minimum(idle, death)
# 1 年で失効
lost_exp = np.where(t_end > 1.0, B, 0.0)
# 年 5% の保管料(失効しない)。亡くなった人の残りは相続される
fee = B * (1 - 0.95 ** t_end)
print('1 年で失効:      発行者の得(失効したポイント) %6.0f 万   持ち主が失った額の割合 %4.1f%%   亡くなった人の残高で発行者の得になったもの %5.0f 万'
      % (lost_exp.sum() / 1e4, 100 * lost_exp.sum() / (n * B), lost_exp[dies].sum() / 1e4))
print('年 5%% の保管料:  発行者の得(保管料)          %6.0f 万   持ち主が失った額の割合 %4.1f%%   亡くなった人の残高で発行者の得になったもの %5.0f 万'
      % (fee.sum() / 1e4, 100 * fee.sum() / (n * B), fee[dies].sum() / 1e4))
```

1 年で失効する規則では、放置した人の 6 割ほどがポイントを全部失い、それはそのまま発行者の得になる。亡くなった人の残高も、すべて発行者のものになる。保管料なら、持ち主が失うのは 1 割ほどで、亡くなった人の残高は相続される(発行者が得るのは、亡くなるまでの保管料だけ)。発行者が得る額は、失効のほうがずっと大きい。**失効の規則は、発行者に、持ち主の忘却と死から利益を得させる。** 作者の指摘は、この意味で正しい。

保管料を取るお金は、経済学では**減価する貨幣**(シルビオ・ゲゼルのスタンプ貨幣。持ち続けると、毎期少しずつ値打ちが減るお金)と呼ばれる。お金を貯め込まずに使わせる仕組みとして、1930 年代の不況の中で、オーストリアのヴェルグルなどで試された。作者の定期休眠ポイントは、それとは目的が逆で、**使わずに持ち続ける権利**を、手数料で買う仕組みである。

### 9.5 今の形

| 作者の構想 | 今の形 |
|---|---|
| 著者の BI(投げ銭だけでは食べていけない) | アイルランドの**芸術家のための基礎所得**(Basic Income for the Arts)。2022 年から、約 2000 人の芸術家に週 325 ユーロを 3 年間払う試行。2025 年 10 月、2026 年からの恒久化が発表された |
| 報酬請求権化(許諾がなくても、報酬を払えば使える) | 日本の**未管理著作物裁定制度**(2026 年 4 月から)。権利者の意思が確かめられない著作物を、補償金を払えば最長 3 年使える。EU の報道出版社の権利(2019 年、作者が 2018 年に論じた「リンク税」) |
| 補償金(間接税で基礎の収入) | 私的録音録画補償金(日本では 2022 年にブルーレイの録画機器も対象になった) |
| 配り方(比例か、平方根か) | 音楽配信の配り方の議論(全体の再生数で配る「プロラタ」と、聴き手ごとに配る「ユーザー中心」)。Spotify の、年 1000 回未満の曲に払わない規則(2024 年〜) |
| 失効の代わりに保管料(定期休眠ポイント) | 資金決済法の前払式支払手段の規制(有効期間 6 か月以内なら、発行の保証金の規制を受けない)。**休眠預金等活用法**(2018 年〜): 10 年以上動きのない預金を、発行者(銀行)の得にせず、民間の公益活動に使う |

最後の行は、作者の「失効を望むことは、持ち主の死を望むこと」への、一つの制度の答えになっている。休眠預金は、銀行のものにはならず、持ち主はあとからでも払い戻しを求められ、使われない間のお金は公益に回る。

### 9.6 この章のまとめ

- 作者の一連の構想の目的は、2007 年の見立て「**間接税(補償金)で基礎の収入を支え、薄く広い直接のパトロネージュで儲けを出す**」にある。これは、BI(基礎)と投げ銭(上乗せ)の形そのものである。
- 2011 年の分配は、**一人の読者が運営する、小さな補償金制度**として読める。匿名のポイントが基金、投げ銭が直接の支払い、ペンションが基礎の給付、ボーナスが上乗せである。
- 投げ銭だけでは、多くの書き手は食べていけない。集めて配るお金を足しても、**使われ方に比例して配る限り、まったく助けにならない**。作者の見立てが意味を持つのは、平方根や均等のように、比例させずに配るときである。
- 定期休眠ポイントの指摘(**失効は、発行者に持ち主の忘却と死から利益を得させる**)は、計算でも確かめられる。今の休眠預金の制度は、その一つの答えである。

## 第 10 章 今どう実現しているか、AI の時代にどう使うか

### 10.1 三つの層の、今の形

各章の終わりに、作者の構想と今の形の対応を書いてきた。まとめると、次のようになる。

| 層 | 作者の構想 | 今の形 | 実現の度合い |
|---|---|---|---|
| 証明 | 時間記録デバイス(2009)、Loaded Magic(2011)、Tweet Sig(2018) | 連鎖する時刻証明、証明書の透明性、Keybase の投稿による証明、セキュアエレメント、デジタルユーロのオフライン払いの設計 | **ほぼ描いた形で実現** |
| 経路 | PP ログイン・PP 振込(2006)、印栞(2011)、PP Authorizer(2017) | パスキー、FedCM、サイトごとの別の利用者番号、オープンバンキングの支払い指図 | **ほぼ実現**(ただし、問題が起きたときだけ身元を明かす手続きはまだない) |
| 価値 | 投げ銭予約(2006)、空スター(2009)、分配とペンション(2011)、No Ad URL(2013) | 投げ銭(note、pixivFANBOX、Nostr の zap)、ギフト記事、遡及的な公共財の資金、二次資金調達、芸術家のための基礎所得 | **人間どうしでは、定額の購読と投げ銭に落ち着いた**。一回ごとの支払いは、読み手が機械になったところで復活した |

実現の順番は、**証明 → 経路 → 価値**だった。価値の層は、ほかの二つの層の上にしか立たないからである。

### 10.2 なぜ当時は実現しなかったか

各章で見たことと、作者自身が当時書いていたことを合わせると、理由は五つにまとめられる。

1. **両面市場の立ち上がり**(第 4 章)。作者は 2006 年に「二社以上がやらねば意味がない」と書いた。利用者とサイトの両方が動かないと始まらない仕組みは、個人には動かせない。
2. **規制**。ポイントの失効の規則(第 9 章)や、前払式支払手段・資金移動の規制。作者は 2026 年に「マイクロペイメントは構想は流行ったんだけど、資金決済法とか規制が強力でとても難しかった」と書いている。
3. **心の取引費用**。一回ごとに「払うか」を決める負担は、金額が小さくても消えない(Szabo 1999、Odlyzko 2003)。人間相手では、定額の購読やまとめ買いが勝った。
4. **証明の問題**(第 1・2 章)。予約が本気か、スターを付けた人が読んだかは、開示や記録だけでは確かめられない。偽りに強いのは、証拠金(本貨の裏付け)と、評判(長い関係)だった。
5. **匿名と責任の両立**(第 4 章)。誰が仲立ちするかが決まらないと、匿名の支払いは、資金洗浄の懸念にぶつかる。

### 10.3 読み手が機械になった

人間向けの一回ごとの支払いは、続かなかった。ウェブの閲覧に応じて少額を書き手に流す Coil(Web Monetization)は、2023 年 3 月にサービスを終えた。お金を払って広告を消す Google Contributor(2014〜2017)と Scroll(2019〜2021)も、単独では続かなかった。

ところが、2025 年、ウェブの最初の仕様から「将来のために予約」とされていたまま使われていなかった HTTP の状態コード **402(Payment Required)** が、二つの形で復活した。

- **Cloudflare の Pay Per Crawl**(2025 年 7 月): AI のクローラーがページを読みに来ると、サイトが一回の読み取りごとの値段を付け、払う意思を示さないクローラーには 402 を返す。Cloudflare が支払いの仲立ちをする。
- **x402**(Coinbase が始めた規格): AI のエージェントが、ウェブの要求一回ごとに、少額を払って結果を受け取る。

読み手が機械になると、3 の「心の取引費用」がなくなる。機械は、一回ごとに払うかどうかを迷わない。作者が 2006 年に描いた「読むたびに少し払う」世界は、**人間の読み手のためではなく、機械の読み手のために**実現しつつある。

そして、お金の向きについても、作者の空スター(読んでもらう側が払う)に近いことが起きている。書き手は、AI の回答に自分の文書が引かれることを望み、そのための工夫(検索の最適化に倣って「生成エンジンの最適化」と呼ばれる)に手間とお金をかけはじめた。**AI に読んでもらうために払う**、という、空スターの逆向きの支払いである。

### 10.4 AI と寄付モデルを統合する

作者は、2026 年 1 月、こう書いた。

> AIには情報の「国境」があるべきだとは思う。ただ、本質的に自由にコピーができる情報になぜお金を払うのか。AIについてもそこが問題になる。案外寄付モデルがここでは適合するのかもしれない。AIならば自分がなぜそう判断したかの根拠も後付けでも網羅的に示すことができる。

そして、「働かずにブログを書いてきた私が有利なのは、ちょっと利益相反な主張だから気が引ける」とも書いている。

私は、**2011 年の分配のプログラムが、そのまま「AI の利用料を、引用された書き手に配る仕組み」の雛形になる**と考える。

| 2011 年の分配 | AI の場合 |
|---|---|
| 基金に入る匿名のポイント | AI の利用料のうち、書き手に回す分 |
| 今期送る基礎ポイント | その期に、AI の回答の根拠として引かれた回数 |
| ボーナス(平方根) | 引かれたことへの上乗せ。平方根で、上位への集中を和らげる |
| ペンション(上限つきの追いつき) | 過去に引かれた(学習や検索に使われた)書き手への、基礎の給付 |
| 1.05 乗の上限、ID 統合の待ち期間、登録の手数料 | 文書や ID を分けて稼ぐことへの備え |
| BI 否認時の返し方 | 配分が取り消されたとき(権利の争い、不正の発覚)の返し方 |

これは、第 9 章の「間接税で基礎、直接のパトロネージュで儲け」の、AI の時代の形である。作者が気にした利益相反(多く書いてきた人が有利になる)には、ペンションの上限(過去の拠出の再評価にとどめ、今の水準を超えさせない)が、そのまま歯止めとして設計されている。

### 10.5 模型

書き手 500 人が、AI の回答の根拠として、毎期 2000 回引かれる。引かれやすさは、ジップの法則(順位の 1.1 乗に反比例)に従う。3 割の書き手は、途中から書き始める。そのうち 10 人は、文書を 8 個に分けて、別々の ID として引かれる(引かれる総数は同じ)。毎期の利用料のうち、書き手に回す分を、次の三通りで配る。

- **比例**: 引かれた回数に比例して配る。
- **平方根**: 引かれた回数の平方根に比例して配る。
- **作者の規則**: 半分を引かれた回数に比例して直接配り、残り半分を基金に入れて、2011 年の分配の計算(`bonus.distribute`、第 5 章で作者のプログラムと一致させたもの)で、ボーナスとペンションとして配る。はてなの手数料はなしとする。

配る額の単位が粗いと、作者の規則の「ボーナスを 10 の倍数に切り捨てる」が強く効く。そこで、一期の額を 30 万(単位が十分に細かい場合)と 3000(粗い場合)の二通りで見る。

```python
%%writefile aidist.py
# AI の利用料を、引用された書き手に配る模型。配り方は、2011 年の分配のプログラム(bonus.distribute)をそのまま使う。
#   基金       = AI の利用料のうち、書き手に回す分(毎期)
#   基礎ポイント = その期に、AI の回答の根拠として引かれた回数(引用 1 回 = 1)
#   ボーナス   = 引用に、平方根で上乗せ
#   ペンション = 過去に引かれた(学習や検索に使われた)書き手への、追いつきの給付
# 書き手には、ふつうの書き手と、文書を細かく分けて(ID や文書を増やして)引用を稼ぐ書き手がいる。
import numpy as np
import bonus


def simulate(T=40, n_writers=500, queries=2000, fund_per_q=3000.0, zipf=1.1, n_split=10, split_k=8,
             rule='author', direct_share=0.5, fee=5, life=None, seed=0):
    """rule:
      'author'   : 直接の配分(引用に比例、direct_share)+ 残りを作者の規則(ボーナス + ペンション)
      'prorata'  : 全部を引用に比例して配る
      'sqrt'     : 全部を引用の平方根に比例して配る(ペンションなし)
    life: 書き手が引かれ続ける期数(None なら終わりなし)。過ぎると引かれなくなり、ペンションだけの対象になる。
    n_split 人の書き手は、文書を split_k 個に分け、それぞれが別の ID として引かれる(引かれる総数は同じ)。
    返す値: 書き手ごとの受け取りの合計(分けた書き手は ID をまとめて)、分けた書き手の取り分、ジニ係数、新しい書き手の取り分。
    """
    rng = np.random.default_rng(seed)
    quality = 1 / np.arange(1, n_writers + 1) ** zipf                     # 引かれやすさ(ジップ)
    born = np.where(rng.random(n_writers) < 0.7, 0, rng.integers(1, T, n_writers))   # 3 割は途中から書き始める
    if life is not None:
        born = rng.integers(0, T - life, n_writers)                         # 引かれ続ける期間が限られるなら、入る時期をばらす
    splitter = np.zeros(n_writers, bool)
    splitter[rng.choice(n_writers, n_split, replace=False)] = True
    owner = {}
    for w in range(n_writers):
        ids = [f'w{w:04d}s{j}' for j in range(split_k)] if splitter[w] else [f'w{w:04d}']
        for i in ids:
            owner[i] = w
    A, C, recv = {}, {}, np.zeros(n_writers)
    fund = 0.0
    for t in range(T):
        live = (born <= t) if life is None else ((born <= t) & (t - born < life))
        if not live.any():
            continue
        p = quality * live; p = p / p.sum()
        cites = rng.multinomial(queries, p)
        BASE = {}
        for w in np.flatnonzero(cites):
            if splitter[w]:
                parts = rng.multinomial(cites[w], np.ones(split_k) / split_k)
                for j, c in enumerate(parts):
                    if c:
                        BASE[f'w{w:04d}s{j}'] = int(c)
            else:
                BASE[f'w{w:04d}'] = int(cites[w])
        pool = fund_per_q
        if rule == 'prorata':
            tot = sum(BASE.values())
            for i, c in BASE.items():
                recv[owner[i]] += pool * c / tot
        elif rule == 'sqrt':
            tot = sum(np.sqrt(c) for c in BASE.values())
            for i, c in BASE.items():
                recv[owner[i]] += pool * np.sqrt(c) / tot
        else:
            tot = sum(BASE.values())
            for i, c in BASE.items():                                       # 直接の配分(引用に比例)
                recv[owner[i]] += direct_share * pool * c / tot
            fund += (1 - direct_share) * pool
            # 作者の規則: 基礎ポイント = 引用の回数、引退という区別はなく、過去に引かれた全員が登録している
            r = bonus.distribute(BASE, dict(A), dict(C), {i: A[i] for i in A}, fund, (), tax=0.0, fee=fee)
            for i, b in r['bonus'].items():
                recv[owner[i]] += b; C[i] = C.get(i, 0) + b
            for i, pp in r['pension'].items():
                recv[owner[i]] += pp - fee; C[i] = C.get(i, 0) + pp
            fund = r['REST']
        for i, c in BASE.items():
            A[i] = A.get(i, 0) + c; C.setdefault(i, 0)
    total = recv.sum()
    late = born > 0
    return dict(recv=recv, split_share=recv[splitter].sum() / total, cite_share=quality[splitter].sum() / quality.sum(),
                gini=_gini(recv), late_share=recv[late].sum() / total, late_frac=late.mean(),
                top1=np.sort(recv)[::-1][:n_writers // 100].sum() / total)


def _gini(x):
    x = np.sort(x); n = len(x)
    return float((2 * np.arange(1, n + 1) - n - 1) @ x / (n * x.sum()))


def summary(seeds=range(4), **kw):
    rs = [simulate(seed=s, **kw) for s in seeds]
    return {k: float(np.mean([r[k] for r in rs])) for k in ('split_share', 'cite_share', 'gini', 'late_share', 'top1')}
```

```python
import aidist
rows = (('比例', dict(rule='prorata')), ('平方根', dict(rule='sqrt')), ('作者の規則', dict(rule='author')))
for F, lab in ((3e5, '単位が細かい(一期 30 万)'), (3000, '単位が粗い(一期 3000)')):
    print(f'{lab}、40 期、4 シードの平均')
    print('  配り方        ジニ係数   上位 1% の取り分   途中から書き始めた人の取り分   文書を分けた人の得(分けないときを 1)')
    for name, kw in rows:
        a = aidist.summary(split_k=1, fund_per_q=F, **kw); b = aidist.summary(split_k=8, fund_per_q=F, **kw)
        print(f'  {name:10s}     {a["gini"]:.2f}          {100 * a["top1"]:5.1f}%                   {100 * a["late_share"]:5.1f}%                          {b["split_share"] / a["split_share"]:.2f}')
    print()
```

単位が十分に細かいとき(上の表):

- **作者の規則は、比例と平方根のあいだ**に分配を置く。上位 1% の取り分は、比例の 4 割強から 3 割ほどに、ジニ係数は 0.80 から 0.66 に下がる。半分を比例で直接配っているので、平方根(0.50)ほどは下がらない。
- 途中から書き始めた人の取り分は、どの配り方でも 16〜17% ほどで、大きくは変わらない。
- **文書を分けると、平方根では 1.5 倍、作者の規則でも 1.3 倍ほど得をする。** 第 6 章では、ID を分けた得は、ペンションの上限で取り返され、10 年ほど待たせればほぼ中立になった。ここで取り返されないのは、書き手が引かれ続けるからである。作者の規則の取り返しは、**ペンションの上限**で働く。引かれ続ける(ボーナスを受け続ける)書き手には、ペンションの出番がなく、取り返しが起きない。

単位が粗いとき(下の表)は、様子が変わる。作者の規則では、少ししか引かれない書き手のボーナスが 10 に届かず 0 になり、分配は比例より偏る(ジニ係数 0.81)。そして、文書を分けると、**かえって損をする**(0.55 倍)。分けた小さな文書は、それぞれ 10 に届かなくなるからである。切り捨ては、ID を分けることへの、意図しなかった歯止めになっている。ただし、その代償は、少ししか引かれない正直な書き手が払う。

```python
print('書き手が引かれ続ける期間を 20 期に限る(引かれなくなった書き手は、ペンションだけの対象)。120 期、単位は細かい、直接の配分なし')
print('  配り方        ジニ係数   文書を分けた人の得')
for name, kw in rows:
    kk = dict(fund_per_q=3e5, T=120, life=20, direct_share=0.0, **kw)
    a = aidist.summary(split_k=1, **kk); b = aidist.summary(split_k=8, **kk)
    print(f'  {name:10s}     {a["gini"]:.2f}        {b["split_share"] / a["split_share"]:.2f}')
```

引かれ続ける期間を限ると(書き手が「引退」すると)、作者の規則は、平方根よりも分けた得を小さくする(上の出力)。ペンションの上限が働きはじめるからである。しかし、第 6 章ほどには打ち消さない。AI の引用は、量が多く、毎期の配分が大きいので、ボーナスで先に得たぶんを、ペンションで取り返しきれない。

### 10.5.1 どれくらいの規模で、経済として回ったか

作者の制度が「経済として回る」(書き手がそれで食べていける)には、どれくらいの規模が要ったか(JRF の 2026 年 10 月の問い)。大まかに見積もる。

書き手の生活費を月 15 万円、払う読者が一人あたり月 500 円を書き手たちに回すとする(はてなポイントは 1 ポイント 1 円)。運営の取り分は 0 とする。すると、**書き手全員を支えるには、書き手一人あたり、払う読者が少なくとも 15 万 / 500 = 300 人要る**。これは、全員に均等に配った場合である。使われ方に偏りがあると、上位の書き手は少ない読者で届くが、多くの書き手が届くには、もっと要る。第 9 章の配り方で計算する。

```python
L, p = 150000, 500                     # 月の生活費、読者一人が月に書き手たちに回す額(円)
print('書き手の 1 割 / 半分が、生活費に届くのに要る「払う読者」の数(書き手一人あたり)。使われ方はジップの法則(1.1)')
print('  書き手の数   配り方                 1 割が届く    半分が届く')
for N in (100, 1000, 10000):
    r = np.arange(1, N + 1); u = 1 / r ** 1.1; u /= u.sum()
    sq = np.sqrt(u) / np.sqrt(u).sum(); eq = np.full(N, 1 / N)
    for lab, inc in (('直接の支払いだけ', u), ('3 割を平方根で配る', 0.7 * u + 0.3 * sq),
                     ('3 割を均等に配る', 0.7 * u + 0.3 * eq), ('全部を均等に配る', eq)):
        srt = np.sort(inc)[::-1]
        print(f'  {N:8d}     {lab:18s}  {L / (p * srt[N // 10 - 1]) / N:8.0f} 人   {L / (p * srt[N // 2 - 1]) / N:8.0f} 人')
```

書き手 1000 人の規模で、直接の支払い(投げ銭)だけだと、半分の書き手が食べていけるには、書き手一人あたり 1500 人ほど、合わせて 150 万人ほどの「払う読者」が要る。集めたお金の 3 割を平方根で配ると、それが 900 人ほど(合わせて 90 万人ほど)に減る。均等に配れば 300 人まで下がる。書き手が増えるほど、偏りのせいで、一人あたりに要る読者は増える。

ケヴィン・ケリーは 2008 年に、作り手は「1000 人の本当のファン」(一人が年に 100 ドルを払う)がいれば食べていける、と書いた。ここでの「300 人が月 500 円」は、それと同じ桁である。問題は、ファンが均等に分かれてはくれないことで、上の表の「直接の支払いだけ」の行が、その現実にあたる。

これと比べると、作者の 2011 年の基金は、規模が桁違いに小さかった。作者は受け取った額を書いていないが、分配の下限が 1000 ポイント(1000 円)で、それを超える受け取りが「(ポイント授受としては)高額」だったことから、四半期に数千ポイントほどの規模だったと思われる。書き手一人を一四半期支える額(45 万円)の 100 分の 1 ほど、書き手 1000 人の半分を支える規模(四半期に 13 億円ほど)の、数十万分の 1 である。**作者が試したのは、経済そのものではなく、経済が回る規模になったときに働くはずの規則の形だった**。そして、その形(平方根の上乗せ、過去の拠出の再評価、あとで取り返すシビル対策)は、規模を変えても意味を持つ(第 6・10 章)。ただし、10.5 節で見たとおり、切り捨ての単位だけは、規模に合わせて移し直す必要がある。

もう一つ、規模とは別の注意がある。作者の制度の中で、ポイントは**どこかから入ってこなければならない**。書き手が受け取ったポイントで、また別の書き手に投げ銭すれば、ポイントは制度の中を回る。しかし、はてなの手数料(5%)がかかるので、一つのポイントが回れるのは、平均で 1 / 0.05 = 20 回ほどまでである。そして、回ることで新しい値打ちが生まれるわけではない。書き手が食べていくには、制度の外の所得(読者の給料や、AI の利用料)から、毎期お金が入ってくる必要がある。作者の匿名の寄付は、その「外からの流れ」の、いちばん小さな見本だった。

### 10.6 提言

ここまでの章の結果から、作者の構想を、これからどう使えるかを提言としてまとめる。

1. **配り方は、比例にしない。** 第 9 章で見たとおり、集めたお金を使われ方に比例して配る限り、書き手の基礎の収入にはならない。平方根や、作者のペンションのような、過去の拠出を今の水準に追いつかせる給付を組み合わせる。AI の利用料の配分(引用への配分)でも同じである。
2. **ID や文書を分けることへの備えは、価値の層だけでは足りない。** 作者の「あとで取り返す」は、引退のある世界(第 6 章)では働いたが、引かれ続ける世界(10.5 節)では弱い。誰がどの文書を書いたか、いつ書いたかの証明(第 3 章の証明の層: 時刻証明、来歴の証明)と、書き手が誰かの保証(第 4 章の経路の層: サイトには匿名、仲立ちする第三者には身元)が要る。**三つの層は、一つの制度として組み合わせて初めて働く。**
3. **偽りに強いのは、本貨の裏付けと、長い関係。** 第 1 章(証拠金)と第 2 章(評判、師弟)の結果は、AI の時代にも変わらない。AI が感想や要約を書ける今、「読んだ」ことの証明は、かえって難しくなった。証明できるのは「応じた」ことまでだと割り切り、続く関係に用途を絞る。
4. **単位の細かさは、設計の一部である。** 作者の規則の切り捨て(10 の倍数)は、基金が乏しいときにペンションを消し(第 6 章)、単位が粗いと少ししか引かれない書き手を締め出す一方で、ID を細かく分けることへの歯止めにもなった(10.5 節)。規則を別の規模に移すときは、切り捨ての単位も移し直す必要がある。
5. **失効ではなく、保管と公益への還流を。** ポイントや配分の受け取り手が見つからないとき、それを運営の得にしない(第 9 章の定期休眠ポイント、休眠預金)。
6. **マイナス金利では、貸倒れの判断が働かない**ことを、設計に入れておく(第 7・8 章)。配分を担保にした貸付を認めるなら、上限を利率ではなく、配分の残高と不確かさで決める。

### 10.7 この章のまとめ

- 作者の三つの層のうち、**証明と経路の層は、ほぼ描いた形で実現した**。価値の層は、人間どうしでは定額と投げ銭に落ち着き、一回ごとの支払いは、**読み手が機械になったところで**(HTTP 402 の復活)実現しつつある。
- 当時実現しなかった理由は、両面市場、規制、心の取引費用、証明、匿名と責任の両立の五つである。作者の構想の中身の誤りではない。
- **2011 年の分配のプログラムは、AI の利用料を引用された書き手に配る仕組みの雛形になる**。模型では、比例で配るより分配の偏りを小さくする。ただし、文書を分けて稼ぐことへの備えは、引かれ続ける世界では弱く、証明と経路の層の助けが要る。

## 第 11 章 まとめ

### 11.1 問いへの答え

第 0 章の六つの問いに、順に答える。

**1. 投げ銭予約は、どういう通貨の制度か。**
予約を紙幣、投げ銭を本貨とする、**部分準備の通貨**である。作者は、紙幣の信用を、送ポイント総額・予約総額・兌換の猶予の開示(今の言葉で準備の証明)で保とうとした。作者が「要証明」と書いた信頼性フィルタは、**証明できない**。受けられないページへの予約は開示のどこにも現れないので、受けられるページで少し払っておけば、最も信頼できる利用者に見える(第 1 章)。偽りに強いのは、情報ではなく、**本貨の裏付け(予約証拠金)**だった。出せる予約は、証拠金 / 率までに限られる。銀行の信用乗数と同じである。

**2. 「読んでもらう側が払う」支払いは、成り立つか。**
空スターは、名宛人・条件・期限・証拠金を持つ**条件付きの支払い**(エスクロー、ハッシュ時間ロック契約)であり、No Ad URL の linker の連なりは**手形の裏書き**である。確かめられるのは「スターを付けた」までで、「読んだ」ではない。作者の見込み「読まない読み手は、だんだん指定されなくなる」は、**評判が私的なら半分しか当たらず、公開すれば当たる**。早い者勝ちは、読まない人を呼び込む(第 2 章)。証明できない条件は、関係の長さで補うしかない。作者が最初に挙げた師弟という用途は、その点で筋がよかった。

**3. 文書がいつ書かれたかの証明は、どう作れるか。**
作者の時間記録デバイス、Loaded Magic、Tweet Sig は、今の連鎖する時刻証明、証明書の透明性、Keybase の投稿による証明に、それぞれ対応する(第 3 章)。時間記録デバイスは、二重の譲渡を見抜けるが、**デバイスの複製は防げない**。P2P の電子マネーには、複製できないハードウェアか、全員で一つの記録を持つこと(Bitcoin)が要る。作者は、Bitcoin の論文の翌年(2009 年)に、前者の道を考えていた。少数のランダムな監査は、大規模な不正を抑える(Loaded Magic の牽制の論理)。

**4. 分配の規則は、ID を分けたりまとめたりする誘因を、狙いどおり抑えられるか。**
抑えられる(第 6 章)。ID を分けるとボーナスは増えるが、ペンションの上限が受け取った上乗せの累計に効くので、長い目で見て取り返される。統合を 10 年ほど待たせれば、分けても分けなくても、生涯の受け取りはほぼ同じになる。作者の「ほどほどの ID の分散と統合」は、この模型では実現している。そして、ペンションは、**過去の書き手の上乗せの倍率を、今期の倍率に追いつかせる**仕組み(過去の拠出の再評価)だった。ただし、その上限は今期の倍率の**最小**で決まるので、一人に大きく送る期には、過去の書き手全員のペンションが止まる。また、ボーナスの切り捨てと組み合わさると、基金が乏しいときにペンションが消える。作者の「原資 1000 未満は繰り越す」規則が、それを防いでいた。

**5. BI が否認されたときの返し方は、何を保証していたか。マイナス金利で何が壊れるか。**
作者の原則(元の債権者が受け取れないものは、新しい債権者も受け取れない)は守られていて、新しい債権者の負担の合計は、**利率一期分(R·X2)を決して超えない**(第 7 章)。作者の閉じた式には誤りがあり、ありもしない「発散」を見せていた。本当に延びるのは、返済の期間である。BI を担保にした貸付の上限 n < 1/r₁ は**自然な借入制約**で、複利では ln 2 / ln(1 + r) に縮む(第 8 章)。そして、**R ≤ 0 では、作者のプログラムはゼロでの割り算で止まり、貸倒れの判定は決して成り立たない**。借り続けることが、ただで使い続けることになる。ポンジ・ゲームの禁止が r ≤ g で破れることの、小さな制度の中での現れである。

**6. 書き手は、どう食べていけばよいのか。**
作者の答えは、2007 年の「**間接税(補償金)で基礎、直接のパトロネージュで儲け**」だった(第 9 章)。2011 年の分配は、それを一人の規模で回した、小さな補償金制度である。ただし、集めたお金を**使われ方に比例して配る限り、基礎の収入にはならない**。平方根や、作者のペンションのような配り方が要る。2011 年の分配のプログラムは、**AI の利用料を、引用された書き手に配る仕組み**の雛形になる(第 10 章)。比例で配るより偏りを小さくするが、文書を分けて稼ぐことへの備えは、引かれ続ける世界では弱く、証明の層と経路の層の助けが要る。

### 11.2 AI 以前の作者のモデルとして

このシリーズは、AI 以前の作者のモデルを、今の AI がどう読むかを主題にしている。この文書の材料について、作者の独創と、拙さを並べておく。

独創と言えるもの:
- **読んでもらう側が払う**、名宛ての条件付きの支払い(空スター、2009)。今も、人間どうしでは広まっていない。
- **ペンションを、過去の拠出の再評価として設計した**こと。今期の水準に追いつかせるという考え方は、公的年金の再評価と同じで、遡及的な公共財の資金(2021 年〜)に先立つ。
- **シビル攻撃を、本人確認ではなく、あとで取り返すことで抑える**設計(1.05 乗の上限と、統合の待ち期間)。
- **暗号やハッシュを信用できなくても働く目印**(ランダムことわざ、2012)。本来読んでいるはずのない人が、それとなく漏洩を知らせられる。カナリアの罠や令状のカナリアと同じ問題を、社会の側の経路で解こうとした。
- 価値・証明・経路の三つの層を、**それぞれ独立に**考え、実際に小さなものを作って動かしたこと(Perl のプログラム、Greasemonkey のスクリプト、ブラウザのアドオン)。

拙さと言えるもの:
- 閉じた式の誤り(第 7 章)、証明されていない信頼性フィルタ(第 1 章)。どちらも、作者自身が「要証明」「専門家から見れば、つっこみどころはいろいろある」と書いていた。
- 両面市場の立ち上がりを、個人が動かそうとしたこと(第 4 章)。作者自身、それが難しいことを書いていた。
- 規則のあいだの、意図していなかった結び付き(最大の受け手が年金を決める、切り捨ての罠。第 6 章)。

作者の拙さは、多くの場合、作者自身が気づいて書き添えていた。そして、作者の規則の細部(1000 の下限など)は、意図していなかったところでも、制度を守っていた。

### 11.3 残る問い

- **AI の時代の「読んだ」の証明**。AI が感想や要約を書ける今、空スターの条件は、何にすればよいか。

### A.2 実行環境

```python
import platform
print('Python', sys.version.split()[0], '/', platform.platform())
print('numpy', np.__version__, '/ matplotlib', matplotlib.__version__)
print('perl', subprocess.run(['perl', '-e', 'print $^V'], capture_output=True, text=True).stdout)
print('CPAN のモジュールの代わりを使ったもの:', ho.prepare_shim()[1] or 'なし')
for name, h in ho.EXPECTED.items():
    print(f'  {name:40s} sha256 {h[:16]}...')
```

作者の分配のプログラムは、ID の順番を Perl のハッシュの順番で決めるので、`PERL_HASH_SEED=0` と `PERL_PERTURB_KEYS=0` で固定して走らせている。Python の書き直しとの一致(第 5 章)は、この条件で確かめた。順番が変わると、√(基礎ポイント) の和の最後の桁が変わり、ボーナスが 10 の倍数の境目にぴったり来たときだけ、結果が変わりうる。

作者の BI 否認時のプログラムは、入力を取らない。第 7.7 節では、定数 `$R` の一行だけを書き換えた写しを、一時ファイルに作って走らせている。作者のファイルそのものは書き換えていない。
